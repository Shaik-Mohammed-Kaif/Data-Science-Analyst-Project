## Total Profit

```sql
SELECT
    SUM(Profit) AS Total_Profit
FROM sales;
```

## Total Quantity

```sql
SELECT
    SUM(Quantity) AS Total_Quantity
FROM sales;
```

## Profit Margin

```sql
SELECT
    ROUND(
        SUM(Profit) / NULLIF(SUM(Sales), 0) * 100,
        2
    ) AS Profit_Margin_Percentage
FROM sales;
```

# 📐 DAX Analytics

A dedicated Measures table is created for reusable business calculations.

## Total Sales

```DAX
Total Sales =
SUM(Sales[Sales])
```

## Total Profit

```DAX
Total Profit =
SUM(Sales[Profit])
```

## Total Quantity

```DAX
Total Quantity =
SUM(Sales[Quantity])
```

## Profit Margin

```DAX
Profit Margin % =
DIVIDE(
    [Total Profit],
    [Total Sales],
    0
)
```

## Average Order Value

```DAX
Average Order Value =
DIVIDE(
    [Total Sales],
    DISTINCTCOUNT(Sales[Order ID]),
    0
)
```

---

# 📅 Time Intelligence

A dedicated Date table is used for reliable time-based analysis.

```text
DimDate
│
├── Date
├── Year
├── Quarter
├── Month
├── Month Number
├── Week
└── Day
```

### Year-over-Year Sales

```DAX
Sales LY =
CALCULATE(
    [Total Sales],
    SAMEPERIODLASTYEAR(
        DimDate[Date]
    )
)
```

### YoY Growth

```DAX
Sales YoY % =
DIVIDE(
    [Total Sales] - [Sales LY],
    [Sales LY],
    0
)
```
