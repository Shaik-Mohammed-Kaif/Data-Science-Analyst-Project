# 📊 Plotly Business Insights

## 01. Data Loading & Library Setup

This notebook starts by importing the required Python libraries
and loading the prepared SuperStore dataset.

The dataset will be used for interactive business visualization
and insight generation using Plotly.

````python
# Import required libraries

import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go

from IPython.display import display
````

````python
# Load the prepared SuperStore dataset

df = pd.read_csv("SuperStore_Feature_Engineered.csv")

display(df.head())
````

````python
# Check dataset dimensions and columns

print("Dataset Shape:", df.shape)
print("\nColumns:")
display(df.columns)
````

````python
# Check data types

display(df.dtypes)
````

````python
display(df.describe().T)
````

# 01. Total Sales KPI

### Business Question

What is the total revenue generated by the business?

This KPI provides the primary revenue indicator for the
SuperStore business.

````python
# Calculate total Sales from the transaction-level Sales column

total_sales = df["Sales"].sum()

print(f"Total Sales: ${total_sales:,.2f}")

# Create an interactive KPI indicator for Total Sales

fig = go.Figure(
    go.Indicator(
        mode="number",
        value=total_sales,
        number={
            "prefix": "$",
            "valueformat": ",.2f"
        },
        title={
            "text": "Total Sales"
        }
    )
)

fig.update_layout(
    title="Executive KPI — Total Sales",
    height=400
)

fig.show()
````

### Business Insight

Total Sales represents the overall revenue generated from all
recorded transactions and serves as the primary financial KPI for
the remaining business analysis.

# 02. Total Profit KPI

### Business Question

What is the total Profit generated by the business?

Total Profit measures the financial value retained after the
cost-related impact represented in the dataset.

````python
# Calculate total Profit from the original transaction-level Profit column

total_profit = df["Profit"].sum()

print(f"Total Profit: ${total_profit:,.2f}")

# Create an interactive KPI indicator for Total Profit

fig = go.Figure(
    go.Indicator(
        mode="number",
        value=total_profit,
        number={
            "prefix": "$",
            "valueformat": ",.2f"
        },
        title={
            "text": "Total Profit"
        }
    )
)

fig.update_layout(
    title="Executive KPI — Total Profit",
    height=400
)

fig.show()
````

### Business Insight

Total Profit provides a direct measure of overall business
profitability and should be evaluated together with Sales to
understand the quality of revenue generation.

# 03. Sales by Category

### Business Question

Which product Category generates the highest Sales?

Category-level aggregation is performed using the original
transaction-level Sales column.

````python
# Aggregate Sales by Category

category_sales = (
    df.groupby("Category", as_index=False)
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Total_Quantity=("Quantity", "sum"),
          Total_Orders=("Order_ID", "nunique")
      )
      .sort_values("Total_Sales", ascending=False)
)

display(category_sales)


# Create an interactive premium bar chart for Category Sales

fig = px.bar(
    category_sales,
    x="Category",
    y="Total_Sales",
    text="Total_Sales",
    color="Category",
    color_discrete_sequence=[
        "#6366F1",
        "#06B6D4",
        "#F59E0B",
        "#EC4899",
        "#10B981"
    ],
    hover_data={
        "Total_Sales": ":,.2f",
        "Total_Profit": ":,.2f",
        "Total_Quantity": ":,.0f",
        "Total_Orders": ":,.0f"
    },
    title="Sales Performance by Category",
    labels={
        "Total_Sales": "Total Sales",
        "Category": "Product Category"
    }
)


# Format value annotations

fig.update_traces(
    texttemplate="$%{text:,.0f}",
    textposition="outside",
    marker_line_width=1.5,
    marker_line_color="#E5E7EB",
    hovertemplate=(
        "<b>%{x}</b><br>"
        "Total Sales: $%{y:,.2f}<br>"
        "Total Profit: $%{customdata[1]:,.2f}<br>"
        "Quantity: %{customdata[2]:,.0f}<br>"
        "Orders: %{customdata[3]:,.0f}"
        "<extra></extra>"
    )
)


# Apply premium dashboard layout

fig.update_layout(
    title={
        "text": "📊 Sales Performance by Category",
        "x": 0.5,
        "xanchor": "center",
        "font": {
            "size": 24,
            "family": "Arial",
            "color": "#F8FAFC"
        }
    },

    paper_bgcolor="#0B1120",
    plot_bgcolor="#111827",

    font={
        "family": "Arial",
        "color": "#E5E7EB"
    },

    height=650,

    margin={
        "l": 80,
        "r": 60,
        "t": 100,
        "b": 80
    },

    xaxis={
        "title": "Product Category",
        "showgrid": False,
        "showline": True,
        "linecolor": "#374151",
        "tickfont": {
            "size": 13,
            "color": "#D1D5DB"
        },
        "title_font": {
            "size": 14,
            "color": "#F8FAFC"
        }
    },

    yaxis={
        "title": "Total Sales",
        "tickprefix": "$",
        "tickformat": ",",
        "showgrid": True,
        "gridcolor": "#1F2937",
        "gridwidth": 1,
        "showline": False,
        "tickfont": {
            "size": 12,
            "color": "#D1D5DB"
        },
        "title_font": {
            "size": 14,
            "color": "#F8FAFC"
        }
    },

    hoverlabel={
        "bgcolor": "#020617",
        "font_size": 13,
        "font_family": "Arial",
        "font_color": "#F8FAFC",
        "bordercolor": "#6366F1"
    },

    legend={
        "title": {
            "text": "Category"
        },
        "font": {
            "color": "#E5E7EB"
        },
        "bgcolor": "rgba(0,0,0,0)"
    }
)


# Add a subtle annotation for the highest Sales category

top_category = category_sales.iloc[0]

fig.add_annotation(
    x=top_category["Category"],
    y=top_category["Total_Sales"],
    text=(
        f"🏆 Top Category<br>"
        f"${top_category['Total_Sales']:,.0f}"
    ),
    showarrow=True,
    arrowhead=2,
    arrowsize=1,
    arrowwidth=2,
    arrowcolor="#F59E0B",
    ax=0,
    ay=-65,
    bgcolor="#020617",
    bordercolor="#F59E0B",
    borderwidth=1,
    borderpad=6,
    font={
        "size": 12,
        "color": "#F8FAFC"
    }
)


# Display the final interactive visualization

fig.show()
````

### Business Insight

The highest-selling Category represents the strongest revenue
contributor. However, Sales should be compared with Profit and
Profit Margin before determining the most valuable Category.

# 04. Profit by Category

### Business Question

Which Category generates the highest Profit?

This analysis compares Category-level profitability using
transaction-level Profit.

````python
# Aggregate Sales and Profit by Category

category_profit = (
    df.groupby("Category", as_index=False)
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Total_Quantity=("Quantity", "sum")
      )
)

# Calculate Profit Margin at the aggregated Category level

category_profit["Profit_Margin_%"] = (
    category_profit["Total_Profit"]
    / category_profit["Total_Sales"]
    * 100
)

# Sort Categories by Profit

category_profit = category_profit.sort_values(
    "Total_Profit",
    ascending=False
)

display(category_profit.round(2))


# Create an interactive premium Profit comparison chart

fig = px.bar(
    category_profit,
    x="Category",
    y="Total_Profit",
    text="Total_Profit",
    color="Category",
    color_discrete_sequence=[
        "#6366F1",
        "#06B6D4",
        "#F59E0B",
        "#EC4899",
        "#10B981"
    ],
    hover_data={
        "Total_Sales": ":,.2f",
        "Total_Profit": ":,.2f",
        "Total_Quantity": ":,.0f",
        "Profit_Margin_%": ":.2f"
    },
    title="Profit Performance by Category",
    labels={
        "Total_Profit": "Total Profit",
        "Category": "Product Category"
    }
)


# Format Profit annotations and hover information

fig.update_traces(
    texttemplate="$%{text:,.0f}",
    textposition="outside",
    marker_line_width=1.5,
    marker_line_color="#E5E7EB",
    hovertemplate=(
        "<b>%{x}</b><br>"
        "Total Profit: $%{y:,.2f}<br>"
        "Total Sales: $%{customdata[0]:,.2f}<br>"
        "Quantity: %{customdata[2]:,.0f}<br>"
        "Profit Margin: %{customdata[3]:.2f}%"
        "<extra></extra>"
    )
)


# Apply premium dashboard-style layout

fig.update_layout(
    title={
        "text": "💰 Profit Performance by Category",
        "x": 0.5,
        "xanchor": "center",
        "font": {
            "size": 24,
            "family": "Arial",
            "color": "#F8FAFC"
        }
    },

    paper_bgcolor="#0B1120",
    plot_bgcolor="#111827",

    font={
        "family": "Arial",
        "color": "#E5E7EB"
    },

    height=650,

    margin={
        "l": 80,
        "r": 60,
        "t": 100,
        "b": 80
    },

    xaxis={
        "title": "Product Category",
        "showgrid": False,
        "showline": True,
        "linecolor": "#374151",
        "tickfont": {
            "size": 13,
            "color": "#D1D5DB"
        },
        "title_font": {
            "size": 14,
            "color": "#F8FAFC"
        }
    },

    yaxis={
        "title": "Total Profit",
        "tickprefix": "$",
        "tickformat": ",",
        "showgrid": True,
        "gridcolor": "#1F2937",
        "gridwidth": 1,
        "showline": False,
        "tickfont": {
            "size": 12,
            "color": "#D1D5DB"
        },
        "title_font": {
            "size": 14,
            "color": "#F8FAFC"
        }
    },

    hoverlabel={
        "bgcolor": "#020617",
        "font_size": 13,
        "font_family": "Arial",
        "font_color": "#F8FAFC",
        "bordercolor": "#6366F1"
    },

    legend={
        "title": {
            "text": "Category"
        },
        "font": {
            "color": "#E5E7EB"
        },
        "bgcolor": "rgba(0,0,0,0)"
    }
)


# Highlight the highest-profit Category

top_profit_category = category_profit.iloc[0]

fig.add_annotation(
    x=top_profit_category["Category"],
    y=top_profit_category["Total_Profit"],
    text=(
        f"🏆 Highest Profit<br>"
        f"${top_profit_category['Total_Profit']:,.0f}"
    ),
    showarrow=True,
    arrowhead=2,
    arrowsize=1,
    arrowwidth=2,
    arrowcolor="#F59E0B",
    ax=0,
    ay=-65,
    bgcolor="#020617",
    bordercolor="#F59E0B",
    borderwidth=1,
    borderpad=6,
    font={
        "size": 12,
        "color": "#F8FAFC"
    }
)


# Display the final interactive visualization

fig.show()
````

### Business Insight

absolute Profit but also by its Profit Margin to determine
operational efficiency.

# 05. Regional Sales & Profitability

### Business Question

Which Regions generate the strongest Sales and Profit?

Regional aggregation helps identify geographically strong and weak
markets.

````python
# Aggregate major business metrics by Region

region_performance = (
    df.groupby("Region", as_index=False)
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Total_Quantity=("Quantity", "sum"),
          Total_Orders=("Order_ID", "nunique"),
          Total_Customers=("Customer_ID", "nunique")
      )
)

# Calculate regional Profit Margin

region_performance["Profit_Margin_%"] = (
    region_performance["Total_Profit"]
    / region_performance["Total_Sales"]
    * 100
)

# Rank Regions by Sales

region_performance = region_performance.sort_values(
    "Total_Sales",
    ascending=False
).reset_index(drop=True)

# Create Sales Rank

region_performance["Sales_Rank"] = (
    region_performance["Total_Sales"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

display(region_performance.round(2))


# Create an interactive premium horizontal Sales visualization

fig = px.bar(
    region_performance.sort_values("Total_Sales"),
    x="Total_Sales",
    y="Region",
    orientation="h",
    text="Total_Sales",
    color="Profit_Margin_%",
    color_continuous_scale=[
        "#312E81",
        "#6366F1",
        "#06B6D4",
        "#10B981"
    ],
    hover_data={
        "Sales_Rank": True,
        "Total_Sales": ":,.2f",
        "Total_Profit": ":,.2f",
        "Profit_Margin_%": ":.2f",
        "Total_Orders": ":,.0f",
        "Total_Customers": ":,.0f",
        "Total_Quantity": ":,.0f"
    },
    title="Regional Sales & Profitability Performance",
    labels={
        "Total_Sales": "Total Sales",
        "Region": "Region",
        "Profit_Margin_%": "Profit Margin (%)"
    }
)


# Format Sales labels and hover information

fig.update_traces(
    texttemplate="$%{text:,.0f}",
    textposition="outside",
    marker_line_width=1.5,
    marker_line_color="#E5E7EB",
    hovertemplate=(
        "<b>%{y}</b><br>"
        "Sales: $%{x:,.2f}<br>"
        "Profit: $%{customdata[2]:,.2f}<br>"
        "Profit Margin: %{customdata[3]:.2f}%<br>"
        "Orders: %{customdata[4]:,.0f}<br>"
        "Customers: %{customdata[5]:,.0f}<br>"
        "Quantity: %{customdata[6]:,.0f}"
        "<extra></extra>"
    )
)


# Apply premium dashboard styling

fig.update_layout(
    title={
        "text": "🌎 Regional Sales & Profitability Performance",
        "x": 0.5,
        "xanchor": "center",
        "font": {
            "size": 24,
            "family": "Arial",
            "color": "#F8FAFC"
        }
    },

    paper_bgcolor="#0B1120",
    plot_bgcolor="#111827",

    font={
        "family": "Arial",
        "color": "#E5E7EB"
    },

    height=650,

    margin={
        "l": 100,
        "r": 80,
        "t": 110,
        "b": 80
    },

    xaxis={
        "title": "Total Sales",
        "tickprefix": "$",
        "tickformat": ",",
        "showgrid": True,
        "gridcolor": "#1F2937",
        "gridwidth": 1,
        "tickfont": {
            "size": 12,
            "color": "#D1D5DB"
        },
        "title_font": {
            "size": 14,
            "color": "#F8FAFC"
        }
    },

    yaxis={
        "title": "Region",
        "showgrid": False,
        "showline": True,
        "linecolor": "#374151",
        "tickfont": {
            "size": 13,
            "color": "#D1D5DB"
        },
        "title_font": {
            "size": 14,
            "color": "#F8FAFC"
        }
    },

    coloraxis_colorbar={
        "title": "Profit<br>Margin %",
        "tickfont": {
            "color": "#D1D5DB"
        },
        "title_font": {
            "color": "#F8FAFC"
        }
    },

    hoverlabel={
        "bgcolor": "#020617",
        "font_size": 13,
        "font_family": "Arial",
        "font_color": "#F8FAFC",
        "bordercolor": "#6366F1"
    }
)


# Highlight the highest-Sales Region

top_region = region_performance.iloc[0]

fig.add_annotation(
    x=top_region["Total_Sales"],
    y=top_region["Region"],
    text=(
        f"🏆 #1 Region<br>"
        f"${top_region['Total_Sales']:,.0f}<br>"
        f"Margin: {top_region['Profit_Margin_%']:.1f}%"
    ),
    showarrow=True,
    arrowhead=2,
    arrowwidth=2,
    arrowcolor="#F59E0B",
    ax=-100,
    ay=-55,
    bgcolor="#020617",
    bordercolor="#F59E0B",
    borderwidth=1,
    borderpad=6,
    font={
        "size": 12,
        "color": "#F8FAFC"
    }
)


# Display the final interactive visualization

fig.show()
````

### Business Insight

Regional Sales identifies the strongest geographic markets.
Profit Margin should additionally be considered to determine
whether high-revenue regions are also financially efficient.

# 07. Top 10 Customers by Sales

### Business Question

Which customers contribute the highest Sales?

````python
# ============================================================
# SUPERSTORE BUSINESS INTELLIGENCE DASHBOARD
# Author: S Mohammed Kaif
# Data Science | Machine Learning | Data Analytics
# ============================================================

import io
import base64

import pandas as pd
import plotly.express as px

from dash import (
    Dash,
    html,
    dcc,
    Input,
    Output,
    State
)


# ============================================================
# 1. LOAD DATA
# ============================================================

DATA_PATH = "SuperStore_Feature_Engineered.csv"

df = pd.read_csv(DATA_PATH)


# ============================================================
# 2. DATA PREPARATION
# ============================================================

if "Order_Date" in df.columns:

    df["Order_Date"] = pd.to_datetime(
        df["Order_Date"],
        errors="coerce"
    )


numeric_columns = [
    "Sales",
    "Quantity",
    "Profit",
    "Shipping_Days",
    "Sales_Per_Unit",
    "Profit_Per_Unit",
    "Profit_Margin",
    "Return_Flag",
    "Order_Year",
    "Order_Month"
]

for column in numeric_columns:

    if column in df.columns:

        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )


# ============================================================
# 3. DASH APP INITIALIZATION
# ============================================================

app = Dash(
    __name__,
    title="SuperStore BI Dashboard"
)

server = app.server


# ============================================================
# 4. MULTI THEME CONFIGURATION
# ============================================================

THEMES = {

    "Ocean Blue": {
        "background": "#F5F7FB",
        "card": "#FFFFFF",
        "text": "#172033",
        "muted": "#64748B",
        "primary": "#2563EB",
        "secondary": "#7C3AED",
        "success": "#059669",
        "warning": "#D97706",
        "danger": "#DC2626",
        "border": "#E2E8F0",
        "header": (
            "linear-gradient("
            "135deg, #0F172A 0%, "
            "#1E3A8A 55%, "
            "#312E81 100%)"
        ),
        "footer": "#0F172A",
        "plot_template": "plotly_white"
    },

    "Premium Cream": {
        "background": "#FAF7F0",
        "card": "#FFFDF8",
        "text": "#29251F",
        "muted": "#756E63",
        "primary": "#8B6F47",
        "secondary": "#A67C52",
        "success": "#557A61",
        "warning": "#B7791F",
        "danger": "#B45353",
        "border": "#E7DED0",
        "header": (
            "linear-gradient("
            "135deg, #29251F 0%, "
            "#6B563A 55%, "
            "#8B6F47 100%)"
        ),
        "footer": "#29251F",
        "plot_template": "plotly_white"
    },

    "Emerald": {
        "background": "#F1FAF7",
        "card": "#FFFFFF",
        "text": "#12372F",
        "muted": "#637A73",
        "primary": "#059669",
        "secondary": "#0D9488",
        "success": "#16A34A",
        "warning": "#D97706",
        "danger": "#DC2626",
        "border": "#D6EEE6",
        "header": (
            "linear-gradient("
            "135deg, #052E2B 0%, "
            "#047857 55%, "
            "#0F766E 100%)"
        ),
        "footer": "#052E2B",
        "plot_template": "plotly_white"
    },

    "Midnight": {
        "background": "#090D18",
        "card": "#111827",
        "text": "#F8FAFC",
        "muted": "#94A3B8",
        "primary": "#60A5FA",
        "secondary": "#A78BFA",
        "success": "#34D399",
        "warning": "#FBBF24",
        "danger": "#F87171",
        "border": "#263449",
        "header": (
            "linear-gradient("
            "135deg, #020617 0%, "
            "#172554 55%, "
            "#312E81 100%)"
        ),
        "footer": "#020617",
        "plot_template": "plotly_dark"
    },

    "Sunset": {
        "background": "#FFF8F3",
        "card": "#FFFFFF",
        "text": "#3B1F18",
        "muted": "#806B63",
        "primary": "#EA580C",
        "secondary": "#C2410C",
        "success": "#059669",
        "warning": "#D97706",
        "danger": "#DC2626",
        "border": "#F1DCD0",
        "header": (
            "linear-gradient("
            "135deg, #431407 0%, "
            "#9A3412 55%, "
            "#C2410C 100%)"
        ),
        "footer": "#431407",
        "plot_template": "plotly_white"
    }
}


# ============================================================
# 5. DEFAULT THEME
# ============================================================

DEFAULT_THEME = THEMES["Ocean Blue"]


# ============================================================
# 6. COMMON CARD STYLE
# ============================================================

def get_card_style(theme):

    return {

        "backgroundColor": theme["card"],

        "border": (
            f"1px solid "
            f"{theme['border']}"
        ),

        "borderRadius": "16px",

        "padding": "20px",

        "boxShadow": (
            "0 4px 18px "
            "rgba(15, 23, 42, 0.06)"
        ),

        "transition": (
            "background-color 0.3s ease, "
            "border-color 0.3s ease"
        )
    }


# ============================================================
# 7. KPI CARD FUNCTION
# ============================================================

def create_kpi_card(
    title,
    value_id,
    subtitle_id=None,
    icon="●",
    accent="#2563EB"
):

    return html.Div(

        [

            html.Div(

                [

                    html.Div(
                        icon,

                        style={
                            "width": "42px",
                            "height": "42px",
                            "borderRadius": "12px",
                            "backgroundColor": f"{accent}15",
                            "color": accent,
                            "display": "flex",
                            "alignItems": "center",
                            "justifyContent": "center",
                            "fontSize": "18px",
                            "fontWeight": "700"
                        }
                    ),

                    html.Div(

                        [

                            html.Div(
                                title,

                                style={
                                    "fontSize": "13px",
                                    "color": DEFAULT_THEME["muted"],
                                    "fontWeight": "600"
                                }
                            ),

                            html.Div(
                                id=value_id,
                                children="—",

                                style={
                                    "fontSize": "27px",
                                    "fontWeight": "800",
                                    "color": DEFAULT_THEME["text"],
                                    "marginTop": "5px"
                                }
                            )

                        ]
                    )

                ],

                style={
                    "display": "flex",
                    "gap": "14px",
                    "alignItems": "center"
                }
            ),

            html.Div(
                id=subtitle_id,
                children="Business performance",

                style={
                    "fontSize": "11px",
                    "color": DEFAULT_THEME["muted"],
                    "marginTop": "16px"
                }
            )

        ],

        style={
            **get_card_style(DEFAULT_THEME),
            "flex": "1",
            "minWidth": "180px"
        }
    )


# ============================================================
# 8. DROPDOWN OPTIONS
# ============================================================

def dropdown_options(column):

    if column not in df.columns:

        return []

    values = (
        df[column]
        .dropna()
        .astype(str)
        .sort_values()
        .unique()
    )

    return [

        {
            "label": value,
            "value": value
        }

        for value in values

    ]


# ============================================================
# 9. LAYOUT
# ============================================================

app.layout = html.Div(

    [

        # Theme state
        dcc.Store(
            id="theme-store",
            data="Ocean Blue"
        ),


        # ====================================================
        # HEADER
        # ====================================================

        html.Div(

            [

                html.Div(

                    [

                        html.Div(
                            "SUPERSTORE",

                            id="header-eyebrow",

                            style={
                                "fontSize": "13px",
                                "fontWeight": "800",
                                "letterSpacing": "3px",
                                "color": "#93C5FD"
                            }
                        ),

                        html.H1(
                            "Business Intelligence Dashboard",

                            id="header-title",

                            style={
                                "margin": "5px 0 0 0",
                                "fontSize": "32px",
                                "fontWeight": "800",
                                "color": "#FFFFFF"
                            }
                        ),

                        html.P(
                            "Sales • Profitability • Customers • Products • Operations",

                            id="header-description",

                            style={
                                "margin": "7px 0 0 0",
                                "color": "#CBD5E1",
                                "fontSize": "14px"
                            }
                        )

                    ]
                ),


                # ====================================================
                # THEME SELECTOR
                # ====================================================

                html.Div(

                    [

                        html.Div(
                            "THEME",

                            style={
                                "fontSize": "10px",
                                "fontWeight": "800",
                                "letterSpacing": "2px",
                                "color": "#CBD5E1",
                                "marginBottom": "6px"
                            }
                        ),

                        dcc.Dropdown(

                            id="theme-selector",

                            options=[

                                {
                                    "label": "🌊 Ocean Blue",
                                    "value": "Ocean Blue"
                                },

                                {
                                    "label": "☕ Premium Cream",
                                    "value": "Premium Cream"
                                },

                                {
                                    "label": "🌿 Emerald",
                                    "value": "Emerald"
                                },

                                {
                                    "label": "🌙 Midnight",
                                    "value": "Midnight"
                                },

                                {
                                    "label": "🌅 Sunset",
                                    "value": "Sunset"
                                }

                            ],

                            value="Ocean Blue",

                            clearable=False,

                            style={
                                "width": "195px"
                            }

                        )

                    ],

                    style={
                        "minWidth": "195px"
                    }
                )

            ],

            id="main-header",

            style={
                "background": DEFAULT_THEME["header"],
                "padding": "30px 38px",
                "display": "flex",
                "justifyContent": "space-between",
                "alignItems": "center",
                "flexWrap": "wrap",
                "gap": "20px"
            }
        ),


        # ====================================================
        # MAIN CONTENT
        # ====================================================

        html.Div(

            [

                # ====================================================
                # FILTER PANEL
                # ====================================================

                html.Div(

                    [

                        html.Div(

                            [

                                html.Div(
                                    "FILTERS",

                                    id="filter-label",

                                    style={
                                        "fontSize": "11px",
                                        "fontWeight": "800",
                                        "letterSpacing": "2px",
                                        "color": DEFAULT_THEME["primary"]
                                    }
                                ),

                                html.H3(
                                    "Business Controls",

                                    id="filter-title",

                                    style={
                                        "margin": "4px 0",
                                        "color": DEFAULT_THEME["text"]
                                    }
                                )

                            ]
                        ),


                        html.Div(

                            [

                                html.Div(

                                    [

                                        html.Label(
                                            "Region",

                                            style={
                                                "fontWeight": "600",
                                                "fontSize": "12px",
                                                "color": DEFAULT_THEME["text"]
                                            }
                                        ),

                                        dcc.Dropdown(
                                            id="region-filter",
                                            options=dropdown_options("Region"),
                                            placeholder="All Regions",
                                            multi=True,
                                            clearable=True
                                        )

                                    ],

                                    style={
                                        "flex": "1",
                                        "minWidth": "180px"
                                    }
                                ),


                                html.Div(

                                    [

                                        html.Label(
                                            "Category",

                                            style={
                                                "fontWeight": "600",
                                                "fontSize": "12px",
                                                "color": DEFAULT_THEME["text"]
                                            }
                                        ),

                                        dcc.Dropdown(
                                            id="category-filter",
                                            options=dropdown_options("Category"),
                                            placeholder="All Categories",
                                            multi=True,
                                            clearable=True
                                        )

                                    ],

                                    style={
                                        "flex": "1",
                                        "minWidth": "180px"
                                    }
                                ),


                                html.Div(

                                    [

                                        html.Label(
                                            "Segment",

                                            style={
                                                "fontWeight": "600",
                                                "fontSize": "12px",
                                                "color": DEFAULT_THEME["text"]
                                            }
                                        ),

                                        dcc.Dropdown(
                                            id="segment-filter",
                                            options=dropdown_options("Segment"),
                                            placeholder="All Segments",
                                            multi=True,
                                            clearable=True
                                        )

                                    ],

                                    style={
                                        "flex": "1",
                                        "minWidth": "180px"
                                    }
                                ),


                                html.Div(

                                    [

                                        html.Label(
                                            "Ship Mode",

                                            style={
                                                "fontWeight": "600",
                                                "fontSize": "12px",
                                                "color": DEFAULT_THEME["text"]
                                            }
                                        ),

                                        dcc.Dropdown(
                                            id="ship-filter",
                                            options=dropdown_options("Ship_Mode"),
                                            placeholder="All Ship Modes",
                                            multi=True,
                                            clearable=True
                                        )

                                    ],

                                    style={
                                        "flex": "1",
                                        "minWidth": "180px"
                                    }
                                )

                            ],

                            style={
                                "display": "flex",
                                "gap": "14px",
                                "flexWrap": "wrap",
                                "marginTop": "18px"
                            }
                        ),


                        html.Div(

                            [

                                html.Button(
                                    "↻ Reset Filters",
                                    id="reset-button",
                                    n_clicks=0,

                                    style={
                                        "border": "none",
                                        "backgroundColor": "#EFF6FF",
                                        "color": "#2563EB",
                                        "padding": "10px 18px",
                                        "borderRadius": "10px",
                                        "fontWeight": "700",
                                        "cursor": "pointer"
                                    }
                                ),


                                html.Button(
                                    "⬇ Download Filtered Data",
                                    id="download-button",
                                    n_clicks=0,

                                    style={
                                        "border": "none",
                                        "backgroundColor": "#0F172A",
                                        "color": "#FFFFFF",
                                        "padding": "10px 18px",
                                        "borderRadius": "10px",
                                        "fontWeight": "700",
                                        "cursor": "pointer"
                                    }
                                ),


                                dcc.Download(
                                    id="download-data"
                                )

                            ],

                            style={
                                "display": "flex",
                                "gap": "10px",
                                "marginTop": "18px",
                                "flexWrap": "wrap"
                            }
                        )

                    ],

                    id="filter-card",

                    style={
                        **get_card_style(DEFAULT_THEME),
                        "marginBottom": "20px"
                    }
                ),


                # ====================================================
                # KPI ROW 1
                # ====================================================

                html.Div(

                    [

                        create_kpi_card(
                            "Total Sales",
                            "kpi-sales",
                            "Revenue generated",
                            "$",
                            "#2563EB"
                        ),

                        create_kpi_card(
                            "Total Profit",
                            "kpi-profit",
                            "Net profitability",
                            "P",
                            "#059669"
                        ),

                        create_kpi_card(
                            "Total Orders",
                            "kpi-orders",
                            "Unique orders",
                            "#",
                            "#7C3AED"
                        ),

                        create_kpi_card(
                            "Customers",
                            "kpi-customers",
                            "Unique customers",
                            "U",
                            "#0891B2"
                        )

                    ],

                    style={
                        "display": "flex",
                        "gap": "16px",
                        "flexWrap": "wrap",
                        "marginBottom": "16px"
                    }
                ),


                # ====================================================
                # KPI ROW 2
                # ====================================================

                html.Div(

                    [

                        create_kpi_card(
                            "Quantity Sold",
                            "kpi-quantity",
                            "Items sold",
                            "Q",
                            "#D97706"
                        ),

                        create_kpi_card(
                            "Profit Margin",
                            "kpi-margin",
                            "Profit efficiency",
                            "%",
                            "#DC2626"
                        ),

                        create_kpi_card(
                            "Average Order Value",
                            "kpi-aov",
                            "Sales per order",
                            "A",
                            "#4F46E5"
                        ),

                        create_kpi_card(
                            "Profit per Order",
                            "kpi-profit-order",
                            "Profit efficiency",
                            "₹",
                            "#0F766E"
                        )

                    ],

                    style={
                        "display": "flex",
                        "gap": "16px",
                        "flexWrap": "wrap",
                        "marginBottom": "20px"
                    }
                ),


                # ====================================================
                # TREND + CATEGORY
                # ====================================================

                html.Div(

                    [

                        html.Div(

                            [

                                html.H3(
                                    "Sales & Profit Trend",
                                    id="monthly-title",

                                    style={
                                        "margin": "0",
                                        "color": DEFAULT_THEME["text"]
                                    }
                                ),

                                html.P(
                                    "Monthly business performance",

                                    style={
                                        "margin": "4px 0",
                                        "color": DEFAULT_THEME["muted"],
                                        "fontSize": "12px"
                                    }
                                ),

                                dcc.Graph(
                                    id="monthly-trend",
                                    config={
                                        "displayModeBar": False
                                    }
                                )

                            ],

                            id="monthly-card",

                            style={
                                **get_card_style(DEFAULT_THEME),
                                "flex": "2",
                                "minWidth": "500px"
                            }
                        ),


                        html.Div(

                            [

                                html.H3(
                                    "Category Performance",

                                    style={
                                        "margin": "0",
                                        "color": DEFAULT_THEME["text"]
                                    }
                                ),

                                html.P(
                                    "Sales contribution by category",

                                    style={
                                        "margin": "4px 0",
                                        "color": DEFAULT_THEME["muted"],
                                        "fontSize": "12px"
                                    }
                                ),

                                dcc.Graph(
                                    id="category-chart",
                                    config={
                                        "displayModeBar": False
                                    }
                                )

                            ],

                            id="category-card",

                            style={
                                **get_card_style(DEFAULT_THEME),
                                "flex": "1",
                                "minWidth": "350px"
                            }
                        )

                    ],

                    style={
                        "display": "flex",
                        "gap": "16px",
                        "flexWrap": "wrap",
                        "marginBottom": "16px"
                    }
                ),


                # ====================================================
                # REGION + SEGMENT
                # ====================================================

                html.Div(

                    [

                        html.Div(

                            [

                                html.H3(
                                    "Regional Performance",

                                    style={
                                        "margin": "0",
                                        "color": DEFAULT_THEME["text"]
                                    }
                                ),

                                html.P(
                                    "Sales and profitability by Region",

                                    style={
                                        "margin": "4px 0",
                                        "color": DEFAULT_THEME["muted"],
                                        "fontSize": "12px"
                                    }
                                ),

                                dcc.Graph(
                                    id="region-chart",
                                    config={
                                        "displayModeBar": False
                                    }
                                )

                            ],

                            id="region-card",

                            style={
                                **get_card_style(DEFAULT_THEME),
                                "flex": "1",
                                "minWidth": "400px"
                            }
                        ),


                        html.Div(

                            [

                                html.H3(
                                    "Segment Performance",

                                    style={
                                        "margin": "0",
                                        "color": DEFAULT_THEME["text"]
                                    }
                                ),

                                html.P(
                                    "Customer segment contribution",

                                    style={
                                        "margin": "4px 0",
                                        "color": DEFAULT_THEME["muted"],
                                        "fontSize": "12px"
                                    }
                                ),

                                dcc.Graph(
                                    id="segment-chart",
                                    config={
                                        "displayModeBar": False
                                    }
                                )

                            ],

                            id="segment-card",

                            style={
                                **get_card_style(DEFAULT_THEME),
                                "flex": "1",
                                "minWidth": "400px"
                            }
                        )

                    ],

                    style={
                        "display": "flex",
                        "gap": "16px",
                        "flexWrap": "wrap",
                        "marginBottom": "16px"
                    }
                ),


                # ====================================================
                # PRODUCT + PAYMENT
                # ====================================================

                html.Div(

                    [

                        html.Div(

                            [

                                html.H3(
                                    "Top 10 Products",

                                    style={
                                        "margin": "0",
                                        "color": DEFAULT_THEME["text"]
                                    }
                                ),

                                html.P(
                                    "Highest revenue-generating products",

                                    style={
                                        "margin": "4px 0",
                                        "color": DEFAULT_THEME["muted"],
                                        "fontSize": "12px"
                                    }
                                ),

                                dcc.Graph(
                                    id="product-chart",
                                    config={
                                        "displayModeBar": False
                                    }
                                )

                            ],

                            id="product-card",

                            style={
                                **get_card_style(DEFAULT_THEME),
                                "flex": "2",
                                "minWidth": "500px"
                            }
                        ),


                        html.Div(

                            [

                                html.H3(
                                    "Payment Mode",

                                    style={
                                        "margin": "0",
                                        "color": DEFAULT_THEME["text"]
                                    }
                                ),

                                html.P(
                                    "Revenue mix by payment channel",

                                    style={
                                        "margin": "4px 0",
                                        "color": DEFAULT_THEME["muted"],
                                        "fontSize": "12px"
                                    }
                                ),

                                dcc.Graph(
                                    id="payment-chart",
                                    config={
                                        "displayModeBar": False
                                    }
                                )

                            ],

                            id="payment-card",

                            style={
                                **get_card_style(DEFAULT_THEME),
                                "flex": "1",
                                "minWidth": "350px"
                            }
                        )

                    ],

                    style={
                        "display": "flex",
                        "gap": "16px",
                        "flexWrap": "wrap",
                        "marginBottom": "20px"
                    }
                ),


                # ====================================================
                # BUSINESS SUMMARY
                # ====================================================

                html.Div(

                    [

                        html.Div(

                            [

                                html.Div(
                                    "EXECUTIVE SUMMARY",

                                    style={
                                        "fontSize": "11px",
                                        "fontWeight": "800",
                                        "letterSpacing": "2px",
                                        "color": DEFAULT_THEME["primary"]
                                    }
                                ),

                                html.H2(
                                    "Business Performance Overview",

                                    style={
                                        "margin": "5px 0 12px",
                                        "color": DEFAULT_THEME["text"]
                                    }
                                ),

                                html.P(
                                    id="business-summary",

                                    children=(
                                        "Select filters to generate "
                                        "the business performance summary."
                                    ),

                                    style={
                                        "color": DEFAULT_THEME["muted"],
                                        "lineHeight": "1.8",
                                        "fontSize": "14px"
                                    }
                                )

                            ],

                            style={
                                "flex": "1"
                            }
                        )

                    ],

                    id="summary-card",

                    style={
                        **get_card_style(DEFAULT_THEME),
                        "marginBottom": "25px"
                    }
                )

            ],

            id="main-content",

            style={
                "maxWidth": "1500px",
                "margin": "0 auto",
                "padding": "24px 25px 35px"
            }
        ),


        # ====================================================
        # FOOTER
        # ====================================================

        html.Footer(

            [

                html.Div(

                    [

                        html.Div(
                            "S MOHAMMED KAIF",

                            style={
                                "fontWeight": "800",
                                "fontSize": "14px",
                                "color": "#FFFFFF",
                                "letterSpacing": "1px"
                            }
                        ),

                        html.Div(
                            "Data Science • Machine Learning • Data Analytics",

                            style={
                                "fontSize": "12px",
                                "color": "#CBD5E1",
                                "marginTop": "5px"
                            }
                        )

                    ]
                ),


                # ====================================================
                # SOCIAL LINKS
                # ====================================================

                html.Div(

                    [

                        html.A(
                            "🔗 LinkedIn",

                            href=(
                                "https://www.linkedin.com/"
                                "in/s-mohammed-kaif-2a500a341/"
                            ),

                            target="_blank",

                            style={
                                "color": "#FFFFFF",
                                "textDecoration": "none",
                                "padding": "9px 16px",
                                "borderRadius": "9px",
                                "border": (
                                    "1px solid "
                                    "rgba(255,255,255,0.20)"
                                ),
                                "backgroundColor": (
                                    "rgba(255,255,255,0.08)"
                                ),
                                "fontSize": "12px",
                                "fontWeight": "700",
                                "display": "inline-block",
                                "marginRight": "8px"
                            }
                        ),


                        html.A(
                            "💻 GitHub",

                            href=(
                                "https://github.com/"
                                "Shaik-Mohammed-Kaif"
                            ),

                            target="_blank",

                            style={
                                "color": "#FFFFFF",
                                "textDecoration": "none",
                                "padding": "9px 16px",
                                "borderRadius": "9px",
                                "border": (
                                    "1px solid "
                                    "rgba(255,255,255,0.20)"
                                ),
                                "backgroundColor": (
                                    "rgba(255,255,255,0.08)"
                                ),
                                "fontSize": "12px",
                                "fontWeight": "700",
                                "display": "inline-block"
                            }
                        )

                    ]
                ),


                html.Div(
                    "SuperStore Business Intelligence Dashboard • 2026",

                    style={
                        "fontSize": "11px",
                        "color": "#94A3B8"
                    }
                )

            ],

            id="main-footer",

            style={
                "backgroundColor": DEFAULT_THEME["footer"],
                "padding": "22px 30px",
                "display": "flex",
                "justifyContent": "space-between",
                "alignItems": "center",
                "gap": "15px",
                "flexWrap": "wrap"
            }
        )

    ],

    id="app-shell",

    style={
        "backgroundColor": DEFAULT_THEME["background"],
        "minHeight": "100vh",
        "fontFamily": "Arial, sans-serif",
        "transition": "all 0.3s ease"
    }
)


# ============================================================
# 10. THEME CALLBACK
# ============================================================

@app.callback(

    Output("theme-store", "data"),

    Input(
        "theme-selector",
        "value"
    )

)

def update_theme(
    selected_theme
):

    return selected_theme


# ============================================================
# 11. RESET FILTERS CALLBACK
# ============================================================

@app.callback(

    Output(
        "region-filter",
        "value"
    ),

    Output(
        "category-filter",
        "value"
    ),

    Output(
        "segment-filter",
        "value"
    ),

    Output(
        "ship-filter",
        "value"
    ),

    Input(
        "reset-button",
        "n_clicks"
    ),

    prevent_initial_call=True
)

def reset_filters(
    n_clicks
):

    return [], [], [], []


# ============================================================
# 12. MAIN DASHBOARD CALLBACK
# ============================================================

@app.callback(

    Output("kpi-sales", "children"),
    Output("kpi-profit", "children"),
    Output("kpi-orders", "children"),
    Output("kpi-customers", "children"),
    Output("kpi-quantity", "children"),
    Output("kpi-margin", "children"),
    Output("kpi-aov", "children"),
    Output("kpi-profit-order", "children"),

    Output("monthly-trend", "figure"),
    Output("category-chart", "figure"),
    Output("region-chart", "figure"),
    Output("segment-chart", "figure"),
    Output("product-chart", "figure"),
    Output("payment-chart", "figure"),
    Output("business-summary", "children"),

    Input("region-filter", "value"),
    Input("category-filter", "value"),
    Input("segment-filter", "value"),
    Input("ship-filter", "value"),
    Input("theme-selector", "value")

)

def update_dashboard(

    selected_regions,
    selected_categories,
    selected_segments,
    selected_ship_modes,
    selected_theme

):

    # ========================================================
    # SELECT THEME
    # ========================================================

    theme = THEMES.get(
        selected_theme,
        DEFAULT_THEME
    )


    # ========================================================
    # FILTER DATA
    # ========================================================

    filtered_df = df.copy()


    if selected_regions:

        filtered_df = filtered_df[
            filtered_df["Region"]
            .astype(str)
            .isin(selected_regions)
        ]


    if selected_categories:

        filtered_df = filtered_df[
            filtered_df["Category"]
            .astype(str)
            .isin(selected_categories)
        ]


    if selected_segments:

        filtered_df = filtered_df[
            filtered_df["Segment"]
            .astype(str)
            .isin(selected_segments)
        ]


    if selected_ship_modes:

        filtered_df = filtered_df[
            filtered_df["Ship_Mode"]
            .astype(str)
            .isin(selected_ship_modes)
        ]


    # ========================================================
    # KPI CALCULATIONS
    # ========================================================

    total_sales = filtered_df["Sales"].sum()

    total_profit = filtered_df["Profit"].sum()

    total_orders = filtered_df["Order_ID"].nunique()

    total_customers = filtered_df["Customer_ID"].nunique()

    total_quantity = filtered_df["Quantity"].sum()


    profit_margin = (

        total_profit
        / total_sales
        * 100

        if total_sales != 0

        else 0

    )


    aov = (

        total_sales
        / total_orders

        if total_orders != 0

        else 0

    )


    profit_per_order = (

        total_profit
        / total_orders

        if total_orders != 0

        else 0

    )


    # ========================================================
    # MONTHLY TREND
    # ========================================================

    if not filtered_df.empty:

        monthly = (

            filtered_df

            .groupby(
                [
                    "Order_Year",
                    "Order_Month"
                ],

                as_index=False
            )

            .agg(

                Sales=("Sales", "sum"),

                Profit=("Profit", "sum")

            )

        )


        monthly["Order_Year"] = (

            pd.to_numeric(
                monthly["Order_Year"],
                errors="coerce"
            )

            .fillna(0)

            .astype(int)

        )


        monthly["Order_Month"] = (

            pd.to_numeric(
                monthly["Order_Month"],
                errors="coerce"
            )

            .fillna(1)

            .astype(int)

        )


        monthly["Period"] = pd.to_datetime(

            dict(

                year=monthly["Order_Year"],

                month=monthly["Order_Month"],

                day=1

            ),

            errors="coerce"

        )


        monthly = monthly.sort_values(
            "Period"
        )


        monthly["Period_Label"] = (

            monthly["Period"]

            .dt.strftime(
                "%b %Y"
            )

        )

    else:

        monthly = pd.DataFrame(

            columns=[
                "Period_Label",
                "Sales",
                "Profit"
            ]

        )


    fig_monthly = px.line(

        monthly,

        x="Period_Label",

        y=[
            "Sales",
            "Profit"
        ],

        markers=True,

        template=theme["plot_template"],

        color_discrete_sequence=[
            theme["primary"],
            theme["success"]
        ],

        labels={
            "value": "Amount",
            "Period_Label": "Period"
        }

    )


    fig_monthly.update_traces(
        line={
            "width": 3
        }
    )


    fig_monthly.update_layout(

        height=390,

        margin=dict(
            l=50,
            r=20,
            t=25,
            b=50
        ),

        legend_title_text="",

        hovermode="x unified",

        paper_bgcolor=theme["card"],

        plot_bgcolor=theme["card"],

        font={
            "color": theme["text"]
        }
    )


    # ========================================================
    # CATEGORY
    # ========================================================

    category = (

        filtered_df

        .groupby(
            "Category",
            as_index=False
        )

        .agg(

            Sales=("Sales", "sum"),

            Profit=("Profit", "sum")

        )

        .sort_values(
            "Sales",
            ascending=False
        )

    )


    fig_category = px.bar(

        category,

        x="Category",

        y="Sales",

        color="Profit",

        template=theme["plot_template"],

        color_continuous_scale=[
            theme["primary"],
            theme["secondary"]
        ],

        labels={
            "Sales": "Sales",
            "Category": ""
        }

    )


    fig_category.update_layout(

        height=390,

        margin=dict(
            l=45,
            r=20,
            t=25,
            b=50
        ),

        coloraxis_colorbar_title="Profit",

        paper_bgcolor=theme["card"],

        plot_bgcolor=theme["card"],

        font={
            "color": theme["text"]
        }
    )


    # ========================================================
    # REGION
    # ========================================================

    region = (

        filtered_df

        .groupby(
            "Region",
            as_index=False
        )

        .agg(

            Sales=("Sales", "sum"),

            Profit=("Profit", "sum")

        )

        .sort_values(
            "Sales",
            ascending=False
        )

    )


    fig_region = px.bar(

        region,

        x="Region",

        y=[
            "Sales",
            "Profit"
        ],

        barmode="group",

        template=theme["plot_template"],

        color_discrete_sequence=[
            theme["primary"],
            theme["success"]
        ]

    )


    fig_region.update_layout(

        height=390,

        margin=dict(
            l=45,
            r=20,
            t=25,
            b=50
        ),

        legend_title_text="",

        paper_bgcolor=theme["card"],

        plot_bgcolor=theme["card"],

        font={
            "color": theme["text"]
        }
    )


    # ========================================================
    # SEGMENT
    # ========================================================

    segment = (

        filtered_df

        .groupby(
            "Segment",
            as_index=False
        )

        .agg(

            Sales=("Sales", "sum"),

            Profit=("Profit", "sum")

        )

    )


    fig_segment = px.pie(

        segment,

        names="Segment",

        values="Sales",

        hole=0.55,

        template=theme["plot_template"],

        color_discrete_sequence=[
            theme["primary"],
            theme["secondary"],
            theme["success"],
            theme["warning"]
        ]

    )


    fig_segment.update_layout(

        height=390,

        margin=dict(
            l=20,
            r=20,
            t=25,
            b=20
        ),

        legend_title_text="",

        paper_bgcolor=theme["card"],

        font={
            "color": theme["text"]
        }
    )


    # ========================================================
    # TOP PRODUCTS
    # ========================================================

    products = (

        filtered_df

        .groupby(
            "Product_Name",
            as_index=False
        )

        .agg(

            Sales=("Sales", "sum"),

            Profit=("Profit", "sum")

        )

        .sort_values(
            "Sales",
            ascending=False
        )

        .head(10)

    )


    products = products.sort_values(
        "Sales"
    )


    fig_products = px.bar(

        products,

        x="Sales",

        y="Product_Name",

        orientation="h",

        color="Profit",

        template=theme["plot_template"],

        color_continuous_scale=[
            theme["primary"],
            theme["secondary"]
        ]

    )


    fig_products.update_layout(

        height=500,

        margin=dict(
            l=10,
            r=30,
            t=25,
            b=45
        ),

        coloraxis_colorbar_title="Profit",

        paper_bgcolor=theme["card"],

        plot_bgcolor=theme["card"],

        font={
            "color": theme["text"]
        }
    )


    # ========================================================
    # PAYMENT MODE
    # ========================================================

    payment = (

        filtered_df

        .groupby(
            "Payment_Mode",
            as_index=False
        )

        .agg(

            Sales=("Sales", "sum"),

            Profit=("Profit", "sum")

        )

        .sort_values(
            "Sales",
            ascending=False
        )

    )


    fig_payment = px.pie(

        payment,

        names="Payment_Mode",

        values="Sales",

        hole=0.58,

        template=theme["plot_template"],

        color_discrete_sequence=[
            theme["primary"],
            theme["secondary"],
            theme["success"],
            theme["warning"],
            theme["danger"]
        ]

    )


    fig_payment.update_layout(

        height=500,

        margin=dict(
            l=10,
            r=10,
            t=25,
            b=20
        ),

        legend_title_text="",

        paper_bgcolor=theme["card"],

        font={
            "color": theme["text"]
        }
    )


    # ========================================================
    # EXECUTIVE SUMMARY
    # ========================================================

    selected_filter_count = sum(

        [

            bool(selected_regions),

            bool(selected_categories),

            bool(selected_segments),

            bool(selected_ship_modes)

        ]

    )


    filter_status = (

        "Filtered view"

        if selected_filter_count > 0

        else "Overall business view"

    )


    summary = (

        f"{filter_status}: the selected dataset contains "

        f"{total_orders:,.0f} unique orders from "

        f"{total_customers:,.0f} customers, generating "

        f"${total_sales:,.2f} in Sales and "

        f"${total_profit:,.2f} in Profit. "

        f"The overall Profit Margin is "

        f"{profit_margin:.2f}%, with an Average Order Value "

        f"of ${aov:,.2f}."

    )


    # ========================================================
    # RETURN EVERYTHING
    # ========================================================

    return (

        f"${total_sales:,.0f}",

        f"${total_profit:,.0f}",

        f"{total_orders:,.0f}",

        f"{total_customers:,.0f}",

        f"{total_quantity:,.0f}",

        f"{profit_margin:.2f}%",

        f"${aov:,.2f}",

        f"${profit_per_order:,.2f}",

        fig_monthly,

        fig_category,

        fig_region,

        fig_segment,

        fig_products,

        fig_payment,

        summary

    )


# ============================================================
# 13. THEME UI CALLBACK
# ============================================================

@app.callback(

    Output(
        "app-shell",
        "style"
    ),

    Output(
        "main-header",
        "style"
    ),

    Output(
        "main-footer",
        "style"
    ),

    Output(
        "filter-card",
        "style"
    ),

    Output(
        "monthly-card",
        "style"
    ),

    Output(
        "category-card",
        "style"
    ),

    Output(
        "region-card",
        "style"
    ),

    Output(
        "segment-card",
        "style"
    ),

    Output(
        "product-card",
        "style"
    ),

    Output(
        "payment-card",
        "style"
    ),

    Output(
        "summary-card",
        "style"
    ),

    Input(
        "theme-selector",
        "value"
    )

)

def apply_theme(
    selected_theme
):

    theme = THEMES.get(
        selected_theme,
        DEFAULT_THEME
    )


    app_style = {

        "backgroundColor":
            theme["background"],

        "minHeight":
            "100vh",

        "fontFamily":
            "Arial, sans-serif",

        "transition":
            "all 0.3s ease"

    }


    header_style = {

        "background":
            theme["header"],

        "padding":
            "30px 38px",

        "display":
            "flex",

        "justifyContent":
            "space-between",

        "alignItems":
            "center",

        "flexWrap":
            "wrap",

        "gap":
            "20px"

    }


    footer_style = {

        "backgroundColor":
            theme["footer"],

        "padding":
            "22px 30px",

        "display":
            "flex",

        "justifyContent":
            "space-between",

        "alignItems":
            "center",

        "gap":
            "15px",

        "flexWrap":
            "wrap"

    }


    card_style = get_card_style(
        theme
    )


    return (

        app_style,

        header_style,

        footer_style,

        {
            **card_style,
            "marginBottom": "20px"
        },

        {
            **card_style,
            "flex": "2",
            "minWidth": "500px"
        },

        {
            **card_style,
            "flex": "1",
            "minWidth": "350px"
        },

        {
            **card_style,
            "flex": "1",
            "minWidth": "400px"
        },

        {
            **card_style,
            "flex": "1",
            "minWidth": "400px"
        },

        {
            **card_style,
            "flex": "2",
            "minWidth": "500px"
        },

        {
            **card_style,
            "flex": "1",
            "minWidth": "350px"
        },

        {
            **card_style,
            "marginBottom": "25px"
        }

    )


# ============================================================
# 14. DOWNLOAD FILTERED DATA
# ============================================================

@app.callback(

    Output(
        "download-data",
        "data"
    ),

    Input(
        "download-button",
        "n_clicks"
    ),

    State(
        "region-filter",
        "value"
    ),

    State(
        "category-filter",
        "value"
    ),

    State(
        "segment-filter",
        "value"
    ),

    State(
        "ship-filter",
        "value"
    ),

    prevent_initial_call=True

)

def download_filtered_data(

    n_clicks,

    selected_regions,

    selected_categories,

    selected_segments,

    selected_ship_modes

):

    filtered_df = df.copy()


    if selected_regions:

        filtered_df = filtered_df[
            filtered_df["Region"]
            .astype(str)
            .isin(selected_regions)
        ]


    if selected_categories:

        filtered_df = filtered_df[
            filtered_df["Category"]
            .astype(str)
            .isin(selected_categories)
        ]


    if selected_segments:

        filtered_df = filtered_df[
            filtered_df["Segment"]
            .astype(str)
            .isin(selected_segments)
        ]


    if selected_ship_modes:

        filtered_df = filtered_df[
            filtered_df["Ship_Mode"]
            .astype(str)
            .isin(selected_ship_modes)
        ]


    return dcc.send_data_frame(

        filtered_df.to_csv,

        "SuperStore_Feature_Engineered.csv",

        index=False

    )


# ============================================================
# 15. RUN SERVER
# ============================================================

if __name__ == "__main__":

    app.run(

        debug=True,

        host="127.0.0.1",

        port=8050

    )
````
