````python
## Load the dataset base on YT Google Cloud API's Using key.

# ==========================================================
# 📚 Import Required Libraries
# ==========================================================

import pandas as pd
from googleapiclient.discovery import build

# ==========================================================
# 🔑 YouTube Data API Key
# ==========================================================

API_KEY = "AIzaSyDlVRCnLKKgvs4lpdWNhoEIwQy0zrR1ojA"

# ==========================================================
# 📥 Load Trending YouTube Videos
# ==========================================================

def get_trending_videos(api_key, max_results=200, region="US"):

    # Build YouTube API Service
    youtube = build(
        serviceName="youtube",
        version="v3",
        developerKey=api_key
    )

    videos = []

    request = youtube.videos().list(
        part="snippet,contentDetails,statistics",
        chart="mostPopular",
        regionCode=region,
        maxResults=50
    )

    while request and len(videos) < max_results:

        response = request.execute()

        for item in response["items"]:

            videos.append({

                "video_id": item["id"],

                "title": item["snippet"]["title"],

                "description": item["snippet"]["description"],

                "published_at": item["snippet"]["publishedAt"],

                "channel_id": item["snippet"]["channelId"],

                "channel_title": item["snippet"]["channelTitle"],

                "category_id": item["snippet"]["categoryId"],

                "tags": item["snippet"].get("tags", []),

                "duration": item["contentDetails"]["duration"],

                "definition": item["contentDetails"]["definition"],

                "caption": item["contentDetails"].get("caption", "false"),

                "view_count": item["statistics"].get("viewCount", 0),

                "like_count": item["statistics"].get("likeCount", 0),

                "favorite_count": item["statistics"].get("favoriteCount", 0),

                "comment_count": item["statistics"].get("commentCount", 0)

            })

        request = youtube.videos().list_next(request, response)

    return pd.DataFrame(videos[:max_results])

# ==========================================================
# 🚀 Load Dataset into DataFrame
# ==========================================================

df = get_trending_videos(
    api_key=API_KEY,
    max_results=200,
    region="US"
)

# ==========================================================
# 📊 Preview Dataset
# ==========================================================

print(f"Dataset Shape : {df.shape}")

df.head()
````

````python
# Extract all URLs from description

df["urls_links"] = df["description"].fillna("").apply(
    lambda x: re.findall(r'https?://[^\s]+', x)
)

df["url_count"] = df["urls_links"].str.len()

df[["description", "urls_links", "url_count"]].head()
````

````python
df["urls_links"] = df["urls_links"].apply(
    lambda x: " | ".join(x) if x else None
)

df[["urls_links", "url_count"]].head()
````

````python
def identify_url_type(url):

    url = url.lower()

    if "instagram.com" in url:
        return "Instagram"

    elif "youtube.com" in url or "youtu.be" in url:
        return "YouTube"

    elif "twitter.com" in url or "x.com" in url:
        return "Twitter/X"

    elif "facebook.com" in url or "fb.com" in url:
        return "Facebook"

    elif "tiktok.com" in url:
        return "TikTok"

    elif "discord.gg" in url or "discord.com" in url:
        return "Discord"

    elif "spotify.com" in url:
        return "Spotify"

    elif "music.apple.com" in url:
        return "Apple Music"

    elif "soundcloud.com" in url:
        return "SoundCloud"

    elif "twitch.tv" in url:
        return "Twitch"

    elif "store.steampowered.com" in url or "steamcommunity.com" in url:
        return "Steam"

    elif "epicgames.com" in url:
        return "Epic Games"

    else:
        return "Other"
````

````python
def extract_url_types(text):

    urls = re.findall(r'https?://[^\s]+', text)

    types = [identify_url_type(url) for url in urls]

    return list(dict.fromkeys(types))


df["url_types"] = df["description"].fillna("").apply(
    extract_url_types
)

df[["urls_links", "url_types"]].head()
````

````python
df["hashtags"] = df["description"].fillna("").apply(
    lambda x: re.findall(r'#[A-Za-z0-9_]+', x)
)

df["hashtag_count"] = df["hashtags"].str.len()

df[["hashtags", "hashtag_count"]].head()
````

````python
df["clean_description"] = (

    df["description"]
    
    .fillna("")
    
    .str.replace(
        r'https?://[^\s]+',
        '',
        regex=True
    )
    
    .str.replace(
        r'#[A-Za-z0-9_]+',
        '',
        regex=True
    )
    
    .str.replace(
        r'\s+',
        ' ',
        regex=True
    )
    
    .str.strip()
)
````

````python
df["tags_clean"] = df["tags"].apply(
    lambda x: x if isinstance(x, list) else []
)

df["tag_count"] = df["tags_clean"].str.len()

df[["tags_clean", "tag_count"]].head()
````

# 🧠 Natural Language Processing (NLP) Pipeline

## Business Objective

The **description** column contains large volumes of unstructured textual information such as promotional content, hashtags, URLs, social media links, creator messages, and video descriptions. Raw text cannot be analyzed directly, so Natural Language Processing (NLP) techniques were applied to convert this unstructured data into meaningful business insights.

The primary objective of this NLP pipeline is to clean, preprocess, transform, and analyze textual data to identify trending topics, important keywords, creator marketing strategies, SEO patterns, and content optimization opportunities.

Throughout this project, ten major NLP modules were implemented, each contributing to a different stage of text analytics and business intelligence.

---

## NLP Modules Implemented

### Module 1 — Text Cleaning
Cleaned raw descriptions by removing unnecessary symbols, extra spaces, line breaks, emojis, and unwanted characters while preserving meaningful textual information.

---

### Module 2 — URL & Metadata Extraction
Extracted URLs, hashtags, and social media links (Instagram, YouTube, Twitter/X, Facebook, TikTok, Discord, Spotify, Apple Music, Twitch, Steam, Epic Games, etc.) into separate structured columns for business analysis.

---

### Module 3 — Tokenization
Split cleaned text into individual words (tokens), making the textual data suitable for Natural Language Processing and feature engineering.

---

### Module 4 — Stopword Removal
Removed common English stopwords (such as "the", "is", "and", "of") to retain only informative and meaningful words for analysis.

---

### Module 5 — Lemmatization
Converted words into their root forms (e.g., "playing" → "play", "games" → "game") to improve text consistency and reduce vocabulary size.

---

### Module 6 — Word Frequency Analysis
Calculated the frequency of each word to identify the most commonly used terms across YouTube video descriptions and understand dominant content themes.

---

### Module 7 — Word Cloud Visualization
Generated a Word Cloud to visually represent the most frequently occurring keywords, providing a quick overview of trending topics and creator content strategies.

---

### Module 8 — Bigram & Trigram Analysis
Extracted two-word and three-word phrases to identify meaningful keyword combinations such as "official trailer", "music video", and "live stream", helping analyze common promotional and SEO phrases.

---

### Module 9 — TF-IDF Keyword Analysis
Applied TF-IDF (Term Frequency–Inverse Document Frequency) to identify the most important and unique keywords instead of simply the most frequent ones. This highlights category-specific and content-specific terms useful for SEO and recommendation systems.

---

### Module 10 — Keyword Intelligence & Category Analysis
Performed category-wise keyword analysis to discover how different YouTube categories use different vocabularies, marketing strategies, and content themes. This supports content optimization, trend analysis, and business decision-making.

---

# Business Value

The NLP pipeline transformed raw YouTube descriptions into structured, analyzable features, enabling advanced text analytics and business intelligence. These processed textual features support SEO optimization, keyword discovery, creator strategy analysis, recommendation systems, content intelligence, dashboard reporting, and data-driven decision-making.

Overall, the NLP workflow significantly enhanced the quality of the dataset by converting unstructured textual information into valuable business insights for YouTube Trending Video Analytics.

````python
# IQR Method for Viral Videos
Q1 = df["view_count"].quantile(0.25)
Q3 = df["view_count"].quantile(0.75)

IQR = Q3 - Q1

upper_limit = Q3 + 1.5 * IQR

viral_analysis = (

    df

    .assign(

        Viral_Video=lambda x: np.where(
            x["view_count"] > upper_limit,
            "Yes",
            "No"
        )

    )

    .groupby("category_name", as_index=False)

    .agg(

        Total_Videos=("video_id","count"),

        Viral_Videos=("Viral_Video", lambda x: (x=="Yes").sum()),

        Average_Views=("view_count","mean"),

        Maximum_Views=("view_count","max")

    )

)

viral_analysis["Viral_%"]=(

    viral_analysis["Viral_Videos"]

    /

    viral_analysis["Total_Videos"]

    *100

).round(2)

viral_analysis = (

    viral_analysis

    .sort_values(

        "Viral_%",

        ascending=False

    )

)

display(viral_analysis)


plt.figure(figsize=(14,8))

plt.hlines(

    y=viral_analysis["category_name"],

    xmin=0,

    xmax=viral_analysis["Viral_%"],

    color="lightgray",

    linewidth=3

)

plt.scatter(

    viral_analysis["Viral_%"],

    viral_analysis["category_name"],

    s=180,

    color="crimson"

)

plt.title(

    "Percentage of Viral Videos by YouTube Category",

    fontsize=18,

    fontweight="bold"

)

plt.xlabel("Viral Videos (%)")

plt.ylabel("Category")

plt.grid(axis="x", alpha=0.3)

plt.tight_layout()

plt.show()
````

## Statistical Method's for Hypothesis Testing

````python
# ==========================================================
# Statistical Test 1
# Independent Samples t-Test
# ==========================================================

from scipy.stats import ttest_ind
import pandas as pd

# HD Videos
hd = (

    df

    .query("definition == 'hd'")

    ["engagement_rate"]

    .dropna()

)

# SD Videos
sd = (

    df

    .query("definition == 'sd'")

    ["engagement_rate"]

    .dropna()

)

# Independent t-Test
t_statistic, p_value = ttest_ind(

    hd,

    sd,

    equal_var=False

)

# Summary Table
summary = pd.DataFrame({

    "Metric":[

        "HD Sample Size",

        "SD Sample Size",

        "HD Mean Engagement",

        "SD Mean Engagement",

        "HD Std",

        "SD Std",

        "T Statistic",

        "P Value"

    ],

    "Value":[

        len(hd),

        len(sd),

        round(hd.mean(),2),

        round(sd.mean(),2),

        round(hd.std(),2),

        round(sd.std(),2),

        round(t_statistic,4),

        round(p_value,6)

    ]

})

alpha = 0.05

if p_value < alpha:

    print("Reject the Null Hypothesis (H₀)")
    print("There is a statistically significant difference between HD and SD engagement.")

else:

    print("Fail to Reject the Null Hypothesis (H₀)")
    print("No statistically significant difference exists between HD and SD engagement.")

display(summary)

plt.figure(figsize=(8,6))

sns.boxplot(

    data=df,

    x="definition",

    y="engagement_rate",

    palette="Set2"

)

sns.stripplot(

    data=df,

    x="definition",

    y="engagement_rate",

    color="black",

    alpha=0.35,

    jitter=True

)

plt.title(

    "HD vs SD Engagement Rate",

    fontsize=16,

    fontweight="bold"

)

plt.xlabel("Video Definition")

plt.ylabel("Engagement Rate")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
````

````python
# ==========================================================
# Statistical Test 2
# One-Way ANOVA
# ==========================================================

from scipy.stats import f_oneway
import pandas as pd

# Create groups for each category
groups = [

    group["engagement_rate"].dropna().values

    for _, group in df.groupby("category_name")

]

# Run ANOVA
f_statistic, p_value = f_oneway(*groups)

# Summary Table

anova_summary = (

    df

    .groupby("category_name", as_index=False)

    .agg(

        Total_Videos=("video_id","count"),

        Average_Engagement=("engagement_rate","mean"),

        Median_Engagement=("engagement_rate","median"),

        Std_Engagement=("engagement_rate","std")

    )

)

anova_summary = (

    anova_summary

    .round(2)

    .sort_values(

        "Average_Engagement",

        ascending=False

    )

)


print("\n")

print(f"F Statistic : {f_statistic:.4f}")

print(f"P Value     : {p_value:.6f}")

alpha = 0.05

if p_value < alpha:

    print("✅ Reject the Null Hypothesis (H₀)")
    print("Audience engagement differs significantly across categories.")

else:

    print("❌ Fail to Reject the Null Hypothesis (H₀)")
    print("No significant difference in engagement across categories.")

display(anova_summary)

plt.figure(figsize=(16,7))

sns.pointplot(

    data=df,

    x="category_name",

    y="engagement_rate",

    estimator="mean",

    errorbar=("ci",95),

    capsize=0.2,

    color="royalblue"

)

plt.title(

    "Average Engagement Rate Across YouTube Categories",

    fontsize=18,

    fontweight="bold"

)

plt.xlabel("YouTube Category")

plt.ylabel("Average Engagement Rate")

plt.xticks(rotation=45)

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
````

````python
# ==========================================================
# Statistical Test 3
# Tukey HSD Post-Hoc Test
# ==========================================================

from statsmodels.stats.multicomp import pairwise_tukeyhsd
import pandas as pd

# Tukey HSD Test
tukey = pairwise_tukeyhsd(

    endog=df["engagement_rate"],

    groups=df["category_name"],

    alpha=0.05

)

# Convert result into DataFrame
tukey_result = pd.DataFrame(

    tukey.summary().data[1:],

    columns=tukey.summary().data[0]

)

display(tukey_result)

plt.figure(figsize=(12,8))

tukey.plot_simultaneous()

plt.title(

    "Tukey HSD Multiple Comparison of Category Engagement",

    fontsize=16,

    fontweight="bold"

)

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
````

````python
# ==========================================================
# Statistical Test 4
# Pearson Correlation Test
# ==========================================================

from scipy.stats import pearsonr
import pandas as pd

# Pearson Correlation
correlation, p_value = pearsonr(

    df["view_count"],

    df["like_count"]

)

# Executive Summary

pearson_summary = pd.DataFrame({

    "Metric":[

        "Correlation Coefficient (r)",

        "P Value",

        "Total Videos",

        "Average Views",

        "Average Likes"

    ],

    "Value":[

        round(correlation,4),

        round(p_value,6),

        len(df),

        round(df["view_count"].mean(),2),

        round(df["like_count"].mean(),2)

    ]

})

alpha = 0.05

print(f"Correlation (r): {correlation:.4f}")
print(f"P-value: {p_value:.6f}")

if p_value < alpha:

    print("\nReject the Null Hypothesis (H₀)")
    print("There is a statistically significant linear relationship between Views and Likes.")

else:

    print("\nFail to Reject the Null Hypothesis (H₀)")
    print("No statistically significant linear relationship exists.")

display(pearson_summary)


plt.figure(figsize=(10,7))

sns.regplot(

    data=df,

    x="view_count",

    y="like_count",

    scatter_kws={

        "alpha":0.35,

        "s":40

    },

    line_kws={

        "color":"red",

        "linewidth":3

    }

)

plt.title(

    "Relationship Between Views and Likes",

    fontsize=18,

    fontweight="bold"

)

plt.xlabel("View Count")

plt.ylabel("Like Count")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
````

````python
# ==========================================================
# Statistical Test 5
# Spearman Rank Correlation
# ==========================================================

from scipy.stats import spearmanr
import pandas as pd

# Spearman Correlation
correlation, p_value = spearmanr(

    df["title_length"],

    df["engagement_rate"]

)

# Executive Summary
spearman_summary = pd.DataFrame({

    "Metric":[

        "Spearman Correlation (ρ)",

        "P Value",

        "Total Videos",

        "Average Title Length",

        "Average Engagement"

    ],

    "Value":[

        round(correlation,4),

        round(p_value,6),

        len(df),

        round(df["title_length"].mean(),2),

        round(df["engagement_rate"].mean(),2)

    ]

})

alpha = 0.05

print(f"Spearman Correlation : {correlation:.4f}")
print(f"P-value              : {p_value:.6f}")

if p_value < alpha:

    print("\n✅ Reject the Null Hypothesis (H₀)")
    print("A statistically significant relationship exists between title length and engagement.")

else:

    print("\n❌ Fail to Reject the Null Hypothesis (H₀)")
    print("No statistically significant relationship exists.")

display(spearman_summary)


plt.figure(figsize=(10,7))

sns.regplot(

    data=df,

    x="title_length",

    y="engagement_rate",

    scatter_kws={

        "alpha":0.5,

        "s":50

    },

    line_kws={

        "color":"red",

        "linewidth":3

    }

)

plt.title(

    "Title Length vs Engagement Rate",

    fontsize=16,

    fontweight="bold"

)

plt.xlabel("Title Length (Characters)")

plt.ylabel("Engagement Rate")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
````

````python
# ==========================================================
# Statistical Test 6
# Shapiro-Wilk Normality Test
# ==========================================================

from scipy.stats import shapiro
import pandas as pd

# Remove missing values
engagement = df["engagement_rate"].dropna()

# Shapiro-Wilk Test
statistic, p_value = shapiro(engagement)

# Executive Summary
normality_summary = pd.DataFrame({

    "Metric":[

        "Sample Size",

        "Average Engagement",

        "Median Engagement",

        "Standard Deviation",

        "Shapiro Statistic",

        "P Value"

    ],

    "Value":[

        len(engagement),

        round(engagement.mean(),2),

        round(engagement.median(),2),

        round(engagement.std(),2),

        round(statistic,4),

        round(p_value,6)

    ]

})

alpha = 0.05

print(f"Shapiro Statistic : {statistic:.4f}")
print(f"P-value           : {p_value:.6f}")

if p_value > alpha:

    print("\n✅ Fail to Reject H₀")
    print("Engagement Rate is approximately normally distributed.")

else:

    print("\n❌ Reject H₀")
    print("Engagement Rate is NOT normally distributed.")
    
display(normality_summary)

plt.figure(figsize=(10,6))

sns.histplot(

    engagement,

    bins=30,

    kde=True,

    color="royalblue"

)

plt.title(

    "Distribution of Engagement Rate",

    fontsize=16,

    fontweight="bold"

)

plt.xlabel("Engagement Rate")

plt.ylabel("Frequency")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
````

````python
# ==========================================================
# Statistical Test 7
# Levene's Test (Homogeneity of Variance)
# ==========================================================

from scipy.stats import levene
import pandas as pd

# Create engagement groups by upload session
groups = [

    group["engagement_rate"].dropna().values

    for _, group in df.groupby("publish_session")

]

# Levene Test
statistic, p_value = levene(*groups)

# Executive Summary
levene_summary = (

    df

    .groupby("publish_session", as_index=False)

    .agg(

        Total_Videos=("video_id","count"),

        Average_Engagement=("engagement_rate","mean"),

        Median_Engagement=("engagement_rate","median"),

        Std_Engagement=("engagement_rate","std")

    )

)

levene_summary = levene_summary.round(2)

print(f"\nLevene Statistic : {statistic:.4f}")
print(f"P Value          : {p_value:.6f}")

alpha = 0.05

print(f"Levene Statistic : {statistic:.4f}")
print(f"P-value          : {p_value:.6f}")

if p_value > alpha:

    print("\n✅ Fail to Reject H₀")
    print("The variances are equal across upload sessions.")

else:

    print("\n❌ Reject H₀")
    print("The variances are not equal across upload sessions.")

display(levene_summary)

plt.figure(figsize=(10,6))

sns.boxplot(

    data=df,

    x="publish_session",

    y="engagement_rate",

    hue="publish_session",

    palette="Set2",

    legend=False

)

plt.title(

    "Engagement Rate Distribution Across Upload Sessions",

    fontsize=16,

    fontweight="bold"

)

plt.xlabel("Upload Session")

plt.ylabel("Engagement Rate")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.show()
````

````python
# ==========================================================
# Statistical Test 8
# Chi-Square Test of Independence
# ==========================================================

from scipy.stats import chi2_contingency
import pandas as pd

# Contingency Table
contingency_table = pd.crosstab(

    df["publish_session"],

    df["duration_category"]

)

# Chi-Square Test
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

alpha = 0.05

print(f"P-value : {p_value:.6f}")

if p_value < alpha:

    print("\n✅ Reject H₀")
    print("There is a significant association between upload session and duration category.")

else:

    print("\n❌ Fail to Reject H₀")
    print("No significant association exists between upload session and duration category.")

print("\n")

print(f"Chi-Square Statistic : {chi2:.4f}")
print(f"Degrees of Freedom   : {dof}")
print(f"P Value              : {p_value:.6f}")

print("Contingency Table\n")
display(contingency_table)

plt.figure(figsize=(8,6))

sns.heatmap(

    contingency_table,

    annot=True,

    fmt="d",

    cmap="YlGnBu",

    linewidths=0.5

)

plt.title(

    "Upload Session vs Duration Category",

    fontsize=16,

    fontweight="bold"

)

plt.xlabel("Duration Category")

plt.ylabel("Upload Session")

plt.tight_layout()

plt.show()
````

````python
# ==========================================================
# Statistical Test 9
# Mann-Whitney U Test
# ==========================================================

from scipy.stats import mannwhitneyu
import pandas as pd

# Caption Enabled Videos
caption_yes = (

    df

    .query("has_caption == True")

    ["engagement_rate"]

    .dropna()

)

# Caption Disabled Videos
caption_no = (

    df

    .query("has_caption == False")

    ["engagement_rate"]

    .dropna()

)

# Mann-Whitney U Test
u_statistic, p_value = mannwhitneyu(

    caption_yes,

    caption_no,

    alternative="two-sided"

)

# Executive Summary
mann_summary = pd.DataFrame({

    "Metric":[

        "Caption Videos",

        "No Caption Videos",

        "Caption Mean Engagement",

        "No Caption Mean Engagement",

        "U Statistic",

        "P Value"

    ],

    "Value":[

        len(caption_yes),

        len(caption_no),

        round(caption_yes.mean(),2),

        round(caption_no.mean(),2),

        round(u_statistic,2),

        round(p_value,6)

    ]

})

alpha = 0.05

print(f"U Statistic : {u_statistic:.2f}")
print(f"P-value     : {p_value:.6f}")

if p_value < alpha:

    print("\n✅ Reject H₀")
    print("Caption availability significantly affects engagement.")

else:

    print("\n❌ Fail to Reject H₀")
    print("No statistically significant difference in engagement.")

display(mann_summary)


plt.figure(figsize=(8,6))

sns.violinplot(

    data=df,

    x="has_caption",

    y="engagement_rate",

    hue="has_caption",

    palette="Set2",

    inner="quartile",

    legend=False

)

plt.title(

    "Engagement Rate: Caption vs No Caption",

    fontsize=16,

    fontweight="bold"

)

plt.xlabel("Has Caption")

plt.ylabel("Engagement Rate")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.show()
````

````python
# ==========================================================
# Statistical Test 10
# Kruskal-Wallis Test
# ==========================================================

from scipy.stats import kruskal
import pandas as pd

# Create groups
groups = [

    group["engagement_rate"].dropna().values

    for _, group in df.groupby("duration_category")

]

# Kruskal-Wallis Test
h_statistic, p_value = kruskal(*groups)

# Executive Summary
kruskal_summary = (

    df

    .groupby("duration_category", as_index=False)

    .agg(

        Total_Videos=("video_id","count"),

        Average_Engagement=("engagement_rate","mean"),

        Median_Engagement=("engagement_rate","median"),

        Std_Engagement=("engagement_rate","std"),

        Average_Views=("view_count","mean")

    )

)

kruskal_summary = (

    kruskal_summary

    .round(2)

    .sort_values(

        "Average_Engagement",

        ascending=False

    )

)

print(f"\nH Statistic : {h_statistic:.4f}")
print(f"P Value     : {p_value:.6f}")

alpha = 0.05

print(f"H Statistic : {h_statistic:.4f}")
print(f"P-value     : {p_value:.6f}")

if p_value < alpha:

    print("\n✅ Reject H₀")
    print("Audience engagement differs significantly across duration categories.")

else:

    print("\n❌ Fail to Reject H₀")
    print("No statistically significant difference in engagement across duration categories.")

display(kruskal_summary)

plt.figure(figsize=(10,6))

sns.boxenplot(

    data=df,

    x="duration_category",

    y="engagement_rate",

    hue="duration_category",

    palette="viridis",

    legend=False

)

plt.title(

    "Audience Engagement Across Video Duration Categories",

    fontsize=16,

    fontweight="bold"

)

plt.xlabel("Duration Category")

plt.ylabel("Engagement Rate")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.show()
````
