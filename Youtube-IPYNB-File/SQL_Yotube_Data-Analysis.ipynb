# ADVANCED STATISTICS — 41 to 50

```python
# 41. Views variance

pd.read_sql_query("""
SELECT
    AVG(view_count * view_count)
    - AVG(view_count) * AVG(view_count)
    AS view_variance
FROM trending_videos;
""", conn)
```

```python
# 42. Views standard deviation

pd.read_sql_query("""
SELECT
    SQRT(
        AVG(view_count * view_count)
        - AVG(view_count) * AVG(view_count)
    ) AS view_std_dev
FROM trending_videos;
""", conn)
```

```python
# 43. Pearson correlation: views vs likes

pd.read_sql_query("""
SELECT
(
    SUM(
        (view_count - v.avg_views)
        * (like_count - v.avg_likes)
    )
)
/
(
    SQRT(
        SUM((view_count - v.avg_views) * (view_count - v.avg_views))
        *
        SUM((like_count - v.avg_likes) * (like_count - v.avg_likes))
    )
) AS pearson_views_likes
FROM trending_videos
CROSS JOIN (
    SELECT
        AVG(view_count) AS avg_views,
        AVG(like_count) AS avg_likes
    FROM trending_videos
) v;
""", conn)
```

```python
# 44. Pearson correlation: views vs comments

pd.read_sql_query("""
SELECT
(
    SUM(
        (view_count - v.avg_views)
        * (comment_count - v.avg_comments)
    )
)
/
(
    SQRT(
        SUM((view_count - v.avg_views) * (view_count - v.avg_views))
        *
        SUM((comment_count - v.avg_comments) * (comment_count - v.avg_comments))
    )
) AS pearson_views_comments
FROM trending_videos
CROSS JOIN (
    SELECT
        AVG(view_count) AS avg_views,
        AVG(comment_count) AS avg_comments
    FROM trending_videos
) v;
""", conn)
```

```python
# 45. Covariance: views vs likes

pd.read_sql_query("""
SELECT
    ROUND(AVG(
        (view_count - v.avg_views)
        *
        (like_count - v.avg_likes)
    ),3) AS covariance_views_likes
FROM trending_videos
CROSS JOIN (
    SELECT
        AVG(view_count) AS avg_views,
        AVG(like_count) AS avg_likes
    FROM trending_videos
) v;
""", conn)
```

```python
# 46. Covariance: views vs comments

pd.read_sql_query("""
SELECT
    AVG(
        (view_count - v.avg_views)
        *
        (comment_count - v.avg_comments)
    ) AS covariance_views_comments
FROM trending_videos
CROSS JOIN (
    SELECT
        AVG(view_count) AS avg_views,
        AVG(comment_count) AS avg_comments
    FROM trending_videos
) v;
""", conn)
```

```python
# 47. Top videos by engagement rate

pd.read_sql_query("""
SELECT
    video_id,
    title,
    channel_title,
    engagement_rate
FROM trending_videos
ORDER BY engagement_rate DESC
LIMIT 10;
""", conn)
```

```python
# 48. Top channels by average views

pd.read_sql_query("""
SELECT
    channel_title,
    COUNT(*) AS total_videos,
    ROUND(AVG(view_count), 2) AS average_views
FROM trending_videos
GROUP BY channel_title
HAVING COUNT(*) >= 2
ORDER BY average_views DESC
LIMIT 10;
""", conn)
```

```python
# 49. Category performance summary

pd.read_sql_query("""
SELECT
    category_name,
    COUNT(*) AS total_videos,
    ROUND(AVG(view_count), 2) AS avg_views,
    ROUND(AVG(like_count), 2) AS avg_likes,
    ROUND(AVG(comment_count), 2) AS avg_comments,
    ROUND(AVG(engagement_rate), 2) AS avg_engagement
FROM trending_videos
GROUP BY category_name
ORDER BY avg_engagement DESC;
""", conn)
```

```python
# 50. Overall performance summary

pd.read_sql_query("""
SELECT
    COUNT(*) AS total_videos,
    COUNT(DISTINCT channel_title) AS total_channels,
    COUNT(DISTINCT category_name) AS total_categories,

    SUM(view_count) AS total_views,

    ROUND(AVG(view_count), 2) AS avg_views,
    ROUND(AVG(like_count), 2) AS avg_likes,
    ROUND(AVG(comment_count), 2) AS avg_comments,

    ROUND(AVG(engagement_rate), 2) AS avg_engagement_rate,
    ROUND(AVG(like_rate), 2) AS avg_like_rate,
    ROUND(AVG(comment_rate), 2) AS avg_comment_rate

FROM trending_videos;
""", conn)
```

```python
# Which videos have views greater than average views?

pd.read_sql_query("""
SELECT
    video_id,
    title,
    view_count
FROM trending_videos
WHERE view_count > (
    SELECT AVG(view_count)
    FROM trending_videos
)
ORDER BY view_count DESC;
""", conn)
```

```python
# Which videos have engagement rate greater than the average engagement rate?

pd.read_sql_query("""
SELECT
    video_id,
    title,
    engagement_rate
FROM trending_videos
WHERE engagement_rate > (
    SELECT AVG(engagement_rate)
    FROM trending_videos
)
ORDER BY engagement_rate DESC;
""", conn)
```

```python
# Calculate the percentage of videos having more than 1 million views.

pd.read_sql_query("""
SELECT
    COUNT(*) AS total_videos,
    SUM(
        CASE
            WHEN view_count > 1000000 THEN 1
            ELSE 0
        END
    ) AS high_view_videos,
    ROUND(
        SUM(
            CASE
                WHEN view_count > 1000000 THEN 1
                ELSE 0
            END
        ) * 100.0 / COUNT(*),
        2
    ) AS high_view_rate
FROM trending_videos;
""", conn)
```

```python
# Calculate the percentage of videos having engagement rate above 5%.

pd.read_sql_query("""
SELECT
    COUNT(*) AS total_videos,
    SUM(
        CASE
            WHEN engagement_rate > 5 THEN 1
            ELSE 0
        END
    ) AS high_engagement_videos,
    ROUND(
        SUM(
            CASE
                WHEN engagement_rate > 5 THEN 1
                ELSE 0
            END
        ) * 100.0 / COUNT(*),
        2
    ) AS high_engagement_rate
FROM trending_videos;
""", conn)
```

```python
# Classify videos based on their views.

pd.read_sql_query("""
SELECT
    video_id,
    title,
    view_count,
    CASE
        WHEN view_count >= 1000000 THEN 'High'
        WHEN view_count >= 100000 THEN 'Medium'
        ELSE 'Low'
    END AS view_performance
FROM trending_videos
ORDER BY view_count DESC;
""", conn)
```

```python
# Find videos having high views but below-average engagement.

pd.read_sql_query("""
SELECT
    video_id,
    title,
    view_count,
    engagement_rate
FROM trending_videos
WHERE view_count > (
    SELECT AVG(view_count)
    FROM trending_videos
)
AND engagement_rate < (
    SELECT AVG(engagement_rate)
    FROM trending_videos
)
ORDER BY view_count DESC;
""", conn)
```

```python
# What is the average number of hashtags for each category?

display(pd.read_sql_query("""
SELECT
    category_name,
    COUNT(*) AS total_videos,
    ROUND(AVG(hashtag_count), 2) AS avg_hashtags
FROM trending_videos
GROUP BY category_name
ORDER BY avg_hashtags DESC
""", conn))
```

```python
# What is the average number of tags for each category?

display(pd.read_sql_query("""
SELECT
    category_name,
    COUNT(*) AS total_videos,
    ROUND(AVG(tag_count), 2) AS avg_tags
FROM trending_videos
GROUP BY category_name
ORDER BY avg_tags DESC
""", conn))
```

```python
# Which categories create longer titles on average?

display(pd.read_sql_query("""
SELECT
    category_name,
    ROUND(AVG(title_length), 2) AS avg_title_length,
    ROUND(AVG(title_word_count), 2) AS avg_title_words
FROM trending_videos
GROUP BY category_name
ORDER BY avg_title_length DESC
""", conn))
```

```python
# Which categories provide longer descriptions on average?

display(pd.read_sql_query("""
SELECT
    category_name,
    ROUND(AVG(description_length), 2) AS avg_description_length,
    ROUND(AVG(description_word_count), 2) AS avg_description_words
FROM trending_videos
GROUP BY category_name
ORDER BY avg_description_length DESC
""", conn))
```

```python
# Does a higher number of hashtags correspond to higher average views?

display(pd.read_sql_query("""
SELECT
    hashtag_count,
    COUNT(*) AS total_videos,
    ROUND(AVG(view_count), 2) AS avg_views
FROM trending_videos
GROUP BY hashtag_count
ORDER BY hashtag_count
""", conn))
```

```python
# Does a higher number of tags correspond to higher average engagement?

display(pd.read_sql_query("""
SELECT
    tag_count,
    COUNT(*) AS total_videos,
    ROUND(AVG(engagement_rate), 2) AS avg_engagement
FROM trending_videos
GROUP BY tag_count
ORDER BY tag_count
""", conn))
```

```python
# Do longer titles receive higher average engagement?

display(pd.read_sql_query("""
SELECT
    CASE
        WHEN title_length < 50 THEN 'Short Title'
        WHEN title_length BETWEEN 50 AND 100 THEN 'Medium Title'
        ELSE 'Long Title'
    END AS title_type,
    COUNT(*) AS total_videos,
    ROUND(AVG(view_count), 2) AS avg_views,
    ROUND(AVG(engagement_rate), 2) AS avg_engagement
FROM trending_videos
GROUP BY title_type
ORDER BY avg_engagement DESC
""", conn))
```

```python
# Do longer descriptions receive higher average views?

display(pd.read_sql_query("""
SELECT
    CASE
        WHEN description_length < 500 THEN 'Short Description'
        WHEN description_length BETWEEN 500 AND 1500 THEN 'Medium Description'
        ELSE 'Long Description'
    END AS description_type,
    COUNT(*) AS total_videos,
    ROUND(AVG(view_count), 2) AS avg_views
FROM trending_videos
GROUP BY description_type
ORDER BY avg_views DESC
""", conn))
```

```python
# Which video duration categories have the highest average engagement?

display(pd.read_sql_query("""
SELECT
    duration_category,
    COUNT(*) AS total_videos,
    ROUND(AVG(engagement_rate), 2) AS avg_engagement
FROM trending_videos
GROUP BY duration_category
ORDER BY avg_engagement DESC
""", conn))
```

```python
# Which duration categories generate the highest average views?

display(pd.read_sql_query("""
SELECT
    duration_category,
    COUNT(*) AS total_videos,
    ROUND(AVG(view_count), 2) AS avg_views
FROM trending_videos
GROUP BY duration_category
ORDER BY avg_views DESC
""", conn))
```

```python
# Which upload sessions have the highest average hashtag usage?

display(pd.read_sql_query("""
SELECT
    publish_session,
    COUNT(*) AS total_videos,
    ROUND(AVG(hashtag_count), 2) AS avg_hashtags,
    ROUND(AVG(tag_count), 2) AS avg_tags
FROM trending_videos
GROUP BY publish_session
ORDER BY avg_hashtags DESC
""", conn))
```

```python
# Which video categories have the highest URL usage?

display(pd.read_sql_query("""
SELECT
    category_name,
    COUNT(*) AS total_videos,
    SUM(CASE WHEN url_count > 0 THEN 1 ELSE 0 END) AS videos_with_urls,
    ROUND(
        SUM(CASE WHEN url_count > 0 THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS url_usage_rate
FROM trending_videos
GROUP BY category_name
ORDER BY url_usage_rate DESC
""", conn))
```

```python
# Which channels use the most hashtags per video on average?

display(pd.read_sql_query("""
SELECT
    channel_title,
    COUNT(*) AS total_videos,
    ROUND(AVG(hashtag_count), 2) AS avg_hashtags_per_video
FROM trending_videos
GROUP BY channel_title
HAVING COUNT(*) >= 2
ORDER BY avg_hashtags_per_video DESC
LIMIT 10
""", conn))
```

```python
# Which channels use the most tags per video on average?

display(pd.read_sql_query("""
SELECT
    channel_title,
    COUNT(*) AS total_videos,
    ROUND(AVG(tag_count), 2) AS avg_tags_per_video
FROM trending_videos
GROUP BY channel_title
HAVING COUNT(*) >= 2
ORDER BY avg_tags_per_video DESC
LIMIT 10
""", conn))
```

```python
# How does title length relate to like rate?

display(pd.read_sql_query("""
SELECT
    CASE
        WHEN title_length < 50 THEN 'Short'
        WHEN title_length BETWEEN 50 AND 100 THEN 'Medium'
        ELSE 'Long'
    END AS title_length_group,
    COUNT(*) AS total_videos,
    ROUND(AVG(like_rate), 2) AS avg_like_rate,
    ROUND(AVG(comment_rate), 2) AS avg_comment_rate
FROM trending_videos
GROUP BY title_length_group
ORDER BY avg_like_rate DESC
""", conn))
```
