## 🎯 Project Overview

This project is a complete end-to-end **Electric Vehicle Intelligence Platform** built using real-world public EV data.

The objective is to transform raw electric vehicle data into meaningful business intelligence through a complete data workflow.

The project covers the complete lifecycle:

**Data Collection → Data Validation → Data Engineering → SQL → Data Cleaning → EDA → Statistics → Visualization → Feature Engineering → Machine Learning → Explainable AI → Dashboard → Deployment**

---

## 💼 Business Problem

The electric vehicle market is developing rapidly across different countries and regions.

However, raw EV data can be difficult to understand because information may contain:

- Multiple countries
- Different vehicle categories
- Different years
- EV sales
- EV stock
- EV adoption rates
- Charging infrastructure
- Energy-related indicators
- Historical observations
- Scenario-based values

Businesses and decision-makers need a structured analytical system that can transform this raw information into useful insights.

### The key questions are:

- Which countries have the highest EV adoption?
- How are EV sales changing over time?
- Which vehicle categories are growing fastest?
- Which regions are leading the EV transition?
- How is charging infrastructure developing?
- Which markets have strong EV growth?
- What relationship exists between EV adoption and charging infrastructure?
- Which factors are associated with EV market development?
- Can historical EV data support useful Machine Learning tasks?

`````python
for col in df.columns:
    count = df[col].value_counts(dropna=False)
    percent = df[col].value_counts(normalize=True, dropna=False) * 100

    result = pd.DataFrame({
        "Count": count,
        "Percentage": percent.round(2)
    })

    print(f"\n{'=' * 60}")
    print(f"Column: {col}")
    display(result)
`````

`````python
for col in df.columns:
    count = df[col].value_counts(dropna=False)
    
    result = pd.DataFrame({
        "Count": count,
        "Percentage": (count / len(df) * 100).round(2),
        "Cumulative %": (count / len(df) * 100).cumsum().round(2)
    })

    print(f"\n{'=' * 60}")
    print(f"Column: {col}")
    display(result)
`````
