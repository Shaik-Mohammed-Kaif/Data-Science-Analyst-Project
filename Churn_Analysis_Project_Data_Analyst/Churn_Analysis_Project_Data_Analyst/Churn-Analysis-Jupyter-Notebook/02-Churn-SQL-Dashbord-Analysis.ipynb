# SQL Business Analysis — Customer Churn & Retention Analytics

## Business Objective

The objective of this SQL analysis is to transform customer, subscription, transaction, and support data into **actionable business intelligence**.

The analysis focuses on understanding **customer behavior, churn drivers, revenue performance, retention opportunities, and financial risk** so that business stakeholders can make data-driven decisions.

The key business question is:

> **Who is churning, why are they churning, how much revenue is at risk, where is the problem concentrated, and which customers should the business prioritize for retention?**

---

## Business Problems to Solve

The SQL analysis will answer the following business questions:

### 1. Customer Health

* How large is the current customer base?
* How many customers are retained versus churned?
* What is the overall churn rate?
* What is the overall retention rate?
* Which customer groups have the highest churn?

### 2. Revenue Performance

* What is the total revenue generated?
* What is the net revenue?
* What is the average revenue per customer?
* Which states generate the most revenue?
* Which contracts generate the highest revenue?
* Which customer segments contribute the most revenue?

### 3. Churn Drivers

* Which contract type has the highest churn?
* Which internet service has the highest churn?
* What are the most common churn categories?
* What are the major churn reasons?
* Does customer tenure influence churn?
* Which customer segments are most vulnerable?

### 4. Geographic Risk

* Which states have the highest customer churn?
* Which states generate the highest revenue?
* Which states have high revenue but poor retention?
* Which geographic regions represent the greatest business risk?

### 5. Customer Value & Retention

* Who are the highest-value customers?
* Which high-value customers have churned?
* Which customers generate the most revenue?
* Which customer groups have high value but low retention?
* Where should retention campaigns be prioritized?

---

# Key Business KPIs

The SQL analysis will calculate and monitor:

| KPI                      | Business Meaning                          |
| ------------------------ | ----------------------------------------- |
| **Total Customers**      | Overall customer base                     |
| **Churned Customers**    | Number of customers lost                  |
| **Churn Rate %**         | Percentage of customers who churned       |
| **Retention Rate %**     | Percentage of customers retained          |
| **Total Revenue**        | Overall revenue generated                 |
| **Net Revenue**          | Revenue after applicable adjustments      |
| **Revenue per Customer** | Average customer value                    |
| **Average Tenure**       | Average customer relationship duration    |
| **Revenue at Risk**      | Revenue associated with churned customers |
| **High-Value Churn**     | Valuable customers lost to churn          |

````python
# Fetch all table names
tables = pd.read_sql(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)

# Print table names and columns
for table_name in tables["name"]:

    print("=" * 60)
    print(f"📋 Table Name : {table_name}")
    print("=" * 60)

    columns = pd.read_sql(
        f"PRAGMA table_info({table_name});",
        conn
    )

    print(columns["name"].to_list())
    print()

# Close connection
conn.close()
````

````python
# Connect Database
conn = sqlite3.connect("customer_data.db")

# Get Table Names
tables = pd.read_sql("""
SELECT name
FROM sqlite_master
WHERE type='table';
""", conn)

print(f"Total Tables : {len(tables)}\n")

for table in tables["name"]:

    print("=" * 70)
    print(f"📂 TABLE : {table.upper()}")
    print("=" * 70)

    schema = pd.read_sql(
        f"PRAGMA table_info({table});",
        conn
    )

    print(schema[["name", "type"]])

    print("\n")

conn.close()
````

## CHURN BUSINESS METRICS

````python
churn_rate = df.Churn_Category.value_counts(normalize = True).mul(100).round(2).map(lambda x : f'{x} %')

print(f"Churn Rate: {churn_rate}")
````

````python
retention_rate = df.Customer_Status.value_counts(normalize = True).mul(100).round(2).map(lambda x : f'{x} %')

print(f"Retention Rate: {retention_rate}")
````

````python
# Overall Customer Status %
status_rate = (
    df["Customer_Status"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

# Overall Churn Rate
churn_rate = status_rate.get("Churned", 0)

# Overall Retention Rate
retention_rate = status_rate.get("Stayed", 0)

#Overall Safe Rate
Safe_rate = status_rate.get("Joined", 0)

# Revenue at Risk = Revenue generated by churned customers
revenue_at_risk = df.loc[
    df["Customer_Status"].eq("Churned"),
    "Total_Revenue"
].sum()

# Total Revenue
total_revenue = df["Total_Revenue"].sum()

# Revenue at Risk %
revenue_at_risk_pct = (
    revenue_at_risk / total_revenue * 100
)

print(f"Overall Churn Rate: {churn_rate:.2f}%")
print(f"Overall Retention Rate: {retention_rate:.2f}%")
print(f"Overall Safe Rate: {Safe_rate:.2f}%")
print(f"Revenue at Risk: ₹{revenue_at_risk:,.2f}")
print(f"Revenue at Risk: {revenue_at_risk_pct:.2f}% of Total Revenue")
````

````python
df.loc[
    df["Customer_Status"].eq("Churned"),
    "Total_Revenue"
].mean()
````

## SORTING + CUMULATIVE ANALYSIS

````python
df.sort_values(
    "Total_Revenue",
    ascending=True
)
````

````python
df_sorted = df.sort_values(
    "Total_Revenue",
    ascending=False
).copy()

df_sorted["Cumulative_Revenue"] = (
    df_sorted["Total_Revenue"].cumsum()
)

display(df_sorted[[
    "Customer_ID",
    "Total_Revenue",
    "Cumulative_Revenue"
]].head(20))
````

````python
df_sorted["Cumulative_Revenue_Pct"] = (
    df_sorted["Cumulative_Revenue"] /
    df_sorted["Total_Revenue"].sum() * 100
)

display(df_sorted.Cumulative_Revenue_Pct)
````

````python
df_sorted["Cumulative_Customers"] = (
    np.arange(1, len(df_sorted) + 1)
)
df_sorted.Cumulative_Customers
````

````python
df_sorted["Revenue_Contribution_Pct"] = (
    df_sorted["Total_Revenue"] /
    df_sorted["Total_Revenue"].sum() * 100
)
df_sorted.Revenue_Contribution_Pct
````

## ROLLING OPERATIONS

````python
df_sorted["Rolling_Avg_Revenue_10"] = (
    df_sorted["Total_Revenue"]
    .rolling(window=10)
    .mean()
)

display(df_sorted[[
    "Customer_ID",
    "Total_Revenue",
    "Rolling_Avg_Revenue_10"
]])
````

## 🎯 Business Analysis Approach

The Customer Churn Intelligence project focuses on understanding
customer behavior from both a customer-retention and financial
perspective.

The analysis does not focus only on the number of churned customers.
Instead, customer churn is evaluated across multiple business
dimensions including:

- Customer demographics
- Customer lifecycle
- Contract structure
- Subscription services
- Payment behavior
- Customer engagement
- Monthly charges
- Total revenue
- Refunds
- Service adoption
- Churn category
- Churn reason

### Business Analysis Framework

Customer Base
        ↓
Customer Segmentation
        ↓
Subscription Behavior
        ↓
Service Usage
        ↓
Customer Engagement
        ↓
Revenue Analysis
        ↓
Churn Analysis
        ↓
Revenue at Risk
        ↓
High-Risk Customer Segments
        ↓
Retention Strategy

### Core Business Objectives

The project aims to answer four major business questions:

1. WHO is churning?
2. WHY are customers churning?
3. WHEN are customers most likely to churn?
4. HOW MUCH revenue is exposed because of churn?

### Decision-Making Perspective

The analysis connects customer-level behavior with business
outcomes.

Customer Behavior → Churn Risk → Revenue Impact → Business Action

This allows stakeholders to move beyond descriptive reporting
and develop data-driven customer retention strategies.

### Key Business Outcomes

The analysis is designed to identify:

- High-churn customer segments
- High-value customers at risk
- Contract types associated with churn
- Customer lifecycle stages with elevated churn
- Services associated with retention or attrition
- Revenue contribution by customer segment
- Revenue associated with churned customers
- Potential revenue at risk
- Customer groups requiring retention campaigns

### Business Value

The final objective is to help organizations:

- Reduce customer attrition
- Protect recurring revenue
- Improve customer lifetime value
- Prioritize retention campaigns
- Improve customer experience
- Optimize subscription strategies
- Identify high-value customers at risk
- Support data-driven management decisions

````python
# ============================================================
# 🎯 BUSINESS QUESTION 1
# ============================================================
# What is the overall customer churn rate, and how does the
# customer base split across Churned, Stayed, and Joined customers?
#
# Business Goal:
# Measure overall customer retention health and understand
# the distribution of customer lifecycle statuses.
# ============================================================

status_count = (
    df["Customer_Status"]
    .value_counts()
    .rename_axis("Customer_Status")
    .reset_index(name="Customer_Count")
)

display(status_count)

status_count["Percentage"] = (
    status_count["Customer_Count"]
    .div(status_count["Customer_Count"].sum())
    .mul(100)
    .round(2)
)

display(status_count)

total_customers = df["Customer_ID"].nunique()

churned_customers = (
    df.loc[
        df["Customer_Status"].eq("Churned"),
        "Customer_ID"
    ].nunique()
)

stayed_customers = (
    df.loc[
        df["Customer_Status"].eq("Stayed"),
        "Customer_ID"
    ].nunique()
)

joined_customers = (
    df.loc[
        df["Customer_Status"].eq("Joined"),
        "Customer_ID"
    ].nunique()
)


churn_rate = (
    churned_customers / total_customers * 100
)

retention_rate = (
    stayed_customers / total_customers * 100
)


print("=" * 70)
print("CUSTOMER CHURN & RETENTION KPI")
print("=" * 70)

print(f"Total Customers     : {total_customers:,}")
print(f"Churned Customers   : {churned_customers:,}")
print(f"Stayed Customers    : {stayed_customers:,}")
print(f"Joined Customers    : {joined_customers:,}")

print(f"Overall Churn Rate  : {churn_rate:.2f}%")
print(f"Retention Rate      : {retention_rate:.2f}%")

print("=" * 70)

summary = pd.DataFrame({
    "Metric": [
        "Total Customers",
        "Churned Customers",
        "Stayed Customers",
        "Joined Customers",
        "Churn Rate",
        "Retention Rate"
    ],
    
    "Value": [
        total_customers,
        churned_customers,
        stayed_customers,
        joined_customers,
        f"{churn_rate:.2f}%",
        f"{retention_rate:.2f}%"
    ]
})

display(summary)

fig = px.bar(
    status_count,
    x="Customer_Status",
    y="Customer_Count",
    text="Customer_Count",
    color="Customer_Status",
    title="Customer Distribution by Status",
    labels={
        "Customer_Status": "Customer Status",
        "Customer_Count": "Number of Customers"
    }
)

fig.update_traces(
    textposition="outside"
)

fig.update_layout(
    template="plotly_white",
    showlegend=False,
    title_x=0.5
)

fig.show()

fig = px.pie(
    status_count,
    names="Customer_Status",
    values="Customer_Count",
    hole=0.55,
    title="Customer Status Distribution"
)

fig.update_traces(
    textinfo="label+percent",
    hovertemplate=(
        "<b>Status:</b> %{label}<br>"
        "<b>Customers:</b> %{value:,}<br>"
        "<b>Share:</b> %{percent}"
        "<extra></extra>"
    )
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()


fig = px.bar(
    status_count.sort_values("Percentage"),
    x="Percentage",
    y="Customer_Status",
    orientation="h",
    text=status_count.sort_values("Percentage")["Percentage"].apply(
        lambda x: f"{x:.2f}%"
    ),
    color="Percentage",
    title="Customer Status Percentage Distribution",
    labels={
        "Percentage": "Customer Share (%)",
        "Customer_Status": "Customer Status"
    }
)

fig.update_traces(
    textposition="outside"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()


revenue_status = (
    df.groupby("Customer_Status", as_index=False)
      .agg(
          Customers=("Customer_ID", "nunique"),
          Total_Revenue=("Total_Revenue", "sum"),
          Avg_Revenue=("Total_Revenue", "mean"),
          Avg_Monthly_Charge=("Monthly_Charge", "mean"),
          Avg_Tenure=("Tenure_in_Months", "mean")
      )
)

revenue_status["Revenue_Share"] = (
    revenue_status["Total_Revenue"]
    .div(revenue_status["Total_Revenue"].sum())
    .mul(100)
    .round(2)
)

display(
    revenue_status.round(2)
)


fig = px.bar(
    revenue_status,
    x="Customer_Status",
    y="Total_Revenue",
    text="Total_Revenue",
    color="Customer_Status",
    title="Total Revenue Contribution by Customer Status",
    labels={
        "Customer_Status": "Customer Status",
        "Total_Revenue": "Total Revenue"
    }
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    template="plotly_white",
    showlegend=False,
    title_x=0.5
)

fig.show()


fig = px.bar(
    revenue_status,
    x="Customer_Status",
    y="Avg_Revenue",
    text="Avg_Revenue",
    color="Avg_Revenue",
    title="Average Customer Revenue by Status",
    labels={
        "Customer_Status": "Customer Status",
        "Avg_Revenue": "Average Revenue"
    }
)

fig.update_traces(
    texttemplate="%{text:,.2f}",
    textposition="outside"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()


churned_revenue = (
    df.loc[
        df["Customer_Status"].eq("Churned"),
        "Total_Revenue"
    ].sum()
)

total_revenue = df["Total_Revenue"].sum()

revenue_at_risk_pct = (
    churned_revenue / total_revenue * 100
)


print("=" * 70)
print("REVENUE RISK ANALYSIS")
print("=" * 70)

print(f"Total Revenue             : {total_revenue:,.2f}")
print(f"Churned Customer Revenue  : {churned_revenue:,.2f}")
print(f"Revenue at Risk (%)       : {revenue_at_risk_pct:.2f}%")

print("=" * 70)


fig = px.pie(
    revenue_status,
    names="Customer_Status",
    values="Total_Revenue",
    hole=0.5,
    title="Revenue Contribution by Customer Status"
)

fig.update_traces(
    textinfo="label+percent",
    hovertemplate=(
        "<b>Status:</b> %{label}<br>"
        "<b>Revenue:</b> %{value:,.2f}<br>"
        "<b>Share:</b> %{percent}"
        "<extra></extra>"
    )
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()


print("\n" + "=" * 70)
print("🧠 BUSINESS INSIGHTS")
print("=" * 70)

if churn_rate >= 30:
    print(
        f"🔴 High churn risk: {churn_rate:.2f}% of customers are churned."
    )

elif churn_rate >= 20:
    print(
        f"🟠 Moderate churn risk: {churn_rate:.2f}% of customers are churned."
    )

else:
    print(
        f"🟢 Relatively controlled churn: {churn_rate:.2f}% of customers are churned."
    )


print(
    f"📊 {stayed_customers:,} customers are currently retained, "
    f"representing {retention_rate:.2f}% of the customer base."
)

print(
    f"💰 Churned customers represent "
    f"{churned_revenue:,.2f} in historical revenue."
)

print(
    f"⚠️ Approximately {revenue_at_risk_pct:.2f}% of total revenue "
    f"is associated with churned customers."
)

print(
    "🎯 Business Action: Prioritize churned/high-risk segments "
    "for retention analysis using contract, tenure, service, "
    "payment and support-related variables."
)

print("=" * 70)
````

````python
# ============================================================
# 🎯 BUSINESS QUESTION 2
# ============================================================

# How does customer tenure influence revenue generation,
# monthly charges, and customer value?

#
# Business Goal:
# Understand whether longer-tenure customers generate higher
# revenue and identify valuable customer lifecycle segments.
# ============================================================


tenure_summary = (
    df.groupby("Tenure_Group", as_index=False)
    .agg(
        Customers=("Customer_ID", "nunique"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Total_Charges=("Total_Charges", "mean"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Refunds=("Total_Refunds", "mean"),
        Avg_Referrals=("Number_of_Referrals", "mean"),
        Churned_Customers=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        )
    )
)

tenure_summary["Churn_Rate"] = (
    tenure_summary["Churned_Customers"]
    .div(tenure_summary["Customers"])
    .mul(100)
    .round(2)
)

tenure_summary["Revenue_Share"] = (
    tenure_summary["Total_Revenue"]
    .div(tenure_summary["Total_Revenue"].sum())
    .mul(100)
    .round(2)
)

display(tenure_summary.round(2))


lifecycle_summary = (
    df.groupby("Customer_Lifecycle", as_index=False)
    .agg(
        Customers=("Customer_ID", "nunique"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Referrals=("Number_of_Referrals", "mean")
    )
)

display(lifecycle_summary.round(2))


tenure_status = (
    pd.crosstab(
        df["Tenure_Group"],
        df["Customer_Status"]
    )
    .reset_index()
)

display(tenure_status)

tenure_status_pct = (
    pd.crosstab(
        df["Tenure_Group"],
        df["Customer_Status"],
        normalize="index"
    )
    .mul(100)
    .round(2)
    .reset_index()
)

display(tenure_status_pct)


highest_revenue_group = tenure_summary.loc[
    tenure_summary["Avg_Revenue"].idxmax(),
    "Tenure_Group"
]

highest_revenue_value = tenure_summary.loc[
    tenure_summary["Avg_Revenue"].idxmax(),
    "Avg_Revenue"
]

highest_churn_group = tenure_summary.loc[
    tenure_summary["Churn_Rate"].idxmax(),
    "Tenure_Group"
]

lowest_churn_group = tenure_summary.loc[
    tenure_summary["Churn_Rate"].idxmin(),
    "Tenure_Group"
]

longest_tenure_group = tenure_summary.loc[
    tenure_summary["Avg_Tenure"].idxmax(),
    "Tenure_Group"
]

kpi_summary = pd.DataFrame({
    "Metric": [
        "Highest Revenue Tenure Group",
        "Highest Average Revenue",
        "Highest Churn Tenure Group",
        "Lowest Churn Tenure Group",
        "Longest Tenure Group"
    ],
    "Value": [
        highest_revenue_group,
        f"{highest_revenue_value:,.2f}",
        highest_churn_group,
        lowest_churn_group,
        longest_tenure_group
    ]
})

display(kpi_summary)

fig = px.bar(
    tenure_summary,
    x="Tenure_Group",
    y="Total_Revenue",
    color="Total_Revenue",
    text="Total_Revenue",
    color_continuous_scale="Viridis",
    title="Total Revenue Contribution by Tenure Group",
    labels={
        "Tenure_Group": "Tenure Group",
        "Total_Revenue": "Total Revenue"
    }
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5
)

fig.show()

fig = px.line(
    tenure_summary,
    x="Tenure_Group",
    y="Avg_Revenue",
    markers=True,
    text="Avg_Revenue",
    title="Average Customer Revenue Across Tenure Groups",
    labels={
        "Tenure_Group": "Tenure Group",
        "Avg_Revenue": "Average Revenue"
    }
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="top center"
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5
)

fig.show()


fig = px.bar(
    tenure_summary.sort_values("Churn_Rate"),
    x="Tenure_Group",
    y="Churn_Rate",
    color="Churn_Rate",
    text="Churn_Rate",
    color_continuous_scale="RdYlGn_r",
    title="Customer Churn Rate by Tenure Group",
    labels={
        "Tenure_Group": "Tenure Group",
        "Churn_Rate": "Churn Rate (%)"
    }
)

fig.update_traces(
    texttemplate="%{text:.2f}%",
    textposition="outside"
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5
)

fig.show()


fig = px.bar(
    tenure_status,
    x="Tenure_Group",
    y=["Churned", "Stayed", "Joined"],
    barmode="group",
    text_auto=True,
    title="Customer Lifecycle Status by Tenure Group",
    labels={
        "Tenure_Group": "Tenure Group",
        "value": "Customers",
        "variable": "Customer Status"
    }
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5
)

fig.show()


status_long = tenure_status_pct.melt(
    id_vars="Tenure_Group",
    var_name="Customer_Status",
    value_name="Percentage"
)

fig = px.bar(
    status_long,
    x="Tenure_Group",
    y="Percentage",
    color="Customer_Status",
    barmode="stack",
    text="Percentage",
    title="Customer Status Percentage by Tenure Group",
    labels={
        "Tenure_Group": "Tenure Group",
        "Percentage": "Customer Share (%)",
        "Customer_Status": "Customer Status"
    }
)

fig.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="inside"
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5
)

fig.show()


plot_df = df[
    [
        "Customer_ID",
        "Tenure_in_Months",
        "Total_Revenue",
        "Monthly_Charge",
        "Customer_Status",
        "Contract"
    ]
].copy()

plot_df["Revenue_Size"] = (
    plot_df["Total_Revenue"]
    .clip(lower=0)
)

fig = px.scatter(
    plot_df,
    x="Tenure_in_Months",
    y="Total_Revenue",
    size="Revenue_Size",
    color="Customer_Status",
    symbol="Contract",
    hover_data=[
        "Customer_ID",
        "Monthly_Charge"
    ],
    opacity=0.65,
    title="Customer Tenure vs Total Revenue",
    labels={
        "Tenure_in_Months": "Tenure (Months)",
        "Total_Revenue": "Total Revenue",
        "Customer_Status": "Customer Status",
        "Contract": "Contract Type"
    }
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5
)

fig.show()


fig = px.scatter(
    df,
    x="Tenure_in_Months",
    y="Monthly_Charge",
    color="Customer_Status",
    trendline="ols",
    opacity=0.55,
    title="Customer Tenure vs Monthly Charge",
    labels={
        "Tenure_in_Months": "Tenure (Months)",
        "Monthly_Charge": "Monthly Charge",
        "Customer_Status": "Customer Status"
    }
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5
)

fig.show()


fig = px.pie(
    tenure_summary,
    names="Tenure_Group",
    values="Total_Revenue",
    hole=0.55,
    title="Revenue Share by Customer Tenure Group"
)

fig.update_traces(
    textinfo="label+percent",
    hovertemplate=(
        "Tenure Group: %{label}<br>"
        "Revenue: %{value:,.2f}<br>"
        "Revenue Share: %{percent}<extra></extra>"
    )
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5
)

fig.show()


heatmap_data = tenure_summary.set_index("Tenure_Group")[
    [
        "Avg_Tenure",
        "Avg_Monthly_Charge",
        "Avg_Revenue",
        "Churn_Rate",
        "Revenue_Share"
    ]
]

fig = px.imshow(
    heatmap_data,
    text_auto=".2f",
    aspect="auto",
    color_continuous_scale="Turbo",
    title="Tenure Group Business KPI Heatmap",
    labels={
        "x": "Business KPI",
        "y": "Tenure Group",
        "color": "Value"
    }
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5
)

fig.show()


print("\n" + "=" * 75)
print("🧠 BUSINESS INSIGHTS — TENURE & REVENUE ANALYSIS")
print("=" * 75)

print(
    f"💰 Highest average-revenue tenure group : "
    f"{highest_revenue_group}"
)

print(
    f"💵 Highest average customer revenue     : "
    f"{highest_revenue_value:,.2f}"
)

print(
    f"🔴 Highest churn-risk tenure group      : "
    f"{highest_churn_group}"
)

print(
    f"🟢 Lowest churn-risk tenure group       : "
    f"{lowest_churn_group}"
)

print(
    f"⏳ Longest-tenure customer group        : "
    f"{longest_tenure_group}"
)

print(
    "🎯 Business Action: Identify early-tenure "
    "customers with high churn and design onboarding, "
    "engagement, and retention campaigns."
)

print(
    "💡 Strategic Focus: Compare tenure with contract, "
    "monthly charges, services, payment method, and "
    "customer status to identify the strongest churn drivers."
)

print("=" * 75)
````

````python
# ============================================================
# 🎯 BUSINESS QUESTION 3
# ============================================================

# How does contract type influence customer churn, retention,
# tenure, monthly charges, and revenue contribution?
#
# Business Goal:
# Identify contract types with higher churn risk and understand
# which contract structure generates stronger customer retention
# and revenue performance.
# ============================================================


contract_summary = (
    df.groupby("Contract", as_index=False)
    .agg(
        Customers=("Customer_ID", "nunique"),
        Avg_Age=("Age", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Total_Charges=("Total_Charges", "mean"),
        Avg_Refunds=("Total_Refunds", "mean"),
        Avg_Referrals=("Number_of_Referrals", "mean")
    )
)

contract_summary["Customer_Share"] = (
    contract_summary["Customers"]
    .div(contract_summary["Customers"].sum())
    .mul(100)
    .round(2)
)

contract_summary["Revenue_Share"] = (
    contract_summary["Total_Revenue"]
    .div(contract_summary["Total_Revenue"].sum())
    .mul(100)
    .round(2)
)

display(contract_summary.round(2))


# ============================================================
# CONTRACT × CUSTOMER STATUS
# ============================================================

contract_status = (
    pd.crosstab(
        df["Contract"],
        df["Customer_Status"]
    )
    .reset_index()
)

display(contract_status)


contract_status_pct = (
    pd.crosstab(
        df["Contract"],
        df["Customer_Status"],
        normalize="index"
    )
    .mul(100)
    .round(2)
    .reset_index()
)

display(contract_status_pct)


# ============================================================
# CHURN RATE BY CONTRACT
# ============================================================

contract_churn = (
    df.assign(
        Is_Churned=df["Customer_Status"].eq("Churned")
    )
    .groupby("Contract", as_index=False)
    .agg(
        Customers=("Customer_ID", "nunique"),
        Churned_Customers=("Is_Churned", "sum"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Total_Revenue=("Total_Revenue", "sum")
    )
)

contract_churn["Churn_Rate"] = (
    contract_churn["Churned_Customers"]
    .div(contract_churn["Customers"])
    .mul(100)
    .round(2)
)

contract_churn["Retention_Rate"] = (
    100 - contract_churn["Churn_Rate"]
).round(2)

display(contract_churn.round(2))


# ============================================================
# CONTRACT KPI SUMMARY
# ============================================================

best_contract = contract_churn.loc[
    contract_churn["Churn_Rate"].idxmin(),
    "Contract"
]

highest_risk_contract = contract_churn.loc[
    contract_churn["Churn_Rate"].idxmax(),
    "Contract"
]

highest_revenue_contract = contract_summary.loc[
    contract_summary["Total_Revenue"].idxmax(),
    "Contract"
]

longest_tenure_contract = contract_summary.loc[
    contract_summary["Avg_Tenure"].idxmax(),
    "Contract"
]

overall_churn_rate = (
    df["Customer_Status"].eq("Churned").mean() * 100
)


contract_kpi = pd.DataFrame({
    "Metric": [
        "Total Customers",
        "Overall Churn Rate",
        "Lowest Churn Contract",
        "Highest Risk Contract",
        "Highest Revenue Contract",
        "Longest Tenure Contract"
    ],
    "Value": [
        df["Customer_ID"].nunique(),
        f"{overall_churn_rate:.2f}%",
        best_contract,
        highest_risk_contract,
        highest_revenue_contract,
        longest_tenure_contract
    ]
})

display(contract_kpi)


# ============================================================
# 📊 VISUAL 1 — CHURN RATE BY CONTRACT
# ============================================================

fig = px.bar(
    contract_churn.sort_values("Churn_Rate"),
    x="Contract",
    y="Churn_Rate",
    text="Churn_Rate",
    color="Churn_Rate",
    color_continuous_scale="Turbo",
    title="Customer Churn Rate by Contract Type",
    labels={
        "Contract": "Contract Type",
        "Churn_Rate": "Churn Rate (%)"
    }
)

fig.update_traces(
    texttemplate="%{text:.2f}%",
    textposition="outside"
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5,
    coloraxis_colorbar_title="Churn %"
)

fig.show()


# ============================================================
# 📊 VISUAL 2 — CUSTOMER STATUS BY CONTRACT
# ============================================================

fig = px.bar(
    contract_status,
    x="Contract",
    y=["Churned", "Stayed", "Joined"],
    barmode="group",
    text_auto=True,
    title="Customer Lifecycle Status by Contract Type",
    labels={
        "Contract": "Contract Type",
        "value": "Customers",
        "variable": "Customer Status"
    }
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5,
    legend_title="Customer Status"
)

fig.show()


# ============================================================
# 📊 VISUAL 3 — STATUS PERCENTAGE
# ============================================================

status_long = contract_status_pct.melt(
    id_vars="Contract",
    var_name="Customer_Status",
    value_name="Percentage"
)

fig = px.bar(
    status_long,
    x="Contract",
    y="Percentage",
    color="Customer_Status",
    barmode="stack",
    text="Percentage",
    title="Customer Status Percentage Distribution by Contract",
    labels={
        "Contract": "Contract Type",
        "Percentage": "Customer Share (%)",
        "Customer_Status": "Customer Status"
    }
)

fig.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="inside"
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5
)

fig.show()


# ============================================================
# 📊 VISUAL 4 — CONTRACT REVENUE CONTRIBUTION
# ============================================================

fig = px.treemap(
    contract_summary,
    path=["Contract"],
    values="Total_Revenue",
    color="Avg_Revenue",
    color_continuous_scale="Viridis",
    title="Revenue Contribution by Contract Type"
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5
)

fig.show()


# ============================================================
# 📊 VISUAL 5 — TENURE VS MONTHLY CHARGE
# ============================================================

fig = px.scatter(
    contract_summary,
    x="Avg_Tenure",
    y="Avg_Monthly_Charge",
    size="Customers",
    color="Contract",
    hover_data=[
        "Total_Revenue",
        "Avg_Revenue",
        "Revenue_Share"
    ],
    title="Average Tenure vs Monthly Charge by Contract",
    labels={
        "Avg_Tenure": "Average Tenure (Months)",
        "Avg_Monthly_Charge": "Average Monthly Charge"
    }
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5
)

fig.show()


# ============================================================
# 📊 VISUAL 6 — CHURN RATE VS RETENTION RATE
# ============================================================

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=contract_churn["Contract"],
        y=contract_churn["Churn_Rate"],
        name="Churn Rate"
    )
)

fig.add_trace(
    go.Bar(
        x=contract_churn["Contract"],
        y=contract_churn["Retention_Rate"],
        name="Retention Rate"
    )
)

fig.update_layout(
    barmode="group",
    template="plotly_dark",
    title="Churn Rate vs Retention Rate by Contract",
    title_x=0.5,
    xaxis_title="Contract Type",
    yaxis_title="Rate (%)"
)

fig.show()


# ============================================================
# 📊 VISUAL 7 — CONTRACT PERFORMANCE HEATMAP
# ============================================================

heatmap_data = contract_churn.set_index("Contract")[
    [
        "Churn_Rate",
        "Retention_Rate",
        "Avg_Tenure",
        "Avg_Monthly_Charge"
    ]
]

fig = px.imshow(
    heatmap_data,
    text_auto=".2f",
    aspect="auto",
    color_continuous_scale="Plasma",
    title="Contract Performance KPI Heatmap",
    labels={
        "x": "Business KPI",
        "y": "Contract Type",
        "color": "Value"
    }
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5
)

fig.show()


# ============================================================
# 💰 REVENUE ANALYSIS
# ============================================================

contract_revenue = (
    df.groupby("Contract", as_index=False)
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Total_Charges=("Total_Charges", "mean")
    )
)

contract_revenue["Revenue_Per_Customer"] = (
    contract_revenue["Total_Revenue"]
    / contract_revenue["Customers"]
)

display(contract_revenue.round(2))


# ============================================================
# 📊 VISUAL 8 — REVENUE PER CUSTOMER
# ============================================================

fig = px.bar(
    contract_revenue.sort_values("Revenue_Per_Customer"),
    x="Contract",
    y="Revenue_Per_Customer",
    color="Revenue_Per_Customer",
    text="Revenue_Per_Customer",
    color_continuous_scale="Mint",
    title="Revenue Generated Per Customer by Contract",
    labels={
        "Contract": "Contract Type",
        "Revenue_Per_Customer": "Revenue per Customer"
    }
)

fig.update_traces(
    texttemplate="%{text:,.2f}",
    textposition="outside"
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5
)

fig.show()


# ============================================================
# 🧠 BUSINESS INSIGHTS
# ============================================================

print("\n" + "=" * 75)
print("🧠 BUSINESS INSIGHTS — CONTRACT ANALYSIS")
print("=" * 75)

print(
    f"📌 Lowest churn contract        : {best_contract}"
)

print(
    f"⚠️ Highest churn-risk contract  : {highest_risk_contract}"
)

print(
    f"💰 Highest revenue contract     : {highest_revenue_contract}"
)

print(
    f"⏳ Longest average tenure       : {longest_tenure_contract}"
)

print(
    f"📊 Overall customer churn rate  : {overall_churn_rate:.2f}%"
)

risk_rate = contract_churn.loc[
    contract_churn["Contract"].eq(highest_risk_contract),
    "Churn_Rate"
].iloc[0]

best_rate = contract_churn.loc[
    contract_churn["Contract"].eq(best_contract),
    "Churn_Rate"
].iloc[0]

print(
    f"🔎 Churn gap between highest-risk "
    f"and lowest-risk contracts: {risk_rate - best_rate:.2f} percentage points"
)

print(
    "🎯 Business Action: Prioritize high-churn contract customers "
    "with targeted retention offers, longer-term plans, loyalty "
    "benefits, and proactive customer engagement."
)

print(
    "💡 Strategic Focus: Compare contract type with tenure, "
    "monthly charges, payment method, services, and support "
    "behavior before designing retention campaigns."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 4
# ============================================================

# Which customer segments generate the strongest revenue,
# retention, and customer value?

# Business Goal:
# Compare revenue segments by customer volume, revenue,
# churn, tenure, and customer value to identify the most
# valuable and risky customer segments.
# ============================================================


segment_summary = (
    df.groupby("Revenue_Segment", dropna=False)
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Service_Count=("Service_Count", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum()),
        Joined=("Customer_Status", lambda x: (x == "Joined").sum())
    )
    .reset_index()
)


segment_summary["Churn_Rate"] = (
    segment_summary["Churned"]
    .div(segment_summary["Customers"])
    .mul(100)
)

segment_summary["Retention_Rate"] = (
    segment_summary["Stayed"]
    .div(segment_summary["Customers"])
    .mul(100)
)

segment_summary["Revenue_Share"] = (
    segment_summary["Total_Revenue"]
    .div(segment_summary["Total_Revenue"].sum())
    .mul(100)
)

segment_summary["Revenue_Per_Customer"] = (
    segment_summary["Total_Revenue"]
    .div(segment_summary["Customers"])
)

display(segment_summary.round(2))


# ============================================================
# CUSTOMER STATUS BY REVENUE SEGMENT
# ============================================================

status_segment = (
    pd.crosstab(
        df["Revenue_Segment"],
        df["Customer_Status"],
        normalize="index"
    )
    .mul(100)
    .round(2)
)

display(status_segment)


# ============================================================
# REVENUE SEGMENT × CONTRACT
# ============================================================

segment_contract = (
    pd.crosstab(
        df["Revenue_Segment"],
        df["Contract"],
        normalize="index"
    )
    .mul(100)
    .round(2)
)

display(segment_contract)


# ============================================================
# REVENUE SEGMENT × TENURE GROUP
# ============================================================

segment_tenure = (
    df.groupby(
        ["Revenue_Segment", "Tenure_Group"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean")
    )
    .reset_index()
)

display(segment_tenure.round(2))


# ============================================================
# REVENUE SEGMENT × STATE
# ============================================================

segment_state = (
    df.groupby(
        ["Revenue_Segment", "State"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean")
    )
    .reset_index()
)

display(segment_state.round(2))


# ============================================================
# VISUAL 1 — REVENUE SEGMENT BUBBLE
# ============================================================

fig = px.scatter(
    segment_summary,
    x="Customers",
    y="Total_Revenue",
    size="Revenue_Per_Customer",
    color="Churn_Rate",
    text="Revenue_Segment",
    hover_data=[
        "Customers",
        "Total_Revenue",
        "Revenue_Per_Customer",
        "Churn_Rate",
        "Retention_Rate",
        "Avg_Tenure"
    ],
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Revenue Segment: Customer Scale vs Revenue"
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Customers",
    yaxis_title="Total Revenue"
)

fig.show()


# ============================================================
# VISUAL 2 — REVENUE SEGMENT PARALLEL CATEGORIES
# ============================================================

fig = px.parallel_categories(
    df,
    dimensions=[
        "Revenue_Segment",
        "Customer_Status",
        "Contract"
    ],
    color="Total_Revenue",
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Revenue Segment → Customer Status → Contract"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


# ============================================================
# VISUAL 3 — REVENUE SEGMENT TREEMAP
# ============================================================

treemap_data = (
    df.groupby(
        ["Revenue_Segment", "Customer_Status"],
        dropna=False
    )
    .agg(
        Revenue=("Total_Revenue", "sum"),
        Customers=("Customer_ID", "nunique")
    )
    .reset_index()
)

fig = px.treemap(
    treemap_data,
    path=[
        "Revenue_Segment",
        "Customer_Status"
    ],
    values="Revenue",
    color="Customers",
    color_continuous_scale="Plasma",
    hover_data=[
        "Revenue",
        "Customers"
    ],
    template="plotly_white",
    title="Revenue Contribution by Segment and Customer Status"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


# ============================================================
# VISUAL 4 — SEGMENT × CONTRACT HEATMAP
# ============================================================

fig = px.imshow(
    segment_contract,
    text_auto=".1f",
    aspect="auto",
    color_continuous_scale="Cividis",
    template="plotly_dark",
    title="Contract Mix Within Revenue Segments",
    labels={
        "x": "Contract",
        "y": "Revenue Segment",
        "color": "Customer Share (%)"
    }
)

fig.update_layout(
    title_x=0.5
)

fig.show()


# ============================================================
# VISUAL 5 — SEGMENT RETENTION POLAR
# ============================================================

fig = px.line_polar(
    segment_summary,
    r="Retention_Rate",
    theta="Revenue_Segment",
    line_close=True,
    markers=True,
    template="plotly_white",
    title="Customer Retention Strength by Revenue Segment"
)

fig.update_traces(
    fill="toself"
)

fig.update_layout(
    title_x=0.5,
    polar=dict(
        radialaxis=dict(
            title="Retention Rate (%)"
        )
    )
)

fig.show()


# ============================================================
# VISUAL 6 — SEGMENT PERFORMANCE MATRIX
# ============================================================

matrix_data = segment_summary[
    [
        "Revenue_Segment",
        "Churn_Rate",
        "Retention_Rate",
        "Avg_Revenue",
        "Avg_Tenure",
        "Revenue_Per_Customer"
    ]
].copy()

matrix_data = matrix_data.set_index(
    "Revenue_Segment"
)

matrix_data = matrix_data.rename(
    columns={
        "Churn_Rate": "Churn",
        "Retention_Rate": "Retention",
        "Avg_Revenue": "Avg Revenue",
        "Avg_Tenure": "Avg Tenure",
        "Revenue_Per_Customer": "Revenue / Customer"
    }
)

fig = px.imshow(
    matrix_data,
    text_auto=".1f",
    aspect="auto",
    color_continuous_scale="RdYlGn",
    template="plotly_white",
    title="Revenue Segment Business Performance Matrix"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Business KPI",
    yaxis_title="Revenue Segment"
)

fig.show()


# ============================================================
# BUSINESS KPI
# ============================================================

highest_revenue_segment = segment_summary.loc[
    segment_summary["Total_Revenue"].idxmax()
]

highest_value_segment = segment_summary.loc[
    segment_summary["Revenue_Per_Customer"].idxmax()
]

highest_churn_segment = segment_summary.loc[
    segment_summary["Churn_Rate"].idxmax()
]

highest_retention_segment = segment_summary.loc[
    segment_summary["Retention_Rate"].idxmax()
]

highest_tenure_segment = segment_summary.loc[
    segment_summary["Avg_Tenure"].idxmax()
]


summary = pd.DataFrame({
    "Business Metric": [
        "Highest Revenue Segment",
        "Highest Revenue per Customer",
        "Highest Churn Segment",
        "Highest Retention Segment",
        "Highest Tenure Segment"
    ],

    "Segment": [
        highest_revenue_segment["Revenue_Segment"],
        highest_value_segment["Revenue_Segment"],
        highest_churn_segment["Revenue_Segment"],
        highest_retention_segment["Revenue_Segment"],
        highest_tenure_segment["Revenue_Segment"]
    ],

    "Value": [
        f"{highest_revenue_segment['Total_Revenue']:,.2f}",
        f"{highest_value_segment['Revenue_Per_Customer']:,.2f}",
        f"{highest_churn_segment['Churn_Rate']:.2f}%",
        f"{highest_retention_segment['Retention_Rate']:.2f}%",
        f"{highest_tenure_segment['Avg_Tenure']:.2f} Months"
    ]
})

display(summary)


# ============================================================
# BUSINESS INSIGHTS
# ============================================================

print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{highest_revenue_segment['Revenue_Segment']} generates the "
    f"highest total revenue at "
    f"{highest_revenue_segment['Total_Revenue']:,.2f}."
)

print(
    f"{highest_value_segment['Revenue_Segment']} has the highest "
    f"revenue per customer at "
    f"{highest_value_segment['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{highest_churn_segment['Revenue_Segment']} has the highest "
    f"churn rate at "
    f"{highest_churn_segment['Churn_Rate']:.2f}%."
)

print(
    f"{highest_retention_segment['Revenue_Segment']} has the strongest "
    f"retention rate at "
    f"{highest_retention_segment['Retention_Rate']:.2f}%."
)

print(
    f"{highest_tenure_segment['Revenue_Segment']} has the highest "
    f"average tenure at "
    f"{highest_tenure_segment['Avg_Tenure']:.2f} months."
)

print(
    "Revenue segments should be evaluated using both customer "
    "value and customer risk rather than revenue alone."
)

print(
    "High-value segments with elevated churn should receive "
    "priority retention campaigns."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 5
# ============================================================

# Do security, protection, and support services improve
# customer retention and revenue performance?

# Business Goal:
# Understand whether customers using additional protection
# and support services generate higher revenue and show better
# retention compared with customers without these services.
# ============================================================


service_summary = (
    df.groupby(
        [
            "Online_Security",
            "Online_Backup",
            "Device_Protection_Plan",
            "Premium_Support"
        ],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Service_Count=("Service_Count", "mean"),
        Avg_Streaming_Services=("Streaming_Service_Count", "mean"),
        Avg_Protection_Services=("Protection_Service_Count", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

service_summary["Churn_Rate"] = (
    service_summary["Churned"]
    .div(service_summary["Customers"])
    .mul(100)
)

service_summary["Retention_Rate"] = (
    service_summary["Stayed"]
    .div(service_summary["Customers"])
    .mul(100)
)

service_summary["Revenue_Per_Customer"] = (
    service_summary["Total_Revenue"]
    .div(service_summary["Customers"])
)

service_summary["Revenue_Share"] = (
    service_summary["Total_Revenue"]
    .div(service_summary["Total_Revenue"].sum())
    .mul(100)
)

service_summary = service_summary.sort_values(
    "Customers",
    ascending=False
)

display(service_summary.round(2))


service_adoption = pd.DataFrame({
    "Service": [
        "Online Security",
        "Online Backup",
        "Device Protection",
        "Premium Support"
    ],

    "Customers_Using": [
        df["Online_Security"].eq("Yes").sum(),
        df["Online_Backup"].eq("Yes").sum(),
        df["Device_Protection_Plan"].eq("Yes").sum(),
        df["Premium_Support"].eq("Yes").sum()
    ]
})

service_adoption["Adoption_Rate"] = (
    service_adoption["Customers_Using"]
    .div(df["Customer_ID"].nunique())
    .mul(100)
)

display(service_adoption.round(2))


security_summary = (
    df.groupby(
        "Online_Security",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

security_summary["Churn_Rate"] = (
    security_summary["Churned"]
    .div(security_summary["Customers"])
    .mul(100)
)

security_summary["Retention_Rate"] = (
    security_summary["Stayed"]
    .div(security_summary["Customers"])
    .mul(100)
)

security_summary["Revenue_Per_Customer"] = (
    security_summary["Total_Revenue"]
    .div(security_summary["Customers"])
)

display(security_summary.round(2))


support_summary = (
    df.groupby(
        "Premium_Support",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

support_summary["Churn_Rate"] = (
    support_summary["Churned"]
    .div(support_summary["Customers"])
    .mul(100)
)

support_summary["Retention_Rate"] = (
    support_summary["Stayed"]
    .div(support_summary["Customers"])
    .mul(100)
)

support_summary["Revenue_Per_Customer"] = (
    support_summary["Total_Revenue"]
    .div(support_summary["Customers"])
)

display(support_summary.round(2))


protection_summary = (
    df.groupby(
        "Device_Protection_Plan",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Service_Count=("Service_Count", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

protection_summary["Churn_Rate"] = (
    protection_summary["Churned"]
    .div(protection_summary["Customers"])
    .mul(100)
)

protection_summary["Retention_Rate"] = (
    protection_summary["Stayed"]
    .div(protection_summary["Customers"])
    .mul(100)
)

protection_summary["Revenue_Per_Customer"] = (
    protection_summary["Total_Revenue"]
    .div(protection_summary["Customers"])
)

display(protection_summary.round(2))


service_status = (
    pd.crosstab(
        df["Service_Count"],
        df["Customer_Status"],
        normalize="index"
    )
    .mul(100)
    .round(2)
)

display(service_status)


fig = px.bar(
    service_adoption.sort_values("Adoption_Rate"),
    x="Adoption_Rate",
    y="Service",
    orientation="h",
    text="Adoption_Rate",
    color="Adoption_Rate",
    color_continuous_scale="Viridis",
    template="plotly_white",
    title="Customer Adoption of Protection and Support Services"
)

fig.update_traces(
    texttemplate="%{text:.2f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Customer Adoption (%)",
    yaxis_title="Service"
)

fig.show()


fig = px.scatter(
    service_summary,
    x="Avg_Service_Count",
    y="Revenue_Per_Customer",
    size="Customers",
    color="Churn_Rate",
    hover_data=[
        "Customers",
        "Total_Revenue",
        "Avg_Tenure",
        "Retention_Rate",
        "Avg_Streaming_Services",
        "Avg_Protection_Services"
    ],
    color_continuous_scale="Turbo",
    template="plotly_dark",
    title="Service Adoption vs Customer Revenue and Churn"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Number of Services Used",
    yaxis_title="Revenue per Customer"
)

fig.show()


fig = px.funnel(
    service_adoption.sort_values(
        "Customers_Using",
        ascending=False
    ),
    y="Service",
    x="Customers_Using",
    text="Customers_Using",
    template="plotly_white",
    title="Service Adoption Funnel"
)

fig.update_traces(
    textposition="inside"
)

fig.update_layout(
    title_x=0.5,
    yaxis_title="Service",
    xaxis_title="Customers Using Service"
)

fig.show()


fig = px.bar(
    security_summary,
    x="Online_Security",
    y="Retention_Rate",
    color="Churn_Rate",
    text="Retention_Rate",
    color_continuous_scale="RdYlGn",
    template="plotly_dark",
    title="Online Security and Customer Retention"
)

fig.update_traces(
    texttemplate="%{text:.2f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Online Security",
    yaxis_title="Retention Rate (%)"
)

fig.show()


fig = px.scatter(
    support_summary,
    x="Avg_Tenure",
    y="Avg_Revenue",
    size="Customers",
    color="Retention_Rate",
    text="Premium_Support",
    color_continuous_scale="Plasma",
    template="plotly_white",
    title="Premium Support: Tenure vs Revenue"
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Tenure (Months)",
    yaxis_title="Average Revenue"
)

fig.show()


fig = px.imshow(
    service_status,
    text_auto=".1f",
    aspect="auto",
    color_continuous_scale="Magma",
    template="plotly_dark",
    title="Customer Status Distribution by Number of Services",
    labels={
        "x": "Customer Status",
        "y": "Number of Services",
        "color": "Customer Share (%)"
    }
)

fig.update_layout(
    title_x=0.5
)

fig.show()


highest_adoption = service_adoption.loc[
    service_adoption["Adoption_Rate"].idxmax()
]

highest_security_retention = security_summary.loc[
    security_summary["Retention_Rate"].idxmax()
]

highest_support_revenue = support_summary.loc[
    support_summary["Avg_Revenue"].idxmax()
]

highest_protection_retention = protection_summary.loc[
    protection_summary["Retention_Rate"].idxmax()
]

highest_service_value = service_summary.loc[
    service_summary["Revenue_Per_Customer"].idxmax()
]


summary = pd.DataFrame({
    "Business Metric": [
        "Highest Service Adoption",
        "Best Security Retention",
        "Highest Support Revenue",
        "Best Protection Retention",
        "Highest Service Combination Value"
    ],

    "Category": [
        highest_adoption["Service"],
        highest_security_retention["Online_Security"],
        highest_support_revenue["Premium_Support"],
        highest_protection_retention["Device_Protection_Plan"],
        " + ".join([
            str(highest_service_value["Online_Security"]),
            str(highest_service_value["Online_Backup"]),
            str(highest_service_value["Device_Protection_Plan"]),
            str(highest_service_value["Premium_Support"])
        ])
    ],

    "Value": [
        f"{highest_adoption['Adoption_Rate']:.2f}%",
        f"{highest_security_retention['Retention_Rate']:.2f}%",
        f"{highest_support_revenue['Avg_Revenue']:,.2f}",
        f"{highest_protection_retention['Retention_Rate']:.2f}%",
        f"{highest_service_value['Revenue_Per_Customer']:,.2f}"
    ]
})

display(summary)


print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{highest_adoption['Service']} has the highest customer adoption "
    f"at {highest_adoption['Adoption_Rate']:.2f}%."
)

print(
    f"Customers with {highest_security_retention['Online_Security']} "
    f"Online Security show the strongest retention rate of "
    f"{highest_security_retention['Retention_Rate']:.2f}%."
)

print(
    f"{highest_support_revenue['Premium_Support']} Premium Support "
    f"customers have the highest average revenue of "
    f"{highest_support_revenue['Avg_Revenue']:,.2f}."
)

print(
    f"{highest_protection_retention['Device_Protection_Plan']} "
    f"Device Protection customers have a retention rate of "
    f"{highest_protection_retention['Retention_Rate']:.2f}%."
)

print(
    f"The strongest service combination generates approximately "
    f"{highest_service_value['Revenue_Per_Customer']:,.2f} revenue "
    f"per customer."
)

print(
    "Protection and support services should be evaluated as "
    "customer-value and retention drivers rather than only as "
    "additional service features."
)

print(
    "Low-adoption services with strong retention or revenue performance "
    "may represent opportunities for targeted upselling."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 6
# ============================================================

# Which payment methods generate the strongest revenue and
# customer retention, and which payment methods are associated
# with higher churn and refund behavior?

# ============================================================


payment_summary = (
    df.groupby("Payment_Method", dropna=False)
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Total_Refunds=("Total_Refunds", "sum"),
        Avg_Refunds=("Total_Refunds", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum()),
        Joined=("Customer_Status", lambda x: (x == "Joined").sum())
    )
    .reset_index()
)

payment_summary["Churn_Rate"] = (
    payment_summary["Churned"]
    .div(payment_summary["Customers"])
    .mul(100)
)

payment_summary["Retention_Rate"] = (
    payment_summary["Stayed"]
    .div(payment_summary["Customers"])
    .mul(100)
)

payment_summary["Revenue_Share"] = (
    payment_summary["Total_Revenue"]
    .div(payment_summary["Total_Revenue"].sum())
    .mul(100)
)

payment_summary["Revenue_Per_Customer"] = (
    payment_summary["Total_Revenue"]
    .div(payment_summary["Customers"])
)

payment_summary["Refund_Rate"] = (
    payment_summary["Total_Refunds"]
    .div(payment_summary["Total_Revenue"])
    .mul(100)
)

display(payment_summary.round(2))


payment_status = (
    pd.crosstab(
        df["Payment_Method"],
        df["Customer_Status"],
        normalize="index"
    )
    .mul(100)
    .round(2)
)

display(payment_status)


payment_contract = (
    pd.crosstab(
        df["Payment_Method"],
        df["Contract"],
        normalize="index"
    )
    .mul(100)
    .round(2)
)

display(payment_contract)


payment_internet = (
    pd.crosstab(
        df["Payment_Method"],
        df["Internet_Type"],
        normalize="index"
    )
    .mul(100)
    .round(2)
)

display(payment_internet)


payment_state = (
    df.groupby(
        ["Payment_Method", "State"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean")
    )
    .reset_index()
)

display(payment_state.round(2))


payment_status_long = (
    payment_status
    .reset_index()
    .melt(
        id_vars="Payment_Method",
        var_name="Customer_Status",
        value_name="Percentage"
    )
)


payment_contract_long = (
    payment_contract
    .reset_index()
    .melt(
        id_vars="Payment_Method",
        var_name="Contract",
        value_name="Percentage"
    )
)


# ============================================================
# VISUAL 1
# ============================================================

fig = px.scatter(
    payment_summary,
    x="Churn_Rate",
    y="Revenue_Per_Customer",
    size="Customers",
    color="Retention_Rate",
    text="Payment_Method",
    hover_data=[
        "Customers",
        "Total_Revenue",
        "Churn_Rate",
        "Retention_Rate",
        "Refund_Rate"
    ],
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Payment Method Risk vs Customer Value"
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Churn Rate (%)",
    yaxis_title="Revenue per Customer"
)

fig.show()


# ============================================================
# VISUAL 2
# ============================================================

fig = px.sunburst(
    payment_status_long,
    path=["Payment_Method", "Customer_Status"],
    values="Percentage",
    color="Percentage",
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Payment Method Customer Lifecycle Structure"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


# ============================================================
# VISUAL 3
# ============================================================

fig = px.parallel_categories(
    df,
    dimensions=[
        "Payment_Method",
        "Contract",
        "Customer_Status"
    ],
    color="Total_Revenue",
    color_continuous_scale="Plasma",
    template="plotly_white",
    title="Payment Method → Contract → Customer Status"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


# ============================================================
# VISUAL 4
# ============================================================

fig = px.imshow(
    payment_contract,
    text_auto=".1f",
    aspect="auto",
    color_continuous_scale="Cividis",
    template="plotly_dark",
    title="Contract Mix by Payment Method",
    labels={
        "x": "Contract",
        "y": "Payment Method",
        "color": "Customer Share (%)"
    }
)

fig.update_layout(
    title_x=0.5
)

fig.show()


# ============================================================
# VISUAL 5
# ============================================================

fig = px.bar(
    payment_summary.sort_values("Refund_Rate"),
    x="Payment_Method",
    y="Refund_Rate",
    color="Avg_Net_Revenue",
    text="Refund_Rate",
    color_continuous_scale="RdYlGn",
    template="plotly_white",
    title="Refund Rate and Net Revenue by Payment Method"
)

fig.update_traces(
    texttemplate="%{text:.2f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Payment Method",
    yaxis_title="Refund Rate (%)"
)

fig.show()


# ============================================================
# VISUAL 6
# ============================================================

fig = px.density_heatmap(
    df,
    x="Payment_Method",
    y="Monthly_Charge",
    z="Total_Revenue",
    histfunc="avg",
    nbinsy=15,
    color_continuous_scale="Magma",
    template="plotly_dark",
    title="Average Revenue Intensity by Payment Method and Monthly Charge"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Payment Method",
    yaxis_title="Monthly Charge"
)

fig.show()


# ============================================================
# BUSINESS KPI
# ============================================================

highest_revenue_payment = payment_summary.loc[
    payment_summary["Total_Revenue"].idxmax()
]

highest_value_payment = payment_summary.loc[
    payment_summary["Revenue_Per_Customer"].idxmax()
]

highest_churn_payment = payment_summary.loc[
    payment_summary["Churn_Rate"].idxmax()
]

highest_retention_payment = payment_summary.loc[
    payment_summary["Retention_Rate"].idxmax()
]

highest_refund_payment = payment_summary.loc[
    payment_summary["Refund_Rate"].idxmax()
]


summary = pd.DataFrame({
    "Business Metric": [
        "Highest Revenue Payment Method",
        "Highest Revenue per Customer",
        "Highest Churn Payment Method",
        "Highest Retention Payment Method",
        "Highest Refund Rate Payment Method"
    ],

    "Payment Method": [
        highest_revenue_payment["Payment_Method"],
        highest_value_payment["Payment_Method"],
        highest_churn_payment["Payment_Method"],
        highest_retention_payment["Payment_Method"],
        highest_refund_payment["Payment_Method"]
    ],

    "Value": [
        f"{highest_revenue_payment['Total_Revenue']:,.2f}",
        f"{highest_value_payment['Revenue_Per_Customer']:,.2f}",
        f"{highest_churn_payment['Churn_Rate']:.2f}%",
        f"{highest_retention_payment['Retention_Rate']:.2f}%",
        f"{highest_refund_payment['Refund_Rate']:.2f}%"
    ]
})

display(summary)


# ============================================================
# BUSINESS INSIGHTS
# ============================================================

print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{highest_revenue_payment['Payment_Method']} generates the "
    f"highest total revenue of "
    f"{highest_revenue_payment['Total_Revenue']:,.2f}."
)

print(
    f"{highest_value_payment['Payment_Method']} has the highest "
    f"revenue per customer at "
    f"{highest_value_payment['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{highest_churn_payment['Payment_Method']} has the highest "
    f"churn rate at "
    f"{highest_churn_payment['Churn_Rate']:.2f}%."
)

print(
    f"{highest_retention_payment['Payment_Method']} has the highest "
    f"retention rate at "
    f"{highest_retention_payment['Retention_Rate']:.2f}%."
)

print(
    f"{highest_refund_payment['Payment_Method']} has the highest "
    f"refund rate at "
    f"{highest_refund_payment['Refund_Rate']:.2f}%."
)

print(
    "Payment methods should be evaluated using revenue, customer "
    "value, churn, retention, and refund behavior together."
)

print(
    "High-value payment segments with elevated churn or refund rates "
    "should be investigated for payment friction and customer experience issues."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 7
# ============================================================

# How does customer referral behavior influence customer value,
# service adoption, tenure, and retention?

# Business Goal:
# Understand whether referred customers are more valuable and
# loyal, and identify customer groups where referral programs
# can create additional growth opportunities.
# ============================================================


referral_summary = (
    df.groupby("Referral_Group", dropna=False)
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Referrals=("Number_of_Referrals", "mean"),
        Avg_Service_Count=("Service_Count", "mean"),
        Avg_Streaming_Services=("Streaming_Service_Count", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum()),
        Joined=("Customer_Status", lambda x: (x == "Joined").sum())
    )
    .reset_index()
)

referral_summary["Churn_Rate"] = (
    referral_summary["Churned"]
    .div(referral_summary["Customers"])
    .mul(100)
)

referral_summary["Retention_Rate"] = (
    referral_summary["Stayed"]
    .div(referral_summary["Customers"])
    .mul(100)
)

referral_summary["Revenue_Share"] = (
    referral_summary["Total_Revenue"]
    .div(referral_summary["Total_Revenue"].sum())
    .mul(100)
)

referral_summary["Revenue_Per_Customer"] = (
    referral_summary["Total_Revenue"]
    .div(referral_summary["Customers"])
)

display(referral_summary.round(2))


referral_status = (
    pd.crosstab(
        df["Referral_Group"],
        df["Customer_Status"],
        normalize="index"
    )
    .mul(100)
    .round(2)
)

display(referral_status)


referral_lifecycle = (
    df.groupby(
        ["Referral_Group", "Customer_Lifecycle"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean")
    )
    .reset_index()
)

display(referral_lifecycle.round(2))


referral_age = (
    df.groupby(
        ["Referral_Group", "Age_Group"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean")
    )
    .reset_index()
)

display(referral_age.round(2))


referral_service = (
    df.groupby("Referral_Group", dropna=False)
    .agg(
        Avg_Service_Count=("Service_Count", "mean"),
        Avg_Streaming_Service_Count=("Streaming_Service_Count", "mean"),
        Avg_Protection_Service_Count=("Protection_Service_Count", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean")
    )
    .reset_index()
)

display(referral_service.round(2))


referral_contract = (
    pd.crosstab(
        df["Referral_Group"],
        df["Contract"],
        normalize="index"
    )
    .mul(100)
    .round(2)
)

display(referral_contract)


fig = px.scatter(
    referral_summary,
    x="Avg_Referrals",
    y="Revenue_Per_Customer",
    size="Customers",
    color="Retention_Rate",
    text="Referral_Group",
    hover_data=[
        "Customers",
        "Total_Revenue",
        "Avg_Tenure",
        "Avg_Service_Count",
        "Churn_Rate"
    ],
    color_continuous_scale="Viridis",
    template="plotly_white",
    title="Referral Activity and Customer Value"
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Number of Referrals",
    yaxis_title="Revenue per Customer"
)

fig.show()


fig = px.sunburst(
    referral_lifecycle,
    path=[
        "Referral_Group",
        "Customer_Lifecycle"
    ],
    values="Customers",
    color="Avg_Revenue",
    color_continuous_scale="Turbo",
    template="plotly_dark",
    title="Referral Groups Across Customer Lifecycle"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


fig = px.area(
    referral_age,
    x="Age_Group",
    y="Avg_Revenue",
    color="Referral_Group",
    markers=True,
    template="plotly_white",
    title="Average Revenue by Age Group and Referral Segment"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Age Group",
    yaxis_title="Average Revenue"
)

fig.show()


fig = px.imshow(
    referral_contract,
    text_auto=".1f",
    aspect="auto",
    color_continuous_scale="Plasma",
    template="plotly_dark",
    title="Contract Preference Across Referral Groups",
    labels={
        "x": "Contract",
        "y": "Referral Group",
        "color": "Customer Share (%)"
    }
)

fig.update_layout(
    title_x=0.5
)

fig.show()


service_long = referral_service.melt(
    id_vars="Referral_Group",
    var_name="Service_Metric",
    value_name="Average_Value"
)

fig = px.bar(
    service_long,
    x="Referral_Group",
    y="Average_Value",
    color="Service_Metric",
    barmode="group",
    template="plotly_white",
    title="Service Adoption Across Referral Groups"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Referral Group",
    yaxis_title="Average Service Usage"
)

fig.show()


fig = px.parallel_categories(
    df,
    dimensions=[
        "Has_Referral",
        "Customer_Lifecycle",
        "Customer_Status",
        "Revenue_Segment"
    ],
    color="Total_Revenue",
    color_continuous_scale="Magma",
    template="plotly_dark",
    title="Referral → Lifecycle → Status → Revenue Segment"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


highest_revenue_referral = referral_summary.loc[
    referral_summary["Total_Revenue"].idxmax()
]

highest_value_referral = referral_summary.loc[
    referral_summary["Revenue_Per_Customer"].idxmax()
]

highest_retention_referral = referral_summary.loc[
    referral_summary["Retention_Rate"].idxmax()
]

highest_churn_referral = referral_summary.loc[
    referral_summary["Churn_Rate"].idxmax()
]

highest_tenure_referral = referral_summary.loc[
    referral_summary["Avg_Tenure"].idxmax()
]


summary = pd.DataFrame({
    "Business Metric": [
        "Highest Total Revenue",
        "Highest Revenue per Customer",
        "Highest Retention",
        "Highest Churn",
        "Highest Average Tenure"
    ],

    "Referral Group": [
        highest_revenue_referral["Referral_Group"],
        highest_value_referral["Referral_Group"],
        highest_retention_referral["Referral_Group"],
        highest_churn_referral["Referral_Group"],
        highest_tenure_referral["Referral_Group"]
    ],

    "Value": [
        f"{highest_revenue_referral['Total_Revenue']:,.2f}",
        f"{highest_value_referral['Revenue_Per_Customer']:,.2f}",
        f"{highest_retention_referral['Retention_Rate']:.2f}%",
        f"{highest_churn_referral['Churn_Rate']:.2f}%",
        f"{highest_tenure_referral['Avg_Tenure']:.2f} Months"
    ]
})

display(summary)


print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{highest_revenue_referral['Referral_Group']} contributes the "
    f"highest total revenue of "
    f"{highest_revenue_referral['Total_Revenue']:,.2f}."
)

print(
    f"{highest_value_referral['Referral_Group']} generates the highest "
    f"revenue per customer at "
    f"{highest_value_referral['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{highest_retention_referral['Referral_Group']} has the strongest "
    f"retention rate of "
    f"{highest_retention_referral['Retention_Rate']:.2f}%."
)

print(
    f"{highest_churn_referral['Referral_Group']} has the highest churn "
    f"rate at "
    f"{highest_churn_referral['Churn_Rate']:.2f}%."
)

print(
    f"{highest_tenure_referral['Referral_Group']} has the highest "
    f"average tenure of "
    f"{highest_tenure_referral['Avg_Tenure']:.2f} months."
)

print(
    "Referral performance should be evaluated using customer value, "
    "retention, tenure, lifecycle stage, and service adoption together."
)

print(
    "High-value referral groups can be targeted for referral programs, "
    "while high-churn referral groups require retention-focused analysis."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 8
# ============================================================

# Which states generate the strongest customer value and revenue,
# and where are there significant geographic performance gaps?

# Business Goal:
# Identify high-value and underperforming states to support
# regional marketing, customer acquisition, and revenue strategy.
# ============================================================


state_summary = (
    df.groupby("State", dropna=False)
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Median_Revenue=("Total_Revenue", "median"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Referrals=("Number_of_Referrals", "mean"),
        Avg_Service_Count=("Service_Count", "mean")
    )
    .reset_index()
)

state_summary["Revenue_Share"] = (
    state_summary["Total_Revenue"]
    .div(state_summary["Total_Revenue"].sum())
    .mul(100)
)

state_summary["Revenue_Per_Customer"] = (
    state_summary["Total_Revenue"]
    .div(state_summary["Customers"])
)

state_summary["Customer_Share"] = (
    state_summary["Customers"]
    .div(state_summary["Customers"].sum())
    .mul(100)
)

state_summary["Revenue_vs_Avg"] = (
    state_summary["Avg_Revenue"]
    - state_summary["Avg_Revenue"].mean()
)

state_summary["Revenue_Index"] = (
    state_summary["Avg_Revenue"]
    .div(state_summary["Avg_Revenue"].mean())
    .mul(100)
)

state_summary = state_summary.sort_values(
    "Total_Revenue",
    ascending=False
)

display(state_summary.round(2))


state_performance = (
    df.groupby("State", dropna=False)
    .agg(
        Customers=("Customer_ID", "nunique"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Revenue_Percentile=("Revenue_Percentile", "mean"),
        Avg_State_Rank=("State_Revenue_Rank", "mean"),
        Avg_Revenue_vs_State=("Revenue_vs_State_Avg", "mean")
    )
    .reset_index()
)

display(state_performance.round(2))


state_lifecycle = (
    df.groupby(
        ["State", "Customer_Lifecycle"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean")
    )
    .reset_index()
)

display(state_lifecycle.round(2))


state_segment = (
    df.groupby(
        ["State", "Revenue_Segment"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean")
    )
    .reset_index()
)

display(state_segment.round(2))


state_summary["Revenue_Contribution_Rank"] = (
    state_summary["Total_Revenue"]
    .rank(
        ascending=False,
        method="dense"
    )
)

state_summary["Customer_Volume_Rank"] = (
    state_summary["Customers"]
    .rank(
        ascending=False,
        method="dense"
    )
)

state_summary["Value_Gap"] = (
    state_summary["Revenue_Per_Customer"]
    - state_summary["Revenue_Per_Customer"].mean()
)

display(
    state_summary[
        [
            "State",
            "Customers",
            "Total_Revenue",
            "Revenue_Share",
            "Revenue_Per_Customer",
            "Customer_Share",
            "Revenue_Contribution_Rank",
            "Customer_Volume_Rank",
            "Value_Gap"
        ]
    ].round(2)
)


fig = px.choropleth(
    state_summary,
    locations="State",
    locationmode="USA-states",
    color="Total_Revenue",
    scope="usa",
    hover_name="State",
    hover_data=[
        "Customers",
        "Avg_Revenue",
        "Revenue_Per_Customer",
        "Revenue_Share"
    ],
    color_continuous_scale="Viridis",
    template="plotly_white",
    title="Geographic Revenue Distribution by State"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


fig = px.treemap(
    state_summary,
    path=["State"],
    values="Total_Revenue",
    color="Revenue_Per_Customer",
    color_continuous_scale="Turbo",
    hover_data=[
        "Customers",
        "Avg_Revenue",
        "Revenue_Share"
    ],
    template="plotly_dark",
    title="State Revenue Contribution and Customer Value"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


fig = px.scatter(
    state_summary,
    x="Customers",
    y="Total_Revenue",
    size="Revenue_Per_Customer",
    color="Revenue_Index",
    text="State",
    hover_data=[
        "Avg_Revenue",
        "Revenue_Share",
        "Avg_Tenure",
        "Avg_Service_Count"
    ],
    color_continuous_scale="Plasma",
    template="plotly_white",
    title="Customer Volume vs State Revenue Performance"
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Number of Customers",
    yaxis_title="Total Revenue"
)

fig.show()


fig = px.bar(
    state_summary.sort_values("Revenue_Per_Customer"),
    x="Revenue_Per_Customer",
    y="State",
    orientation="h",
    color="Revenue_Index",
    text="Revenue_Per_Customer",
    color_continuous_scale="RdYlGn",
    template="plotly_dark",
    title="Revenue per Customer by State"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Revenue per Customer",
    yaxis_title="State"
)

fig.show()


lifecycle_pivot = (
    state_lifecycle
    .pivot(
        index="State",
        columns="Customer_Lifecycle",
        values="Customers"
    )
    .fillna(0)
)

fig = px.imshow(
    lifecycle_pivot,
    text_auto=True,
    aspect="auto",
    color_continuous_scale="Magma",
    template="plotly_white",
    title="Customer Lifecycle Concentration Across States",
    labels={
        "x": "Customer Lifecycle",
        "y": "State",
        "color": "Customers"
    }
)

fig.update_layout(
    title_x=0.5
)

fig.show()


fig = px.scatter(
    state_summary,
    x="Avg_Service_Count",
    y="Revenue_Per_Customer",
    size="Customers",
    color="Avg_Tenure",
    text="State",
    hover_data=[
        "Total_Revenue",
        "Revenue_Share",
        "Avg_Revenue",
        "Avg_Referrals"
    ],
    color_continuous_scale="Cividis",
    template="plotly_dark",
    title="Service Adoption and Customer Value by State"
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Service Count",
    yaxis_title="Revenue per Customer"
)

fig.show()


highest_revenue_state = state_summary.loc[
    state_summary["Total_Revenue"].idxmax()
]

highest_value_state = state_summary.loc[
    state_summary["Revenue_Per_Customer"].idxmax()
]

largest_customer_state = state_summary.loc[
    state_summary["Customers"].idxmax()
]

highest_index_state = state_summary.loc[
    state_summary["Revenue_Index"].idxmax()
]

lowest_value_state = state_summary.loc[
    state_summary["Revenue_Per_Customer"].idxmin()
]


summary = pd.DataFrame({
    "Business Metric": [
        "Highest Total Revenue",
        "Highest Revenue per Customer",
        "Largest Customer Base",
        "Highest Revenue Index",
        "Lowest Revenue per Customer"
    ],

    "State": [
        highest_revenue_state["State"],
        highest_value_state["State"],
        largest_customer_state["State"],
        highest_index_state["State"],
        lowest_value_state["State"]
    ],

    "Value": [
        f"{highest_revenue_state['Total_Revenue']:,.2f}",
        f"{highest_value_state['Revenue_Per_Customer']:,.2f}",
        f"{largest_customer_state['Customers']:,}",
        f"{highest_index_state['Revenue_Index']:.2f}",
        f"{lowest_value_state['Revenue_Per_Customer']:,.2f}"
    ]
})

display(summary)


print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{highest_revenue_state['State']} generates the highest total "
    f"revenue of {highest_revenue_state['Total_Revenue']:,.2f}."
)

print(
    f"{highest_value_state['State']} has the highest revenue per "
    f"customer at {highest_value_state['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{largest_customer_state['State']} has the largest customer "
    f"base with {largest_customer_state['Customers']:,} customers."
)

print(
    f"{highest_index_state['State']} has the strongest revenue index "
    f"at {highest_index_state['Revenue_Index']:.2f}."
)

print(
    f"{lowest_value_state['State']} has the lowest revenue per "
    f"customer at {lowest_value_state['Revenue_Per_Customer']:,.2f}."
)

print(
    "States with high customer volume but relatively low revenue per "
    "customer should be evaluated for upselling and value-expansion opportunities."
)

print(
    "High-value states can be prioritized for customer retention, "
    "premium service adoption, and targeted regional campaigns."
)

print(
    "Regional performance should be evaluated using both customer "
    "volume and customer value rather than total revenue alone."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 9
# ============================================================

# How does customer lifecycle stage influence tenure,
# revenue generation, and churn risk?

# Business Goal:
# Identify which customer lifecycle stages create the strongest
# revenue value and which stages require business attention.
# ============================================================


lifecycle_summary = (
    df.groupby(
        "Customer_Lifecycle",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Charge_Difference=("Charge_Difference", "mean"),
        Avg_Refund_Rate=("Refund_Rate", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum()),
        Joined=("Customer_Status", lambda x: (x == "Joined").sum())
    )
    .reset_index()
)

lifecycle_summary["Churn_Rate"] = (
    lifecycle_summary["Churned"]
    .div(lifecycle_summary["Customers"])
    .mul(100)
)

lifecycle_summary["Retention_Rate"] = (
    lifecycle_summary["Stayed"]
    .div(lifecycle_summary["Customers"])
    .mul(100)
)

lifecycle_summary["Joined_Rate"] = (
    lifecycle_summary["Joined"]
    .div(lifecycle_summary["Customers"])
    .mul(100)
)

lifecycle_summary["Revenue_Per_Customer"] = (
    lifecycle_summary["Total_Revenue"]
    .div(lifecycle_summary["Customers"])
)

lifecycle_summary["Revenue_Share"] = (
    lifecycle_summary["Total_Revenue"]
    .div(lifecycle_summary["Total_Revenue"].sum())
    .mul(100)
)

lifecycle_summary = lifecycle_summary.sort_values(
    "Revenue_Per_Customer",
    ascending=False
)

display(lifecycle_summary.round(2))


tenure_lifecycle = (
    df.groupby(
        "Tenure_Group",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

tenure_lifecycle["Churn_Rate"] = (
    tenure_lifecycle["Churned"]
    .div(tenure_lifecycle["Customers"])
    .mul(100)
)

tenure_lifecycle["Retention_Rate"] = (
    tenure_lifecycle["Stayed"]
    .div(tenure_lifecycle["Customers"])
    .mul(100)
)

tenure_lifecycle["Revenue_Per_Customer"] = (
    tenure_lifecycle["Total_Revenue"]
    .div(tenure_lifecycle["Customers"])
)

display(tenure_lifecycle.round(2))


age_lifecycle = (
    df.groupby(
        "Age_Group",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Refund_Rate=("Refund_Rate", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

age_lifecycle["Churn_Rate"] = (
    age_lifecycle["Churned"]
    .div(age_lifecycle["Customers"])
    .mul(100)
)

age_lifecycle["Retention_Rate"] = (
    age_lifecycle["Stayed"]
    .div(age_lifecycle["Customers"])
    .mul(100)
)

display(age_lifecycle.round(2))


revenue_segment_summary = (
    df.groupby(
        "Revenue_Segment",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

revenue_segment_summary["Churn_Rate"] = (
    revenue_segment_summary["Churned"]
    .div(revenue_segment_summary["Customers"])
    .mul(100)
)

revenue_segment_summary["Retention_Rate"] = (
    revenue_segment_summary["Stayed"]
    .div(revenue_segment_summary["Customers"])
    .mul(100)
)

revenue_segment_summary["Revenue_Share"] = (
    revenue_segment_summary["Total_Revenue"]
    .div(revenue_segment_summary["Total_Revenue"].sum())
    .mul(100)
)

display(revenue_segment_summary.round(2))


fig = px.bar(
    lifecycle_summary.sort_values("Revenue_Per_Customer"),
    x="Revenue_Per_Customer",
    y="Customer_Lifecycle",
    orientation="h",
    text="Revenue_Per_Customer",
    color="Churn_Rate",
    color_continuous_scale="RdYlGn_r",
    template="plotly_white",
    title="Revenue Value by Customer Lifecycle"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Revenue per Customer",
    yaxis_title="Customer Lifecycle"
)

fig.show()


fig = px.line(
    tenure_lifecycle,
    x="Tenure_Group",
    y="Churn_Rate",
    markers=True,
    template="plotly_dark",
    title="Churn Risk Across Customer Tenure"
)

fig.update_traces(
    line_width=3,
    marker_size=9
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Tenure Group",
    yaxis_title="Churn Rate (%)"
)

fig.show()


fig = px.scatter(
    lifecycle_summary,
    x="Avg_Tenure",
    y="Avg_Net_Revenue",
    size="Customers",
    color="Churn_Rate",
    hover_data=[
        "Customer_Lifecycle",
        "Customers",
        "Revenue_Share",
        "Retention_Rate"
    ],
    color_continuous_scale="Plasma",
    template="plotly_white",
    title="Lifecycle Tenure vs Net Revenue"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Tenure (Months)",
    yaxis_title="Average Net Revenue"
)

fig.show()


fig = px.bar(
    age_lifecycle.sort_values("Retention_Rate"),
    x="Age_Group",
    y="Retention_Rate",
    color="Avg_Revenue",
    text="Retention_Rate",
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Retention Performance Across Age Groups"
)

fig.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Age Group",
    yaxis_title="Retention Rate (%)"
)

fig.show()


fig = px.scatter(
    revenue_segment_summary,
    x="Avg_Tenure",
    y="Avg_Revenue",
    size="Customers",
    color="Revenue_Share",
    text="Revenue_Segment",
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Revenue Segment: Tenure vs Average Revenue"
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Tenure (Months)",
    yaxis_title="Average Revenue"
)

fig.show()


fig = px.bar(
    revenue_segment_summary.sort_values("Revenue_Share"),
    x="Revenue_Share",
    y="Revenue_Segment",
    orientation="h",
    text="Revenue_Share",
    color="Churn_Rate",
    color_continuous_scale="RdYlGn_r",
    template="plotly_dark",
    title="Revenue Contribution and Churn Exposure by Segment"
)

fig.update_traces(
    texttemplate="%{text:.2f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Revenue Share (%)",
    yaxis_title="Revenue Segment"
)

fig.show()


highest_lifecycle_value = lifecycle_summary.loc[
    lifecycle_summary["Revenue_Per_Customer"].idxmax()
]

highest_lifecycle_churn = lifecycle_summary.loc[
    lifecycle_summary["Churn_Rate"].idxmax()
]

best_tenure_retention = tenure_lifecycle.loc[
    tenure_lifecycle["Retention_Rate"].idxmax()
]

highest_age_churn = age_lifecycle.loc[
    age_lifecycle["Churn_Rate"].idxmax()
]

highest_revenue_segment = revenue_segment_summary.loc[
    revenue_segment_summary["Total_Revenue"].idxmax()
]


business_summary = pd.DataFrame({
    "Business Metric": [
        "Highest Lifecycle Revenue Value",
        "Highest Lifecycle Churn Risk",
        "Best Tenure Retention",
        "Highest Age-Group Churn",
        "Largest Revenue Segment"
    ],

    "Category": [
        highest_lifecycle_value["Customer_Lifecycle"],
        highest_lifecycle_churn["Customer_Lifecycle"],
        best_tenure_retention["Tenure_Group"],
        highest_age_churn["Age_Group"],
        highest_revenue_segment["Revenue_Segment"]
    ],

    "Value": [
        f"{highest_lifecycle_value['Revenue_Per_Customer']:,.2f}",
        f"{highest_lifecycle_churn['Churn_Rate']:.2f}%",
        f"{best_tenure_retention['Retention_Rate']:.2f}%",
        f"{highest_age_churn['Churn_Rate']:.2f}%",
        f"{highest_revenue_segment['Total_Revenue']:,.2f}"
    ]
})

display(business_summary)


print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{highest_lifecycle_value['Customer_Lifecycle']} customers "
    f"generate the highest revenue per customer at "
    f"{highest_lifecycle_value['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{highest_lifecycle_churn['Customer_Lifecycle']} has the highest "
    f"lifecycle churn rate at "
    f"{highest_lifecycle_churn['Churn_Rate']:.2f}%."
)

print(
    f"{best_tenure_retention['Tenure_Group']} has the strongest "
    f"retention rate at "
    f"{best_tenure_retention['Retention_Rate']:.2f}%."
)

print(
    f"{highest_age_churn['Age_Group']} records the highest churn "
    f"rate among age groups at "
    f"{highest_age_churn['Churn_Rate']:.2f}%."
)

print(
    f"{highest_revenue_segment['Revenue_Segment']} contributes the "
    f"largest total revenue of "
    f"{highest_revenue_segment['Total_Revenue']:,.2f}."
)

print(
    "Lifecycle and tenure analysis can help identify when customers "
    "become most valuable and when retention intervention is most needed."
)

print(
    "Business teams should prioritize high-value lifecycle groups "
    "with elevated churn risk to protect revenue while improving "
    "long-term customer retention."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 10
# ============================================================

# Which customer segments generate the highest revenue,
# and where are revenue leakage and inefficient charges concentrated?

# Business Goal:
# Evaluate revenue quality using Net Revenue, Refunds,
# Extra Charges, Long Distance Charges and Revenue Ratios.
# ============================================================


revenue_quality = (
    df.groupby(
        "Revenue_Segment",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Total_Refunds=("Total_Refunds", "sum"),
        Extra_Data_Charges=("Total_Extra_Data_Charges", "sum"),
        Long_Distance_Charges=("Total_Long_Distance_Charges", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Refund_Rate=("Refund_Rate", "mean"),
        Avg_Extra_Charge_Ratio=("Extra_Charge_Ratio", "mean"),
        Avg_Charge_Revenue_Ratio=("Charge_Revenue_Ratio", "mean")
    )
    .reset_index()
)

revenue_quality["Revenue_Loss"] = (
    revenue_quality["Total_Revenue"]
    - revenue_quality["Net_Revenue"]
)

revenue_quality["Net_Revenue_Rate"] = (
    revenue_quality["Net_Revenue"]
    .div(revenue_quality["Total_Revenue"])
    .mul(100)
)

revenue_quality["Refund_Share"] = (
    revenue_quality["Total_Refunds"]
    .div(revenue_quality["Total_Revenue"])
    .mul(100)
)

revenue_quality["Extra_Charge_Share"] = (
    (
        revenue_quality["Extra_Data_Charges"]
        + revenue_quality["Long_Distance_Charges"]
    )
    .div(revenue_quality["Total_Revenue"])
    .mul(100)
)

revenue_quality["Revenue_Share"] = (
    revenue_quality["Total_Revenue"]
    .div(revenue_quality["Total_Revenue"].sum())
    .mul(100)
)

display(revenue_quality.round(2))


payment_quality = (
    df.groupby(
        "Payment_Method",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Total_Refunds=("Total_Refunds", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Refund_Rate=("Refund_Rate", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean")
    )
    .reset_index()
)

payment_quality["Net_Revenue_Rate"] = (
    payment_quality["Net_Revenue"]
    .div(payment_quality["Total_Revenue"])
    .mul(100)
)

payment_quality["Refund_Rate_Percentage"] = (
    payment_quality["Total_Refunds"]
    .div(payment_quality["Total_Revenue"])
    .mul(100)
)

display(payment_quality.round(2))


state_quality = (
    df.groupby(
        "State",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Total_Refunds=("Total_Refunds", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Refund_Rate=("Refund_Rate", "mean"),
        Avg_Extra_Charge_Ratio=("Extra_Charge_Ratio", "mean")
    )
    .reset_index()
)

state_quality["Revenue_Loss"] = (
    state_quality["Total_Revenue"]
    - state_quality["Net_Revenue"]
)

state_quality["Net_Revenue_Rate"] = (
    state_quality["Net_Revenue"]
    .div(state_quality["Total_Revenue"])
    .mul(100)
)

state_quality["Refund_Share"] = (
    state_quality["Total_Refunds"]
    .div(state_quality["Total_Revenue"])
    .mul(100)
)

display(
    state_quality
    .sort_values("Revenue_Loss", ascending=False)
    .round(2)
)


charge_quality = (
    df.groupby(
        "Customer_Lifecycle",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Total_Charges=("Total_Charges", "mean"),
        Avg_Total_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Refunds=("Total_Refunds", "mean"),
        Avg_Extra_Charges=("Total_Extra_Charges", "mean"),
        Avg_Charge_Difference=("Charge_Difference", "mean"),
        Avg_Charge_Revenue_Ratio=("Charge_Revenue_Ratio", "mean"),
        Avg_Revenue_to_Charge_Ratio=("Revenue_to_Charge_Ratio", "mean")
    )
    .reset_index()
)

display(charge_quality.round(2))


fig = px.treemap(
    revenue_quality,
    path=["Revenue_Segment"],
    values="Total_Revenue",
    color="Net_Revenue_Rate",
    color_continuous_scale="RdYlGn",
    hover_data=[
        "Customers",
        "Net_Revenue",
        "Revenue_Loss",
        "Refund_Share"
    ],
    template="plotly_white",
    title="Revenue Portfolio and Net Revenue Quality"
)

fig.update_layout(title_x=0.5)

fig.show()


fig = px.scatter(
    revenue_quality,
    x="Total_Revenue",
    y="Revenue_Loss",
    size="Customers",
    color="Refund_Share",
    hover_name="Revenue_Segment",
    color_continuous_scale="Reds",
    template="plotly_dark",
    title="Revenue Scale vs Revenue Leakage"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Total Revenue",
    yaxis_title="Revenue Leakage"
)

fig.show()


fig = px.bar(
    payment_quality.sort_values("Net_Revenue_Rate"),
    x="Payment_Method",
    y="Net_Revenue_Rate",
    color="Refund_Rate_Percentage",
    text="Net_Revenue_Rate",
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Net Revenue Efficiency by Payment Method"
)

fig.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Payment Method",
    yaxis_title="Net Revenue Rate (%)"
)

fig.show()


fig = px.scatter(
    state_quality,
    x="Avg_Revenue",
    y="Avg_Net_Revenue",
    size="Customers",
    color="Refund_Share",
    hover_name="State",
    color_continuous_scale="Magma",
    template="plotly_dark",
    title="State-Level Revenue Quality"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Revenue",
    yaxis_title="Average Net Revenue"
)

fig.show()


fig = px.bar(
    charge_quality.sort_values("Avg_Charge_Difference"),
    x="Customer_Lifecycle",
    y="Avg_Charge_Difference",
    color="Avg_Revenue_to_Charge_Ratio",
    text="Avg_Charge_Difference",
    color_continuous_scale="Viridis",
    template="plotly_white",
    title="Charge Difference Across Customer Lifecycle"
)

fig.update_traces(
    texttemplate="%{text:,.2f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Customer Lifecycle",
    yaxis_title="Average Charge Difference"
)

fig.show()


fig = px.scatter(
    charge_quality,
    x="Avg_Extra_Charges",
    y="Avg_Net_Revenue",
    size="Customers",
    color="Avg_Charge_Revenue_Ratio",
    hover_name="Customer_Lifecycle",
    color_continuous_scale="Plasma",
    template="plotly_dark",
    title="Extra Charges vs Net Revenue Efficiency"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Extra Charges",
    yaxis_title="Average Net Revenue"
)

fig.show()


highest_revenue_segment = revenue_quality.loc[
    revenue_quality["Total_Revenue"].idxmax()
]

highest_leakage_segment = revenue_quality.loc[
    revenue_quality["Revenue_Loss"].idxmax()
]

highest_refund_segment = revenue_quality.loc[
    revenue_quality["Refund_Share"].idxmax()
]

best_payment_method = payment_quality.loc[
    payment_quality["Net_Revenue_Rate"].idxmax()
]

highest_state_leakage = state_quality.loc[
    state_quality["Revenue_Loss"].idxmax()
]

best_lifecycle_efficiency = charge_quality.loc[
    charge_quality["Avg_Revenue_to_Charge_Ratio"].idxmax()
]


business_summary = pd.DataFrame({
    "Business_Metric": [
        "Highest Revenue Segment",
        "Highest Revenue Leakage Segment",
        "Highest Refund Exposure Segment",
        "Most Efficient Payment Method",
        "Highest State Revenue Leakage",
        "Best Revenue-to-Charge Lifecycle"
    ],

    "Category": [
        highest_revenue_segment["Revenue_Segment"],
        highest_leakage_segment["Revenue_Segment"],
        highest_refund_segment["Revenue_Segment"],
        best_payment_method["Payment_Method"],
        highest_state_leakage["State"],
        best_lifecycle_efficiency["Customer_Lifecycle"]
    ],

    "Value": [
        f"{highest_revenue_segment['Total_Revenue']:,.2f}",
        f"{highest_leakage_segment['Revenue_Loss']:,.2f}",
        f"{highest_refund_segment['Refund_Share']:.2f}%",
        f"{best_payment_method['Net_Revenue_Rate']:.2f}%",
        f"{highest_state_leakage['Revenue_Loss']:,.2f}",
        f"{best_lifecycle_efficiency['Avg_Revenue_to_Charge_Ratio']:.2f}"
    ]
})

display(business_summary)


print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{highest_revenue_segment['Revenue_Segment']} generates the "
    f"highest total revenue of "
    f"{highest_revenue_segment['Total_Revenue']:,.2f}."
)

print(
    f"{highest_leakage_segment['Revenue_Segment']} has the largest "
    f"revenue leakage of "
    f"{highest_leakage_segment['Revenue_Loss']:,.2f}."
)

print(
    f"{highest_refund_segment['Revenue_Segment']} has the highest "
    f"refund exposure at "
    f"{highest_refund_segment['Refund_Share']:.2f}% of revenue."
)

print(
    f"{best_payment_method['Payment_Method']} shows the strongest "
    f"net revenue efficiency at "
    f"{best_payment_method['Net_Revenue_Rate']:.2f}%."
)

print(
    f"{highest_state_leakage['State']} records the highest state-level "
    f"revenue leakage of "
    f"{highest_state_leakage['Revenue_Loss']:,.2f}."
)

print(
    f"{best_lifecycle_efficiency['Customer_Lifecycle']} has the strongest "
    f"revenue-to-charge ratio at "
    f"{best_lifecycle_efficiency['Avg_Revenue_to_Charge_Ratio']:.2f}."
)

print(
    "Revenue quality should be evaluated beyond total sales by "
    "tracking refunds, extra charges, net revenue and charge efficiency."
)

print(
    "Business teams should investigate high-leakage segments and "
    "payment or geographic groups where revenue is generated but "
    "a comparatively smaller portion converts into net revenue."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 11
# ============================================================

# Which contract types have the highest churn risk and
# how does contract commitment influence revenue and tenure?

# Business Goal:
# Identify contract structures associated with customer risk,
# revenue value and long-term customer commitment.
# ============================================================


contract_summary = (
    df.groupby(
        "Contract",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum()),
        Joined=("Customer_Status", lambda x: (x == "Joined").sum())
    )
    .reset_index()
)

contract_summary["Churn_Rate"] = (
    contract_summary["Churned"]
    .div(contract_summary["Customers"])
    .mul(100)
)

contract_summary["Retention_Rate"] = (
    contract_summary["Stayed"]
    .div(contract_summary["Customers"])
    .mul(100)
)

contract_summary["Joined_Rate"] = (
    contract_summary["Joined"]
    .div(contract_summary["Customers"])
    .mul(100)
)

contract_summary["Revenue_Per_Customer"] = (
    contract_summary["Total_Revenue"]
    .div(contract_summary["Customers"])
)

contract_summary["Net_Revenue_Rate"] = (
    contract_summary["Net_Revenue"]
    .div(contract_summary["Total_Revenue"])
    .mul(100)
)

contract_summary["Revenue_Share"] = (
    contract_summary["Total_Revenue"]
    .div(contract_summary["Total_Revenue"].sum())
    .mul(100)
)

display(contract_summary.round(2))


commitment_summary = (
    df.groupby(
        "Is_Long_Term_Contract",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

commitment_summary["Churn_Rate"] = (
    commitment_summary["Churned"]
    .div(commitment_summary["Customers"])
    .mul(100)
)

commitment_summary["Retention_Rate"] = (
    commitment_summary["Stayed"]
    .div(commitment_summary["Customers"])
    .mul(100)
)

commitment_summary["Revenue_Per_Customer"] = (
    commitment_summary["Total_Revenue"]
    .div(commitment_summary["Customers"])
)

display(commitment_summary.round(2))


contract_status = (
    pd.crosstab(
        df["Contract"],
        df["Customer_Status"]
    )
    .reset_index()
)

for col in ["Churned", "Stayed", "Joined"]:
    if col not in contract_status.columns:
        contract_status[col] = 0

contract_status["Total_Customers"] = (
    contract_status[["Churned", "Stayed", "Joined"]]
    .sum(axis=1)
)

contract_status["Churn_Rate"] = (
    contract_status["Churned"]
    .div(contract_status["Total_Customers"])
    .mul(100)
)

contract_status["Retention_Rate"] = (
    contract_status["Stayed"]
    .div(contract_status["Total_Customers"])
    .mul(100)
)

display(contract_status.round(2))


contract_tenure = (
    df.groupby(
        ["Contract", "Is_Long_Term_Contract"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum())
    )
    .reset_index()
)

contract_tenure["Churn_Rate"] = (
    contract_tenure["Churned"]
    .div(contract_tenure["Customers"])
    .mul(100)
)

display(contract_tenure.round(2))


contract_revenue = (
    df.groupby(
        "Contract",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean")
    )
    .reset_index()
)

contract_revenue["Revenue_Per_Customer"] = (
    contract_revenue["Total_Revenue"]
    .div(contract_revenue["Customers"])
)

contract_revenue["Net_Revenue_Rate"] = (
    contract_revenue["Net_Revenue"]
    .div(contract_revenue["Total_Revenue"])
    .mul(100)
)

display(contract_revenue.round(2))


fig = px.bar(
    contract_summary.sort_values("Churn_Rate"),
    x="Contract",
    y="Churn_Rate",
    color="Revenue_Per_Customer",
    text="Churn_Rate",
    color_continuous_scale="RdYlGn_r",
    template="plotly_white",
    title="Churn Risk by Contract Type"
)

fig.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Contract Type",
    yaxis_title="Churn Rate (%)"
)

fig.show()


fig = px.sunburst(
    contract_status,
    path=["Contract"],
    values="Total_Customers",
    color="Churn_Rate",
    color_continuous_scale="Reds",
    template="plotly_dark",
    title="Customer Portfolio and Contract Churn Exposure"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


fig = px.scatter(
    contract_summary,
    x="Avg_Tenure",
    y="Revenue_Per_Customer",
    size="Customers",
    color="Churn_Rate",
    hover_name="Contract",
    hover_data=[
        "Retention_Rate",
        "Net_Revenue_Rate",
        "Revenue_Share"
    ],
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Contract Tenure vs Customer Revenue Value"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Tenure (Months)",
    yaxis_title="Revenue per Customer"
)

fig.show()


fig = px.bar(
    commitment_summary,
    x="Is_Long_Term_Contract",
    y="Revenue_Per_Customer",
    color="Churn_Rate",
    text="Revenue_Per_Customer",
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Long-Term Commitment and Revenue Value"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Long-Term Contract Indicator",
    yaxis_title="Revenue per Customer"
)

fig.show()


fig = px.scatter(
    contract_tenure,
    x="Avg_Tenure",
    y="Avg_Net_Revenue",
    size="Customers",
    color="Churn_Rate",
    symbol="Is_Long_Term_Contract",
    hover_name="Contract",
    color_continuous_scale="Plasma",
    template="plotly_white",
    title="Contract Commitment, Tenure and Net Revenue"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Tenure (Months)",
    yaxis_title="Average Net Revenue"
)

fig.show()


fig = px.bar(
    contract_revenue.sort_values("Net_Revenue_Rate"),
    x="Contract",
    y="Net_Revenue_Rate",
    color="Avg_Tenure",
    text="Net_Revenue_Rate",
    color_continuous_scale="Aggrnyl",
    template="plotly_dark",
    title="Net Revenue Efficiency by Contract"
)

fig.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Contract Type",
    yaxis_title="Net Revenue Rate (%)"
)

fig.show()


highest_churn_contract = contract_summary.loc[
    contract_summary["Churn_Rate"].idxmax()
]

lowest_churn_contract = contract_summary.loc[
    contract_summary["Churn_Rate"].idxmin()
]

highest_revenue_contract = contract_summary.loc[
    contract_summary["Revenue_Per_Customer"].idxmax()
]

highest_tenure_contract = contract_summary.loc[
    contract_summary["Avg_Tenure"].idxmax()
]

best_net_revenue_contract = contract_revenue.loc[
    contract_revenue["Net_Revenue_Rate"].idxmax()
]

best_retention_commitment = commitment_summary.loc[
    commitment_summary["Retention_Rate"].idxmax()
]


business_summary = pd.DataFrame({
    "Business_Metric": [
        "Highest Churn Contract",
        "Lowest Churn Contract",
        "Highest Revenue per Customer",
        "Highest Average Tenure",
        "Best Net Revenue Efficiency",
        "Best Retention Commitment"
    ],

    "Category": [
        highest_churn_contract["Contract"],
        lowest_churn_contract["Contract"],
        highest_revenue_contract["Contract"],
        highest_tenure_contract["Contract"],
        best_net_revenue_contract["Contract"],
        str(best_retention_commitment["Is_Long_Term_Contract"])
    ],

    "Value": [
        f"{highest_churn_contract['Churn_Rate']:.2f}%",
        f"{lowest_churn_contract['Churn_Rate']:.2f}%",
        f"{highest_revenue_contract['Revenue_Per_Customer']:,.2f}",
        f"{highest_tenure_contract['Avg_Tenure']:.2f} months",
        f"{best_net_revenue_contract['Net_Revenue_Rate']:.2f}%",
        f"{best_retention_commitment['Retention_Rate']:.2f}%"
    ]
})

display(business_summary)


print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{highest_churn_contract['Contract']} customers have the highest "
    f"churn rate at {highest_churn_contract['Churn_Rate']:.2f}%."
)

print(
    f"{lowest_churn_contract['Contract']} customers have the lowest "
    f"churn rate at {lowest_churn_contract['Churn_Rate']:.2f}%."
)

print(
    f"{highest_revenue_contract['Contract']} generates the highest "
    f"revenue per customer at "
    f"{highest_revenue_contract['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{highest_tenure_contract['Contract']} customers show the highest "
    f"average tenure of {highest_tenure_contract['Avg_Tenure']:.2f} months."
)

print(
    f"{best_net_revenue_contract['Contract']} provides the strongest "
    f"net revenue efficiency at "
    f"{best_net_revenue_contract['Net_Revenue_Rate']:.2f}%."
)

print(
    f"Long-term commitment category "
    f"{best_retention_commitment['Is_Long_Term_Contract']} has the "
    f"strongest retention rate at "
    f"{best_retention_commitment['Retention_Rate']:.2f}%."
)

print(
    "Contract structure can influence customer stability, lifetime "
    "value and revenue quality, making contract-level monitoring "
    "important for retention planning."
)

print(
    "High-churn contract groups should receive targeted retention "
    "offers, while high-value long-term customers should be protected "
    "through loyalty and renewal strategies."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 12
# ============================================================

# Which customer demographic segments generate the strongest
# revenue value, and where are the major demographic opportunities?

# Business Goal:
# Compare customer value across gender, marital status, age group
# and state to identify high-value and underperforming segments.
# ============================================================


demographic_summary = (
    df.groupby(
        ["Gender", "Married"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Refund_Rate=("Refund_Rate", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

demographic_summary["Churn_Rate"] = (
    demographic_summary["Churned"]
    .div(demographic_summary["Customers"])
    .mul(100)
)

demographic_summary["Retention_Rate"] = (
    demographic_summary["Stayed"]
    .div(demographic_summary["Customers"])
    .mul(100)
)

demographic_summary["Revenue_Per_Customer"] = (
    demographic_summary["Total_Revenue"]
    .div(demographic_summary["Customers"])
)

demographic_summary["Net_Revenue_Rate"] = (
    demographic_summary["Net_Revenue"]
    .div(demographic_summary["Total_Revenue"])
    .mul(100)
)

display(demographic_summary.round(2))


age_summary = (
    df.groupby(
        "Age_Group",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Refund_Rate=("Refund_Rate", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

age_summary["Churn_Rate"] = (
    age_summary["Churned"]
    .div(age_summary["Customers"])
    .mul(100)
)

age_summary["Retention_Rate"] = (
    age_summary["Stayed"]
    .div(age_summary["Customers"])
    .mul(100)
)

age_summary["Revenue_Per_Customer"] = (
    age_summary["Total_Revenue"]
    .div(age_summary["Customers"])
)

age_summary["Net_Revenue_Rate"] = (
    age_summary["Net_Revenue"]
    .div(age_summary["Total_Revenue"])
    .mul(100)
)

display(age_summary.round(2))


gender_summary = (
    df.groupby(
        "Gender",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

gender_summary["Churn_Rate"] = (
    gender_summary["Churned"]
    .div(gender_summary["Customers"])
    .mul(100)
)

gender_summary["Retention_Rate"] = (
    gender_summary["Stayed"]
    .div(gender_summary["Customers"])
    .mul(100)
)

gender_summary["Revenue_Per_Customer"] = (
    gender_summary["Total_Revenue"]
    .div(gender_summary["Customers"])
)

display(gender_summary.round(2))


state_demographic = (
    df.groupby(
        "State",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Refund_Rate=("Refund_Rate", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

state_demographic["Churn_Rate"] = (
    state_demographic["Churned"]
    .div(state_demographic["Customers"])
    .mul(100)
)

state_demographic["Retention_Rate"] = (
    state_demographic["Stayed"]
    .div(state_demographic["Customers"])
    .mul(100)
)

state_demographic["Revenue_Per_Customer"] = (
    state_demographic["Total_Revenue"]
    .div(state_demographic["Customers"])
)

state_demographic["Revenue_Share"] = (
    state_demographic["Total_Revenue"]
    .div(state_demographic["Total_Revenue"].sum())
    .mul(100)
)

display(
    state_demographic
    .sort_values("Total_Revenue", ascending=False)
    .round(2)
)


fig = px.bar(
    demographic_summary,
    x="Gender",
    y="Revenue_Per_Customer",
    color="Married",
    barmode="group",
    text="Revenue_Per_Customer",
    template="plotly_white",
    title="Customer Value by Gender and Marital Status"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Gender",
    yaxis_title="Revenue per Customer"
)

fig.show()


fig = px.scatter(
    age_summary,
    x="Avg_Tenure",
    y="Avg_Revenue",
    size="Customers",
    color="Churn_Rate",
    text="Age_Group",
    color_continuous_scale="Turbo",
    template="plotly_dark",
    title="Age Group: Tenure vs Revenue Value"
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Tenure (Months)",
    yaxis_title="Average Revenue"
)

fig.show()


fig = px.pie(
    gender_summary,
    names="Gender",
    values="Total_Revenue",
    hole=0.55,
    template="plotly_white",
    title="Revenue Contribution by Gender"
)

fig.update_traces(
    textinfo="label+percent",
    hovertemplate=(
        "Gender: %{label}<br>"
        "Revenue: %{value:,.2f}<br>"
        "Share: %{percent}<extra></extra>"
    )
)

fig.update_layout(
    title_x=0.5
)

fig.show()


fig = px.bar(
    age_summary.sort_values("Retention_Rate"),
    x="Age_Group",
    y="Retention_Rate",
    color="Avg_Net_Revenue",
    text="Retention_Rate",
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Retention Performance by Age Group"
)

fig.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Age Group",
    yaxis_title="Retention Rate (%)"
)

fig.show()


state_plot = (
    state_demographic
    .nlargest(12, "Total_Revenue")
    .sort_values("Total_Revenue")
)

fig = px.bar(
    state_plot,
    x="Total_Revenue",
    y="State",
    orientation="h",
    color="Revenue_Per_Customer",
    text="Total_Revenue",
    color_continuous_scale="Plasma",
    template="plotly_white",
    title="Top States by Revenue Contribution"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Total Revenue",
    yaxis_title="State"
)

fig.show()


fig = px.scatter(
    state_demographic,
    x="Revenue_Per_Customer",
    y="Retention_Rate",
    size="Customers",
    color="Churn_Rate",
    hover_name="State",
    hover_data=[
        "Total_Revenue",
        "Avg_Tenure",
        "Avg_Refund_Rate"
    ],
    color_continuous_scale="RdYlGn_r",
    template="plotly_dark",
    title="State Customer Value vs Retention"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Revenue per Customer",
    yaxis_title="Retention Rate (%)"
)

fig.show()


highest_demographic_value = demographic_summary.loc[
    demographic_summary["Revenue_Per_Customer"].idxmax()
]

highest_age_revenue = age_summary.loc[
    age_summary["Revenue_Per_Customer"].idxmax()
]

best_age_retention = age_summary.loc[
    age_summary["Retention_Rate"].idxmax()
]

highest_gender_revenue = gender_summary.loc[
    gender_summary["Revenue_Per_Customer"].idxmax()
]

highest_state_revenue = state_demographic.loc[
    state_demographic["Total_Revenue"].idxmax()
]

best_state_retention = state_demographic.loc[
    state_demographic["Retention_Rate"].idxmax()
]


business_summary = pd.DataFrame({
    "Business_Metric": [
        "Highest Demographic Revenue Value",
        "Highest Age-Group Revenue Value",
        "Best Age-Group Retention",
        "Highest Gender Revenue Value",
        "Highest Revenue State",
        "Best State Retention"
    ],

    "Category": [
        f"{highest_demographic_value['Gender']} / "
        f"{highest_demographic_value['Married']}",
        highest_age_revenue["Age_Group"],
        best_age_retention["Age_Group"],
        highest_gender_revenue["Gender"],
        highest_state_revenue["State"],
        best_state_retention["State"]
    ],

    "Value": [
        f"{highest_demographic_value['Revenue_Per_Customer']:,.2f}",
        f"{highest_age_revenue['Revenue_Per_Customer']:,.2f}",
        f"{best_age_retention['Retention_Rate']:.2f}%",
        f"{highest_gender_revenue['Revenue_Per_Customer']:,.2f}",
        f"{highest_state_revenue['Total_Revenue']:,.2f}",
        f"{best_state_retention['Retention_Rate']:.2f}%"
    ]
})

display(business_summary)


print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{highest_demographic_value['Gender']} / "
    f"{highest_demographic_value['Married']} customers generate the "
    f"highest revenue per customer at "
    f"{highest_demographic_value['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{highest_age_revenue['Age_Group']} has the highest revenue per "
    f"customer among age groups at "
    f"{highest_age_revenue['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{best_age_retention['Age_Group']} shows the strongest age-group "
    f"retention at {best_age_retention['Retention_Rate']:.2f}%."
)

print(
    f"{highest_gender_revenue['Gender']} generates the highest "
    f"revenue per customer at "
    f"{highest_gender_revenue['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{highest_state_revenue['State']} contributes the largest total "
    f"revenue of {highest_state_revenue['Total_Revenue']:,.2f}."
)

print(
    f"{best_state_retention['State']} has the strongest state-level "
    f"retention rate at {best_state_retention['Retention_Rate']:.2f}%."
)

print(
    "Demographic analysis helps identify customer groups that combine "
    "strong revenue value with healthy retention."
)

print(
    "Business teams can use these segments for targeted campaigns, "
    "regional growth planning and differentiated customer strategies."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 13
# ============================================================

# Which payment methods and billing preferences are associated
# with stronger revenue performance and customer retention?

# Business Goal:
# Evaluate payment behavior to identify profitable payment
# segments, retention patterns and potential billing risks.
# ============================================================

payment_summary = (
    df.groupby("Payment_Method", dropna=False)
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Refund_Rate=("Refund_Rate", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

payment_summary["Churn_Rate"] = (
    payment_summary["Churned"]
    .div(payment_summary["Customers"])
    .mul(100)
)

payment_summary["Retention_Rate"] = (
    payment_summary["Stayed"]
    .div(payment_summary["Customers"])
    .mul(100)
)

payment_summary["Revenue_Per_Customer"] = (
    payment_summary["Total_Revenue"]
    .div(payment_summary["Customers"])
)

payment_summary["Net_Revenue_Rate"] = (
    payment_summary["Net_Revenue"]
    .div(payment_summary["Total_Revenue"])
    .mul(100)
)

payment_summary["Revenue_Share"] = (
    payment_summary["Total_Revenue"]
    .div(payment_summary["Total_Revenue"].sum())
    .mul(100)
)

display(payment_summary.round(2))


billing_summary = (
    df.groupby(
        ["Is_Automatic_Payment", "Paperless_Billing"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Refund_Rate=("Refund_Rate", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

billing_summary["Churn_Rate"] = (
    billing_summary["Churned"]
    .div(billing_summary["Customers"])
    .mul(100)
)

billing_summary["Retention_Rate"] = (
    billing_summary["Stayed"]
    .div(billing_summary["Customers"])
    .mul(100)
)

billing_summary["Revenue_Per_Customer"] = (
    billing_summary["Total_Revenue"]
    .div(billing_summary["Customers"])
)

display(billing_summary.round(2))


payment_status = (
    df.groupby(
        ["Payment_Method", "Customer_Status"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Charge=("Monthly_Charge", "mean")
    )
    .reset_index()
)

display(payment_status.round(2))


fig = px.bar(
    payment_summary.sort_values("Revenue_Per_Customer"),
    x="Revenue_Per_Customer",
    y="Payment_Method",
    orientation="h",
    color="Churn_Rate",
    text="Revenue_Per_Customer",
    color_continuous_scale="RdYlGn_r",
    template="plotly_white",
    title="Revenue per Customer by Payment Method"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Revenue per Customer",
    yaxis_title="Payment Method"
)

fig.show()


fig = px.scatter(
    payment_summary,
    x="Avg_Monthly_Charge",
    y="Revenue_Per_Customer",
    size="Customers",
    color="Retention_Rate",
    text="Payment_Method",
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Payment Method: Monthly Charge vs Customer Value"
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Monthly Charge",
    yaxis_title="Revenue per Customer"
)

fig.show()


fig = px.bar(
    payment_summary.sort_values("Retention_Rate"),
    x="Payment_Method",
    y="Retention_Rate",
    color="Revenue_Per_Customer",
    text="Retention_Rate",
    color_continuous_scale="Plasma",
    template="plotly_white",
    title="Retention Rate by Payment Method"
)

fig.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Payment Method",
    yaxis_title="Retention Rate (%)"
)

fig.show()


fig = px.treemap(
    payment_summary,
    path=["Payment_Method"],
    values="Total_Revenue",
    color="Net_Revenue_Rate",
    color_continuous_scale="Turbo",
    template="plotly_dark",
    title="Revenue Contribution and Net Revenue Efficiency"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


fig = px.bar(
    billing_summary,
    x="Is_Automatic_Payment",
    y="Revenue_Per_Customer",
    color="Paperless_Billing",
    barmode="group",
    text="Revenue_Per_Customer",
    template="plotly_white",
    title="Customer Revenue by Payment Automation and Billing Preference"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Automatic Payment",
    yaxis_title="Revenue per Customer"
)

fig.show()


fig = px.density_heatmap(
    billing_summary,
    x="Is_Automatic_Payment",
    y="Paperless_Billing",
    z="Retention_Rate",
    text_auto=".1f",
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Retention Rate: Payment Automation vs Paperless Billing"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Automatic Payment",
    yaxis_title="Paperless Billing"
)

fig.show()


best_payment_value = payment_summary.loc[
    payment_summary["Revenue_Per_Customer"].idxmax()
]

best_payment_retention = payment_summary.loc[
    payment_summary["Retention_Rate"].idxmax()
]

highest_payment_revenue = payment_summary.loc[
    payment_summary["Total_Revenue"].idxmax()
]

highest_payment_churn = payment_summary.loc[
    payment_summary["Churn_Rate"].idxmax()
]

best_billing_value = billing_summary.loc[
    billing_summary["Revenue_Per_Customer"].idxmax()
]

best_billing_retention = billing_summary.loc[
    billing_summary["Retention_Rate"].idxmax()
]


business_summary = pd.DataFrame({
    "Business_Metric": [
        "Highest Revenue per Customer Payment Method",
        "Best Retention Payment Method",
        "Largest Revenue Contribution Payment Method",
        "Highest Churn Payment Method",
        "Highest Value Billing Segment",
        "Best Retention Billing Segment"
    ],

    "Segment": [
        best_payment_value["Payment_Method"],
        best_payment_retention["Payment_Method"],
        highest_payment_revenue["Payment_Method"],
        highest_payment_churn["Payment_Method"],
        f"{best_billing_value['Is_Automatic_Payment']} / "
        f"{best_billing_value['Paperless_Billing']}",
        f"{best_billing_retention['Is_Automatic_Payment']} / "
        f"{best_billing_retention['Paperless_Billing']}"
    ],

    "Value": [
        f"{best_payment_value['Revenue_Per_Customer']:,.2f}",
        f"{best_payment_retention['Retention_Rate']:.2f}%",
        f"{highest_payment_revenue['Total_Revenue']:,.2f}",
        f"{highest_payment_churn['Churn_Rate']:.2f}%",
        f"{best_billing_value['Revenue_Per_Customer']:,.2f}",
        f"{best_billing_retention['Retention_Rate']:.2f}%"
    ]
})

display(business_summary)


print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{best_payment_value['Payment_Method']} generates the highest "
    f"revenue per customer at "
    f"{best_payment_value['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{best_payment_retention['Payment_Method']} has the strongest "
    f"retention rate at "
    f"{best_payment_retention['Retention_Rate']:.2f}%."
)

print(
    f"{highest_payment_revenue['Payment_Method']} contributes the "
    f"largest total revenue of "
    f"{highest_payment_revenue['Total_Revenue']:,.2f}."
)

print(
    f"{highest_payment_churn['Payment_Method']} has the highest "
    f"churn rate at "
    f"{highest_payment_churn['Churn_Rate']:.2f}%."
)

print(
    f"The highest-value billing combination is "
    f"{best_billing_value['Is_Automatic_Payment']} automatic payment "
    f"and {best_billing_value['Paperless_Billing']} paperless billing, "
    f"with {best_billing_value['Revenue_Per_Customer']:,.2f} revenue "
    f"per customer."
)

print(
    f"The strongest billing combination for retention is "
    f"{best_billing_retention['Is_Automatic_Payment']} automatic payment "
    f"and {best_billing_retention['Paperless_Billing']} paperless billing, "
    f"with {best_billing_retention['Retention_Rate']:.2f}% retention."
)

print(
    "Payment behavior can be used to identify high-value billing "
    "segments and payment groups requiring churn intervention."
)

print(
    "Business Action: Promote reliable payment options and target "
    "high-churn payment segments with retention or billing incentives."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 15
# ============================================================

# How does customer referral behavior influence customer value,
# revenue contribution and retention across the customer lifecycle?

# Business Goal:
# Measure the business impact of referrals and identify referral
# segments that generate stronger customer value and retention.
# ============================================================


referral_summary = (
    df.groupby(
        ["Has_Referral", "Referral_Group"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Referrals=("Number_of_Referrals", "sum"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

referral_summary["Churn_Rate"] = (
    referral_summary["Churned"]
    .div(referral_summary["Customers"])
    .mul(100)
)

referral_summary["Retention_Rate"] = (
    referral_summary["Stayed"]
    .div(referral_summary["Customers"])
    .mul(100)
)

referral_summary["Revenue_Per_Customer"] = (
    referral_summary["Total_Revenue"]
    .div(referral_summary["Customers"])
)

referral_summary["Net_Revenue_Per_Customer"] = (
    referral_summary["Net_Revenue"]
    .div(referral_summary["Customers"])
)

referral_summary["Revenue_Share"] = (
    referral_summary["Total_Revenue"]
    .div(referral_summary["Total_Revenue"].sum())
    .mul(100)
)

display(referral_summary.round(2))


lifecycle_referral = (
    df.groupby(
        ["Customer_Lifecycle", "Has_Referral"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Referrals=("Number_of_Referrals", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

lifecycle_referral["Churn_Rate"] = (
    lifecycle_referral["Churned"]
    .div(lifecycle_referral["Customers"])
    .mul(100)
)

lifecycle_referral["Retention_Rate"] = (
    lifecycle_referral["Stayed"]
    .div(lifecycle_referral["Customers"])
    .mul(100)
)

lifecycle_referral["Revenue_Per_Customer"] = (
    lifecycle_referral["Total_Revenue"]
    .div(lifecycle_referral["Customers"])
)

display(lifecycle_referral.round(2))


referral_status = (
    df.groupby(
        ["Referral_Group", "Customer_Status"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum")
    )
    .reset_index()
)

display(referral_status.round(2))


fig = px.bar(
    referral_summary.sort_values("Revenue_Per_Customer"),
    x="Revenue_Per_Customer",
    y="Referral_Group",
    color="Has_Referral",
    orientation="h",
    text="Revenue_Per_Customer",
    barmode="group",
    template="plotly_white",
    title="Revenue per Customer by Referral Segment"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Revenue per Customer",
    yaxis_title="Referral Group"
)

fig.show()


fig = px.scatter(
    referral_summary,
    x="Total_Referrals",
    y="Revenue_Per_Customer",
    size="Customers",
    color="Retention_Rate",
    text="Referral_Group",
    hover_data=[
        "Has_Referral",
        "Churn_Rate",
        "Avg_Tenure"
    ],
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Referral Activity vs Customer Revenue"
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Total Referrals",
    yaxis_title="Revenue per Customer"
)

fig.show()


fig = px.bar(
    referral_summary.sort_values("Retention_Rate"),
    x="Referral_Group",
    y="Retention_Rate",
    color="Has_Referral",
    text="Retention_Rate",
    barmode="group",
    template="plotly_white",
    title="Retention Rate Across Referral Segments"
)

fig.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Referral Group",
    yaxis_title="Retention Rate (%)"
)

fig.show()


fig = px.treemap(
    referral_summary,
    path=["Has_Referral", "Referral_Group"],
    values="Total_Revenue",
    color="Revenue_Share",
    color_continuous_scale="Plasma",
    template="plotly_dark",
    title="Revenue Contribution of Referral Segments"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


fig = px.bar(
    lifecycle_referral,
    x="Customer_Lifecycle",
    y="Revenue_Per_Customer",
    color="Has_Referral",
    barmode="group",
    text="Revenue_Per_Customer",
    template="plotly_white",
    title="Customer Value Across Lifecycle and Referral Status"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Customer Lifecycle",
    yaxis_title="Revenue per Customer"
)

fig.show()


fig = px.scatter(
    lifecycle_referral,
    x="Avg_Referrals",
    y="Retention_Rate",
    size="Customers",
    color="Revenue_Per_Customer",
    symbol="Has_Referral",
    hover_data=[
        "Customer_Lifecycle",
        "Churn_Rate",
        "Avg_Tenure"
    ],
    color_continuous_scale="Turbo",
    template="plotly_dark",
    title="Referral Engagement vs Lifecycle Retention"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Referrals",
    yaxis_title="Retention Rate (%)"
)

fig.show()


best_referral_value = referral_summary.loc[
    referral_summary["Revenue_Per_Customer"].idxmax()
]

best_referral_retention = referral_summary.loc[
    referral_summary["Retention_Rate"].idxmax()
]

highest_referral_revenue = referral_summary.loc[
    referral_summary["Total_Revenue"].idxmax()
]

highest_referral_churn = referral_summary.loc[
    referral_summary["Churn_Rate"].idxmax()
]

best_lifecycle_value = lifecycle_referral.loc[
    lifecycle_referral["Revenue_Per_Customer"].idxmax()
]

best_lifecycle_retention = lifecycle_referral.loc[
    lifecycle_referral["Retention_Rate"].idxmax()
]


business_summary = pd.DataFrame({
    "Business_Metric": [
        "Highest Referral Customer Value",
        "Best Referral Retention",
        "Largest Referral Revenue Contribution",
        "Highest Referral Churn",
        "Highest Lifecycle Customer Value",
        "Best Lifecycle Retention"
    ],

    "Segment": [
        f"{best_referral_value['Has_Referral']} / "
        f"{best_referral_value['Referral_Group']}",

        f"{best_referral_retention['Has_Referral']} / "
        f"{best_referral_retention['Referral_Group']}",

        f"{highest_referral_revenue['Has_Referral']} / "
        f"{highest_referral_revenue['Referral_Group']}",

        f"{highest_referral_churn['Has_Referral']} / "
        f"{highest_referral_churn['Referral_Group']}",

        f"{best_lifecycle_value['Customer_Lifecycle']} / "
        f"{best_lifecycle_value['Has_Referral']}",

        f"{best_lifecycle_retention['Customer_Lifecycle']} / "
        f"{best_lifecycle_retention['Has_Referral']}"
    ],

    "Value": [
        f"{best_referral_value['Revenue_Per_Customer']:,.2f}",
        f"{best_referral_retention['Retention_Rate']:.2f}%",
        f"{highest_referral_revenue['Total_Revenue']:,.2f}",
        f"{highest_referral_churn['Churn_Rate']:.2f}%",
        f"{best_lifecycle_value['Revenue_Per_Customer']:,.2f}",
        f"{best_lifecycle_retention['Retention_Rate']:.2f}%"
    ]
})

display(business_summary)


print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{best_referral_value['Has_Referral']} / "
    f"{best_referral_value['Referral_Group']} generates the highest "
    f"revenue per customer at "
    f"{best_referral_value['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{best_referral_retention['Has_Referral']} / "
    f"{best_referral_retention['Referral_Group']} has the strongest "
    f"retention rate at "
    f"{best_referral_retention['Retention_Rate']:.2f}%."
)

print(
    f"{highest_referral_revenue['Has_Referral']} / "
    f"{highest_referral_revenue['Referral_Group']} contributes the "
    f"largest total revenue of "
    f"{highest_referral_revenue['Total_Revenue']:,.2f}."
)

print(
    f"{highest_referral_churn['Has_Referral']} / "
    f"{highest_referral_churn['Referral_Group']} has the highest "
    f"churn rate at "
    f"{highest_referral_churn['Churn_Rate']:.2f}%."
)

print(
    f"The highest-value lifecycle/referral combination is "
    f"{best_lifecycle_value['Customer_Lifecycle']} with "
    f"{best_lifecycle_value['Has_Referral']} referral status, "
    f"generating "
    f"{best_lifecycle_value['Revenue_Per_Customer']:,.2f} "
    f"revenue per customer."
)

print(
    f"The strongest lifecycle/referral retention combination has "
    f"a retention rate of "
    f"{best_lifecycle_retention['Retention_Rate']:.2f}%."
)

print(
    "Referral behavior can help identify customer groups with "
    "higher commercial value and stronger long-term engagement."
)

print(
    "Business Action: Strengthen referral programs in high-value "
    "segments while targeting low-retention referral groups "
    "with tailored engagement strategies."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 16
# ============================================================

# How does customer revenue and retention performance change
# across the customer tenure timeline?

# Business Goal:
# Analyze revenue, customer value, churn and retention behavior
# across tenure to identify important customer lifecycle stages.
# ============================================================

import pandas as pd
import numpy as np
import plotly.express as px


time_df = (
    df[
        [
            "Customer_ID",
            "Tenure_in_Months",
            "Customer_Status",
            "Total_Revenue",
            "Net_Revenue",
            "Revenue_Per_Month",
            "Monthly_Charge",
            "Total_Refunds"
        ]
    ]
    .dropna(subset=["Tenure_in_Months"])
    .copy()
)

time_df["Tenure_in_Months"] = pd.to_numeric(
    time_df["Tenure_in_Months"],
    errors="coerce"
)

time_df = time_df.dropna(
    subset=["Tenure_in_Months"]
)

time_df["Tenure_in_Months"] = (
    time_df["Tenure_in_Months"]
    .astype(int)
)


tenure_summary = (
    time_df
    .groupby("Tenure_in_Months")
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Revenue_Per_Month=("Revenue_Per_Month", "mean"),
        Total_Refunds=("Total_Refunds", "sum"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
    .sort_values("Tenure_in_Months")
)

tenure_summary["Churn_Rate"] = (
    tenure_summary["Churned"]
    .div(tenure_summary["Customers"])
    .mul(100)
)

tenure_summary["Retention_Rate"] = (
    tenure_summary["Stayed"]
    .div(tenure_summary["Customers"])
    .mul(100)
)

tenure_summary["Revenue_Per_Customer"] = (
    tenure_summary["Total_Revenue"]
    .div(tenure_summary["Customers"])
)

tenure_summary["Net_Revenue_Per_Customer"] = (
    tenure_summary["Net_Revenue"]
    .div(tenure_summary["Customers"])
)

tenure_summary["Refund_Rate"] = (
    tenure_summary["Total_Refunds"]
    .div(tenure_summary["Total_Revenue"])
    .mul(100)
    .replace([np.inf, -np.inf], np.nan)
)

display(
    tenure_summary.round(2)
)


tenure_summary["Rolling_Avg_Revenue"] = (
    tenure_summary["Avg_Revenue"]
    .rolling(
        window=6,
        min_periods=1
    )
    .mean()
)

tenure_summary["Rolling_Avg_Churn"] = (
    tenure_summary["Churn_Rate"]
    .rolling(
        window=6,
        min_periods=1
    )
    .mean()
)

tenure_summary["Rolling_Avg_Retention"] = (
    tenure_summary["Retention_Rate"]
    .rolling(
        window=6,
        min_periods=1
    )
    .mean()
)

display(
    tenure_summary[
        [
            "Tenure_in_Months",
            "Avg_Revenue",
            "Rolling_Avg_Revenue",
            "Churn_Rate",
            "Rolling_Avg_Churn",
            "Retention_Rate",
            "Rolling_Avg_Retention"
        ]
    ].round(2)
)


fig = px.line(
    tenure_summary,
    x="Tenure_in_Months",
    y=[
        "Avg_Revenue",
        "Avg_Net_Revenue",
        "Avg_Revenue_Per_Month"
    ],
    markers=True,
    template="plotly_white",
    title="Customer Revenue Trend Across Tenure"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Customer Tenure (Months)",
    yaxis_title="Revenue"
)

fig.show()


fig = px.line(
    tenure_summary,
    x="Tenure_in_Months",
    y=[
        "Churn_Rate",
        "Retention_Rate"
    ],
    markers=True,
    template="plotly_dark",
    title="Churn and Retention Trend Across Tenure"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Customer Tenure (Months)",
    yaxis_title="Rate (%)"
)

fig.show()


fig = px.area(
    tenure_summary,
    x="Tenure_in_Months",
    y="Total_Revenue",
    markers=True,
    template="plotly_white",
    title="Cumulative Revenue Contribution Across Tenure"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Customer Tenure (Months)",
    yaxis_title="Total Revenue"
)

fig.show()


fig = px.line(
    tenure_summary,
    x="Tenure_in_Months",
    y=[
        "Avg_Revenue",
        "Rolling_Avg_Revenue"
    ],
    markers=True,
    template="plotly_dark",
    title="Average Revenue and Rolling Revenue Trend"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Customer Tenure (Months)",
    yaxis_title="Revenue"
)

fig.show()


fig = px.scatter(
    tenure_summary,
    x="Tenure_in_Months",
    y="Revenue_Per_Customer",
    size="Customers",
    color="Churn_Rate",
    hover_data=[
        "Retention_Rate",
        "Avg_Monthly_Charge",
        "Refund_Rate"
    ],
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Customer Value and Churn Risk Across Tenure"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Customer Tenure (Months)",
    yaxis_title="Revenue per Customer"
)

fig.show()


fig = px.line(
    tenure_summary,
    x="Tenure_in_Months",
    y=[
        "Refund_Rate",
        "Avg_Monthly_Charge"
    ],
    markers=True,
    template="plotly_dark",
    title="Refund and Monthly Charge Behavior Across Tenure"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Customer Tenure (Months)",
    yaxis_title="Metric Value"

)

fig.show()


peak_revenue = tenure_summary.loc[
    tenure_summary["Revenue_Per_Customer"].idxmax()
]

peak_retention = tenure_summary.loc[
    tenure_summary["Retention_Rate"].idxmax()
]

peak_churn = tenure_summary.loc[
    tenure_summary["Churn_Rate"].idxmax()
]

peak_total_revenue = tenure_summary.loc[
    tenure_summary["Total_Revenue"].idxmax()
]

lowest_refund = tenure_summary.loc[
    tenure_summary["Refund_Rate"].idxmin()
]

highest_refund = tenure_summary.loc[
    tenure_summary["Refund_Rate"].idxmax()
]


business_summary = pd.DataFrame({
    "Metric": [
        "Highest Revenue per Customer",
        "Highest Retention",
        "Highest Churn",
        "Highest Total Revenue",
        "Lowest Refund Rate",
        "Highest Refund Rate"
    ],

    "Tenure_Month": [
        peak_revenue["Tenure_in_Months"],
        peak_retention["Tenure_in_Months"],
        peak_churn["Tenure_in_Months"],
        peak_total_revenue["Tenure_in_Months"],
        lowest_refund["Tenure_in_Months"],
        highest_refund["Tenure_in_Months"]
    ],

    "Value": [
        f"{peak_revenue['Revenue_Per_Customer']:,.2f}",
        f"{peak_retention['Retention_Rate']:.2f}%",
        f"{peak_churn['Churn_Rate']:.2f}%",
        f"{peak_total_revenue['Total_Revenue']:,.2f}",
        f"{lowest_refund['Refund_Rate']:.2f}%",
        f"{highest_refund['Refund_Rate']:.2f}%"
    ]
})

display(business_summary)


print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"Highest revenue per customer occurs at "
    f"{int(peak_revenue['Tenure_in_Months'])} months of tenure, "
    f"with {peak_revenue['Revenue_Per_Customer']:,.2f} "
    f"revenue per customer."
)

print(
    f"Highest retention is observed at "
    f"{int(peak_retention['Tenure_in_Months'])} months, "
    f"reaching {peak_retention['Retention_Rate']:.2f}%."
)

print(
    f"Highest churn is observed at "
    f"{int(peak_churn['Tenure_in_Months'])} months, "
    f"reaching {peak_churn['Churn_Rate']:.2f}%."
)

print(
    f"The tenure point generating the highest total revenue is "
    f"{int(peak_total_revenue['Tenure_in_Months'])} months, "
    f"with {peak_total_revenue['Total_Revenue']:,.2f}."
)

print(
    f"Refund rate ranges from "
    f"{lowest_refund['Refund_Rate']:.2f}% to "
    f"{highest_refund['Refund_Rate']:.2f}% across tenure."
)

print(
    "The rolling revenue and retention trends help identify "
    "whether customer value and loyalty are strengthening or weakening "
    "as tenure increases."
)

print(
    "Business Action: Focus retention campaigns on tenure stages "
    "where churn rises, while strengthening engagement at stages "
    "where customer value and retention are highest."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 17
# ============================================================

# How does customer value and revenue efficiency vary across
# customer lifecycle and revenue segments?

# Business Goal:
# Identify lifecycle stages and revenue segments that generate
# strong customer value, profitability and retention.
# ============================================================

lifecycle_summary = (
    df.groupby(
        ["Customer_Lifecycle", "Revenue_Segment"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Charge_Revenue_Ratio=("Charge_Revenue_Ratio", "mean"),
        Avg_Revenue_Per_Month=("Revenue_Per_Month", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

lifecycle_summary["Churn_Rate"] = (
    lifecycle_summary["Churned"]
    .div(lifecycle_summary["Customers"])
    .mul(100)
)

lifecycle_summary["Retention_Rate"] = (
    lifecycle_summary["Stayed"]
    .div(lifecycle_summary["Customers"])
    .mul(100)
)

lifecycle_summary["Revenue_Per_Customer"] = (
    lifecycle_summary["Total_Revenue"]
    .div(lifecycle_summary["Customers"])
)

lifecycle_summary["Net_Revenue_Per_Customer"] = (
    lifecycle_summary["Net_Revenue"]
    .div(lifecycle_summary["Customers"])
)

lifecycle_summary["Revenue_Share"] = (
    lifecycle_summary["Total_Revenue"]
    .div(lifecycle_summary["Total_Revenue"].sum())
    .mul(100)
)

display(lifecycle_summary.round(2))


segment_summary = (
    df.groupby("Revenue_Segment", dropna=False)
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Charge_Revenue_Ratio=("Charge_Revenue_Ratio", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

segment_summary["Churn_Rate"] = (
    segment_summary["Churned"]
    .div(segment_summary["Customers"])
    .mul(100)
)

segment_summary["Retention_Rate"] = (
    segment_summary["Stayed"]
    .div(segment_summary["Customers"])
    .mul(100)
)

segment_summary["Revenue_Per_Customer"] = (
    segment_summary["Total_Revenue"]
    .div(segment_summary["Customers"])
)

segment_summary["Net_Revenue_Per_Customer"] = (
    segment_summary["Net_Revenue"]
    .div(segment_summary["Customers"])
)

display(segment_summary.round(2))


fig = px.bar(
    lifecycle_summary,
    x="Customer_Lifecycle",
    y="Revenue_Per_Customer",
    color="Revenue_Segment",
    barmode="group",
    text="Revenue_Per_Customer",
    template="plotly_white",
    title="Revenue per Customer by Lifecycle and Revenue Segment"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Customer Lifecycle",
    yaxis_title="Revenue per Customer"
)

fig.show()


fig = px.scatter(
    lifecycle_summary,
    x="Avg_Tenure",
    y="Net_Revenue_Per_Customer",
    size="Customers",
    color="Revenue_Segment",
    symbol="Customer_Lifecycle",
    hover_data=[
        "Churn_Rate",
        "Retention_Rate",
        "Avg_Charge_Revenue_Ratio"
    ],
    template="plotly_dark",
    title="Customer Tenure vs Net Revenue Efficiency"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Tenure (Months)",
    yaxis_title="Net Revenue per Customer"
)

fig.show()


fig = px.bar(
    segment_summary.sort_values("Retention_Rate"),
    x="Revenue_Segment",
    y="Retention_Rate",
    text="Retention_Rate",
    color="Churn_Rate",
    color_continuous_scale="RdYlBu",
    template="plotly_white",
    title="Retention Performance Across Revenue Segments"
)

fig.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Revenue Segment",
    yaxis_title="Retention Rate (%)"
)

fig.show()


fig = px.treemap(
    lifecycle_summary,
    path=["Customer_Lifecycle", "Revenue_Segment"],
    values="Total_Revenue",
    color="Net_Revenue_Per_Customer",
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Revenue Contribution by Lifecycle and Segment"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


fig = px.scatter(
    segment_summary,
    x="Avg_Charge_Revenue_Ratio",
    y="Net_Revenue_Per_Customer",
    size="Customers",
    color="Revenue_Segment",
    hover_data=[
        "Churn_Rate",
        "Retention_Rate",
        "Avg_Tenure"
    ],
    template="plotly_white",
    title="Revenue Efficiency vs Net Customer Value"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Charge-Revenue Ratio",
    yaxis_title="Net Revenue per Customer"
)

fig.show()


fig = px.bar(
    lifecycle_summary.sort_values("Churn_Rate"),
    x="Customer_Lifecycle",
    y="Churn_Rate",
    color="Revenue_Segment",
    barmode="group",
    text="Churn_Rate",
    template="plotly_dark",
    title="Churn Risk Across Customer Lifecycle"
)

fig.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Customer Lifecycle",
    yaxis_title="Churn Rate (%)"
)

fig.show()


best_value = lifecycle_summary.loc[
    lifecycle_summary["Net_Revenue_Per_Customer"].idxmax()
]

best_retention = lifecycle_summary.loc[
    lifecycle_summary["Retention_Rate"].idxmax()
]

highest_churn = lifecycle_summary.loc[
    lifecycle_summary["Churn_Rate"].idxmax()
]

highest_revenue = lifecycle_summary.loc[
    lifecycle_summary["Total_Revenue"].idxmax()
]

best_efficiency = lifecycle_summary.loc[
    lifecycle_summary["Avg_Charge_Revenue_Ratio"].idxmax()
]


business_summary = pd.DataFrame({
    "Business_Metric": [
        "Highest Net Revenue per Customer",
        "Highest Retention",
        "Highest Churn",
        "Highest Total Revenue",
        "Highest Charge-Revenue Ratio"
    ],

    "Lifecycle": [
        best_value["Customer_Lifecycle"],
        best_retention["Customer_Lifecycle"],
        highest_churn["Customer_Lifecycle"],
        highest_revenue["Customer_Lifecycle"],
        best_efficiency["Customer_Lifecycle"]
    ],

    "Revenue_Segment": [
        best_value["Revenue_Segment"],
        best_retention["Revenue_Segment"],
        highest_churn["Revenue_Segment"],
        highest_revenue["Revenue_Segment"],
        best_efficiency["Revenue_Segment"]
    ],

    "Value": [
        f"{best_value['Net_Revenue_Per_Customer']:,.2f}",
        f"{best_retention['Retention_Rate']:.2f}%",
        f"{highest_churn['Churn_Rate']:.2f}%",
        f"{highest_revenue['Total_Revenue']:,.2f}",
        f"{best_efficiency['Avg_Charge_Revenue_Ratio']:.2f}"
    ]
})

display(business_summary)


print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{best_value['Customer_Lifecycle']} / "
    f"{best_value['Revenue_Segment']} has the highest net revenue "
    f"per customer at "
    f"{best_value['Net_Revenue_Per_Customer']:,.2f}."
)

print(
    f"{best_retention['Customer_Lifecycle']} / "
    f"{best_retention['Revenue_Segment']} records the strongest "
    f"retention rate of "
    f"{best_retention['Retention_Rate']:.2f}%."
)

print(
    f"{highest_churn['Customer_Lifecycle']} / "
    f"{highest_churn['Revenue_Segment']} has the highest churn rate "
    f"at {highest_churn['Churn_Rate']:.2f}%."
)

print(
    f"{highest_revenue['Customer_Lifecycle']} / "
    f"{highest_revenue['Revenue_Segment']} contributes the highest "
    f"total revenue of {highest_revenue['Total_Revenue']:,.2f}."
)

print(
    f"{best_efficiency['Customer_Lifecycle']} / "
    f"{best_efficiency['Revenue_Segment']} has the highest average "
    f"charge-revenue ratio of "
    f"{best_efficiency['Avg_Charge_Revenue_Ratio']:.2f}."
)

print(
    "The analysis highlights which lifecycle stages combine "
    "customer value, revenue efficiency and retention strength."
)

print(
    "Business Action: Prioritize high-value lifecycle segments "
    "for expansion while creating targeted retention strategies "
    "for high-churn segments."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 18
# ============================================================

# Which states have the strongest customer base, revenue
# performance and retention, and where are the major business risks?

# Business Goal:
# Compare geographic customer performance to identify high-value
# markets and states requiring retention or revenue improvement.
# ============================================================

state_summary = (
    df.groupby("State", dropna=False)
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

state_summary["Churn_Rate"] = (
    state_summary["Churned"]
    .div(state_summary["Customers"])
    .mul(100)
)

state_summary["Retention_Rate"] = (
    state_summary["Stayed"]
    .div(state_summary["Customers"])
    .mul(100)
)

state_summary["Revenue_Per_Customer"] = (
    state_summary["Total_Revenue"]
    .div(state_summary["Customers"])
)

state_summary["Net_Revenue_Per_Customer"] = (
    state_summary["Net_Revenue"]
    .div(state_summary["Customers"])
)

state_summary["Revenue_Share"] = (
    state_summary["Total_Revenue"]
    .div(state_summary["Total_Revenue"].sum())
    .mul(100)
)

display(state_summary.round(2))


state_status = (
    df.groupby(
        ["State", "Customer_Status"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum")
    )
    .reset_index()
)

display(state_status.round(2))


top_states = (
    state_summary
    .nlargest(10, "Total_Revenue")
    .sort_values("Total_Revenue")
)

fig = px.bar(
    top_states,
    x="Total_Revenue",
    y="State",
    orientation="h",
    text="Total_Revenue",
    color="Revenue_Per_Customer",
    color_continuous_scale="Viridis",
    template="plotly_white",
    title="Top 10 States by Total Revenue"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Total Revenue",
    yaxis_title="State"
)

fig.show()


fig = px.scatter(
    state_summary,
    x="Customers",
    y="Revenue_Per_Customer",
    size="Total_Revenue",
    color="Churn_Rate",
    hover_data=[
        "State",
        "Retention_Rate",
        "Avg_Tenure",
        "Net_Revenue_Per_Customer"
    ],
    color_continuous_scale="Turbo",
    template="plotly_dark",
    title="Customer Base Size vs Revenue per Customer"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Number of Customers",
    yaxis_title="Revenue per Customer"
)

fig.show()


risk_states = (
    state_summary
    .nlargest(10, "Churn_Rate")
    .sort_values("Churn_Rate")
)

fig = px.bar(
    risk_states,
    x="Churn_Rate",
    y="State",
    orientation="h",
    text="Churn_Rate",
    color="Revenue_Per_Customer",
    color_continuous_scale="RdYlBu",
    template="plotly_white",
    title="Top 10 States by Churn Risk"
)

fig.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Churn Rate (%)",
    yaxis_title="State"
)

fig.show()


fig = px.bar(
    state_summary.nlargest(10, "Revenue_Per_Customer"),
    x="State",
    y="Revenue_Per_Customer",
    text="Revenue_Per_Customer",
    color="Retention_Rate",
    color_continuous_scale="Plasma",
    template="plotly_dark",
    title="Top States by Revenue per Customer"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="State",
    yaxis_title="Revenue per Customer"
)

fig.show()


fig = px.scatter(
    state_summary,
    x="Avg_Tenure",
    y="Retention_Rate",
    size="Customers",
    color="Net_Revenue_Per_Customer",
    hover_data=[
        "State",
        "Churn_Rate",
        "Revenue_Per_Customer"
    ],
    color_continuous_scale="Viridis",
    template="plotly_white",
    title="State Tenure vs Retention Performance"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Tenure (Months)",
    yaxis_title="Retention Rate (%)"
)

fig.show()


fig = px.treemap(
    state_summary,
    path=["State"],
    values="Total_Revenue",
    color="Net_Revenue_Per_Customer",
    color_continuous_scale="Magma",
    template="plotly_dark",
    title="State Revenue Contribution and Customer Value"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


highest_revenue_state = state_summary.loc[
    state_summary["Total_Revenue"].idxmax()
]

highest_value_state = state_summary.loc[
    state_summary["Revenue_Per_Customer"].idxmax()
]

highest_churn_state = state_summary.loc[
    state_summary["Churn_Rate"].idxmax()
]

highest_retention_state = state_summary.loc[
    state_summary["Retention_Rate"].idxmax()
]

highest_net_value_state = state_summary.loc[
    state_summary["Net_Revenue_Per_Customer"].idxmax()
]


business_summary = pd.DataFrame({
    "Business_Metric": [
        "Highest Total Revenue",
        "Highest Revenue per Customer",
        "Highest Churn Risk",
        "Highest Retention",
        "Highest Net Revenue per Customer"
    ],

    "State": [
        highest_revenue_state["State"],
        highest_value_state["State"],
        highest_churn_state["State"],
        highest_retention_state["State"],
        highest_net_value_state["State"]
    ],

    "Value": [
        f"{highest_revenue_state['Total_Revenue']:,.2f}",
        f"{highest_value_state['Revenue_Per_Customer']:,.2f}",
        f"{highest_churn_state['Churn_Rate']:.2f}%",
        f"{highest_retention_state['Retention_Rate']:.2f}%",
        f"{highest_net_value_state['Net_Revenue_Per_Customer']:,.2f}"
    ]
})

display(business_summary)


print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{highest_revenue_state['State']} generates the highest total "
    f"revenue at {highest_revenue_state['Total_Revenue']:,.2f}."
)

print(
    f"{highest_value_state['State']} has the highest revenue per "
    f"customer at {highest_value_state['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{highest_churn_state['State']} has the highest churn rate "
    f"at {highest_churn_state['Churn_Rate']:.2f}%."
)

print(
    f"{highest_retention_state['State']} records the strongest "
    f"retention rate at {highest_retention_state['Retention_Rate']:.2f}%."
)

print(
    f"{highest_net_value_state['State']} generates the highest net "
    f"revenue per customer at "
    f"{highest_net_value_state['Net_Revenue_Per_Customer']:,.2f}."
)

print(
    "Geographic performance shows that customer volume and customer "
    "value are not necessarily concentrated in the same states."
)

print(
    "Business Action: Protect high-revenue markets, investigate "
    "high-churn states and develop state-level retention strategies "
    "based on customer value and tenure."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 19
# ============================================================

# How do payment behaviors affect customer retention,
# revenue generation and refund exposure?

# Business Goal:
# Identify payment methods and billing behaviors associated
# with stronger retention, higher customer value and lower
# revenue leakage.
# ============================================================

payment_summary = (
    df.groupby(
        ["Payment_Method", "Is_Automatic_Payment"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Total_Refunds=("Total_Refunds", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Refund_Rate=("Refund_Rate", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

payment_summary["Churn_Rate"] = (
    payment_summary["Churned"]
    .div(payment_summary["Customers"])
    .mul(100)
)

payment_summary["Retention_Rate"] = (
    payment_summary["Stayed"]
    .div(payment_summary["Customers"])
    .mul(100)
)

payment_summary["Revenue_Per_Customer"] = (
    payment_summary["Total_Revenue"]
    .div(payment_summary["Customers"])
)

payment_summary["Refund_Per_Customer"] = (
    payment_summary["Total_Refunds"]
    .div(payment_summary["Customers"])
)

payment_summary["Net_Revenue_Per_Customer"] = (
    payment_summary["Net_Revenue"]
    .div(payment_summary["Customers"])
)

payment_summary["Revenue_Share"] = (
    payment_summary["Total_Revenue"]
    .div(payment_summary["Total_Revenue"].sum())
    .mul(100)
)

display(payment_summary.round(2))


billing_summary = (
    df.groupby("Paperless_Billing", dropna=False)
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Total_Refunds=("Total_Refunds", "sum"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Refund_Rate=("Refund_Rate", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

billing_summary["Churn_Rate"] = (
    billing_summary["Churned"]
    .div(billing_summary["Customers"])
    .mul(100)
)

billing_summary["Retention_Rate"] = (
    billing_summary["Stayed"]
    .div(billing_summary["Customers"])
    .mul(100)
)

billing_summary["Revenue_Per_Customer"] = (
    billing_summary["Total_Revenue"]
    .div(billing_summary["Customers"])
)

billing_summary["Net_Revenue_Per_Customer"] = (
    billing_summary["Net_Revenue"]
    .div(billing_summary["Customers"])
)

display(billing_summary.round(2))


fig = px.bar(
    payment_summary,
    x="Payment_Method",
    y="Revenue_Per_Customer",
    color="Is_Automatic_Payment",
    barmode="group",
    text="Revenue_Per_Customer",
    template="plotly_white",
    title="Revenue per Customer by Payment Method"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Payment Method",
    yaxis_title="Revenue per Customer"
)

fig.show()


fig = px.scatter(
    payment_summary,
    x="Churn_Rate",
    y="Net_Revenue_Per_Customer",
    size="Customers",
    color="Payment_Method",
    symbol="Is_Automatic_Payment",
    hover_data=[
        "Retention_Rate",
        "Avg_Refund_Rate",
        "Refund_Per_Customer"
    ],
    template="plotly_dark",
    title="Payment Behavior: Churn Risk vs Customer Value"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Churn Rate (%)",
    yaxis_title="Net Revenue per Customer"
)

fig.show()


fig = px.bar(
    payment_summary.sort_values("Refund_Per_Customer"),
    x="Refund_Per_Customer",
    y="Payment_Method",
    color="Is_Automatic_Payment",
    orientation="h",
    text="Refund_Per_Customer",
    template="plotly_white",
    title="Refund Exposure per Customer by Payment Method"
)

fig.update_traces(
    texttemplate="%{text:,.2f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Refund per Customer",
    yaxis_title="Payment Method"
)

fig.show()


fig = px.bar(
    billing_summary,
    x="Paperless_Billing",
    y="Retention_Rate",
    color="Churn_Rate",
    text="Retention_Rate",
    color_continuous_scale="Turbo",
    template="plotly_dark",
    title="Retention Performance by Billing Preference"
)

fig.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Paperless Billing",
    yaxis_title="Retention Rate (%)"
)

fig.show()


fig = px.scatter(
    billing_summary,
    x="Avg_Refund_Rate",
    y="Net_Revenue_Per_Customer",
    size="Customers",
    color="Paperless_Billing",
    hover_data=[
        "Churn_Rate",
        "Retention_Rate",
        "Total_Refunds"
    ],
    template="plotly_white",
    title="Refund Rate vs Net Revenue by Billing Preference"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Refund Rate",
    yaxis_title="Net Revenue per Customer"
)

fig.show()


fig = px.treemap(
    payment_summary,
    path=["Payment_Method", "Is_Automatic_Payment"],
    values="Total_Revenue",
    color="Net_Revenue_Per_Customer",
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Revenue Contribution by Payment Behavior"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


best_payment_value = payment_summary.loc[
    payment_summary["Net_Revenue_Per_Customer"].idxmax()
]

highest_payment_churn = payment_summary.loc[
    payment_summary["Churn_Rate"].idxmax()
]

highest_payment_refund = payment_summary.loc[
    payment_summary["Refund_Per_Customer"].idxmax()
]

best_payment_retention = payment_summary.loc[
    payment_summary["Retention_Rate"].idxmax()
]

best_billing_retention = billing_summary.loc[
    billing_summary["Retention_Rate"].idxmax()
]

highest_billing_refund = billing_summary.loc[
    billing_summary["Avg_Refund_Rate"].idxmax()
]


business_summary = pd.DataFrame({
    "Business_Metric": [
        "Highest Net Revenue per Customer",
        "Highest Payment Churn",
        "Highest Refund per Customer",
        "Highest Payment Retention",
        "Best Billing Retention",
        "Highest Billing Refund Rate"
    ],

    "Payment_Method": [
        best_payment_value["Payment_Method"],
        highest_payment_churn["Payment_Method"],
        highest_payment_refund["Payment_Method"],
        best_payment_retention["Payment_Method"],
        "All Payment Methods",
        "All Payment Methods"
    ],

    "Automatic_Payment": [
        best_payment_value["Is_Automatic_Payment"],
        highest_payment_churn["Is_Automatic_Payment"],
        highest_payment_refund["Is_Automatic_Payment"],
        best_payment_retention["Is_Automatic_Payment"],
        "-",
        "-"
    ],

    "Value": [
        f"{best_payment_value['Net_Revenue_Per_Customer']:,.2f}",
        f"{highest_payment_churn['Churn_Rate']:.2f}%",
        f"{highest_payment_refund['Refund_Per_Customer']:,.2f}",
        f"{best_payment_retention['Retention_Rate']:.2f}%",
        f"{best_billing_retention['Retention_Rate']:.2f}%",
        f"{highest_billing_refund['Avg_Refund_Rate']:.2f}"
    ]
})

display(business_summary)


print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{best_payment_value['Payment_Method']} payment users with "
    f"automatic payment = {best_payment_value['Is_Automatic_Payment']} "
    f"generate the highest net revenue per customer at "
    f"{best_payment_value['Net_Revenue_Per_Customer']:,.2f}."
)

print(
    f"{highest_payment_churn['Payment_Method']} with automatic payment "
    f"= {highest_payment_churn['Is_Automatic_Payment']} has the highest "
    f"churn rate at {highest_payment_churn['Churn_Rate']:.2f}%."
)

print(
    f"{highest_payment_refund['Payment_Method']} with automatic payment "
    f"= {highest_payment_refund['Is_Automatic_Payment']} has the highest "
    f"refund exposure per customer at "
    f"{highest_payment_refund['Refund_Per_Customer']:,.2f}."
)

print(
    f"{best_payment_retention['Payment_Method']} with automatic payment "
    f"= {best_payment_retention['Is_Automatic_Payment']} records the "
    f"highest payment-group retention at "
    f"{best_payment_retention['Retention_Rate']:.2f}%."
)

print(
    f"{best_billing_retention['Paperless_Billing']} paperless-billing "
    f"group has the strongest retention rate at "
    f"{best_billing_retention['Retention_Rate']:.2f}%."
)

print(
    f"{highest_billing_refund['Paperless_Billing']} billing group has "
    f"the highest average refund rate at "
    f"{highest_billing_refund['Avg_Refund_Rate']:.2f}."
)

print(
    "Payment behavior reveals opportunities to improve retention, "
    "reduce refund exposure and increase customer revenue efficiency."
)

print(
    "Business Action: Promote reliable payment channels, investigate "
    "high-refund payment groups and optimize automatic-payment adoption "
    "where it is associated with stronger customer value."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 20
# ============================================================

# How do customer referrals and value deals influence customer
# acquisition, revenue generation and retention?

# Business Goal:
# Measure the business impact of referrals and value deals
# across customer value, revenue and retention.
# ============================================================

referral_summary = (
    df.groupby(
        ["Referral_Group", "Has_Referral"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Referrals=("Number_of_Referrals", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

referral_summary["Churn_Rate"] = (
    referral_summary["Churned"]
    .div(referral_summary["Customers"])
    .mul(100)
)

referral_summary["Retention_Rate"] = (
    referral_summary["Stayed"]
    .div(referral_summary["Customers"])
    .mul(100)
)

referral_summary["Revenue_Per_Customer"] = (
    referral_summary["Total_Revenue"]
    .div(referral_summary["Customers"])
)

referral_summary["Net_Revenue_Per_Customer"] = (
    referral_summary["Net_Revenue"]
    .div(referral_summary["Customers"])
)

referral_summary["Revenue_Share"] = (
    referral_summary["Total_Revenue"]
    .div(referral_summary["Total_Revenue"].sum())
    .mul(100)
)

display(referral_summary.round(2))


deal_summary = (
    df.groupby(
        ["Value_Deal", "Has_Value_Deal"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=("Customer_Status", lambda x: (x == "Churned").sum()),
        Stayed=("Customer_Status", lambda x: (x == "Stayed").sum())
    )
    .reset_index()
)

deal_summary["Churn_Rate"] = (
    deal_summary["Churned"]
    .div(deal_summary["Customers"])
    .mul(100)
)

deal_summary["Retention_Rate"] = (
    deal_summary["Stayed"]
    .div(deal_summary["Customers"])
    .mul(100)
)

deal_summary["Revenue_Per_Customer"] = (
    deal_summary["Total_Revenue"]
    .div(deal_summary["Customers"])
)

deal_summary["Net_Revenue_Per_Customer"] = (
    deal_summary["Net_Revenue"]
    .div(deal_summary["Customers"])
)

display(deal_summary.round(2))


fig = px.bar(
    referral_summary,
    x="Referral_Group",
    y="Revenue_Per_Customer",
    color="Has_Referral",
    barmode="group",
    text="Revenue_Per_Customer",
    template="plotly_white",
    title="Revenue per Customer by Referral Group"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Referral Group",
    yaxis_title="Revenue per Customer"
)

fig.show()


fig = px.scatter(
    referral_summary,
    x="Avg_Referrals",
    y="Net_Revenue_Per_Customer",
    size="Customers",
    color="Retention_Rate",
    symbol="Has_Referral",
    hover_data=[
        "Referral_Group",
        "Churn_Rate",
        "Revenue_Share"
    ],
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Referral Activity vs Customer Net Revenue"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Number of Referrals",
    yaxis_title="Net Revenue per Customer"
)

fig.show()


fig = px.bar(
    referral_summary.sort_values("Retention_Rate"),
    x="Retention_Rate",
    y="Referral_Group",
    color="Has_Referral",
    orientation="h",
    text="Retention_Rate",
    template="plotly_white",
    title="Retention Rate Across Referral Groups"
)

fig.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Retention Rate (%)",
    yaxis_title="Referral Group"
)

fig.show()


fig = px.bar(
    deal_summary,
    x="Value_Deal",
    y="Net_Revenue_Per_Customer",
    color="Has_Value_Deal",
    barmode="group",
    text="Net_Revenue_Per_Customer",
    template="plotly_dark",
    title="Net Revenue per Customer by Value Deal"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Value Deal",
    yaxis_title="Net Revenue per Customer"
)

fig.show()


fig = px.scatter(
    deal_summary,
    x="Avg_Tenure",
    y="Retention_Rate",
    size="Customers",
    color="Value_Deal",
    hover_data=[
        "Churn_Rate",
        "Revenue_Per_Customer",
        "Net_Revenue_Per_Customer"
    ],
    template="plotly_white",
    title="Value Deal, Tenure and Retention Relationship"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Tenure (Months)",
    yaxis_title="Retention Rate (%)"
)

fig.show()


fig = px.treemap(
    referral_summary,
    path=["Referral_Group", "Has_Referral"],
    values="Total_Revenue",
    color="Revenue_Per_Customer",
    color_continuous_scale="Plasma",
    template="plotly_dark",
    title="Revenue Contribution from Referral Segments"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


best_referral_value = referral_summary.loc[
    referral_summary["Net_Revenue_Per_Customer"].idxmax()
]

best_referral_retention = referral_summary.loc[
    referral_summary["Retention_Rate"].idxmax()
]

highest_referral_churn = referral_summary.loc[
    referral_summary["Churn_Rate"].idxmax()
]

best_deal_value = deal_summary.loc[
    deal_summary["Net_Revenue_Per_Customer"].idxmax()
]

best_deal_retention = deal_summary.loc[
    deal_summary["Retention_Rate"].idxmax()
]

highest_deal_revenue = deal_summary.loc[
    deal_summary["Total_Revenue"].idxmax()
]


business_summary = pd.DataFrame({
    "Business_Metric": [
        "Highest Referral Net Revenue per Customer",
        "Highest Referral Retention",
        "Highest Referral Churn",
        "Highest Value Deal Net Revenue per Customer",
        "Highest Value Deal Retention",
        "Highest Value Deal Total Revenue"
    ],

    "Segment": [
        best_referral_value["Referral_Group"],
        best_referral_retention["Referral_Group"],
        highest_referral_churn["Referral_Group"],
        best_deal_value["Value_Deal"],
        best_deal_retention["Value_Deal"],
        highest_deal_revenue["Value_Deal"]
    ],

    "Value": [
        f"{best_referral_value['Net_Revenue_Per_Customer']:,.2f}",
        f"{best_referral_retention['Retention_Rate']:.2f}%",
        f"{highest_referral_churn['Churn_Rate']:.2f}%",
        f"{best_deal_value['Net_Revenue_Per_Customer']:,.2f}",
        f"{best_deal_retention['Retention_Rate']:.2f}%",
        f"{highest_deal_revenue['Total_Revenue']:,.2f}"
    ]
})

display(business_summary)


print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{best_referral_value['Referral_Group']} referral customers "
    f"generate the highest net revenue per customer at "
    f"{best_referral_value['Net_Revenue_Per_Customer']:,.2f}."
)

print(
    f"{best_referral_retention['Referral_Group']} referral segment "
    f"has the strongest retention rate at "
    f"{best_referral_retention['Retention_Rate']:.2f}%."
)

print(
    f"{highest_referral_churn['Referral_Group']} referral segment "
    f"has the highest churn rate at "
    f"{highest_referral_churn['Churn_Rate']:.2f}%."
)

print(
    f"{best_deal_value['Value_Deal']} value-deal segment generates "
    f"the highest net revenue per customer at "
    f"{best_deal_value['Net_Revenue_Per_Customer']:,.2f}."
)

print(
    f"{best_deal_retention['Value_Deal']} value-deal segment records "
    f"the strongest retention rate at "
    f"{best_deal_retention['Retention_Rate']:.2f}%."
)

print(
    f"{highest_deal_revenue['Value_Deal']} contributes the highest "
    f"total revenue at {highest_deal_revenue['Total_Revenue']:,.2f}."
)

print(
    "Referral activity and value deals can influence both customer "
    "value and long-term retention."
)

print(
    "Business Action: Strengthen high-value referral programs, "
    "evaluate value-deal effectiveness and target low-retention "
    "segments with optimized offers."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 21
# ============================================================

# How does customer revenue, customer concentration and churn
# vary across Indian states?

# Business Goal:
# Identify high-value states, revenue concentration,
# customer distribution and state-level churn opportunities.

# ============================================================

import pandas as pd
import plotly.express as px
import requests


# ============================================================
# STATE-LEVEL BUSINESS SUMMARY
# ============================================================

state_summary = (
    df.groupby("State", dropna=False)
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Service_Count=("Service_Count", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        ),
        Stayed=(
            "Customer_Status",
            lambda x: (x == "Stayed").sum()
        )
    )
    .reset_index()
)

state_summary["Churn_Rate"] = (
    state_summary["Churned"]
    .div(state_summary["Customers"])
    .mul(100)
)

state_summary["Retention_Rate"] = (
    state_summary["Stayed"]
    .div(state_summary["Customers"])
    .mul(100)
)

state_summary["Revenue_Per_Customer"] = (
    state_summary["Total_Revenue"]
    .div(state_summary["Customers"])
)

state_summary["Revenue_Share"] = (
    state_summary["Total_Revenue"]
    .div(state_summary["Total_Revenue"].sum())
    .mul(100)
)

state_summary["Customer_Share"] = (
    state_summary["Customers"]
    .div(state_summary["Customers"].sum())
    .mul(100)
)

state_summary["Revenue_vs_Avg"] = (
    state_summary["Total_Revenue"]
    - state_summary["Total_Revenue"].mean()
)

state_summary = state_summary.sort_values(
    "Total_Revenue",
    ascending=False
)

display(
    state_summary.round(2)
)


# ============================================================
# STATE MAP DATA PREPARATION
# ============================================================

state_map = state_summary.copy()

state_map["State"] = (
    state_map["State"]
    .astype(str)
    .str.strip()
    .str.title()
)

numeric_cols = [
    "Total_Revenue",
    "Customers",
    "Revenue_Per_Customer",
    "Net_Revenue",
    "Churn_Rate",
    "Retention_Rate"
]

for col in numeric_cols:
    state_map[col] = pd.to_numeric(
        state_map[col],
        errors="coerce"
    ).fillna(0)


state_name_map = {
    "Andhra Pradesh": "Andhra Pradesh",
    "Arunachal Pradesh": "Arunachal Pradesh",
    "Assam": "Assam",
    "Bihar": "Bihar",
    "Chhattisgarh": "Chhattisgarh",
    "Goa": "Goa",
    "Gujarat": "Gujarat",
    "Haryana": "Haryana",
    "Himachal Pradesh": "Himachal Pradesh",
    "Jharkhand": "Jharkhand",
    "Karnataka": "Karnataka",
    "Kerala": "Kerala",
    "Madhya Pradesh": "Madhya Pradesh",
    "Maharashtra": "Maharashtra",
    "Manipur": "Manipur",
    "Meghalaya": "Meghalaya",
    "Mizoram": "Mizoram",
    "Nagaland": "Nagaland",
    "Odisha": "Odisha",
    "Orissa": "Odisha",
    "Punjab": "Punjab",
    "Rajasthan": "Rajasthan",
    "Sikkim": "Sikkim",
    "Tamil Nadu": "Tamil Nadu",
    "Telangana": "Telangana",
    "Tripura": "Tripura",
    "Uttar Pradesh": "Uttar Pradesh",
    "Uttarakhand": "Uttarakhand",
    "West Bengal": "West Bengal",
    "Delhi": "Delhi",
    "Jammu And Kashmir": "Jammu and Kashmir"
}

state_map["State"] = state_map["State"].replace(
    state_name_map
)


# ============================================================
# INDIA GEOJSON
# ============================================================

india_geojson_url = (
    "https://raw.githubusercontent.com/"
    "geohacker/india/master/state/india_state.geojson"
)

response = requests.get(
    india_geojson_url,
    timeout=30
)

response.raise_for_status()

india_geojson = response.json()


geo_states = [
    feature["properties"]["NAME_1"]
    for feature in india_geojson["features"]
]

matched_states = set(
    state_map["State"]
).intersection(
    set(geo_states)
)

state_map = state_map[
    state_map["State"].isin(matched_states)
].copy()


print("=" * 75)
print("INDIA STATE GEOGRAPHIC DATA CHECK")
print("=" * 75)

print(
    f"Dataset States : "
    f"{state_summary['State'].nunique()}"
)

print(
    f"Matched States : "
    f"{len(matched_states)}"
)

print(
    f"Mapped Customers : "
    f"{state_map['Customers'].sum():,.0f}"
)

print(
    f"Mapped Revenue : "
    f"{state_map['Total_Revenue'].sum():,.2f}"
)

print("=" * 75)


# ============================================================
# KPI SUMMARY
# ============================================================

top_revenue_state = state_summary.loc[
    state_summary["Total_Revenue"].idxmax()
]

top_customer_state = state_summary.loc[
    state_summary["Customers"].idxmax()
]

top_rpc_state = state_summary.loc[
    state_summary["Revenue_Per_Customer"].idxmax()
]

highest_churn_state = state_summary.loc[
    state_summary["Churn_Rate"].idxmax()
]

lowest_churn_state = state_summary.loc[
    state_summary["Churn_Rate"].idxmin()
]

highest_retention_state = state_summary.loc[
    state_summary["Retention_Rate"].idxmax()
]

print("=" * 75)
print("STATE-WISE BUSINESS KPI")
print("=" * 75)

print(
    f"Total States          : "
    f"{state_summary['State'].nunique()}"
)

print(
    f"Total Customers       : "
    f"{state_summary['Customers'].sum():,.0f}"
)

print(
    f"Total Revenue         : "
    f"{state_summary['Total_Revenue'].sum():,.2f}"
)

print(
    f"Top Revenue State     : "
    f"{top_revenue_state['State']}"
)

print(
    f"Top Revenue           : "
    f"{top_revenue_state['Total_Revenue']:,.2f}"
)

print(
    f"Top Customer State    : "
    f"{top_customer_state['State']}"
)

print(
    f"Top Customer Count    : "
    f"{top_customer_state['Customers']:,.0f}"
)

print(
    f"Highest Revenue/Customer : "
    f"{top_rpc_state['State']}"
)

print(
    f"Revenue/Customer      : "
    f"{top_rpc_state['Revenue_Per_Customer']:,.2f}"
)

print(
    f"Highest Churn State   : "
    f"{highest_churn_state['State']}"
)

print(
    f"Highest Churn Rate    : "
    f"{highest_churn_state['Churn_Rate']:.2f}%"
)

print(
    f"Lowest Churn State    : "
    f"{lowest_churn_state['State']}"
)

print(
    f"Lowest Churn Rate     : "
    f"{lowest_churn_state['Churn_Rate']:.2f}%"
)

print(
    f"Highest Retention     : "
    f"{highest_retention_state['State']}"
)

print(
    f"Retention Rate        : "
    f"{highest_retention_state['Retention_Rate']:.2f}%"
)

print("=" * 75)


# ============================================================
# VISUAL 1 — INDIA REVENUE CHOROPLETH
# ============================================================

fig = px.choropleth(
    state_map,
    geojson=india_geojson,
    featureidkey="properties.NAME_1",
    locations="State",
    color="Total_Revenue",
    color_continuous_scale="Turbo",
    hover_name="State",
    custom_data=[
        "Customers",
        "Total_Revenue",
        "Net_Revenue",
        "Revenue_Per_Customer",
        "Churn_Rate",
        "Retention_Rate"
    ],
    title="Interactive State-wise Total Revenue — India"
)

fig.update_traces(
    hovertemplate=(
        "<b>%{location}</b><br><br>"
        "Customers: %{customdata[0]:,}<br>"
        "Total Revenue: ₹%{customdata[1]:,.2f}<br>"
        "Net Revenue: ₹%{customdata[2]:,.2f}<br>"
        "Revenue / Customer: ₹%{customdata[3]:,.2f}<br>"
        "Churn Rate: %{customdata[4]:.2f}%<br>"
        "Retention Rate: %{customdata[5]:.2f}%"
        "<extra></extra>"
    ),
    marker_line_width=0.8
)

fig.update_geos(
    fitbounds="locations",
    visible=False,
    showcountries=True,
    countrycolor="black",
    showland=True,
    landcolor="lightgray",
    showlakes=True,
    lakecolor="lightblue"
)

fig.update_layout(
    title_x=0.5,
    height=750,
    margin=dict(
        l=20,
        r=20,
        t=80,
        b=20
    ),
    coloraxis_colorbar=dict(
        title="Total Revenue"
    )
)

fig.show()


# ============================================================
# VISUAL 2 — STATE REVENUE RANKING
# ============================================================

top_states = (
    state_summary
    .sort_values("Total_Revenue", ascending=True)
    .tail(15)
)

fig = px.bar(
    top_states,
    x="Total_Revenue",
    y="State",
    orientation="h",
    color="Revenue_Per_Customer",
    text="Total_Revenue",
    color_continuous_scale="Plasma",
    title="Top 15 States by Total Revenue",
    template="plotly_white",
    hover_data=[
        "Customers",
        "Revenue_Share",
        "Churn_Rate"
    ]
)

fig.update_traces(
    texttemplate="₹%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Total Revenue",
    yaxis_title="State",
    height=650
)

fig.show()


# ============================================================
# VISUAL 3 — CUSTOMER VS REVENUE
# ============================================================

fig = px.scatter(
    state_summary,
    x="Customers",
    y="Total_Revenue",
    size="Revenue_Per_Customer",
    color="Churn_Rate",
    text="State",
    color_continuous_scale="Turbo",
    template="plotly_dark",
    title="State Customer Base vs Total Revenue",
    hover_data=[
        "Revenue_Per_Customer",
        "Revenue_Share",
        "Retention_Rate"
    ]
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Number of Customers",
    yaxis_title="Total Revenue",
    coloraxis_colorbar=dict(
        title="Churn Rate (%)"
    )
)

fig.show()


# ============================================================
# VISUAL 4 — REVENUE PER CUSTOMER
# ============================================================

rpc_data = (
    state_summary
    .sort_values(
        "Revenue_Per_Customer",
        ascending=True
    )
    .tail(15)
)

fig = px.bar(
    rpc_data,
    x="Revenue_Per_Customer",
    y="State",
    orientation="h",
    color="Churn_Rate",
    text="Revenue_Per_Customer",
    color_continuous_scale="RdYlGn_r",
    title="Top States by Revenue per Customer",
    template="plotly_white",
    hover_data=[
        "Customers",
        "Total_Revenue",
        "Retention_Rate"
    ]
)

fig.update_traces(
    texttemplate="₹%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Revenue per Customer",
    yaxis_title="State"
)

fig.show()


# ============================================================
# VISUAL 5 — STATE CHURN VS RETENTION
# ============================================================

churn_data = (
    state_summary
    .sort_values("Churn_Rate", ascending=False)
)

fig = px.scatter(
    churn_data,
    x="Churn_Rate",
    y="Retention_Rate",
    size="Customers",
    color="Total_Revenue",
    text="State",
    color_continuous_scale="Viridis",
    template="plotly_white",
    title="State-wise Churn and Retention Positioning",
    hover_data=[
        "Customers",
        "Total_Revenue",
        "Revenue_Per_Customer"
    ]
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Churn Rate (%)",
    yaxis_title="Retention Rate (%)",
    coloraxis_colorbar=dict(
        title="Total Revenue"
    )
)

fig.show()


# ============================================================
# VISUAL 6 — REVENUE SHARE TREEMAP
# ============================================================

treemap_data = (
    state_summary
    .sort_values("Total_Revenue", ascending=False)
    .head(20)
)

fig = px.treemap(
    treemap_data,
    path=["State"],
    values="Total_Revenue",
    color="Revenue_Per_Customer",
    color_continuous_scale="Turbo",
    template="plotly_dark",
    title="State Revenue Contribution and Customer Value",
    hover_data=[
        "Customers",
        "Revenue_Share",
        "Churn_Rate",
        "Retention_Rate"
    ]
)

fig.update_layout(
    title_x=0.5,
    height=700
)

fig.show()


# ============================================================
# BUSINESS INSIGHTS
# ============================================================

print("\n" + "=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"State {top_revenue_state['State']} generates the highest "
    f"total revenue of ₹{top_revenue_state['Total_Revenue']:,.2f}."
)

print(
    f"State {top_customer_state['State']} has the largest customer "
    f"base with {top_customer_state['Customers']:,.0f} customers."
)

print(
    f"State {top_rpc_state['State']} has the highest revenue per "
    f"customer at ₹{top_rpc_state['Revenue_Per_Customer']:,.2f}."
)

print(
    f"State {highest_churn_state['State']} has the highest churn "
    f"rate at {highest_churn_state['Churn_Rate']:.2f}%."
)

print(
    f"State {lowest_churn_state['State']} has the lowest churn "
    f"rate at {lowest_churn_state['Churn_Rate']:.2f}%."
)

print(
    f"State {highest_retention_state['State']} has the strongest "
    f"retention rate at "
    f"{highest_retention_state['Retention_Rate']:.2f}%."
)

print(
    "Geographic revenue concentration helps identify states "
    "that contribute disproportionately to overall business revenue."
)

print(
    "High-revenue states with elevated churn should be prioritized "
    "for retention campaigns because they represent both customer "
    "and revenue risk."
)

print(
    "High revenue-per-customer states can be targeted for premium "
    "plans, cross-selling and service expansion."
)

print(
    "Business Action: Combine state-level revenue, customer density, "
    "revenue per customer and churn metrics to prioritize regional "
    "growth and retention strategies."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 22
# ============================================================

# How does customer tenure and lifecycle stage influence
# revenue value, retention and churn risk?

# Business Goal:
# Identify lifecycle stages where customers generate the
# strongest value and where retention intervention is needed.
# ============================================================

plot_df = df.copy()

plot_df["Plot_Revenue"] = (
    pd.to_numeric(
        plot_df["Total_Revenue"],
        errors="coerce"
    )
    .fillna(0)
)

plot_df["Plot_Net_Revenue"] = (
    pd.to_numeric(
        plot_df["Net_Revenue"],
        errors="coerce"
    )
    .fillna(0)
)

plot_df["Plot_Monthly_Charge"] = (
    pd.to_numeric(
        plot_df["Monthly_Charge"],
        errors="coerce"
    )
    .abs()
    .fillna(1)
    .clip(lower=1)
)

plot_df["Tenure_in_Months"] = pd.to_numeric(
    plot_df["Tenure_in_Months"],
    errors="coerce"
)

plot_df["Tenure_in_Months"] = (
    plot_df["Tenure_in_Months"]
    .fillna(
        plot_df["Tenure_in_Months"].median()
    )
)

lifecycle_summary = (
    plot_df.groupby(
        ["Tenure_Group", "Customer_Lifecycle"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Plot_Revenue", "sum"),
        Net_Revenue=("Plot_Net_Revenue", "sum"),
        Avg_Revenue=("Plot_Revenue", "mean"),
        Avg_Net_Revenue=("Plot_Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        ),
        Stayed=(
            "Customer_Status",
            lambda x: (x == "Stayed").sum()
        )
    )
    .reset_index()
)

lifecycle_summary["Churn_Rate"] = (
    lifecycle_summary["Churned"]
    .div(lifecycle_summary["Customers"])
    .mul(100)
)

lifecycle_summary["Retention_Rate"] = (
    lifecycle_summary["Stayed"]
    .div(lifecycle_summary["Customers"])
    .mul(100)
)

lifecycle_summary["Revenue_Per_Customer"] = (
    lifecycle_summary["Total_Revenue"]
    .div(lifecycle_summary["Customers"])
)

lifecycle_summary["Net_Revenue_Per_Customer"] = (
    lifecycle_summary["Net_Revenue"]
    .div(lifecycle_summary["Customers"])
)

lifecycle_summary["Revenue_Share"] = (
    lifecycle_summary["Total_Revenue"]
    .div(
        lifecycle_summary["Total_Revenue"].sum()
    )
    .mul(100)
)

display(
    lifecycle_summary.round(2)
)


tenure_summary = (
    plot_df.groupby(
        "Tenure_Group",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Plot_Revenue", "sum"),
        Net_Revenue=("Plot_Net_Revenue", "sum"),
        Avg_Revenue=("Plot_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        ),
        Stayed=(
            "Customer_Status",
            lambda x: (x == "Stayed").sum()
        )
    )
    .reset_index()
)

tenure_summary["Churn_Rate"] = (
    tenure_summary["Churned"]
    .div(tenure_summary["Customers"])
    .mul(100)
)

tenure_summary["Retention_Rate"] = (
    tenure_summary["Stayed"]
    .div(tenure_summary["Customers"])
    .mul(100)
)

tenure_summary["Revenue_Per_Customer"] = (
    tenure_summary["Total_Revenue"]
    .div(tenure_summary["Customers"])
)

tenure_summary["Net_Revenue_Per_Customer"] = (
    tenure_summary["Net_Revenue"]
    .div(tenure_summary["Customers"])
)

display(
    tenure_summary.round(2)
)


fig = px.bar(
    tenure_summary,
    x="Tenure_Group",
    y="Revenue_Per_Customer",
    color="Retention_Rate",
    text="Revenue_Per_Customer",
    color_continuous_scale="Viridis",
    template="plotly_white",
    title="Revenue per Customer Across Tenure Groups"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Tenure Group",
    yaxis_title="Revenue per Customer"
)

fig.show()


scatter_df = plot_df[
    [
        "Tenure_in_Months",
        "Plot_Net_Revenue",
        "Plot_Monthly_Charge",
        "Customer_Lifecycle",
        "Customer_Status",
        "Total_Revenue",
        "Revenue_Per_Month"
    ]
].copy()

scatter_df = scatter_df.dropna(
    subset=[
        "Tenure_in_Months",
        "Plot_Net_Revenue"
    ]
)

scatter_df["Plot_Monthly_Charge"] = (
    scatter_df["Plot_Monthly_Charge"]
    .abs()
    .clip(lower=1)
)

fig = px.scatter(
    scatter_df,
    x="Tenure_in_Months",
    y="Plot_Net_Revenue",
    color="Customer_Lifecycle",
    size="Plot_Monthly_Charge",
    hover_data=[
        "Customer_Status",
        "Total_Revenue",
        "Revenue_Per_Month"
    ],
    template="plotly_dark",
    title="Customer Tenure vs Net Revenue by Lifecycle"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Tenure in Months",
    yaxis_title="Net Revenue"
)

fig.show()


lifecycle_heatmap = (
    lifecycle_summary
    .pivot(
        index="Customer_Lifecycle",
        columns="Tenure_Group",
        values="Retention_Rate"
    )
)

fig = px.imshow(
    lifecycle_heatmap,
    text_auto=".1f",
    aspect="auto",
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Retention Rate by Lifecycle and Tenure"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Tenure Group",
    yaxis_title="Customer Lifecycle"
)

fig.show()


churn_plot = lifecycle_summary.copy()

churn_plot["Plot_Customers"] = (
    churn_plot["Customers"]
    .clip(lower=1)
)

fig = px.scatter(
    churn_plot,
    x="Avg_Tenure",
    y="Churn_Rate",
    size="Plot_Customers",
    color="Revenue_Per_Customer",
    hover_name="Customer_Lifecycle",
    hover_data=[
        "Tenure_Group",
        "Retention_Rate",
        "Customers",
        "Net_Revenue_Per_Customer"
    ],
    color_continuous_scale="Plasma",
    template="plotly_dark",
    title="Tenure, Churn Risk and Customer Value"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Tenure (Months)",
    yaxis_title="Churn Rate (%)"
)

fig.show()


value_plot = tenure_summary.copy()

value_plot["Plot_Customers"] = (
    value_plot["Customers"]
    .clip(lower=1)
)

fig = px.scatter(
    value_plot,
    x="Avg_Tenure",
    y="Net_Revenue_Per_Customer",
    size="Plot_Customers",
    color="Churn_Rate",
    hover_name="Tenure_Group",
    hover_data=[
        "Retention_Rate",
        "Revenue_Per_Customer",
        "Total_Revenue"
    ],
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Tenure vs Net Customer Value"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Tenure (Months)",
    yaxis_title="Net Revenue per Customer"
)

fig.show()


fig = px.treemap(
    lifecycle_summary,
    path=[
        "Customer_Lifecycle",
        "Tenure_Group"
    ],
    values="Total_Revenue",
    color="Net_Revenue_Per_Customer",
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Revenue Contribution Across Customer Lifecycle"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


best_tenure_value = tenure_summary.loc[
    tenure_summary["Net_Revenue_Per_Customer"].idxmax()
]

highest_tenure_churn = tenure_summary.loc[
    tenure_summary["Churn_Rate"].idxmax()
]

best_lifecycle_value = lifecycle_summary.loc[
    lifecycle_summary[
        "Net_Revenue_Per_Customer"
    ].idxmax()
]

best_lifecycle_retention = lifecycle_summary.loc[
    lifecycle_summary[
        "Retention_Rate"
    ].idxmax()
]

highest_lifecycle_churn = lifecycle_summary.loc[
    lifecycle_summary[
        "Churn_Rate"
    ].idxmax()
]

largest_revenue_lifecycle = lifecycle_summary.loc[
    lifecycle_summary[
        "Total_Revenue"
    ].idxmax()
]


business_summary = pd.DataFrame({
    "Business_Metric": [
        "Highest Tenure Net Revenue per Customer",
        "Highest Tenure Churn",
        "Highest Lifecycle Net Revenue per Customer",
        "Highest Lifecycle Retention",
        "Highest Lifecycle Churn",
        "Highest Lifecycle Total Revenue"
    ],
    "Segment": [
        best_tenure_value["Tenure_Group"],
        highest_tenure_churn["Tenure_Group"],
        best_lifecycle_value["Customer_Lifecycle"],
        best_lifecycle_retention["Customer_Lifecycle"],
        highest_lifecycle_churn["Customer_Lifecycle"],
        largest_revenue_lifecycle["Customer_Lifecycle"]
    ],
    "Value": [
        f"{best_tenure_value['Net_Revenue_Per_Customer']:,.2f}",
        f"{highest_tenure_churn['Churn_Rate']:.2f}%",
        f"{best_lifecycle_value['Net_Revenue_Per_Customer']:,.2f}",
        f"{best_lifecycle_retention['Retention_Rate']:.2f}%",
        f"{highest_lifecycle_churn['Churn_Rate']:.2f}%",
        f"{largest_revenue_lifecycle['Total_Revenue']:,.2f}"
    ]
})

display(business_summary)


print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{best_tenure_value['Tenure_Group']} has the highest "
    f"net revenue per customer at "
    f"{best_tenure_value['Net_Revenue_Per_Customer']:,.2f}."
)

print(
    f"{highest_tenure_churn['Tenure_Group']} has the highest "
    f"tenure-group churn rate at "
    f"{highest_tenure_churn['Churn_Rate']:.2f}%."
)

print( 
    f"{best_lifecycle_value['Customer_Lifecycle']} generates "
    f"the highest net revenue per customer at "
    f"{best_lifecycle_value['Net_Revenue_Per_Customer']:,.2f}."
)

print(
    f"{best_lifecycle_retention['Customer_Lifecycle']} has "
    f"the strongest retention rate at "
    f"{best_lifecycle_retention['Retention_Rate']:.2f}%."
)

print(
    f"{highest_lifecycle_churn['Customer_Lifecycle']} has "
    f"the highest churn rate at "
    f"{highest_lifecycle_churn['Churn_Rate']:.2f}%."
)

print(
    f"{largest_revenue_lifecycle['Customer_Lifecycle']} "
    f"contributes the highest total revenue at "
    f"{largest_revenue_lifecycle['Total_Revenue']:,.2f}."
)

print(
    "Customer tenure and lifecycle stage reveal where "
    "customer value strengthens and where churn risk "
    "becomes concentrated."
)

print(
    "Business Action: Build lifecycle-specific retention "
    "strategies, strengthen early-stage engagement and "
    "protect high-value long-tenure customers."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 23
# ============================================================

# Which payment methods and billing preferences generate
# stronger customer value while maintaining lower churn?

# Business Goal:
# Evaluate payment behavior, revenue contribution and churn
# to identify financially valuable and retention-friendly
# billing strategies.
# ============================================================

plot_df = df.copy()

plot_df["Monthly_Charge"] = pd.to_numeric(
    plot_df["Monthly_Charge"],
    errors="coerce"
).fillna(0)

plot_df["Total_Revenue"] = pd.to_numeric(
    plot_df["Total_Revenue"],
    errors="coerce"
).fillna(0)

plot_df["Net_Revenue"] = pd.to_numeric(
    plot_df["Net_Revenue"],
    errors="coerce"
).fillna(0)

payment_summary = (
    plot_df.groupby(
        "Payment_Method",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        ),
        Stayed=(
            "Customer_Status",
            lambda x: (x == "Stayed").sum()
        )
    )
    .reset_index()
)

payment_summary["Churn_Rate"] = (
    payment_summary["Churned"]
    .div(payment_summary["Customers"])
    .mul(100)
)

payment_summary["Retention_Rate"] = (
    payment_summary["Stayed"]
    .div(payment_summary["Customers"])
    .mul(100)
)

payment_summary["Revenue_Per_Customer"] = (
    payment_summary["Total_Revenue"]
    .div(payment_summary["Customers"])
)

payment_summary["Revenue_Share"] = (
    payment_summary["Total_Revenue"]
    .div(payment_summary["Total_Revenue"].sum())
    .mul(100)
)

display(
    payment_summary
    .sort_values("Total_Revenue", ascending=False)
    .round(2)
)


billing_summary = (
    plot_df.groupby(
        ["Is_Automatic_Payment", "Paperless_Billing"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        ),
        Stayed=(
            "Customer_Status",
            lambda x: (x == "Stayed").sum()
        )
    )
    .reset_index()
)

billing_summary["Churn_Rate"] = (
    billing_summary["Churned"]
    .div(billing_summary["Customers"])
    .mul(100)
)

billing_summary["Retention_Rate"] = (
    billing_summary["Stayed"]
    .div(billing_summary["Customers"])
    .mul(100)
)

billing_summary["Revenue_Per_Customer"] = (
    billing_summary["Total_Revenue"]
    .div(billing_summary["Customers"])
)

display(
    billing_summary.round(2)
)


# ============================================================
# KPI SUMMARY
# ============================================================

top_payment = payment_summary.loc[
    payment_summary["Total_Revenue"].idxmax()
]

best_payment_value = payment_summary.loc[
    payment_summary["Revenue_Per_Customer"].idxmax()
]

lowest_payment_churn = payment_summary.loc[
    payment_summary["Churn_Rate"].idxmin()
]

highest_payment_churn = payment_summary.loc[
    payment_summary["Churn_Rate"].idxmax()
]

print("=" * 75)
print("PAYMENT & BILLING PERFORMANCE")
print("=" * 75)

print(
    f"Highest Revenue Payment Method : "
    f"{top_payment['Payment_Method']}"
)

print(
    f"Highest Payment Revenue        : "
    f"{top_payment['Total_Revenue']:,.2f}"
)

print(
    f"Highest Revenue/Customer       : "
    f"{best_payment_value['Payment_Method']} "
    f"({best_payment_value['Revenue_Per_Customer']:,.2f})"
)

print(
    f"Lowest Churn Payment Method    : "
    f"{lowest_payment_churn['Payment_Method']} "
    f"({lowest_payment_churn['Churn_Rate']:.2f}%)"
)

print(
    f"Highest Churn Payment Method   : "
    f"{highest_payment_churn['Payment_Method']} "
    f"({highest_payment_churn['Churn_Rate']:.2f}%)"
)

print("=" * 75)


# ============================================================
# VISUAL 1
# ============================================================

fig = px.bar(
    payment_summary.sort_values(
        "Total_Revenue",
        ascending=True
    ),
    x="Total_Revenue",
    y="Payment_Method",
    orientation="h",
    color="Revenue_Share",
    text="Total_Revenue",
    color_continuous_scale="Viridis",
    template="plotly_white",
    title="Total Revenue by Payment Method"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Total Revenue",
    yaxis_title="Payment Method"
)

fig.show()


# ============================================================
# VISUAL 2
# ============================================================

fig = px.bar(
    payment_summary.sort_values(
        "Churn_Rate",
        ascending=True
    ),
    x="Payment_Method",
    y="Churn_Rate",
    color="Churn_Rate",
    text="Churn_Rate",
    color_continuous_scale="Reds",
    template="plotly_dark",
    title="Churn Rate by Payment Method"
)

fig.update_traces(
    texttemplate="%{text:.2f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Payment Method",
    yaxis_title="Churn Rate (%)"
)

fig.show()


# ============================================================
# VISUAL 3
# ============================================================

fig = px.scatter(
    payment_summary,
    x="Avg_Monthly_Charge",
    y="Revenue_Per_Customer",
    size="Customers",
    color="Payment_Method",
    hover_data=[
        "Total_Revenue",
        "Net_Revenue",
        "Churn_Rate",
        "Retention_Rate"
    ],
    template="plotly_white",
    title="Monthly Charge vs Revenue per Customer"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Monthly Charge",
    yaxis_title="Revenue per Customer"
)

fig.show()


# ============================================================
# VISUAL 4
# ============================================================

billing_heatmap = (
    billing_summary
    .pivot(
        index="Is_Automatic_Payment",
        columns="Paperless_Billing",
        values="Churn_Rate"
    )
)

fig = px.imshow(
    billing_heatmap,
    text_auto=".2f",
    aspect="auto",
    color_continuous_scale="RdYlGn_r",
    template="plotly_white",
    title="Churn Rate: Automatic Payment × Paperless Billing"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Paperless Billing",
    yaxis_title="Automatic Payment"
)

fig.show()


# ============================================================
# VISUAL 5
# ============================================================

fig = px.sunburst(
    plot_df,
    path=[
        "Payment_Method",
        "Is_Automatic_Payment",
        "Paperless_Billing"
    ],
    values="Total_Revenue",
    color="Net_Revenue",
    color_continuous_scale="Turbo",
    template="plotly_dark",
    title="Revenue Structure by Payment & Billing Preferences"
)

fig.update_layout(
    title_x=0.5
)

fig.show()


# ============================================================
# VISUAL 6
# ============================================================

fig = px.bar(
    billing_summary,
    x="Is_Automatic_Payment",
    y="Revenue_Per_Customer",
    color="Paperless_Billing",
    barmode="group",
    text="Revenue_Per_Customer",
    template="plotly_white",
    title="Revenue per Customer by Billing Preferences"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Automatic Payment",
    yaxis_title="Revenue per Customer"
)

fig.show()


# ============================================================
# BUSINESS INSIGHTS
# ============================================================

print("=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{top_payment['Payment_Method']} generates the highest "
    f"overall payment-method revenue of "
    f"{top_payment['Total_Revenue']:,.2f}."
)

print(
    f"{best_payment_value['Payment_Method']} has the highest "
    f"revenue per customer at "
    f"{best_payment_value['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{lowest_payment_churn['Payment_Method']} has the lowest "
    f"churn rate at "
    f"{lowest_payment_churn['Churn_Rate']:.2f}%."
)

print(
    f"{highest_payment_churn['Payment_Method']} has the highest "
    f"churn rate at "
    f"{highest_payment_churn['Churn_Rate']:.2f}%."
)

best_billing = billing_summary.loc[
    billing_summary["Revenue_Per_Customer"].idxmax()
]

print(
    f"The strongest billing combination by revenue per customer "
    f"is Automatic Payment = "
    f"{best_billing['Is_Automatic_Payment']} and Paperless Billing = "
    f"{best_billing['Paperless_Billing']}."
)

print(
    "Business Action: Promote payment methods and billing "
    "combinations associated with higher customer value and "
    "lower churn while investigating high-churn payment groups."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 24
# ============================================================

# How do payment methods and billing preferences influence
# customer churn, revenue, refunds and customer value?

# Business Goal:
# Identify payment and billing patterns associated with
# customer value, revenue leakage and churn risk.

# ============================================================

import pandas as pd
import plotly.express as px


analysis_df = df.copy()

numeric_cols = [
    "Monthly_Charge",
    "Total_Revenue",
    "Net_Revenue",
    "Total_Refunds",
    "Refund_Rate",
    "Total_Extra_Charges"
]

for col in numeric_cols:
    analysis_df[col] = pd.to_numeric(
        analysis_df[col],
        errors="coerce"
    ).fillna(0)


# ============================================================
# PAYMENT METHOD SUMMARY
# ============================================================

payment_summary = (
    analysis_df.groupby(
        "Payment_Method",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Total_Refunds=("Total_Refunds", "sum"),
        Extra_Charges=("Total_Extra_Charges", "sum"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        ),
        Stayed=(
            "Customer_Status",
            lambda x: (x == "Stayed").sum()
        )
    )
    .reset_index()
)

payment_summary["Churn_Rate"] = (
    payment_summary["Churned"]
    .div(payment_summary["Customers"])
    .mul(100)
)

payment_summary["Retention_Rate"] = (
    payment_summary["Stayed"]
    .div(payment_summary["Customers"])
    .mul(100)
)

payment_summary["Revenue_Per_Customer"] = (
    payment_summary["Total_Revenue"]
    .div(payment_summary["Customers"])
)

payment_summary["Refund_Per_Customer"] = (
    payment_summary["Total_Refunds"]
    .div(payment_summary["Customers"])
)

payment_summary["Revenue_Share"] = (
    payment_summary["Total_Revenue"]
    .div(payment_summary["Total_Revenue"].sum())
    .mul(100)
)

display(
    payment_summary
    .sort_values("Total_Revenue", ascending=False)
    .round(2)
)


# ============================================================
# AUTOMATIC PAYMENT SUMMARY
# ============================================================

auto_summary = (
    analysis_df.groupby(
        "Is_Automatic_Payment",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Total_Refunds=("Total_Refunds", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        ),
        Stayed=(
            "Customer_Status",
            lambda x: (x == "Stayed").sum()
        )
    )
    .reset_index()
)

auto_summary["Churn_Rate"] = (
    auto_summary["Churned"]
    .div(auto_summary["Customers"])
    .mul(100)
)

auto_summary["Retention_Rate"] = (
    auto_summary["Stayed"]
    .div(auto_summary["Customers"])
    .mul(100)
)

auto_summary["Revenue_Per_Customer"] = (
    auto_summary["Total_Revenue"]
    .div(auto_summary["Customers"])
)

auto_summary["Refund_Per_Customer"] = (
    auto_summary["Total_Refunds"]
    .div(auto_summary["Customers"])
)

display(
    auto_summary.round(2)
)


# ============================================================
# BILLING PREFERENCE SUMMARY
# ============================================================

billing_summary = (
    analysis_df.groupby(
        ["Paperless_Billing", "Is_Paperless"],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Total_Refunds=("Total_Refunds", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        ),
        Stayed=(
            "Customer_Status",
            lambda x: (x == "Stayed").sum()
        )
    )
    .reset_index()
)

billing_summary["Churn_Rate"] = (
    billing_summary["Churned"]
    .div(billing_summary["Customers"])
    .mul(100)
)

billing_summary["Retention_Rate"] = (
    billing_summary["Stayed"]
    .div(billing_summary["Customers"])
    .mul(100)
)

billing_summary["Revenue_Per_Customer"] = (
    billing_summary["Total_Revenue"]
    .div(billing_summary["Customers"])
)

display(
    billing_summary.round(2)
)


# ============================================================
# PAYMENT + BILLING COMBINATION
# ============================================================

payment_billing = (
    analysis_df.groupby(
        [
            "Payment_Method",
            "Is_Automatic_Payment"
        ],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Total_Refunds=("Total_Refunds", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        )
    )
    .reset_index()
)

payment_billing["Churn_Rate"] = (
    payment_billing["Churned"]
    .div(payment_billing["Customers"])
    .mul(100)
)

payment_billing["Revenue_Per_Customer"] = (
    payment_billing["Total_Revenue"]
    .div(payment_billing["Customers"])
)

display(
    payment_billing
    .sort_values("Total_Revenue", ascending=False)
    .round(2)
)


# ============================================================
# KPI ANALYSIS
# ============================================================

highest_revenue_payment = payment_summary.loc[
    payment_summary["Total_Revenue"].idxmax()
]

highest_value_payment = payment_summary.loc[
    payment_summary["Revenue_Per_Customer"].idxmax()
]

highest_churn_payment = payment_summary.loc[
    payment_summary["Churn_Rate"].idxmax()
]

highest_refund_payment = payment_summary.loc[
    payment_summary["Refund_Per_Customer"].idxmax()
]

best_auto_group = auto_summary.loc[
    auto_summary["Churn_Rate"].idxmin()
]

worst_auto_group = auto_summary.loc[
    auto_summary["Churn_Rate"].idxmax()
]


print("=" * 75)
print("PAYMENT & BILLING BUSINESS KPI")
print("=" * 75)

print(
    f"Highest Revenue Payment Method : "
    f"{highest_revenue_payment['Payment_Method']}"
)

print(
    f"Revenue Contribution            : "
    f"₹{highest_revenue_payment['Total_Revenue']:,.2f}"
)

print(
    f"Highest Revenue/Customer        : "
    f"{highest_value_payment['Payment_Method']}"
)

print(
    f"Revenue/Customer                : "
    f"₹{highest_value_payment['Revenue_Per_Customer']:,.2f}"
)

print(
    f"Highest Churn Payment Method    : "
    f"{highest_churn_payment['Payment_Method']}"
)

print(
    f"Churn Rate                      : "
    f"{highest_churn_payment['Churn_Rate']:.2f}%"
)

print(
    f"Highest Refund/Customer         : "
    f"{highest_refund_payment['Payment_Method']}"
)

print(
    f"Refund/Customer                 : "
    f"₹{highest_refund_payment['Refund_Per_Customer']:,.2f}"
)

print(
    f"Lowest Churn Auto-Payment Group : "
    f"{best_auto_group['Is_Automatic_Payment']}"
)

print(
    f"Lowest Churn Rate               : "
    f"{best_auto_group['Churn_Rate']:.2f}%"
)

print(
    f"Highest Churn Auto-Payment Group: "
    f"{worst_auto_group['Is_Automatic_Payment']}"
)

print(
    f"Highest Churn Rate              : "
    f"{worst_auto_group['Churn_Rate']:.2f}%"
)

print("=" * 75)


# ============================================================
# VISUAL 1
# ============================================================

fig = px.bar(
    payment_summary.sort_values(
        "Total_Revenue",
        ascending=True
    ),
    x="Total_Revenue",
    y="Payment_Method",
    orientation="h",
    color="Revenue_Per_Customer",
    text="Total_Revenue",
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Total Revenue Contribution by Payment Method",
    hover_data=[
        "Customers",
        "Net_Revenue",
        "Revenue_Share",
        "Churn_Rate"
    ]
)

fig.update_traces(
    texttemplate="₹%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Total Revenue",
    yaxis_title="Payment Method"
)

fig.show()


# ============================================================
# VISUAL 2
# ============================================================

fig = px.scatter(
    payment_summary,
    x="Churn_Rate",
    y="Revenue_Per_Customer",
    size="Customers",
    color="Total_Refunds",
    text="Payment_Method",
    color_continuous_scale="Plasma",
    template="plotly_dark",
    title="Payment Method: Churn Risk vs Customer Value",
    hover_data=[
        "Total_Revenue",
        "Net_Revenue",
        "Refund_Per_Customer",
        "Retention_Rate"
    ]
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Churn Rate (%)",
    yaxis_title="Revenue per Customer",
    coloraxis_colorbar=dict(
        title="Total Refunds"
    )
)

fig.show()


# ============================================================
# VISUAL 3
# ============================================================

fig = px.bar(
    auto_summary,
    x="Is_Automatic_Payment",
    y=[
        "Churn_Rate",
        "Retention_Rate"
    ],
    barmode="group",
    text_auto=".2f",
    template="plotly_white",
    title="Automatic Payment and Customer Retention"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Automatic Payment",
    yaxis_title="Rate (%)"
)

fig.show()


# ============================================================
# VISUAL 4
# ============================================================

fig = px.scatter(
    auto_summary,
    x="Avg_Monthly_Charge",
    y="Revenue_Per_Customer",
    size="Customers",
    color="Refund_Per_Customer",
    text="Is_Automatic_Payment",
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Automatic Payment: Charges vs Customer Value",
    hover_data=[
        "Total_Revenue",
        "Net_Revenue",
        "Churn_Rate",
        "Retention_Rate"
    ]
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Monthly Charge",
    yaxis_title="Revenue per Customer",
    coloraxis_colorbar=dict(
        title="Refund / Customer"
    )
)

fig.show()


# ============================================================
# VISUAL 5
# ============================================================

fig = px.bar(
    payment_billing,
    x="Payment_Method",
    y="Churn_Rate",
    color="Is_Automatic_Payment",
    barmode="group",
    text="Churn_Rate",
    template="plotly_white",
    title="Churn Rate by Payment Method and Automatic Payment",
    hover_data=[
        "Customers",
        "Total_Revenue",
        "Revenue_Per_Customer"
    ]
)

fig.update_traces(
    texttemplate="%{text:.2f}%",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Payment Method",
    yaxis_title="Churn Rate (%)"
)

fig.show()


# ============================================================
# VISUAL 6
# ============================================================

fig = px.treemap(
    payment_summary,
    path=["Payment_Method"],
    values="Total_Revenue",
    color="Churn_Rate",
    color_continuous_scale="RdYlGn_r",
    template="plotly_dark",
    title="Revenue Structure and Churn Risk by Payment Method",
    hover_data=[
        "Customers",
        "Revenue_Per_Customer",
        "Total_Refunds",
        "Refund_Per_Customer",
        "Retention_Rate"
    ]
)

fig.update_layout(
    title_x=0.5,
    height=650
)

fig.show()


# ============================================================
# BUSINESS INSIGHTS
# ============================================================

print("\n" + "=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{highest_revenue_payment['Payment_Method']} generates the "
    f"highest total revenue at "
    f"₹{highest_revenue_payment['Total_Revenue']:,.2f}."
)

print(
    f"{highest_value_payment['Payment_Method']} has the highest "
    f"revenue per customer at "
    f"₹{highest_value_payment['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{highest_churn_payment['Payment_Method']} shows the highest "
    f"payment-method churn rate of "
    f"{highest_churn_payment['Churn_Rate']:.2f}%."
)

print(
    f"{highest_refund_payment['Payment_Method']} has the highest "
    f"refund per customer at "
    f"₹{highest_refund_payment['Refund_Per_Customer']:,.2f}."
)

print(
    f"The automatic-payment group with the lowest churn is "
    f"{best_auto_group['Is_Automatic_Payment']} at "
    f"{best_auto_group['Churn_Rate']:.2f}%."
)

print(
    f"The automatic-payment group with the highest churn is "
    f"{worst_auto_group['Is_Automatic_Payment']} at "
    f"{worst_auto_group['Churn_Rate']:.2f}%."
)

print(
    "Payment-method performance should be evaluated using both "
    "customer value and churn rather than revenue alone."
)

print(
    "Refund-heavy payment segments may indicate revenue leakage "
    "and should be investigated alongside billing behavior."
)

print(
    "Automatic-payment behavior can help identify billing patterns "
    "associated with stronger or weaker customer retention."
)

print(
    "Business Action: Focus retention and billing optimization on "
    "payment segments with high churn, high refunds and strong "
    "revenue exposure."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 25
# ============================================================

# Which customer value groups generate the strongest revenue,
# net revenue and retention performance?

# Business Goal:
# Identify high-value customers, revenue efficiency,
# customer lifetime value and retention opportunities.

# ============================================================

import pandas as pd
import plotly.express as px


analysis_df = df.copy()


# ============================================================
# DATA PREPARATION
# ============================================================

numeric_cols = [
    "Age",
    "Monthly_Charge",
    "Total_Revenue",
    "Net_Revenue",
    "Total_Extra_Charges",
    "Tenure_in_Months"
]

for col in numeric_cols:
    analysis_df[col] = pd.to_numeric(
        analysis_df[col],
        errors="coerce"
    )

analysis_df["Age"] = analysis_df["Age"].fillna(
    analysis_df["Age"].median()
)

for col in [
    "Monthly_Charge",
    "Total_Revenue",
    "Net_Revenue",
    "Total_Extra_Charges",
    "Tenure_in_Months"
]:
    analysis_df[col] = analysis_df[col].fillna(0)


analysis_df["Revenue_Efficiency"] = (
    analysis_df["Net_Revenue"]
    .div(
        analysis_df["Total_Revenue"].replace(0, pd.NA)
    )
    .mul(100)
    .fillna(0)
)

analysis_df["Revenue_per_Month"] = (
    analysis_df["Total_Revenue"]
    .div(
        analysis_df["Tenure_in_Months"].replace(0, pd.NA)
    )
    .fillna(0)
)


# ============================================================
# AGE GROUP
# ============================================================

analysis_df["Age_Group"] = pd.cut(
    analysis_df["Age"],
    bins=[0, 25, 35, 45, 55, 65, 120],
    labels=[
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "56-65",
        "66+"
    ],
    include_lowest=True
)


# ============================================================
# CUSTOMER VALUE TIER
# ============================================================

analysis_df["Revenue_Value_Tier"] = pd.qcut(
    analysis_df["Total_Revenue"],
    q=4,
    labels=[
        "Low Value",
        "Medium Value",
        "High Value",
        "Very High Value"
    ],
    duplicates="drop"
)


# ============================================================
# VALUE TIER SUMMARY
# ============================================================

value_summary = (
    analysis_df.groupby(
        "Revenue_Value_Tier",
        observed=True,
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Extra_Charges=("Total_Extra_Charges", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        ),
        Stayed=(
            "Customer_Status",
            lambda x: (x == "Stayed").sum()
        )
    )
    .reset_index()
)

value_summary["Churn_Rate"] = (
    value_summary["Churned"]
    .div(value_summary["Customers"])
    .mul(100)
)

value_summary["Retention_Rate"] = (
    value_summary["Stayed"]
    .div(value_summary["Customers"])
    .mul(100)
)

value_summary["Revenue_Per_Customer"] = (
    value_summary["Total_Revenue"]
    .div(value_summary["Customers"])
)

value_summary["Net_Revenue_Per_Customer"] = (
    value_summary["Net_Revenue"]
    .div(value_summary["Customers"])
)

value_summary["Revenue_Efficiency"] = (
    value_summary["Net_Revenue"]
    .div(
        value_summary["Total_Revenue"]
        .replace(0, pd.NA)
    )
    .mul(100)
    .fillna(0)
)

value_summary["Revenue_Share"] = (
    value_summary["Total_Revenue"]
    .div(value_summary["Total_Revenue"].sum())
    .mul(100)
)

display(
    value_summary.round(2)
)


# ============================================================
# AGE GROUP SUMMARY
# ============================================================

age_summary = (
    analysis_df.groupby(
        "Age_Group",
        observed=True,
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        ),
        Stayed=(
            "Customer_Status",
            lambda x: (x == "Stayed").sum()
        )
    )
    .reset_index()
)

age_summary["Churn_Rate"] = (
    age_summary["Churned"]
    .div(age_summary["Customers"])
    .mul(100)
)

age_summary["Retention_Rate"] = (
    age_summary["Stayed"]
    .div(age_summary["Customers"])
    .mul(100)
)

age_summary["Revenue_Per_Customer"] = (
    age_summary["Total_Revenue"]
    .div(age_summary["Customers"])
)

age_summary["Revenue_Efficiency"] = (
    age_summary["Net_Revenue"]
    .div(
        age_summary["Total_Revenue"]
        .replace(0, pd.NA)
    )
    .mul(100)
    .fillna(0)
)

display(
    age_summary.round(2)
)


# ============================================================
# GENDER + MARITAL STATUS SUMMARY
# ============================================================

demographic_summary = (
    analysis_df.groupby(
        [
            "Gender",
            "Married"
        ],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        ),
        Stayed=(
            "Customer_Status",
            lambda x: (x == "Stayed").sum()
        )
    )
    .reset_index()
)

demographic_summary["Churn_Rate"] = (
    demographic_summary["Churned"]
    .div(demographic_summary["Customers"])
    .mul(100)
)

demographic_summary["Retention_Rate"] = (
    demographic_summary["Stayed"]
    .div(demographic_summary["Customers"])
    .mul(100)
)

demographic_summary["Revenue_Per_Customer"] = (
    demographic_summary["Total_Revenue"]
    .div(demographic_summary["Customers"])
)

display(
    demographic_summary.round(2)
)


# ============================================================
# STATE SUMMARY
# ============================================================

state_summary = (
    analysis_df.groupby(
        "State",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        ),
        Stayed=(
            "Customer_Status",
            lambda x: (x == "Stayed").sum()
        )
    )
    .reset_index()
)

state_summary["Churn_Rate"] = (
    state_summary["Churned"]
    .div(state_summary["Customers"])
    .mul(100)
)

state_summary["Retention_Rate"] = (
    state_summary["Stayed"]
    .div(state_summary["Customers"])
    .mul(100)
)

state_summary["Revenue_Per_Customer"] = (
    state_summary["Total_Revenue"]
    .div(state_summary["Customers"])
)

state_summary["Revenue_Share"] = (
    state_summary["Total_Revenue"]
    .div(state_summary["Total_Revenue"].sum())
    .mul(100)
)

display(
    state_summary
    .sort_values(
        "Total_Revenue",
        ascending=False
    )
    .round(2)
)


# ============================================================
# KPI
# ============================================================

top_value_tier = value_summary.loc[
    value_summary["Total_Revenue"].idxmax()
]

highest_customer_value = value_summary.loc[
    value_summary["Revenue_Per_Customer"].idxmax()
]

best_efficiency_tier = value_summary.loc[
    value_summary["Revenue_Efficiency"].idxmax()
]

highest_churn_tier = value_summary.loc[
    value_summary["Churn_Rate"].idxmax()
]

top_age_revenue = age_summary.loc[
    age_summary["Total_Revenue"].idxmax()
]

top_state_revenue = state_summary.loc[
    state_summary["Total_Revenue"].idxmax()
]

best_state_value = state_summary.loc[
    state_summary["Revenue_Per_Customer"].idxmax()
]


print("=" * 75)
print("CUSTOMER VALUE & REVENUE KPI")
print("=" * 75)

print(
    f"Top Revenue Value Tier    : "
    f"{top_value_tier['Revenue_Value_Tier']}"
)

print(
    f"Total Revenue             : "
    f"₹{top_value_tier['Total_Revenue']:,.2f}"
)

print(
    f"Highest Customer Value    : "
    f"{highest_customer_value['Revenue_Value_Tier']}"
)

print(
    f"Revenue / Customer        : "
    f"₹{highest_customer_value['Revenue_Per_Customer']:,.2f}"
)

print(
    f"Best Revenue Efficiency   : "
    f"{best_efficiency_tier['Revenue_Value_Tier']}"
)

print(
    f"Revenue Efficiency        : "
    f"{best_efficiency_tier['Revenue_Efficiency']:.2f}%"
)

print(
    f"Highest Churn Value Tier  : "
    f"{highest_churn_tier['Revenue_Value_Tier']}"
)

print(
    f"Churn Rate                : "
    f"{highest_churn_tier['Churn_Rate']:.2f}%"
)

print(
    f"Top Age Group Revenue     : "
    f"{top_age_revenue['Age_Group']}"
)

print(
    f"Top Revenue State         : "
    f"{top_state_revenue['State']}"
)

print(
    f"Best State Revenue/Customer: "
    f"{best_state_value['State']}"
)

print("=" * 75)


# ============================================================
# VISUAL 1
# ============================================================

fig = px.bar(
    value_summary,
    x="Revenue_Value_Tier",
    y="Total_Revenue",
    color="Revenue_Efficiency",
    text="Total_Revenue",
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Total Revenue by Customer Value Tier",
    hover_data=[
        "Customers",
        "Revenue_Per_Customer",
        "Churn_Rate",
        "Retention_Rate"
    ]
)

fig.update_traces(
    texttemplate="₹%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Customer Value Tier",
    yaxis_title="Total Revenue",
    coloraxis_colorbar_title="Revenue Efficiency (%)"
)

fig.show()


# ============================================================
# VISUAL 2
# ============================================================

fig = px.scatter(
    value_summary,
    x="Revenue_Per_Customer",
    y="Retention_Rate",
    size="Customers",
    color="Churn_Rate",
    text="Revenue_Value_Tier",
    color_continuous_scale="RdYlGn_r",
    template="plotly_dark",
    title="Customer Value vs Retention Performance",
    hover_data=[
        "Total_Revenue",
        "Net_Revenue",
        "Avg_Tenure",
        "Revenue_Efficiency"
    ]
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Revenue per Customer",
    yaxis_title="Retention Rate (%)",
    coloraxis_colorbar_title="Churn Rate (%)"
)

fig.show()


# ============================================================
# VISUAL 3
# ============================================================

fig = px.bar(
    age_summary,
    x="Age_Group",
    y="Revenue_Per_Customer",
    color="Churn_Rate",
    text="Revenue_Per_Customer",
    color_continuous_scale="Viridis",
    template="plotly_white",
    title="Revenue per Customer by Age Group",
    hover_data=[
        "Customers",
        "Total_Revenue",
        "Net_Revenue",
        "Avg_Tenure",
        "Retention_Rate"
    ]
)

fig.update_traces(
    texttemplate="₹%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Age Group",
    yaxis_title="Revenue per Customer",
    coloraxis_colorbar_title="Churn Rate (%)"
)

fig.show()


# ============================================================
# VISUAL 4
# ============================================================

fig = px.scatter(
    age_summary,
    x="Avg_Tenure",
    y="Revenue_Per_Customer",
    size="Customers",
    color="Revenue_Efficiency",
    text="Age_Group",
    color_continuous_scale="Plasma",
    template="plotly_dark",
    title="Customer Tenure vs Revenue Value",
    hover_data=[
        "Total_Revenue",
        "Net_Revenue",
        "Churn_Rate",
        "Retention_Rate"
    ]
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Tenure (Months)",
    yaxis_title="Revenue per Customer",
    coloraxis_colorbar_title="Revenue Efficiency (%)"
)

fig.show()


# ============================================================
# VISUAL 5
# ============================================================

fig = px.bar(
    demographic_summary,
    x="Gender",
    y="Total_Revenue",
    color="Married",
    barmode="group",
    text="Total_Revenue",
    template="plotly_white",
    title="Revenue Performance by Gender and Marital Status",
    hover_data=[
        "Customers",
        "Revenue_Per_Customer",
        "Churn_Rate",
        "Retention_Rate"
    ]
)

fig.update_traces(
    texttemplate="₹%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Gender",
    yaxis_title="Total Revenue",
    legend_title="Married"
)

fig.show()


# ============================================================
# VISUAL 6
# ============================================================

state_plot = state_summary.copy()

state_plot["Plot_Size"] = (
    state_plot["Customers"]
    .abs()
    .clip(lower=1)
)

fig = px.scatter(
    state_plot,
    x="Total_Revenue",
    y="Revenue_Per_Customer",
    size="Plot_Size",
    color="Churn_Rate",
    text="State",
    color_continuous_scale="Turbo",
    template="plotly_dark",
    title="State Revenue vs Customer Value",
    hover_data=[
        "Customers",
        "Net_Revenue",
        "Revenue_Share",
        "Retention_Rate",
        "Avg_Tenure"
    ]
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Total Revenue",
    yaxis_title="Revenue per Customer",
    coloraxis_colorbar_title="Churn Rate (%)"
)

fig.show()


# ============================================================
# BUSINESS INSIGHTS
# ============================================================

print("\n" + "=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{top_value_tier['Revenue_Value_Tier']} contributes the highest "
    f"revenue at ₹{top_value_tier['Total_Revenue']:,.2f}."
)

print(
    f"{highest_customer_value['Revenue_Value_Tier']} has the highest "
    f"revenue per customer at "
    f"₹{highest_customer_value['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{best_efficiency_tier['Revenue_Value_Tier']} delivers the "
    f"highest revenue efficiency at "
    f"{best_efficiency_tier['Revenue_Efficiency']:.2f}%."
)

print(
    f"{highest_churn_tier['Revenue_Value_Tier']} has the highest "
    f"churn rate at "
    f"{highest_churn_tier['Churn_Rate']:.2f}%."
)

print(
    f"{top_age_revenue['Age_Group']} generates the highest total "
    f"revenue among age groups."
)

print(
    f"{top_state_revenue['State']} generates the highest state-level "
    f"revenue at "
    f"₹{top_state_revenue['Total_Revenue']:,.2f}."
)

print(
    f"{best_state_value['State']} has the strongest revenue per "
    f"customer among states."
)

print(
    "High-value customers should receive stronger retention and "
    "cross-selling strategies."
)

print(
    "High-value groups with high churn represent important "
    "revenue-at-risk opportunities."
)

print(
    "Revenue efficiency helps identify customer groups that "
    "generate stronger net revenue relative to total revenue."
)

print(
    "Business Action: Prioritize high-value customers for "
    "retention, premium services and targeted upselling."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 26
# ============================================================

# How does referral activity influence customer value,
# revenue generation and churn risk?

import pandas as pd
import plotly.express as px


analysis_df = df.copy()


# ============================================================
# DATA PREPARATION
# ============================================================

required_cols = [
    "Customer_ID",
    "Number_of_Referrals",
    "Customer_Status",
    "Total_Revenue",
    "Net_Revenue",
    "Monthly_Charge",
    "Tenure_in_Months",
    "State",
    "Gender",
    "Married"
]

missing_cols = [
    col for col in required_cols
    if col not in analysis_df.columns
]

if missing_cols:
    raise ValueError(
        f"Missing columns in dataframe: {missing_cols}"
    )


numeric_cols = [
    "Number_of_Referrals",
    "Total_Revenue",
    "Net_Revenue",
    "Monthly_Charge",
    "Tenure_in_Months"
]

for col in numeric_cols:
    analysis_df[col] = pd.to_numeric(
        analysis_df[col],
        errors="coerce"
    ).fillna(0)


analysis_df["Number_of_Referrals"] = (
    analysis_df["Number_of_Referrals"]
    .clip(lower=0)
)


# ============================================================
# REFERRAL GROUP
# ============================================================

analysis_df["Referral_Group"] = pd.cut(
    analysis_df["Number_of_Referrals"],
    bins=[
        -1,
        0,
        2,
        5,
        float("inf")
    ],
    labels=[
        "No Referrals",
        "1-2 Referrals",
        "3-5 Referrals",
        "6+ Referrals"
    ]
)


# ============================================================
# REFERRAL SUMMARY
# ============================================================

referral_summary = (
    analysis_df.groupby(
        "Referral_Group",
        observed=True
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Net_Revenue=("Net_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Referrals=("Number_of_Referrals", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        ),
        Stayed=(
            "Customer_Status",
            lambda x: (x == "Stayed").sum()
        )
    )
    .reset_index()
)


referral_summary["Churn_Rate"] = (
    referral_summary["Churned"]
    .div(referral_summary["Customers"])
    .mul(100)
)


referral_summary["Retention_Rate"] = (
    referral_summary["Stayed"]
    .div(referral_summary["Customers"])
    .mul(100)
)


referral_summary["Revenue_Per_Customer"] = (
    referral_summary["Total_Revenue"]
    .div(referral_summary["Customers"])
)


referral_summary["Net_Revenue_Per_Customer"] = (
    referral_summary["Net_Revenue"]
    .div(referral_summary["Customers"])
)


referral_summary["Revenue_Share"] = (
    referral_summary["Total_Revenue"]
    .div(referral_summary["Total_Revenue"].sum())
    .mul(100)
)


referral_summary["Customer_Share"] = (
    referral_summary["Customers"]
    .div(referral_summary["Customers"].sum())
    .mul(100)
)


referral_summary["Revenue_Efficiency"] = (
    referral_summary["Net_Revenue"]
    .div(
        referral_summary["Total_Revenue"]
        .replace(0, pd.NA)
    )
    .mul(100)
    .fillna(0)
)


display(
    referral_summary.round(2)
)


# ============================================================
# REFERRAL + CUSTOMER STATUS
# ============================================================

referral_status = (
    analysis_df.groupby(
        [
            "Referral_Group",
            "Customer_Status"
        ],
        observed=True
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean")
    )
    .reset_index()
)


display(
    referral_status
    .sort_values(
        "Total_Revenue",
        ascending=False
    )
    .round(2)
)


# ============================================================
# REFERRAL + STATE
# ============================================================

referral_state = (
    analysis_df.groupby(
        [
            "Referral_Group",
            "State"
        ],
        observed=True,
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        )
    )
    .reset_index()
)


referral_state["Churn_Rate"] = (
    referral_state["Churned"]
    .div(referral_state["Customers"])
    .mul(100)
)


referral_state["Revenue_Per_Customer"] = (
    referral_state["Total_Revenue"]
    .div(referral_state["Customers"])
)


display(
    referral_state
    .sort_values(
        "Total_Revenue",
        ascending=False
    )
    .head(20)
    .round(2)
)


# ============================================================
# REFERRAL + GENDER
# ============================================================

referral_gender = (
    analysis_df.groupby(
        [
            "Referral_Group",
            "Gender"
        ],
        observed=True,
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Churned=(
            "Customer_Status",
            lambda x: (x == "Churned").sum()
        )
    )
    .reset_index()
)


referral_gender["Churn_Rate"] = (
    referral_gender["Churned"]
    .div(referral_gender["Customers"])
    .mul(100)
)


referral_gender["Revenue_Per_Customer"] = (
    referral_gender["Total_Revenue"]
    .div(referral_gender["Customers"])
)


display(
    referral_gender.round(2)
)


# ============================================================
# KPI
# ============================================================

top_revenue_group = referral_summary.loc[
    referral_summary["Total_Revenue"].idxmax()
]


highest_value_group = referral_summary.loc[
    referral_summary["Revenue_Per_Customer"].idxmax()
]


best_retention_group = referral_summary.loc[
    referral_summary["Retention_Rate"].idxmax()
]


highest_churn_group = referral_summary.loc[
    referral_summary["Churn_Rate"].idxmax()
]


best_efficiency_group = referral_summary.loc[
    referral_summary["Revenue_Efficiency"].idxmax()
]


largest_customer_group = referral_summary.loc[
    referral_summary["Customers"].idxmax()
]


print("=" * 75)
print("REFERRAL PERFORMANCE KPI")
print("=" * 75)

print(
    f"Top Revenue Referral     : "
    f"{top_revenue_group['Referral_Group']}"
)

print(
    f"Total Revenue            : "
    f"₹{top_revenue_group['Total_Revenue']:,.2f}"
)

print(
    f"Highest Customer Value   : "
    f"{highest_value_group['Referral_Group']}"
)

print(
    f"Revenue / Customer       : "
    f"₹{highest_value_group['Revenue_Per_Customer']:,.2f}"
)

print(
    f"Best Retention           : "
    f"{best_retention_group['Referral_Group']}"
)

print(
    f"Retention Rate           : "
    f"{best_retention_group['Retention_Rate']:.2f}%"
)

print(
    f"Highest Churn            : "
    f"{highest_churn_group['Referral_Group']}"
)

print(
    f"Churn Rate               : "
    f"{highest_churn_group['Churn_Rate']:.2f}%"
)

print(
    f"Best Revenue Efficiency  : "
    f"{best_efficiency_group['Referral_Group']}"
)

print(
    f"Revenue Efficiency       : "
    f"{best_efficiency_group['Revenue_Efficiency']:.2f}%"
)

print(
    f"Largest Customer Group   : "
    f"{largest_customer_group['Referral_Group']}"
)

print(
    f"Customers                : "
    f"{largest_customer_group['Customers']:,}"
)

print("=" * 75)


# ============================================================
# VISUAL 1
# ============================================================

fig = px.bar(
    referral_summary,
    x="Referral_Group",
    y="Total_Revenue",
    color="Revenue_Efficiency",
    text="Total_Revenue",
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Total Revenue by Referral Group",
    hover_data=[
        "Customers",
        "Revenue_Per_Customer",
        "Churn_Rate",
        "Retention_Rate",
        "Revenue_Share"
    ]
)

fig.update_traces(
    texttemplate="₹%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Referral Group",
    yaxis_title="Total Revenue",
    coloraxis_colorbar_title="Revenue Efficiency (%)"
)

fig.show()


# ============================================================
# VISUAL 2
# ============================================================

fig = px.scatter(
    referral_summary,
    x="Customers",
    y="Revenue_Per_Customer",
    size="Customers",
    color="Churn_Rate",
    text="Referral_Group",
    color_continuous_scale="RdYlGn_r",
    template="plotly_dark",
    title="Referral Volume vs Customer Revenue Value",
    hover_data=[
        "Total_Revenue",
        "Net_Revenue",
        "Retention_Rate",
        "Avg_Tenure"
    ]
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Customers",
    yaxis_title="Revenue per Customer",
    coloraxis_colorbar_title="Churn Rate (%)"
)

fig.show()


# ============================================================
# VISUAL 3
# ============================================================

fig = px.bar(
    referral_summary,
    x="Referral_Group",
    y=[
        "Churn_Rate",
        "Retention_Rate"
    ],
    barmode="group",
    text_auto=".2f",
    template="plotly_white",
    title="Referral Group Retention vs Churn"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Referral Group",
    yaxis_title="Rate (%)"
)

fig.show()


# ============================================================
# VISUAL 4
# ============================================================

fig = px.scatter(
    referral_summary,
    x="Avg_Tenure",
    y="Revenue_Per_Customer",
    size="Customers",
    color="Revenue_Efficiency",
    text="Referral_Group",
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Referral Customer Tenure vs Revenue Value",
    hover_data=[
        "Total_Revenue",
        "Net_Revenue",
        "Churn_Rate",
        "Retention_Rate"
    ]
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Tenure (Months)",
    yaxis_title="Revenue per Customer",
    coloraxis_colorbar_title="Revenue Efficiency (%)"
)

fig.show()


# ============================================================
# VISUAL 5
# ============================================================

fig = px.sunburst(
    referral_status,
    path=[
        "Referral_Group",
        "Customer_Status"
    ],
    values="Customers",
    color="Total_Revenue",
    color_continuous_scale="Plasma",
    template="plotly_white",
    title="Customer Status Distribution by Referral Group",
    hover_data=[
        "Avg_Revenue",
        "Avg_Tenure"
    ]
)

fig.update_layout(
    title_x=0.5,
    height=700
)

fig.show()


# ============================================================
# VISUAL 6
# ============================================================

referral_state_plot = (
    referral_state
    .sort_values(
        "Total_Revenue",
        ascending=False
    )
    .head(20)
    .copy()
)

referral_state_plot["Plot_Size"] = (
    referral_state_plot["Customers"]
    .abs()
    .clip(lower=1)
)

fig = px.scatter(
    referral_state_plot,
    x="Total_Revenue",
    y="Revenue_Per_Customer",
    size="Plot_Size",
    color="Churn_Rate",
    text="State",
    symbol="Referral_Group",
    color_continuous_scale="Turbo",
    template="plotly_dark",
    title="Referral Performance Across Top Revenue States",
    hover_data=[
        "Referral_Group",
        "State",
        "Customers",
        "Net_Revenue",
        "Churn_Rate"
    ]
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Total Revenue",
    yaxis_title="Revenue per Customer",
    coloraxis_colorbar_title="Churn Rate (%)"
)

fig.show()


# ============================================================
# BUSINESS INSIGHTS
# ============================================================

print("\n" + "=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{top_revenue_group['Referral_Group']} generates the highest "
    f"total revenue of "
    f"₹{top_revenue_group['Total_Revenue']:,.2f}."
)

print(
    f"{highest_value_group['Referral_Group']} has the highest "
    f"revenue per customer at "
    f"₹{highest_value_group['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{best_retention_group['Referral_Group']} has the strongest "
    f"retention rate of "
    f"{best_retention_group['Retention_Rate']:.2f}%."
)

print(
    f"{highest_churn_group['Referral_Group']} has the highest "
    f"churn rate of "
    f"{highest_churn_group['Churn_Rate']:.2f}%."
)

print(
    f"{best_efficiency_group['Referral_Group']} provides the "
    f"highest revenue efficiency of "
    f"{best_efficiency_group['Revenue_Efficiency']:.2f}%."
)

print(
    f"{largest_customer_group['Referral_Group']} contributes the "
    f"largest customer population with "
    f"{largest_customer_group['Customers']:,} customers."
)

print(
    "Referral volume should be evaluated together with customer "
    "value, retention and churn rather than customer count alone."
)

print(
    "High-referral groups with strong revenue and retention can "
    "be prioritized for referral and loyalty programs."
)

print(
    "High-referral groups with high churn should receive targeted "
    "retention strategies before increasing acquisition investment."
)

print(
    "State-level referral analysis helps identify geographic areas "
    "where referral-driven customers create stronger business value."
)

print(
    "Business Action: Invest more in referral sources that combine "
    "high customer value, strong retention and healthy revenue efficiency."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 27
# ============================================================

# Which churn categories and reasons create the greatest
# revenue exposure and customer retention risk?

# Business Goal:
# Identify the major sources of revenue loss among churned
# customers and prioritize retention actions.

# ============================================================

import pandas as pd
import plotly.express as px


analysis_df = df.copy()


# ============================================================
# DATA PREPARATION
# ============================================================

required_cols = [
    "Customer_ID",
    "Customer_Status",
    "Churn_Category",
    "Churn_Reason",
    "Total_Revenue",
    "Net_Revenue",
    "Monthly_Charge",
    "Tenure_in_Months",
    "Contract",
    "Payment_Method"
]

missing_cols = [
    col for col in required_cols
    if col not in analysis_df.columns
]

if missing_cols:
    raise ValueError(
        f"Missing columns in dataframe: {missing_cols}"
    )


for col in [
    "Total_Revenue",
    "Net_Revenue",
    "Monthly_Charge",
    "Tenure_in_Months"
]:

    analysis_df[col] = pd.to_numeric(
        analysis_df[col],
        errors="coerce"
    ).fillna(0)


analysis_df["Total_Revenue"] = (
    analysis_df["Total_Revenue"]
    .clip(lower=0)
)

analysis_df["Net_Revenue"] = (
    analysis_df["Net_Revenue"]
    .clip(lower=0)
)

analysis_df["Monthly_Charge"] = (
    analysis_df["Monthly_Charge"]
    .clip(lower=0)
)

analysis_df["Tenure_in_Months"] = (
    analysis_df["Tenure_in_Months"]
    .clip(lower=0)
)


churned_df = analysis_df[
    analysis_df["Customer_Status"]
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("churned")
].copy()


if churned_df.empty:
    raise ValueError(
        "No churned customers found in Customer_Status."
    )


# ============================================================
# CHURN CATEGORY SUMMARY
# ============================================================

category_summary = (
    churned_df.groupby(
        "Churn_Category",
        dropna=False
    )
    .agg(
        Churned_Customers=("Customer_ID", "nunique"),
        Revenue_Exposure=("Total_Revenue", "sum"),
        Net_Revenue_Exposure=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean")
    )
    .reset_index()
)


category_summary["Revenue_Share"] = (
    category_summary["Revenue_Exposure"]
    .div(
        category_summary["Revenue_Exposure"].sum()
    )
    .mul(100)
)


category_summary["Revenue_Per_Customer"] = (
    category_summary["Revenue_Exposure"]
    .div(
        category_summary["Churned_Customers"]
    )
)


display(
    category_summary
    .sort_values(
        "Revenue_Exposure",
        ascending=False
    )
    .round(2)
)


# ============================================================
# CHURN REASON SUMMARY
# ============================================================

reason_summary = (
    churned_df.groupby(
        "Churn_Reason",
        dropna=False
    )
    .agg(
        Churned_Customers=("Customer_ID", "nunique"),
        Revenue_Exposure=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean")
    )
    .reset_index()
)


reason_summary["Revenue_Share"] = (
    reason_summary["Revenue_Exposure"]
    .div(
        reason_summary["Revenue_Exposure"].sum()
    )
    .mul(100)
)


reason_summary["Revenue_Per_Customer"] = (
    reason_summary["Revenue_Exposure"]
    .div(
        reason_summary["Churned_Customers"]
    )
)


display(
    reason_summary
    .sort_values(
        "Revenue_Exposure",
        ascending=False
    )
    .round(2)
)


# ============================================================
# CONTRACT + PAYMENT METHOD
# ============================================================

contract_payment = (
    churned_df.groupby(
        [
            "Contract",
            "Payment_Method"
        ],
        dropna=False
    )
    .agg(
        Churned_Customers=("Customer_ID", "nunique"),
        Revenue_Exposure=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean")
    )
    .reset_index()
)


contract_payment["Revenue_Per_Customer"] = (
    contract_payment["Revenue_Exposure"]
    .div(
        contract_payment["Churned_Customers"]
    )
)


display(
    contract_payment
    .sort_values(
        "Revenue_Exposure",
        ascending=False
    )
    .head(20)
    .round(2)
)


# ============================================================
# TENURE RISK GROUP
# ============================================================

churned_df["Tenure_Group"] = pd.cut(
    churned_df["Tenure_in_Months"],
    bins=[
        -1,
        6,
        12,
        24,
        36,
        float("inf")
    ],
    labels=[
        "0-6 Months",
        "7-12 Months",
        "13-24 Months",
        "25-36 Months",
        "37+ Months"
    ]
)


tenure_summary = (
    churned_df.groupby(
        "Tenure_Group",
        observed=True
    )
    .agg(
        Churned_Customers=("Customer_ID", "nunique"),
        Revenue_Exposure=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean")
    )
    .reset_index()
)


tenure_summary["Revenue_Per_Customer"] = (
    tenure_summary["Revenue_Exposure"]
    .div(
        tenure_summary["Churned_Customers"]
    )
)


display(
    tenure_summary.round(2)
)


# ============================================================
# CHURN CATEGORY + REASON
# ============================================================

category_reason = (
    churned_df.groupby(
        [
            "Churn_Category",
            "Churn_Reason"
        ],
        dropna=False
    )
    .agg(
        Churned_Customers=("Customer_ID", "nunique"),
        Revenue_Exposure=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean")
    )
    .reset_index()
)


category_reason["Revenue_Per_Customer"] = (
    category_reason["Revenue_Exposure"]
    .div(
        category_reason["Churned_Customers"]
    )
)


display(
    category_reason
    .sort_values(
        "Revenue_Exposure",
        ascending=False
    )
    .head(20)
    .round(2)
)


# ============================================================
# KPI
# ============================================================

top_category = category_summary.loc[
    category_summary["Revenue_Exposure"].idxmax()
]


top_reason = reason_summary.loc[
    reason_summary["Revenue_Exposure"].idxmax()
]


highest_value_reason = reason_summary.loc[
    reason_summary["Revenue_Per_Customer"].idxmax()
]


top_tenure_group = tenure_summary.loc[
    tenure_summary["Revenue_Exposure"].idxmax()
]


top_contract_payment = contract_payment.loc[
    contract_payment["Revenue_Exposure"].idxmax()
]


total_revenue_exposure = (
    churned_df["Total_Revenue"].sum()
)


total_churned_customers = (
    churned_df["Customer_ID"].nunique()
)


avg_churned_revenue = (
    churned_df["Total_Revenue"].mean()
)


print("=" * 75)
print("CHURN REVENUE EXPOSURE KPI")
print("=" * 75)

print(
    f"Total Churned Customers     : "
    f"{total_churned_customers:,}"
)

print(
    f"Revenue Exposure            : "
    f"₹{total_revenue_exposure:,.2f}"
)

print(
    f"Average Revenue Lost/Customer: "
    f"₹{avg_churned_revenue:,.2f}"
)

print(
    f"Highest Risk Category       : "
    f"{top_category['Churn_Category']}"
)

print(
    f"Category Revenue Exposure   : "
    f"₹{top_category['Revenue_Exposure']:,.2f}"
)

print(
    f"Highest Risk Reason         : "
    f"{top_reason['Churn_Reason']}"
)

print(
    f"Reason Revenue Exposure     : "
    f"₹{top_reason['Revenue_Exposure']:,.2f}"
)

print(
    f"Highest Value Churn Reason  : "
    f"{highest_value_reason['Churn_Reason']}"
)

print(
    f"Revenue / Churned Customer  : "
    f"₹{highest_value_reason['Revenue_Per_Customer']:,.2f}"
)

print(
    f"Highest Revenue Tenure Group: "
    f"{top_tenure_group['Tenure_Group']}"
)

print(
    f"Revenue Exposure            : "
    f"₹{top_tenure_group['Revenue_Exposure']:,.2f}"
)

print("=" * 75)


# ============================================================
# VISUAL 1
# ============================================================

fig = px.bar(
    category_summary.sort_values(
        "Revenue_Exposure",
        ascending=True
    ),
    x="Revenue_Exposure",
    y="Churn_Category",
    orientation="h",
    color="Revenue_Share",
    text="Revenue_Exposure",
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Revenue Exposure by Churn Category",
    hover_data=[
        "Churned_Customers",
        "Avg_Revenue",
        "Avg_Tenure",
        "Revenue_Per_Customer"
    ]
)

fig.update_traces(
    texttemplate="₹%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Revenue Exposure",
    yaxis_title="Churn Category",
    coloraxis_colorbar_title="Revenue Share (%)"
)

fig.show()


# ============================================================
# VISUAL 2
# ============================================================

top_reasons_plot = (
    reason_summary
    .sort_values(
        "Revenue_Exposure",
        ascending=False
    )
    .head(12)
    .sort_values(
        "Revenue_Exposure",
        ascending=True
    )
)

fig = px.bar(
    top_reasons_plot,
    x="Revenue_Exposure",
    y="Churn_Reason",
    orientation="h",
    color="Avg_Monthly_Charge",
    text="Revenue_Exposure",
    color_continuous_scale="Plasma",
    template="plotly_dark",
    title="Top Churn Reasons by Revenue Exposure",
    hover_data=[
        "Churned_Customers",
        "Revenue_Share",
        "Revenue_Per_Customer",
        "Avg_Tenure"
    ]
)

fig.update_traces(
    texttemplate="₹%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Revenue Exposure",
    yaxis_title="Churn Reason",
    coloraxis_colorbar_title="Avg Monthly Charge"
)

fig.show()


# ============================================================
# VISUAL 3
# ============================================================

fig = px.scatter(
    reason_summary,
    x="Avg_Tenure",
    y="Revenue_Per_Customer",
    size="Churned_Customers",
    color="Revenue_Exposure",
    text="Churn_Reason",
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Churn Reason: Tenure vs Revenue at Risk",
    hover_data=[
        "Churned_Customers",
        "Avg_Revenue",
        "Avg_Monthly_Charge",
        "Revenue_Share"
    ]
)

fig.update_traces(
    textposition="top center"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Average Tenure (Months)",
    yaxis_title="Revenue per Churned Customer",
    coloraxis_colorbar_title="Revenue Exposure"
)

fig.show()


# ============================================================
# VISUAL 4
# ============================================================

fig = px.bar(
    tenure_summary,
    x="Tenure_Group",
    y="Revenue_Exposure",
    color="Revenue_Per_Customer",
    text="Revenue_Exposure",
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Revenue Exposure Across Churned Customer Tenure",
    hover_data=[
        "Churned_Customers",
        "Avg_Revenue",
        "Avg_Monthly_Charge"
    ]
)

fig.update_traces(
    texttemplate="₹%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Tenure Group",
    yaxis_title="Revenue Exposure",
    coloraxis_colorbar_title="Revenue / Customer"
)

fig.show()


# ============================================================
# VISUAL 5
# ============================================================

contract_plot = (
    contract_payment
    .sort_values(
        "Revenue_Exposure",
        ascending=False
    )
    .head(15)
    .copy()
)

contract_plot["Contract_Payment"] = (
    contract_plot["Contract"].astype(str)
    + " | "
    + contract_plot["Payment_Method"].astype(str)
)

fig = px.treemap(
    contract_plot,
    path=[
        "Contract",
        "Payment_Method"
    ],
    values="Revenue_Exposure",
    color="Revenue_Per_Customer",
    color_continuous_scale="RdYlGn",
    template="plotly_white",
    title="Revenue Exposure by Contract and Payment Method",
    hover_data=[
        "Churned_Customers",
        "Avg_Revenue",
        "Avg_Tenure"
    ]
)

fig.update_layout(
    title_x=0.5,
    height=700
)

fig.show()


# ============================================================
# VISUAL 6
# ============================================================

heatmap_data = (
    category_reason
    .pivot_table(
        index="Churn_Category",
        columns="Churn_Reason",
        values="Revenue_Exposure",
        aggfunc="sum",
        fill_value=0
    )
)

fig = px.imshow(
    heatmap_data,
    text_auto=".0f",
    aspect="auto",
    color_continuous_scale="Magma",
    template="plotly_dark",
    title="Churn Category vs Churn Reason Revenue Exposure",
    labels={
        "x": "Churn Reason",
        "y": "Churn Category",
        "color": "Revenue Exposure"
    }
)

fig.update_layout(
    title_x=0.5,
    height=650
)

fig.show()


# ============================================================
# BUSINESS INSIGHTS
# ============================================================

print("\n" + "=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{top_category['Churn_Category']} contributes the highest "
    f"revenue exposure of "
    f"₹{top_category['Revenue_Exposure']:,.2f}."
)

print(
    f"{top_reason['Churn_Reason']} is the churn reason associated "
    f"with the highest revenue exposure of "
    f"₹{top_reason['Revenue_Exposure']:,.2f}."
)

print(
    f"{highest_value_reason['Churn_Reason']} has the highest "
    f"revenue exposure per churned customer at "
    f"₹{highest_value_reason['Revenue_Per_Customer']:,.2f}."
)

print(
    f"{top_tenure_group['Tenure_Group']} represents the tenure group "
    f"with the largest churn-related revenue exposure."
)

print(
    f"{top_contract_payment['Contract']} contract with "
    f"{top_contract_payment['Payment_Method']} payment method "
    f"shows the highest revenue exposure among the analyzed "
    f"contract-payment combinations."
)

print(
    "Revenue exposure identifies where churn is financially most "
    "damaging, not simply where the number of churned customers is highest."
)

print(
    "High-value churn reasons should receive priority because "
    "each lost customer represents greater revenue risk."
)

print(
    "Early-tenure churn can indicate onboarding, pricing or "
    "expectation problems and should be addressed quickly."
)

print(
    "Contract and payment combinations with high revenue exposure "
    "can be targeted with customized retention offers."
)

print(
    "Business Action: Prioritize retention resources using revenue "
    "exposure, churn reason, customer value and tenure together."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 28
# ============================================================

# Which customer groups generate the strongest revenue,
# customer value and retention performance?

# Business Goal:
# Compare customers using available revenue, tenure,
# payment and status metrics to identify valuable groups.

# ============================================================

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go


# ============================================================
# DATA COPY
# ============================================================

analysis_df = df.copy()


# ============================================================
# REQUIRED COLUMNS
# ============================================================

required_cols = [
    "Customer_ID",
    "Customer_Status",
    "Total_Revenue",
    "Net_Revenue",
    "Monthly_Charge",
    "Tenure_in_Months"
]

missing_cols = [
    col for col in required_cols
    if col not in analysis_df.columns
]

if missing_cols:
    raise ValueError(
        f"Missing columns in dataframe: {missing_cols}"
    )


# ============================================================
# NUMERIC CLEANING
# ============================================================

numeric_cols = [
    "Total_Revenue",
    "Net_Revenue",
    "Monthly_Charge",
    "Tenure_in_Months"
]

for col in numeric_cols:

    analysis_df[col] = pd.to_numeric(
        analysis_df[col],
        errors="coerce"
    ).fillna(0)

    analysis_df[col] = (
        analysis_df[col]
        .clip(lower=0)
    )


# ============================================================
# STATUS CLEANING
# ============================================================

analysis_df["Customer_Status"] = (
    analysis_df["Customer_Status"]
    .astype(str)
    .str.strip()
    .replace({
        "": "Unknown",
        "nan": "Unknown",
        "None": "Unknown"
    })
)


# ============================================================
# CREATE CUSTOMER VALUE GROUPS
# ============================================================

customer_summary = (
    analysis_df
    .groupby(
        "Customer_ID",
        as_index=False
    )
    .agg(
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Monthly_Charge=("Monthly_Charge", "mean"),
        Tenure_in_Months=("Tenure_in_Months", "mean"),
        Customer_Status=(
            "Customer_Status",
            lambda x: (
                x.mode().iloc[0]
                if not x.mode().empty
                else "Unknown"
            )
        )
    )
)


# ============================================================
# CUSTOMER VALUE TIER
# ============================================================

customer_summary["Value_Tier"] = pd.qcut(
    customer_summary["Total_Revenue"],
    q=4,
    labels=[
        "Low Value",
        "Medium Value",
        "High Value",
        "Very High Value"
    ],
    duplicates="drop"
)


# ============================================================
# CUSTOMER LIFETIME VALUE METRICS
# ============================================================

customer_summary["Revenue_Per_Month"] = (
    customer_summary["Total_Revenue"]
    .div(
        customer_summary["Tenure_in_Months"]
        .replace(0, np.nan)
    )
    .fillna(0)
)


customer_summary["Net_Revenue_Rate"] = (
    customer_summary["Net_Revenue"]
    .div(
        customer_summary["Total_Revenue"]
        .replace(0, np.nan)
    )
    .mul(100)
    .fillna(0)
)


# ============================================================
# VALUE TIER SUMMARY
# ============================================================

tier_summary = (
    customer_summary
    .groupby(
        "Value_Tier",
        observed=True
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Revenue_Per_Month=("Revenue_Per_Month", "mean")
    )
    .reset_index()
)


# ============================================================
# REVENUE SHARE
# ============================================================

tier_summary["Revenue_Share"] = (
    tier_summary["Total_Revenue"]
    .div(
        tier_summary["Total_Revenue"].sum()
    )
    .mul(100)
    .fillna(0)
)


tier_summary["Net_Revenue_Rate"] = (
    tier_summary["Net_Revenue"]
    .div(
        tier_summary["Total_Revenue"]
        .replace(0, np.nan)
    )
    .mul(100)
    .fillna(0)
)


# ============================================================
# STATUS ANALYSIS
# ============================================================

status_lower = (
    customer_summary["Customer_Status"]
    .astype(str)
    .str.lower()
    .str.strip()
)


customer_summary["Is_Churned"] = (
    status_lower == "churned"
).astype(int)


customer_summary["Is_Stayed"] = (
    status_lower == "stayed"
).astype(int)


tier_status = (
    customer_summary
    .groupby(
        "Value_Tier",
        observed=True
    )
    .agg(
        Churned=("Is_Churned", "sum"),
        Stayed=("Is_Stayed", "sum"),
        Customers=("Customer_ID", "nunique")
    )
    .reset_index()
)


tier_status["Churn_Rate"] = (
    tier_status["Churned"]
    .div(
        tier_status["Customers"]
        .replace(0, np.nan)
    )
    .mul(100)
    .fillna(0)
)


tier_status["Retention_Rate"] = (
    tier_status["Stayed"]
    .div(
        tier_status["Customers"]
        .replace(0, np.nan)
    )
    .mul(100)
    .fillna(0)
)


tier_summary = tier_summary.merge(
    tier_status[
        [
            "Value_Tier",
            "Churned",
            "Stayed",
            "Churn_Rate",
            "Retention_Rate"
        ]
    ],
    on="Value_Tier",
    how="left"
)


# ============================================================
# DISPLAY SUMMARY
# ============================================================

display(
    tier_summary.round(2)
)


# ============================================================
# KPI
# ============================================================

top_revenue_tier = tier_summary.loc[
    tier_summary["Total_Revenue"].idxmax()
]


highest_value_tier = tier_summary.loc[
    tier_summary["Avg_Revenue"].idxmax()
]


best_retention_tier = tier_summary.loc[
    tier_summary["Retention_Rate"].idxmax()
]


highest_churn_tier = tier_summary.loc[
    tier_summary["Churn_Rate"].idxmax()
]


best_efficiency_tier = tier_summary.loc[
    tier_summary["Net_Revenue_Rate"].idxmax()
]


print("=" * 75)
print("CUSTOMER VALUE ANALYSIS")
print("=" * 75)

print(
    f"Top Revenue Tier        : "
    f"{top_revenue_tier['Value_Tier']}"
)

print(
    f"Total Revenue           : "
    f"₹{top_revenue_tier['Total_Revenue']:,.2f}"
)

print(
    f"Highest Avg Revenue     : "
    f"{highest_value_tier['Value_Tier']}"
)

print(
    f"Average Revenue         : "
    f"₹{highest_value_tier['Avg_Revenue']:,.2f}"
)

print(
    f"Best Retention Tier     : "
    f"{best_retention_tier['Value_Tier']}"
)

print(
    f"Retention Rate          : "
    f"{best_retention_tier['Retention_Rate']:.2f}%"
)

print(
    f"Highest Churn Tier      : "
    f"{highest_churn_tier['Value_Tier']}"
)

print(
    f"Churn Rate              : "
    f"{highest_churn_tier['Churn_Rate']:.2f}%"
)

print(
    f"Best Net Revenue Tier   : "
    f"{best_efficiency_tier['Value_Tier']}"
)

print(
    f"Net Revenue Rate        : "
    f"{best_efficiency_tier['Net_Revenue_Rate']:.2f}%"
)

print("=" * 75)


# ============================================================
# VISUAL 1 — REVENUE BY VALUE TIER
# ============================================================

fig = px.bar(
    tier_summary,
    x="Value_Tier",
    y="Total_Revenue",
    color="Revenue_Share",
    text="Revenue_Share",
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Total Revenue Contribution by Customer Value Tier",
    hover_data=[
        "Customers",
        "Net_Revenue",
        "Avg_Revenue",
        "Revenue_Share"
    ]
)


fig.update_traces(
    texttemplate="%{text:.2f}%",
    textposition="outside"
)


fig.update_layout(
    title_x=0.5,
    xaxis_title="Customer Value Tier",
    yaxis_title="Total Revenue",
    coloraxis_colorbar_title="Revenue Share (%)"
)

fig.show()


# ============================================================
# VISUAL 2 — REVENUE LOLLIPOP
# ============================================================

lollipop_df = (
    tier_summary
    .sort_values(
        "Avg_Revenue",
        ascending=True
    )
)


fig = go.Figure()


fig.add_trace(
    go.Scatter(
        x=lollipop_df["Avg_Revenue"],
        y=lollipop_df["Value_Tier"].astype(str),
        mode="lines",
        line=dict(width=3),
        hoverinfo="skip"
    )
)


fig.add_trace(
    go.Scatter(
        x=lollipop_df["Avg_Revenue"],
        y=lollipop_df["Value_Tier"].astype(str),
        mode="markers+text",
        marker=dict(size=16),
        text=[
            f"₹{x:,.0f}"
            for x in lollipop_df["Avg_Revenue"]
        ],
        textposition="middle right",
        customdata=lollipop_df[
            [
                "Customers",
                "Avg_Tenure",
                "Retention_Rate",
                "Churn_Rate"
            ]
        ],
        hovertemplate=(
            "<b>%{y}</b><br>"
            "Average Revenue: ₹%{x:,.2f}<br>"
            "Customers: %{customdata[0]:,}<br>"
            "Avg Tenure: %{customdata[1]:.1f} months<br>"
            "Retention: %{customdata[2]:.2f}%<br>"
            "Churn: %{customdata[3]:.2f}%"
            "<extra></extra>"
        )
    )
)


fig.update_layout(
    title="Average Customer Revenue — Lollipop",
    title_x=0.5,
    template="plotly_white",
    height=600,
    xaxis_title="Average Revenue",
    yaxis_title="Customer Value Tier"
)

fig.show()


# ============================================================
# VISUAL 3 — RETENTION VS CHURN
# ============================================================

fig = px.bar(
    tier_summary,
    x="Value_Tier",
    y=[
        "Retention_Rate",
        "Churn_Rate"
    ],
    barmode="group",
    text_auto=".2f",
    template="plotly_dark",
    title="Retention and Churn by Customer Value Tier"
)


fig.update_layout(
    title_x=0.5,
    xaxis_title="Customer Value Tier",
    yaxis_title="Rate (%)"
)

fig.show()


# ============================================================
# VISUAL 4 — TENURE VS CUSTOMER VALUE
# ============================================================

fig = px.scatter(
    customer_summary,
    x="Tenure_in_Months",
    y="Total_Revenue",
    size="Monthly_Charge",
    color="Value_Tier",
    symbol="Customer_Status",
    hover_name="Customer_ID",
    hover_data=[
        "Net_Revenue",
        "Revenue_Per_Month",
        "Monthly_Charge"
    ],
    template="plotly_white",
    title="Customer Tenure vs Lifetime Revenue",
    color_discrete_sequence=px.colors.qualitative.Set2
)


fig.update_layout(
    title_x=0.5,
    xaxis_title="Tenure (Months)",
    yaxis_title="Total Revenue"
)

fig.show()


# ============================================================
# VISUAL 5 — VALUE TIER TREEMAP
# ============================================================

fig = px.treemap(
    tier_summary,
    path=["Value_Tier"],
    values="Total_Revenue",
    color="Revenue_Share",
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Customer Value Revenue Concentration",
    hover_data=[
        "Customers",
        "Avg_Revenue",
        "Avg_Tenure",
        "Retention_Rate",
        "Churn_Rate"
    ]
)


fig.update_layout(
    title_x=0.5,
    height=700
)

fig.show()


# ============================================================
# VISUAL 6 — NET REVENUE EFFICIENCY
# ============================================================

efficiency_df = (
    tier_summary
    .sort_values(
        "Net_Revenue_Rate",
        ascending=True
    )
)


fig = px.bar(
    efficiency_df,
    x="Net_Revenue_Rate",
    y="Value_Tier",
    orientation="h",
    color="Avg_Revenue_Per_Month",
    text="Net_Revenue_Rate",
    color_continuous_scale="Viridis",
    template="plotly_dark",
    title="Net Revenue Efficiency by Customer Value Tier",
    hover_data=[
        "Total_Revenue",
        "Net_Revenue",
        "Avg_Revenue",
        "Avg_Tenure",
        "Revenue_Share"
    ]
)


fig.update_traces(
    texttemplate="%{text:.2f}%",
    textposition="outside"
)


fig.update_layout(
    title_x=0.5,
    xaxis_title="Net Revenue Efficiency (%)",
    yaxis_title="Customer Value Tier",
    coloraxis_colorbar_title="Avg Revenue / Month"
)

fig.show()


# ============================================================
# BUSINESS INSIGHTS
# ============================================================

print("\n" + "=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"{top_revenue_tier['Value_Tier']} contributes the highest "
    f"total revenue of "
    f"₹{top_revenue_tier['Total_Revenue']:,.2f}."
)

print(
    f"{highest_value_tier['Value_Tier']} has the highest average "
    f"customer revenue of "
    f"₹{highest_value_tier['Avg_Revenue']:,.2f}."
)

print(
    f"{best_retention_tier['Value_Tier']} has the highest retention "
    f"rate of "
    f"{best_retention_tier['Retention_Rate']:.2f}%."
)

print(
    f"{highest_churn_tier['Value_Tier']} has the highest churn rate "
    f"of "
    f"{highest_churn_tier['Churn_Rate']:.2f}%."
)

print(
    f"{best_efficiency_tier['Value_Tier']} has the strongest net "
    f"revenue efficiency at "
    f"{best_efficiency_tier['Net_Revenue_Rate']:.2f}%."
)

print(
    "High-value customers should receive stronger retention "
    "strategies because their revenue impact is higher."
)

print(
    "High-churn value tiers should be investigated for potential "
    "revenue leakage and customer retention risk."
)

print(
    "Long-tenure customers with strong revenue contribution represent "
    "important long-term customer relationships."
)

print(
    "Business Action: Prioritize retention according to customer "
    "value, revenue contribution and churn risk."
)

print("=" * 75)
````

````python
# ============================================================
# BUSINESS QUESTION 29
# ============================================================

# Which churn categories and reasons create the highest
# revenue exposure and customer retention risk?

# Business Goal:
# Identify the major causes of customer loss, measure their
# revenue impact and locate high-risk contract/internet groups.

# ============================================================

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go


# ============================================================
# DATA PREPARATION
# ============================================================

analysis_df = df.copy()


required_cols = [
    "Customer_ID",
    "Customer_Status",
    "Contract",
    "Internet_Type",
    "Churn_Category",
    "Churn_Reason",
    "Monthly_Charge",
    "Total_Revenue",
    "Tenure_in_Months"
]


missing_cols = [
    col for col in required_cols
    if col not in analysis_df.columns
]


if missing_cols:
    raise ValueError(
        f"Missing columns in dataframe: {missing_cols}"
    )


# ============================================================
# NUMERIC CLEANING
# ============================================================

numeric_cols = [
    "Monthly_Charge",
    "Total_Revenue",
    "Tenure_in_Months"
]


for col in numeric_cols:

    analysis_df[col] = pd.to_numeric(
        analysis_df[col],
        errors="coerce"
    ).fillna(0)

    analysis_df[col] = (
        analysis_df[col]
        .clip(lower=0)
    )


# ============================================================
# TEXT CLEANING
# ============================================================

text_cols = [
    "Customer_Status",
    "Contract",
    "Internet_Type",
    "Churn_Category",
    "Churn_Reason"
]


for col in text_cols:

    analysis_df[col] = (
        analysis_df[col]
        .astype(str)
        .str.strip()
        .replace({
            "": "Unknown",
            "nan": "Unknown",
            "None": "Unknown"
        })
    )


# ============================================================
# CHURN FLAG
# ============================================================

analysis_df["Is_Churned"] = (
    analysis_df["Customer_Status"]
    .str.lower()
    .eq("churned")
).astype(int)


analysis_df["Is_Stayed"] = (
    analysis_df["Customer_Status"]
    .str.lower()
    .eq("stayed")
).astype(int)


# ============================================================
# CHURN DATASET
# ============================================================

churn_df = analysis_df[
    analysis_df["Is_Churned"] == 1
].copy()


if churn_df.empty:

    raise ValueError(
        "No churned customers found in Customer_Status."
    )


# ============================================================
# CHURN CATEGORY SUMMARY
# ============================================================

category_summary = (
    churn_df
    .groupby(
        "Churn_Category",
        dropna=False
    )
    .agg(
        Churned_Customers=("Customer_ID", "nunique"),
        Revenue_Exposure=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean")
    )
    .reset_index()
)


category_summary["Revenue_Share"] = (
    category_summary["Revenue_Exposure"]
    .div(
        category_summary["Revenue_Exposure"].sum()
    )
    .mul(100)
    .fillna(0)
)


category_summary["Revenue_Per_Churned_Customer"] = (
    category_summary["Revenue_Exposure"]
    .div(
        category_summary["Churned_Customers"]
        .replace(0, np.nan)
    )
    .fillna(0)
)


# ============================================================
# CHURN REASON SUMMARY
# ============================================================

reason_summary = (
    churn_df
    .groupby(
        "Churn_Reason",
        dropna=False
    )
    .agg(
        Churned_Customers=("Customer_ID", "nunique"),
        Revenue_Exposure=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean")
    )
    .reset_index()
)


reason_summary["Revenue_Share"] = (
    reason_summary["Revenue_Exposure"]
    .div(
        reason_summary["Revenue_Exposure"].sum()
    )
    .mul(100)
    .fillna(0)
)


reason_summary["Revenue_Per_Customer"] = (
    reason_summary["Revenue_Exposure"]
    .div(
        reason_summary["Churned_Customers"]
        .replace(0, np.nan)
    )
    .fillna(0)
)


# ============================================================
# CONTRACT RISK SUMMARY
# ============================================================

contract_summary = (
    analysis_df
    .groupby(
        "Contract",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Churned=("Is_Churned", "sum"),
        Stayed=("Is_Stayed", "sum"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean")
    )
    .reset_index()
)


contract_summary["Churn_Rate"] = (
    contract_summary["Churned"]
    .div(
        contract_summary["Customers"]
        .replace(0, np.nan)
    )
    .mul(100)
    .fillna(0)
)


contract_summary["Retention_Rate"] = (
    contract_summary["Stayed"]
    .div(
        contract_summary["Customers"]
        .replace(0, np.nan)
    )
    .mul(100)
    .fillna(0)
)


contract_summary["Revenue_Per_Customer"] = (
    contract_summary["Total_Revenue"]
    .div(
        contract_summary["Customers"]
        .replace(0, np.nan)
    )
    .fillna(0)
)


# ============================================================
# INTERNET TYPE RISK SUMMARY
# ============================================================

internet_summary = (
    analysis_df
    .groupby(
        "Internet_Type",
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Churned=("Is_Churned", "sum"),
        Stayed=("Is_Stayed", "sum"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean")
    )
    .reset_index()
)


internet_summary["Churn_Rate"] = (
    internet_summary["Churned"]
    .div(
        internet_summary["Customers"]
        .replace(0, np.nan)
    )
    .mul(100)
    .fillna(0)
)


internet_summary["Retention_Rate"] = (
    internet_summary["Stayed"]
    .div(
        internet_summary["Customers"]
        .replace(0, np.nan)
    )
    .mul(100)
    .fillna(0)
)


# ============================================================
# CONTRACT + INTERNET RISK
# ============================================================

risk_matrix = (
    analysis_df
    .groupby(
        [
            "Contract",
            "Internet_Type"
        ],
        dropna=False
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Churned=("Is_Churned", "sum"),
        Total_Revenue=("Total_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean")
    )
    .reset_index()
)


risk_matrix["Churn_Rate"] = (
    risk_matrix["Churned"]
    .div(
        risk_matrix["Customers"]
        .replace(0, np.nan)
    )
    .mul(100)
    .fillna(0)
)


risk_matrix["Revenue_Per_Customer"] = (
    risk_matrix["Total_Revenue"]
    .div(
        risk_matrix["Customers"]
        .replace(0, np.nan)
    )
    .fillna(0)
)


# ============================================================
# SUMMARY TABLES
# ============================================================

print("CHURN CATEGORY ANALYSIS")

display(
    category_summary
    .sort_values(
        "Revenue_Exposure",
        ascending=False
    )
    .round(2)
)


print("CHURN REASON ANALYSIS")

display(
    reason_summary
    .sort_values(
        "Revenue_Exposure",
        ascending=False
    )
    .round(2)
)


print("CONTRACT RISK ANALYSIS")

display(
    contract_summary
    .sort_values(
        "Churn_Rate",
        ascending=False
    )
    .round(2)
)


print("INTERNET TYPE RISK ANALYSIS")

display(
    internet_summary
    .sort_values(
        "Churn_Rate",
        ascending=False
    )
    .round(2)
)


print("CONTRACT + INTERNET RISK")

display(
    risk_matrix
    .sort_values(
        "Churn_Rate",
        ascending=False
    )
    .round(2)
)


# ============================================================
# KPI
# ============================================================

top_category = category_summary.loc[
    category_summary["Revenue_Exposure"].idxmax()
]


top_reason = reason_summary.loc[
    reason_summary["Revenue_Exposure"].idxmax()
]


highest_contract_risk = contract_summary.loc[
    contract_summary["Churn_Rate"].idxmax()
]


highest_internet_risk = internet_summary.loc[
    internet_summary["Churn_Rate"].idxmax()
]


highest_combination_risk = risk_matrix.loc[
    risk_matrix["Churn_Rate"].idxmax()
]


total_churned = (
    churn_df["Customer_ID"]
    .nunique()
)


total_revenue_exposure = (
    churn_df["Total_Revenue"]
    .sum()
)


overall_churn_rate = (
    total_churned
    / analysis_df["Customer_ID"].nunique()
    * 100
)


print("=" * 80)
print("CHURN REVENUE RISK KPI")
print("=" * 80)

print(
    f"Total Churned Customers     : "
    f"{total_churned:,}"
)

print(
    f"Overall Churn Rate          : "
    f"{overall_churn_rate:.2f}%"
)

print(
    f"Revenue Exposure            : "
    f"₹{total_revenue_exposure:,.2f}"
)

print(
    f"Top Churn Category          : "
    f"{top_category['Churn_Category']}"
)

print(
    f"Category Revenue Exposure   : "
    f"₹{top_category['Revenue_Exposure']:,.2f}"
)

print(
    f"Top Churn Reason            : "
    f"{top_reason['Churn_Reason']}"
)

print(
    f"Reason Revenue Exposure     : "
    f"₹{top_reason['Revenue_Exposure']:,.2f}"
)

print(
    f"Highest Contract Risk       : "
    f"{highest_contract_risk['Contract']}"
)

print(
    f"Contract Churn Rate         : "
    f"{highest_contract_risk['Churn_Rate']:.2f}%"
)

print(
    f"Highest Internet Risk       : "
    f"{highest_internet_risk['Internet_Type']}"
)

print(
    f"Internet Churn Rate         : "
    f"{highest_internet_risk['Churn_Rate']:.2f}%"
)

print(
    f"Highest Risk Combination    : "
    f"{highest_combination_risk['Contract']} + "
    f"{highest_combination_risk['Internet_Type']}"
)

print(
    f"Combination Churn Rate      : "
    f"{highest_combination_risk['Churn_Rate']:.2f}%"
)

print("=" * 80)


# ============================================================
# VISUAL 1 — CHURN CATEGORY REVENUE LOLLIPOP
# ============================================================

lollipop_category = (
    category_summary
    .sort_values(
        "Revenue_Exposure",
        ascending=True
    )
)


fig = go.Figure()


fig.add_trace(
    go.Scatter(
        x=lollipop_category["Revenue_Exposure"],
        y=lollipop_category["Churn_Category"],
        mode="lines",
        line=dict(width=3),
        hoverinfo="skip"
    )
)


fig.add_trace(
    go.Scatter(
        x=lollipop_category["Revenue_Exposure"],
        y=lollipop_category["Churn_Category"],
        mode="markers+text",
        marker=dict(size=16),
        text=[
            f"₹{x:,.0f}"
            for x in lollipop_category["Revenue_Exposure"]
        ],
        textposition="middle right",
        customdata=lollipop_category[
            [
                "Churned_Customers",
                "Revenue_Share",
                "Revenue_Per_Churned_Customer"
            ]
        ],
        hovertemplate=(
            "<b>%{y}</b><br>"
            "Revenue Exposure: ₹%{x:,.2f}<br>"
            "Churned Customers: %{customdata[0]:,}<br>"
            "Revenue Share: %{customdata[1]:.2f}%<br>"
            "Revenue / Churned Customer: "
            "₹%{customdata[2]:,.2f}"
            "<extra></extra>"
        )
    )
)


fig.update_layout(
    title="Revenue Exposure by Churn Category — Lollipop",
    title_x=0.5,
    template="plotly_white",
    height=600,
    xaxis_title="Revenue Exposure",
    yaxis_title="Churn Category"
)

fig.show()


# ============================================================
# VISUAL 2 — CHURN REASON TREEMAP
# ============================================================

fig = px.treemap(
    reason_summary,
    path=["Churn_Reason"],
    values="Revenue_Exposure",
    color="Churned_Customers",
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Revenue Exposure by Churn Reason",
    hover_data=[
        "Churned_Customers",
        "Avg_Revenue",
        "Avg_Monthly_Charge",
        "Avg_Tenure",
        "Revenue_Share"
    ]
)


fig.update_layout(
    title_x=0.5,
    height=700
)

fig.show()


# ============================================================
# VISUAL 3 — CONTRACT CHURN RISK
# ============================================================

fig = px.bar(
    contract_summary.sort_values(
        "Churn_Rate",
        ascending=True
    ),
    x="Churn_Rate",
    y="Contract",
    orientation="h",
    color="Revenue_Per_Customer",
    text="Churn_Rate",
    color_continuous_scale="Plasma",
    template="plotly_dark",
    title="Contract Type Churn Risk",
    hover_data=[
        "Customers",
        "Churned",
        "Stayed",
        "Retention_Rate",
        "Total_Revenue",
        "Revenue_Per_Customer"
    ]
)


fig.update_traces(
    texttemplate="%{text:.2f}%",
    textposition="outside"
)


fig.update_layout(
    title_x=0.5,
    xaxis_title="Churn Rate (%)",
    yaxis_title="Contract",
    coloraxis_colorbar_title="Revenue / Customer"
)

fig.show()


# ============================================================
# VISUAL 4 — CONTRACT + INTERNET HEATMAP
# ============================================================

heatmap_data = risk_matrix.pivot_table(
    index="Contract",
    columns="Internet_Type",
    values="Churn_Rate",
    aggfunc="mean"
)


fig = px.imshow(
    heatmap_data,
    text_auto=".2f",
    aspect="auto",
    color_continuous_scale="Turbo",
    template="plotly_white",
    title="Contract × Internet Type Churn Risk Heatmap"
)


fig.update_layout(
    title_x=0.5,
    height=600,
    xaxis_title="Internet Type",
    yaxis_title="Contract",
    coloraxis_colorbar_title="Churn Rate (%)"
)

fig.show()


# ============================================================
# VISUAL 5 — REVENUE VS TENURE OF CHURNED CUSTOMERS
# ============================================================

scatter_churn = churn_df.copy()


scatter_churn["Plot_Monthly_Charge"] = (
    scatter_churn["Monthly_Charge"]
    .clip(lower=1)
)


fig = px.scatter(
    scatter_churn,
    x="Tenure_in_Months",
    y="Total_Revenue",
    size="Plot_Monthly_Charge",
    color="Churn_Category",
    symbol="Internet_Type",
    hover_name="Customer_ID",
    hover_data=[
        "Churn_Reason",
        "Contract",
        "Monthly_Charge",
        "Net_Revenue"
    ],
    color_discrete_sequence=px.colors.qualitative.Set2,
    template="plotly_white",
    title="Churned Customer Revenue Exposure by Tenure"
)


fig.update_layout(
    title_x=0.5,
    xaxis_title="Tenure (Months)",
    yaxis_title="Total Revenue"
)

fig.show()


# ============================================================
# VISUAL 6 — REVENUE EXPOSURE BY CHURN REASON
# ============================================================

reason_plot = (
    reason_summary
    .sort_values(
        "Revenue_Exposure",
        ascending=False
    )
    .head(12)
    .copy()
)

fig = px.bar(
    reason_plot,
    x="Churn_Reason",
    y="Revenue_Exposure",
    color="Avg_Monthly_Charge",
    text="Revenue_Exposure",
    color_continuous_scale="Viridis",
    template="plotly_white",
    title="Top Churn Reasons by Revenue Exposure",
    hover_data=[
        "Churned_Customers",
        "Avg_Revenue",
        "Avg_Tenure",
        "Revenue_Per_Customer",
        "Revenue_Share",
        "Avg_Monthly_Charge"
    ]
)

fig.update_traces(
    texttemplate="₹%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    height=650,
    xaxis_title="Churn Reason",
    yaxis_title="Revenue Exposure",
    coloraxis_colorbar=dict(
        title="Avg Monthly Charge"
    ),
    xaxis=dict(
        tickangle=-35
    ),
    margin=dict(
        l=40,
        r=40,
        t=90,
        b=160
    )
)

fig.show()

# ============================================================
# BUSINESS INSIGHTS
# ============================================================

print("\n" + "=" * 80)
print("BUSINESS INSIGHTS")
print("=" * 80)

print(
    f"{top_category['Churn_Category']} creates the highest "
    f"revenue exposure of "
    f"₹{top_category['Revenue_Exposure']:,.2f}."
)

print(
    f"{top_reason['Churn_Reason']} is the leading churn reason "
    f"by revenue exposure with "
    f"₹{top_reason['Revenue_Exposure']:,.2f}."
)

print(
    f"{highest_contract_risk['Contract']} has the highest contract "
    f"churn rate at "
    f"{highest_contract_risk['Churn_Rate']:.2f}%."
)

print(
    f"{highest_internet_risk['Internet_Type']} has the highest "
    f"internet-type churn rate at "
    f"{highest_internet_risk['Churn_Rate']:.2f}%."
)

print(
    f"The highest-risk combination is "
    f"{highest_combination_risk['Contract']} + "
    f"{highest_combination_risk['Internet_Type']} "
    f"with a churn rate of "
    f"{highest_combination_risk['Churn_Rate']:.2f}%."
)

print(
    "Revenue exposure is more important than churn volume alone "
    "because losing high-value customers creates greater financial impact."
)

print(
    "High-revenue churn reasons should receive priority in retention "
    "programs and customer recovery campaigns."
)

print(
    "Contract and internet combinations with elevated churn should "
    "be investigated for pricing, service quality and customer-fit issues."
)

print(
    "High-value churned customers with longer tenure represent "
    "important revenue recovery opportunities."
)

print(
    "Business Action: Prioritize retention strategies using churn "
    "reason, revenue exposure, contract risk and internet-service risk."
)

print("=" * 80)
````

````python
# ============================================================
# BUSINESS QUESTION 30
# ============================================================

# Which customers generate the highest lifetime value,
# and how concentrated is the company's revenue among
# high-value customers?

# Business Goal:
# Identify high-value customers, understand lifetime value
# distribution, and measure revenue concentration for
# retention and customer value strategies.

# ============================================================

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go


# ============================================================
# DATA PREPARATION
# ============================================================

analysis_df = df.copy()

required_cols = [
    "Customer_ID",
    "Total_Revenue",
    "Net_Revenue",
    "Tenure_in_Months",
    "Monthly_Charge",
    "Customer_Status"
]

missing_cols = [
    col for col in required_cols
    if col not in analysis_df.columns
]

if missing_cols:
    raise ValueError(
        f"Missing columns in dataframe: {missing_cols}"
    )


numeric_cols = [
    "Total_Revenue",
    "Net_Revenue",
    "Tenure_in_Months",
    "Monthly_Charge"
]

for col in numeric_cols:
    analysis_df[col] = pd.to_numeric(
        analysis_df[col],
        errors="coerce"
    ).fillna(0)

    analysis_df[col] = (
        analysis_df[col]
        .clip(lower=0)
    )


analysis_df["Customer_Status"] = (
    analysis_df["Customer_Status"]
    .astype(str)
    .str.strip()
)


# ============================================================
# CUSTOMER VALUE DATASET
# ============================================================

customer_value = (
    analysis_df
    .groupby(
        "Customer_ID",
        as_index=False
    )
    .agg(
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Tenure_in_Months=("Tenure_in_Months", "mean"),
        Monthly_Charge=("Monthly_Charge", "mean"),
        Customer_Status=(
            "Customer_Status",
            lambda x: x.mode().iloc[0]
            if not x.mode().empty
            else "Unknown"
        )
    )
)


customer_value["Revenue_Per_Month"] = (
    customer_value["Total_Revenue"]
    .div(
        customer_value["Tenure_in_Months"]
        .replace(0, np.nan)
    )
    .fillna(0)
)


customer_value["Net_Revenue_Rate"] = (
    customer_value["Net_Revenue"]
    .div(
        customer_value["Total_Revenue"]
        .replace(0, np.nan)
    )
    .mul(100)
    .fillna(0)
)


# ============================================================
# VALUE RANK
# ============================================================

customer_value = customer_value.sort_values(
    "Total_Revenue",
    ascending=False
).reset_index(drop=True)


customer_value["Revenue_Rank"] = (
    customer_value.index + 1
)


customer_value["Revenue_Share"] = (
    customer_value["Total_Revenue"]
    .div(
        customer_value["Total_Revenue"].sum()
    )
    .mul(100)
)


customer_value["Cumulative_Revenue_Share"] = (
    customer_value["Revenue_Share"]
    .cumsum()
)


customer_value["Customer_Share"] = (
    customer_value["Revenue_Rank"]
    .div(len(customer_value))
    .mul(100)
)


# ============================================================
# VALUE TIERS
# ============================================================

customer_value["Value_Tier"] = pd.qcut(
    customer_value["Total_Revenue"],
    q=4,
    labels=[
        "Low Value",
        "Medium Value",
        "High Value",
        "Very High Value"
    ],
    duplicates="drop"
)


tier_summary = (
    customer_value
    .groupby(
        "Value_Tier",
        observed=True
    )
    .agg(
        Customers=("Customer_ID", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Net_Revenue=("Net_Revenue", "sum"),
        Avg_Revenue=("Total_Revenue", "mean"),
        Avg_Tenure=("Tenure_in_Months", "mean"),
        Avg_Monthly_Charge=("Monthly_Charge", "mean"),
        Avg_Revenue_Per_Month=("Revenue_Per_Month", "mean")
    )
    .reset_index()
)


tier_summary["Revenue_Share"] = (
    tier_summary["Total_Revenue"]
    .div(
        tier_summary["Total_Revenue"].sum()
    )
    .mul(100)
)


display(
    tier_summary.round(2)
)


# ============================================================
# TOP CUSTOMER SUMMARY
# ============================================================

top_10 = (
    customer_value
    .head(10)
    .copy()
)


top_20 = (
    customer_value
    .head(20)
    .copy()
)


top_10_revenue_share = (
    top_10["Total_Revenue"].sum()
    / customer_value["Total_Revenue"].sum()
    * 100
)


top_20_revenue_share = (
    top_20["Total_Revenue"].sum()
    / customer_value["Total_Revenue"].sum()
    * 100
)


# ============================================================
# KPI
# ============================================================

highest_value_customer = (
    customer_value.iloc[0]
)


highest_tenure_customer = customer_value.loc[
    customer_value["Tenure_in_Months"].idxmax()
]


highest_monthly_value = customer_value.loc[
    customer_value["Revenue_Per_Month"].idxmax()
]


highest_net_value = customer_value.loc[
    customer_value["Net_Revenue"].idxmax()
]


highest_value_tier = tier_summary.loc[
    tier_summary["Total_Revenue"].idxmax()
]


print("=" * 75)
print("CUSTOMER LIFETIME VALUE KPI")
print("=" * 75)

print(
    f"Highest Value Customer     : "
    f"{highest_value_customer['Customer_ID']}"
)

print(
    f"Customer Revenue           : "
    f"₹{highest_value_customer['Total_Revenue']:,.2f}"
)

print(
    f"Highest Revenue / Month    : "
    f"₹{highest_monthly_value['Revenue_Per_Month']:,.2f}"
)

print(
    f"Customer                   : "
    f"{highest_monthly_value['Customer_ID']}"
)

print(
    f"Highest Net Revenue        : "
    f"₹{highest_net_value['Net_Revenue']:,.2f}"
)

print(
    f"Top 10 Revenue Share       : "
    f"{top_10_revenue_share:.2f}%"
)

print(
    f"Top 20 Revenue Share       : "
    f"{top_20_revenue_share:.2f}%"
)

print(
    f"Highest Revenue Tier       : "
    f"{highest_value_tier['Value_Tier']}"
)

print(
    f"Tier Revenue               : "
    f"₹{highest_value_tier['Total_Revenue']:,.2f}"
)

print("=" * 75)


# ============================================================
# VISUAL 1 — LOLLIPOP TOP CUSTOMERS
# ============================================================

lollipop_df = (
    customer_value
    .head(15)
    .sort_values(
        "Total_Revenue",
        ascending=True
    )
    .copy()
)


fig = go.Figure()


fig.add_trace(
    go.Scatter(
        x=lollipop_df["Total_Revenue"],
        y=lollipop_df["Customer_ID"].astype(str),
        mode="lines",
        line=dict(width=3),
        hoverinfo="skip"
    )
)


fig.add_trace(
    go.Scatter(
        x=lollipop_df["Total_Revenue"],
        y=lollipop_df["Customer_ID"].astype(str),
        mode="markers+text",
        marker=dict(size=14),
        text=[
            f"₹{x:,.0f}"
            for x in lollipop_df["Total_Revenue"]
        ],
        textposition="middle right",
        customdata=lollipop_df[
            [
                "Tenure_in_Months",
                "Monthly_Charge",
                "Net_Revenue",
                "Customer_Status"
            ]
        ],
        hovertemplate=(
            "<b>Customer %{y}</b><br>"
            "Total Revenue: ₹%{x:,.2f}<br>"
            "Tenure: %{customdata[0]:.1f} Months<br>"
            "Monthly Charge: ₹%{customdata[1]:,.2f}<br>"
            "Net Revenue: ₹%{customdata[2]:,.2f}<br>"
            "Status: %{customdata[3]}"
            "<extra></extra>"
        )
    )
)


fig.update_layout(
    title="Top 15 Customers by Lifetime Revenue — Lollipop",
    title_x=0.5,
    template="plotly_white",
    height=650,
    xaxis_title="Total Revenue",
    yaxis_title="Customer ID"
)

fig.show()


# ============================================================
# VISUAL 2 — PARETO REVENUE CONCENTRATION
# ============================================================

pareto_df = customer_value.copy()


fig = go.Figure()


fig.add_trace(
    go.Bar(
        x=pareto_df["Revenue_Rank"],
        y=pareto_df["Total_Revenue"],
        name="Customer Revenue",
        hovertemplate=(
            "Customer Rank: %{x}<br>"
            "Revenue: ₹%{y:,.2f}"
            "<extra></extra>"
        )
    )
)


fig.add_trace(
    go.Scatter(
        x=pareto_df["Revenue_Rank"],
        y=pareto_df["Cumulative_Revenue_Share"],
        name="Cumulative Revenue %",
        yaxis="y2",
        mode="lines",
        hovertemplate=(
            "Customer Rank: %{x}<br>"
            "Cumulative Revenue: %{y:.2f}%"
            "<extra></extra>"
        )
    )
)


fig.update_layout(
    title="Revenue Concentration — Pareto Analysis",
    title_x=0.5,
    template="plotly_white",
    height=700,
    xaxis_title="Customer Revenue Rank",
    yaxis_title="Individual Customer Revenue",
    yaxis2=dict(
        title="Cumulative Revenue Share (%)",
        overlaying="y",
        side="right",
        range=[0, 100]
    ),
    hovermode="x unified"
)

fig.show()


# ============================================================
# VISUAL 3 — REVENUE VS TENURE
# ============================================================

fig = px.scatter(
    customer_value,
    x="Tenure_in_Months",
    y="Total_Revenue",
    size="Monthly_Charge",
    color="Value_Tier",
    symbol="Customer_Status",
    hover_name="Customer_ID",
    hover_data=[
        "Net_Revenue",
        "Revenue_Per_Month",
        "Monthly_Charge",
        "Revenue_Share"
    ],
    template="plotly_white",
    title="Customer Lifetime Revenue vs Tenure",
    color_discrete_sequence=px.colors.qualitative.Set2
)


fig.update_layout(
    title_x=0.5,
    xaxis_title="Tenure (Months)",
    yaxis_title="Total Revenue",
    legend_title="Customer Value Tier"
)

fig.show()


# ============================================================
# VISUAL 4 — VALUE TIER DISTRIBUTION
# ============================================================

fig = px.box(
    customer_value,
    x="Value_Tier",
    y="Total_Revenue",
    color="Value_Tier",
    points="outliers",
    hover_data=[
        "Customer_ID",
        "Tenure_in_Months",
        "Monthly_Charge",
        "Customer_Status"
    ],
    template="plotly_dark",
    title="Revenue Distribution Across Customer Value Tiers"
)


fig.update_layout(
    title_x=0.5,
    xaxis_title="Customer Value Tier",
    yaxis_title="Total Revenue",
    showlegend=False
)

fig.show()


# ============================================================
# VISUAL 5 — REVENUE DISTRIBUTION
# ============================================================

fig = px.histogram(
    customer_value,
    x="Total_Revenue",
    color="Customer_Status",
    nbins=40,
    marginal="box",
    hover_data=[
        "Tenure_in_Months",
        "Monthly_Charge",
        "Net_Revenue"
    ],
    template="plotly_white",
    title="Customer Lifetime Revenue Distribution"
)


fig.update_layout(
    title_x=0.5,
    xaxis_title="Total Revenue",
    yaxis_title="Number of Customers",
    bargap=0.05
)

fig.show()


# ============================================================
# VISUAL 6 — VALUE TIER REVENUE SHARE
# ============================================================

funnel_df = (
    tier_summary
    .sort_values(
        "Total_Revenue",
        ascending=False
    )
    .copy()
)

fig = px.funnel(
    funnel_df,
    y="Value_Tier",
    x="Total_Revenue",
    text="Total_Revenue",
    template="plotly_white",
    title="Customer Value Tier Revenue Contribution",
    hover_data=[
        "Customers",
        "Revenue_Share",
        "Avg_Revenue",
        "Net_Revenue",
        "Avg_Tenure",
        "Avg_Monthly_Charge"
    ]
)

fig.update_traces(
    texttemplate="₹%{x:,.0f}",
    textposition="inside"
)

fig.update_layout(
    title_x=0.5,
    height=650,
    xaxis_title="Total Revenue",
    yaxis_title="Customer Value Tier",
    margin=dict(
        l=40,
        r=40,
        t=90,
        b=50
    )
)

fig.show()

# ============================================================
# BUSINESS INSIGHTS
# ============================================================

print("\n" + "=" * 75)
print("BUSINESS INSIGHTS")
print("=" * 75)

print(
    f"Customer {highest_value_customer['Customer_ID']} has the "
    f"highest lifetime revenue at "
    f"₹{highest_value_customer['Total_Revenue']:,.2f}."
)

print(
    f"The top 10 customers contribute "
    f"{top_10_revenue_share:.2f}% of total revenue."
)

print(
    f"The top 20 customers contribute "
    f"{top_20_revenue_share:.2f}% of total revenue."
)

print(
    f"The {highest_value_tier['Value_Tier']} tier contributes the "
    f"largest revenue amount of "
    f"₹{highest_value_tier['Total_Revenue']:,.2f}."
)

print(
    f"Customer {highest_monthly_value['Customer_ID']} has the "
    f"highest revenue generated per tenure month at "
    f"₹{highest_monthly_value['Revenue_Per_Month']:,.2f}."
)

print(
    "Revenue concentration shows whether the business depends "
    "heavily on a small group of high-value customers."
)

print(
    "High-value customers should receive stronger retention "
    "and relationship-management strategies."
)

print(
    "Long-tenure customers with high revenue represent valuable "
    "lifetime relationships and should be protected from churn."
)

print(
    "A highly concentrated revenue base creates financial risk "
    "if a small number of major customers leave."
)

print(
    "Business Action: Prioritize retention based on customer "
    "lifetime value rather than customer count alone."
)

print("=" * 75)
````

# SQL Business Analysis — Customer Churn & Retention Analytics

## Business Objective

The objective of this SQL analysis is to transform customer, subscription, transaction, and support data into **actionable business intelligence**.

The analysis focuses on understanding **customer behavior, churn drivers, revenue performance, retention opportunities, and financial risk** so that business stakeholders can make data-driven decisions.

The key business question is:

> **Who is churning, why are they churning, how much revenue is at risk, where is the problem concentrated, and which customers should the business prioritize for retention?**

### 1. Customer Health

* How large is the current customer base?
* How many customers are retained versus churned?

* Which customer groups have the highest churn?

* Which customer segments contribute the most revenue?

* Which customer segments are most vulnerable?

````python
# Q1 — Business Question:
# What is the total size of the customer base?

pd.read_sql_query("""
SELECT
    COUNT(DISTINCT customer_id) AS total_customers
FROM customer_data;
""", conn)
````

````python
# Q2 — Business Question:
# How many customers have left the business?

pd.read_sql_query("""
SELECT
    COUNT(DISTINCT customer_id) AS churned_customers
FROM customer_data
WHERE customer_status = 'Churned';
""", conn)
````

````python
# Q3 — Business Question:
# How many customers are still retained?

pd.read_sql_query("""
SELECT
    COUNT(DISTINCT customer_id) AS retained_customers
FROM customer_data
WHERE customer_status = 'Stayed';
""", conn)
````

````python
# Q4 — Business Question:
# What percentage of customers have churned?

pd.read_sql_query("""
SELECT
    ROUND(
        100.0 *
        SUM(CASE WHEN customer_status = 'Churned' THEN 1 ELSE 0 END)
        / COUNT(*),
        2
    ) AS churn_rate
FROM customer_data;
""", conn)
````

````python
# Q5 — Business Question:
# What percentage of customers have been retained?

pd.read_sql_query("""
SELECT
    ROUND(
        100.0 *
        SUM(CASE WHEN customer_status = 'Stayed' THEN 1 ELSE 0 END)
        / COUNT(*),
        2
    ) AS retention_rate
FROM customer_data;
""", conn)
````

````python
# Q6 — Business Question:
# How are customers distributed between churned and retained groups?

pd.read_sql_query("""
SELECT
    customer_status,
    COUNT(*) AS customers
FROM customer_data
GROUP BY customer_status
ORDER BY customers DESC;
""", conn)
````

````python
# Which customer lifecycle groups have the highest churn rate?

pd.read_sql_query("""
SELECT
    customer_lifecycle,
    COUNT(DISTINCT customer_id) AS customers,
    SUM(
        CASE
            WHEN customer_status = 'Churned' THEN 1
            ELSE 0
        END
    ) AS churned_customers,
    ROUND(
        100.0 * SUM(
            CASE
                WHEN customer_status = 'Churned' THEN 1
                ELSE 0
            END
        ) / COUNT(DISTINCT customer_id),
        2
    ) AS churn_rate
FROM customer_data
GROUP BY customer_lifecycle
ORDER BY churn_rate DESC;
""", conn)
````

````python
# Q32 — Business Question:

# Which multiple-line groups generate the highest revenue?

pd.read_sql_query("""
SELECT
    multiple_lines,
    COUNT(DISTINCT customer_id) AS customers,
    ROUND(SUM(total_revenue), 2) AS total_revenue,
    ROUND(AVG(total_revenue), 2) AS avg_total_revenue
FROM customer_data
GROUP BY multiple_lines
ORDER BY total_revenue DESC;
""", conn)
````

````python
# Q34 — Business Question:

# Which customer groups have the highest extra data charges?

pd.read_sql_query("""
SELECT
    customer_lifecycle,
    COUNT(DISTINCT customer_id) AS customers,
    ROUND(SUM(total_extra_data_charges), 2) AS total_extra_data_charges,
    ROUND(AVG(total_extra_data_charges), 2) AS avg_extra_data_charges
FROM customer_data
GROUP BY customer_lifecycle
ORDER BY total_extra_data_charges DESC;
""", conn)
````

````python
# Q35 — Business Question:

# Which customer groups have the highest total extra charges?

pd.read_sql_query("""
SELECT
    revenue_segment,
    COUNT(DISTINCT customer_id) AS customers,
    ROUND(SUM(total_extra_charges), 2) AS total_extra_charges,
    ROUND(AVG(total_extra_charges), 2) AS avg_extra_charges
FROM customer_data
GROUP BY revenue_segment
ORDER BY total_extra_charges DESC;
""", conn)
````

````python
# Q36 — Business Question:

# Which revenue segments generate the highest net revenue?

pd.read_sql_query("""
SELECT
    revenue_segment,
    COUNT(DISTINCT customer_id) AS customers,
    ROUND(SUM(net_revenue), 2) AS total_net_revenue,
    ROUND(AVG(net_revenue), 2) AS avg_net_revenue
FROM customer_data
GROUP BY revenue_segment
ORDER BY total_net_revenue DESC;
""", conn)
````

````python
# Q37 — Business Question:

# Which customer groups have the highest average charge difference?

pd.read_sql_query("""
SELECT
    tenure_group,
    COUNT(DISTINCT customer_id) AS customers,
    ROUND(AVG(charge_difference), 2) AS avg_charge_difference
FROM customer_data
GROUP BY tenure_group
ORDER BY avg_charge_difference DESC;
""", conn)
````

````python
# Q38 — Business Question:

# Which customer groups have the highest extra-charge ratio?

pd.read_sql_query("""
SELECT
    revenue_segment,
    COUNT(DISTINCT customer_id) AS customers,
    ROUND(AVG(extra_charge_ratio), 2) AS avg_extra_charge_ratio
FROM customer_data
GROUP BY revenue_segment
ORDER BY avg_extra_charge_ratio DESC;
""", conn)
````

````python
# Q39 — Business Question:

# Which customer groups have the highest charge-to-revenue ratio?

pd.read_sql_query("""
SELECT
    customer_lifecycle,
    COUNT(DISTINCT customer_id) AS customers,
    ROUND(AVG(charge_revenue_ratio), 2) AS avg_charge_revenue_ratio
FROM customer_data
GROUP BY customer_lifecycle
ORDER BY avg_charge_revenue_ratio DESC;
""", conn)
````

````python
# Q40 — Business Question:

# Which customer groups have the highest revenue-to-charge ratio?

pd.read_sql_query("""
SELECT
    customer_lifecycle,
    COUNT(DISTINCT customer_id) AS customers,
    ROUND(AVG(revenue_to_charge_ratio), 2) AS avg_revenue_to_charge_ratio
FROM customer_data
GROUP BY customer_lifecycle
ORDER BY avg_revenue_to_charge_ratio DESC;
""", conn)
````

````python
# Q41 — Business Question:

# Which states have above-average revenue but below-average customer retention?

pd.read_sql_query("""
WITH state_metrics AS (
    SELECT
        state,
        COUNT(DISTINCT customer_id) AS customers,
        ROUND(SUM(total_revenue), 2) AS total_revenue,
        ROUND(
            100.0 * SUM(
                CASE
                    WHEN customer_status != 'Churned' THEN 1
                    ELSE 0
                END
            ) / COUNT(DISTINCT customer_id),
            2
        ) AS retention_rate
    FROM customer_data
    GROUP BY state
),
benchmarks AS (
    SELECT
        AVG(total_revenue) AS avg_state_revenue,
        AVG(retention_rate) AS avg_retention_rate
    FROM state_metrics
)
SELECT
    s.state,
    s.customers,
    s.total_revenue,
    s.retention_rate
FROM state_metrics s
CROSS JOIN benchmarks b
WHERE s.total_revenue > b.avg_state_revenue
  AND s.retention_rate < b.avg_retention_rate
ORDER BY s.total_revenue DESC;
""", conn)
````

````python
# Q42 — Business Question:

# Which customer segments create the highest revenue exposure from churn?

pd.read_sql_query("""
SELECT
    revenue_segment,
    COUNT(DISTINCT customer_id) AS churned_customers,
    ROUND(SUM(total_revenue), 2) AS revenue_exposure,
    ROUND(AVG(total_revenue), 2) AS avg_churned_customer_revenue
FROM customer_data
WHERE customer_status = 'Churned'
GROUP BY revenue_segment
ORDER BY revenue_exposure DESC;
""", conn)
````

````python
# Q43 — Business Question:

# Which churn reasons have the highest financial impact?

pd.read_sql_query("""
SELECT
    churn_reason,
    COUNT(DISTINCT customer_id) AS churned_customers,
    ROUND(SUM(total_revenue), 2) AS revenue_exposure,
    ROUND(AVG(total_revenue), 2) AS avg_revenue_per_churned_customer
FROM customer_data
WHERE customer_status = 'Churned'
  AND churn_reason IS NOT NULL
GROUP BY churn_reason
ORDER BY revenue_exposure DESC;
""", conn)
````

````python
# Q44 — Business Question:

# Which contract and internet combinations have the highest churn risk?

pd.read_sql_query("""
SELECT
    contract,
    internet_type,
    COUNT(DISTINCT customer_id) AS customers,
    SUM(
        CASE
            WHEN customer_status = 'Churned' THEN 1
            ELSE 0
        END
    ) AS churned_customers,
    ROUND(
        100.0 * SUM(
            CASE
                WHEN customer_status = 'Churned' THEN 1
                ELSE 0
            END
        ) / COUNT(DISTINCT customer_id),
        2
    ) AS churn_rate
FROM customer_data
GROUP BY contract, internet_type
HAVING COUNT(DISTINCT customer_id) >= 10
ORDER BY churn_rate DESC;
""", conn)
````

````python
# Q45 — Business Question:

# Which customer lifecycle groups combine high churn with high revenue exposure?

pd.read_sql_query("""
WITH lifecycle_metrics AS (
    SELECT
        customer_lifecycle,
        COUNT(DISTINCT customer_id) AS customers,
        SUM(
            CASE
                WHEN customer_status = 'Churned' THEN 1
                ELSE 0
            END
        ) AS churned_customers,
        SUM(
            CASE
                WHEN customer_status = 'Churned'
                THEN total_revenue
                ELSE 0
            END
        ) AS revenue_exposure
    FROM customer_data
    GROUP BY customer_lifecycle
)
SELECT
    customer_lifecycle,
    customers,
    churned_customers,
    ROUND(
        100.0 * churned_customers / customers,
        2
    ) AS churn_rate,
    ROUND(revenue_exposure, 2) AS revenue_exposure
FROM lifecycle_metrics
ORDER BY revenue_exposure DESC, churn_rate DESC;
""", conn)
````

````python
# Q46 — Business Question:

# Who are the top 20 high-value churned customers based on revenue?

pd.read_sql_query("""
SELECT
    customer_id,
    state,
    contract,
    tenure_in_months,
    total_revenue,
    monthly_charge,
    revenue_segment,
    churn_category,
    churn_reason
FROM customer_data
WHERE customer_status = 'Churned'
ORDER BY total_revenue DESC
LIMIT 20;
""", conn)
````

````python
# Q47 — Business Question:

# Which states rank highest in churned revenue exposure?

pd.read_sql_query("""
WITH state_churn AS (
    SELECT
        state,
        COUNT(DISTINCT customer_id) AS churned_customers,
        SUM(total_revenue) AS revenue_exposure
    FROM customer_data
    WHERE customer_status = 'Churned'
    GROUP BY state
)
SELECT
    state,
    churned_customers,
    ROUND(revenue_exposure, 2) AS revenue_exposure,
    RANK() OVER (
        ORDER BY revenue_exposure DESC
    ) AS revenue_risk_rank
FROM state_churn
ORDER BY revenue_risk_rank;
""", conn)
````

````python
# Q48 — Business Question:

# Which customers are high-value and also have high monthly charges?

pd.read_sql_query("""
WITH customer_benchmarks AS (
    SELECT
        AVG(total_revenue) AS avg_revenue,
        AVG(monthly_charge) AS avg_monthly_charge
    FROM customer_data
)
SELECT
    c.customer_id,
    c.state,
    c.contract,
    c.monthly_charge,
    c.total_revenue,
    c.customer_status,
    c.revenue_segment
FROM customer_data c
CROSS JOIN customer_benchmarks b
WHERE c.total_revenue > b.avg_revenue
  AND c.monthly_charge > b.avg_monthly_charge
ORDER BY c.total_revenue DESC;
""", conn)
````

````python
# Q49 — Business Question:

# Which customer segments have the highest combined churn and revenue risk?

pd.read_sql_query("""
WITH segment_risk AS (
    SELECT
        revenue_segment,
        COUNT(DISTINCT customer_id) AS customers,
        SUM(
            CASE
                WHEN customer_status = 'Churned' THEN 1
                ELSE 0
            END
        ) AS churned_customers,
        SUM(
            CASE
                WHEN customer_status = 'Churned'
                THEN total_revenue
                ELSE 0
            END
        ) AS revenue_at_risk
    FROM customer_data
    GROUP BY revenue_segment
)
SELECT
    revenue_segment,
    customers,
    churned_customers,
    ROUND(
        100.0 * churned_customers / customers,
        2
    ) AS churn_rate,
    ROUND(revenue_at_risk, 2) AS revenue_at_risk,
    ROUND(
        100.0 * revenue_at_risk /
        SUM(revenue_at_risk) OVER (),
        2
    ) AS revenue_risk_share
FROM segment_risk
ORDER BY revenue_at_risk DESC;
""", conn)
````

````python
# Q50 — Business Question:

# Which customers should be prioritized first for retention campaigns?

pd.read_sql_query("""
WITH customer_risk AS (
    SELECT
        customer_id,
        state,
        contract,
        tenure_in_months,
        total_revenue,
        monthly_charge,
        customer_status,
        revenue_segment,
        churn_category,
        churn_reason,

        CASE
            WHEN customer_status = 'Churned' THEN 3
            WHEN total_revenue >= (
                SELECT AVG(total_revenue)
                FROM customer_data
            ) THEN 2
            ELSE 1
        END AS risk_priority

    FROM customer_data
)
SELECT
    customer_id,
    state,
    contract,
    tenure_in_months,
    ROUND(total_revenue, 2) AS total_revenue,
    ROUND(monthly_charge, 2) AS monthly_charge,
    revenue_segment,
    customer_status,
    churn_category,
    churn_reason,
    risk_priority,

    CASE
        WHEN risk_priority = 3
            THEN 'Critical Retention Priority'
        WHEN risk_priority = 2
            THEN 'High Retention Priority'
        ELSE 'Standard Priority'
    END AS retention_priority

FROM customer_risk
ORDER BY
    risk_priority DESC,
    total_revenue DESC
LIMIT 50;
""", conn)
````

# 📊 AI-Powered Customer Churn & Retention Analytics Dashboard

## 1. Dashboard Business Purpose

The primary objective of this dashboard is to **understand, monitor, and reduce customer churn**.

The dashboard transforms customer-level data into actionable business insights, enabling management to quickly identify:

* How many customers are currently active?
* How many customers have churned?
* What is the overall **churn rate**?
* What is the impact of churn on revenue?
* Which customer groups have the highest risk?
* Which contract types have higher churn?
* Why are customers leaving?
* Which states or regions are performing better or worse?
* What is the relationship between customer tenure and revenue?
* Which areas should be prioritized to improve customer retention?

---

# 2. Business Problem

Customer churn is a major business challenge for subscription-based businesses.

When customers continuously leave, the business can experience:

**Customer Loss → Revenue Loss → Higher Acquisition Costs → Lower Customer Lifetime Value**

Therefore, it is not enough to simply ask:

> **“How many customers have churned?”**

The business also needs to understand:

> **“Who churned, why did they churn, how much revenue is associated with the churn, and what actions can be taken to retain similar customers?”**

---

# 3. What Business Questions Will the Dashboard Answer?

The dashboard is designed around five major business areas.

### A. Customer Health

**“How healthy is our current customer base?”**

Key metrics:

* Total Customers
* Churned Customers
* Stayed Customers
* Churn Rate
* Retention Rate

---

### B. Revenue Health

**“What is the financial impact of customer churn?”**

Key metrics:

* Total Revenue
* Net Revenue
* Revenue per Customer
* Average Monthly Charge
* Revenue at Risk

---

### C. Churn Drivers

**“Why are customers leaving?”**

The dashboard analyzes:

* Churn Category
* Churn Reason
* Contract Type
* Internet Type
* Customer Value
* Customer Tenure

---

### D. Geographic Performance

**“Which states are performing well and which states have higher business risk?”**

The dashboard analyzes:

* State Revenue
* Customer Count by State
* State Churn Rate
* State Retention Rate
* Revenue per Customer

An **interactive India choropleth map** can be used to provide a geographic view of business performance.

---

### E. Retention Opportunities

**“Which customers or segments should the business prioritize for retention?”**

The dashboard can help identify:

* High-value churned customers
* High-revenue customers
* Long-tenure customers
* High-risk contract types
* High-risk states
* High-risk service combinations

````python
# ================================================================
# AI-POWERED CUSTOMER CHURN & RETENTION ANALYTICS DASHBOARD
# DASH + PLOTLY + PANDAS
# FIXED VERSION — see "FIX #" comments for every bug that was corrected
# ================================================================

# ================================================================
# 1. INSTALLATION
# ================================================================
# python -m pip install --upgrade dash plotly pandas numpy
# ================================================================

# ================================================================
# 2. IMPORTS
# ================================================================

import os
import glob
import warnings

import numpy as np
import pandas as pd

from dash import Dash, dcc, html, Input, Output, State, ctx   # FIX: ctx imported once at top, not inside callback

import plotly.express as px
import plotly.graph_objects as go

warnings.filterwarnings("ignore")


# ================================================================
# 3. DATA FILE
# ================================================================

DATA_FILE = "Churn_Customer_Data.csv"

if not os.path.exists(DATA_FILE):
    csv_files = glob.glob("*.csv")
    if len(csv_files) > 0:
        churn_files = [f for f in csv_files
                       if any(word in f.lower() for word in ["churn", "customer", "telecom"])]
        DATA_FILE = churn_files[0] if churn_files else csv_files[0]

print("Using dataset:", DATA_FILE)


# ================================================================
# 4. LOAD DATA
# ================================================================

try:
    df = pd.read_csv(DATA_FILE, low_memory=False)
except Exception as e:
    raise FileNotFoundError(
        f"\nCould not load CSV file.\nCurrent file: {DATA_FILE}\n\n"
        f"Place your CSV in the same folder or change DATA_FILE.\n\nOriginal error:\n{e}"
    )


# ================================================================
# 5. CLEAN COLUMN NAMES
# ================================================================

df.columns = df.columns.astype(str).str.strip()

for col in df.select_dtypes(include=["object"]).columns:
    df[col] = df[col].astype(str).str.strip()

print("\nDataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())


# ================================================================
# 6. COLUMN FINDER
# ================================================================

def find_column(dataframe, candidates):
    lookup = {str(c).strip().lower(): c for c in dataframe.columns}
    for candidate in candidates:
        key = str(candidate).strip().lower()
        if key in lookup:
            return lookup[key]
    return None


# ================================================================
# 7. AUTOMATIC COLUMN DETECTION
# ================================================================

CUSTOMER_COL = find_column(df, ["customer_id", "Customer_ID", "Customer ID", "customerid", "id"])

REVENUE_COL = find_column(df, ["Total_Revenue", "Total Revenue", "total_revenue", "Revenue", "revenue",
                                 "Payment_Amount", "Payment Amount", "payment_amount"])

NET_REVENUE_COL = find_column(df, ["Net_Revenue", "Net Revenue", "net_revenue"])

# FIX #1: "Tenure_in_Months" was missing from the candidate list, so on this dataset
# TENURE_COL always resolved to None -> Average Tenure KPI showed 0.0 and the
# "Customer Tenure vs Revenue" chart was permanently empty. Added the real variants.
TENURE_COL = find_column(df, [
    "Tenure_in_Months", "Tenure in Months", "tenure_in_months",
    "Tenure_Months", "Tenure Months", "tenure_months", "Tenure", "tenure",
])

MONTHLY_CHARGE_COL = find_column(df, ["Monthly_Charges", "Monthly Charges", "monthly_charges",
                                        "Monthly_Charge", "Monthly Charge"])

STATE_COL = find_column(df, ["State", "state", "Customer_State", "Customer State"])

CONTRACT_COL = find_column(df, ["Contract_Type", "Contract Type", "contract_type", "Contract", "contract"])

# FIX #2: candidate order previously put "Internet_Service" (a plain Yes/No flag)
# ahead of "Internet_Type" (Fiber Optic / DSL / Cable / No Type). Since this dataset
# has BOTH columns, the code was silently picking the useless binary one, so the
# "Contract x Internet Churn Risk" heatmap only ever showed 2 columns instead of 4.
# Internet_Type now takes priority.
INTERNET_COL = find_column(df, ["Internet_Type", "Internet Type", "Internet_Service", "Internet Service",
                                  "internet_service"])

STATUS_COL = find_column(df, ["Customer_Status", "Customer Status", "customer_status", "Churn", "churn",
                                "Churn_Status", "Churn Status"])

CHURN_REASON_COL = find_column(df, ["Churn_Reason", "Churn Reason", "churn_reason"])
CHURN_CATEGORY_COL = find_column(df, ["Churn_Category", "Churn Category", "churn_category"])
PAYMENT_STATUS_COL = find_column(df, ["Payment_Status", "Payment Status", "payment_status"])
PAYMENT_METHOD_COL = find_column(df, ["Payment_Method", "Payment Method", "payment_method"])
TRANSACTION_CHANNEL_COL = find_column(df, ["Transaction_Channel", "Transaction Channel", "transaction_channel"])
DISCOUNT_COL = find_column(df, ["Discount_Applied", "Discount Applied", "discount_applied"])
RECHARGE_COL = find_column(df, ["Recharge_Frequency", "Recharge Frequency", "recharge_frequency"])

# FIX #3 (the "month KPI" bug): value tier / cohort column used for the fallback trend below
TENURE_GROUP_COL = find_column(df, ["Tenure_Group", "Tenure Group", "tenure_group"])

DATE_COL = find_column(df, [
    "Billing_Date", "Billing Date", "billing_date", "Transaction_Date", "Transaction Date",
    "transaction_date", "Date", "date", "Subscription_Start_Date", "Subscription Start Date",
    "subscription_start_date", "Registration_Date", "Registration Date", "registration_date",
])


# ================================================================
# 8. DETECT DATE AUTOMATICALLY — FIX #3 (root cause of the "month KPI" glitch)
# ================================================================
# ORIGINAL BUG: the auto-detector ran pd.to_datetime() on EVERY column, including
# pure numeric ones. pandas silently interprets an int64 column as nanoseconds-
# since-epoch, so a column like "Age" (34, 45, 51 ...) converts into a batch of
# "valid" timestamps that are all ~1970-01-01. That column then got selected as
# DATE_COL. Every row then collapsed into the SAME calendar month, so "Monthly
# Revenue Trend" / "Monthly Churn Rate Trend" showed a single flat point instead
# of a real trend -> this is exactly the "month kpi issue" you were seeing.
#
# FIX: only attempt auto-detection on text/object columns (real dates are strings
# in a CSV), and sanity-check that the parsed years fall in a believable range.

def _looks_like_real_dates(series, min_year=1990, max_year=2035):
    parsed = pd.to_datetime(series, errors="coerce")
    valid = parsed.dropna()
    if valid.empty:
        return False, parsed
    in_range = valid.dt.year.between(min_year, max_year)
    return in_range.mean() > 0.95, parsed


if DATE_COL is None:
    for col in df.columns:
        if df[col].dtype != object:          # <-- skip numeric columns entirely
            continue
        looks_ok, converted = _looks_like_real_dates(df[col])
        n_valid = converted.notna().sum()
        threshold = max(5, int(len(df) * 0.30))
        if looks_ok and n_valid >= threshold:
            DATE_COL = col
            break

HAS_REAL_DATE = DATE_COL is not None
print("Detected date column:", DATE_COL if HAS_REAL_DATE else "None (will use tenure-cohort fallback)")


# ================================================================
# 9. HELPERS
# ================================================================

def numeric_series(dataframe, column):
    if column is None:
        return pd.Series(0, index=dataframe.index, dtype=float)
    return pd.to_numeric(dataframe[column], errors="coerce").fillna(0)


CHURN_TOKENS = ["yes", "y", "true", "1", "churn", "churned", "left", "lost"]
# FIX #4: Retention was previously hard-coded as `100 - churn_rate`, which the
# master prompt explicitly forbids when a real Customer_Status column exists —
# it also silently folds brand-new "Joined" customers into "retained", inflating
# the number. Retention is now computed independently from actual "stayed"-type
# status values, exactly like churn is.
STAYED_TOKENS = ["stayed", "active", "retained", "no", "0", "false", "n"]


def churn_boolean(dataframe):
    if STATUS_COL is None:
        return pd.Series(False, index=dataframe.index)
    values = dataframe[STATUS_COL].astype(str).str.strip().str.lower()
    return values.isin(CHURN_TOKENS)


def retained_boolean(dataframe):
    if STATUS_COL is None:
        return pd.Series(False, index=dataframe.index)
    values = dataframe[STATUS_COL].astype(str).str.strip().str.lower()
    return values.isin(STAYED_TOKENS)


# ================================================================
# 10. PREPARE INTERNAL DATA
# ================================================================

work_df = df.copy()

work_df["_Revenue"] = numeric_series(work_df, REVENUE_COL)
work_df["_NetRevenue"] = numeric_series(work_df, NET_REVENUE_COL) if NET_REVENUE_COL else work_df["_Revenue"]
work_df["_Tenure"] = numeric_series(work_df, TENURE_COL)
work_df["_MonthlyCharge"] = numeric_series(work_df, MONTHLY_CHARGE_COL)
work_df["_IsChurned"] = churn_boolean(work_df)
work_df["_IsRetained"] = retained_boolean(work_df)

if HAS_REAL_DATE:
    work_df["_Date"] = pd.to_datetime(work_df[DATE_COL], errors="coerce")
else:
    work_df["_Date"] = pd.NaT

TENURE_ORDER = ["0-6 Months", "7-12 Months", "13-24 Months", "25-48 Months", "49-60 Months", "61+ Months"]


# ================================================================
# 11. EMPTY FIGURE
# ================================================================

def empty_figure(title, message="No data available"):
    fig = go.Figure()
    fig.add_annotation(text=message, x=0.5, y=0.5, xref="paper", yref="paper",
                        showarrow=False, font=dict(size=15, color="#94A3B8"))
    fig.update_layout(
        title=dict(text=title, x=0.5), template="plotly_dark", height=380,
        margin=dict(l=30, r=30, t=60, b=30),
        paper_bgcolor="#0B1220", plot_bgcolor="#0B1220",
        xaxis=dict(visible=False), yaxis=dict(visible=False),
    )
    return fig


# ================================================================
# 12. FILTER OPTIONS
# ================================================================

def get_options(dataframe, column):
    if column is None:
        return [{"label": "All", "value": "__ALL__"}]
    values = (dataframe[column].dropna().astype(str).str.strip()
              .replace("", np.nan).dropna().unique().tolist())
    values = sorted(values, key=lambda x: x.lower())
    return [{"label": "All", "value": "__ALL__"}] + [{"label": str(v), "value": str(v)} for v in values]


STATE_OPTIONS = get_options(work_df, STATE_COL)
CONTRACT_OPTIONS = get_options(work_df, CONTRACT_COL)
INTERNET_OPTIONS = get_options(work_df, INTERNET_COL)
STATUS_OPTIONS = get_options(work_df, STATUS_COL)


# ================================================================
# 13. THEME
# ================================================================

BG, CARD, CARD_2, BORDER = "#07111F", "#0D1B2A", "#102033", "#203653"
TEXT, MUTED = "#F8FAFC", "#94A3B8"
CYAN, BLUE, PURPLE, GREEN, ORANGE, RED = "#22D3EE", "#3B82F6", "#A78BFA", "#34D399", "#F59E0B", "#F87171"


def graph_layout(title):
    return dict(
        title=dict(text=title, x=0.5, xanchor="center", font=dict(size=17, color=TEXT)),
        template="plotly_dark", paper_bgcolor=CARD, plot_bgcolor=CARD, font=dict(color=TEXT),
        margin=dict(l=55, r=30, t=65, b=65),
        hoverlabel=dict(bgcolor="#111827", font_size=12),
        legend=dict(bgcolor="rgba(0,0,0,0)"),
    )


def kpi_card(icon, title, value, subtitle, accent):
    return html.Div(className="kpi-card", children=[
        html.Div(icon, className="kpi-icon", style={"borderColor": accent, "color": accent}),
        html.Div(className="kpi-content", children=[
            html.Div(title, className="kpi-title"),
            html.Div(value, className="kpi-value"),
            html.Div(subtitle, className="kpi-subtitle"),
        ]),
    ])


# ================================================================
# 14. INDIA GEOJSON (choropleth, with automatic bar-chart fallback)
# ================================================================
# FIX #5: the original code never attempted a real India map at all — it always
# drew the ranked-bar fallback. We now genuinely try the choropleth first and
# fall back automatically (and silently, no crash) if the GeoJSON can't be
# fetched (e.g. no internet access in the runtime environment).

_INDIA_GEOJSON_CACHE = {"data": None, "tried": False}


def get_india_geojson():
    if _INDIA_GEOJSON_CACHE["tried"]:
        return _INDIA_GEOJSON_CACHE["data"]
    _INDIA_GEOJSON_CACHE["tried"] = True
    try:
        import urllib.request, json
        url = "https://raw.githubusercontent.com/geohacker/india/master/state/india_state.geojson"
        with urllib.request.urlopen(url, timeout=6) as resp:
            _INDIA_GEOJSON_CACHE["data"] = json.loads(resp.read().decode())
    except Exception:
        _INDIA_GEOJSON_CACHE["data"] = None
    return _INDIA_GEOJSON_CACHE["data"]


# ================================================================
# 15. APP
# ================================================================

app = Dash(__name__, suppress_callback_exceptions=True)
app.title = "AI Customer Churn Analytics"

app.index_string = """
<!DOCTYPE html>
<html>
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1">
<title>AI Customer Churn Analytics</title>
<style>
* { box-sizing: border-box; }
html, body {
    margin: 0; padding: 0;
    background: radial-gradient(circle at 85% 5%, #102B4C 0%, #07111F 35%, #050B14 100%);
    color: #F8FAFC; font-family: Inter, Arial, Helvetica, sans-serif;
}
body { min-height: 100vh; }
.app-container { width: 100%; max-width: 1920px; margin: auto; padding: 18px; }

.header {
    background: linear-gradient(135deg, #0D1B2A, #0A1625);
    border: 1px solid #203653; border-radius: 20px; padding: 18px 24px; margin-bottom: 15px;
    display: flex; align-items: center; justify-content: space-between;
    box-shadow: 0 12px 40px rgba(0,0,0,.25);
}
.brand { display: flex; align-items: center; gap: 14px; }
.brand-icon {
    width: 56px; height: 56px; border-radius: 16px; display: flex; align-items: center;
    justify-content: center; font-size: 28px;
    background: linear-gradient(135deg, #1677FF, #8B5CF6);
    box-shadow: 0 0 28px rgba(34,211,238,.20);
}
.title { font-size: 25px; font-weight: 900; }
.title-gradient {
    background: linear-gradient(90deg, #FFFFFF, #22D3EE, #A78BFA);
    -webkit-background-clip: text; -webkit-text-fill-color: transparent;
}
.subtitle { color: #94A3B8; margin-top: 5px; font-size: 13px; }
.data-note { color: #64748B; font-size: 10.5px; margin-top: 3px; font-style: italic; }

.filter-bar {
    background: linear-gradient(145deg, #0D1B2A, #091522);
    border: 1px solid #203653; border-radius: 18px; padding: 15px; margin-bottom: 15px;
    display: grid; grid-template-columns: repeat(4, 1fr) 130px; gap: 12px; align-items: end;
}
.filter-label {
    display: block; color: #94A3B8; font-size: 10px; font-weight: 800;
    margin-bottom: 6px; text-transform: uppercase; letter-spacing: .7px;
}
.reset-button {
    width: 100%; height: 40px; border: none; border-radius: 10px;
    background: linear-gradient(135deg, #1677FF, #7C3AED); color: white;
    font-weight: 800; cursor: pointer; transition: .2s ease;
}
.reset-button:hover { transform: translateY(-2px); box-shadow: 0 8px 25px rgba(59,130,246,.35); }

.kpi-grid { display: grid; grid-template-columns: repeat(4, 1fr); gap: 12px; margin-bottom: 15px; }
.kpi-card {
    min-height: 120px; background: linear-gradient(145deg, #102033, #0B1726);
    border: 1px solid #203653; border-radius: 17px; padding: 17px;
    display: flex; align-items: center; justify-content: center; text-align: center;
    gap: 12px; transition: .2s ease;
}
.kpi-card:hover { transform: translateY(-3px); box-shadow: 0 12px 30px rgba(0,0,0,.30); }
.kpi-icon {
    width: 52px; min-width: 52px; height: 52px; border: 2px solid; border-radius: 50%;
    display: flex; align-items: center; justify-content: center; font-size: 22px; background: #07111F;
}
.kpi-content { text-align: center; }
.kpi-title { color: #94A3B8; font-size: 10px; font-weight: 800; text-transform: uppercase; letter-spacing: .5px; }
.kpi-value { color: #FFFFFF; font-size: 22px; font-weight: 900; margin-top: 5px; }
.kpi-subtitle { color: #64748B; font-size: 10px; margin-top: 4px; }

.chart-grid { display: grid; grid-template-columns: repeat(2, minmax(0, 1fr)); gap: 15px; }
.chart-card {
    min-width: 0; background: linear-gradient(145deg, #0D1B2A, #091522);
    border: 1px solid #203653; border-radius: 18px; overflow: hidden;
    box-shadow: 0 8px 25px rgba(0,0,0,.18); transition: border-color .2s ease;
}
.chart-card.full { grid-column: 1 / -1; }
.chart-card:hover { border-color: #2C4B70; }

.footer { text-align: center; color: #64748B; font-size: 11px; padding: 18px; }

.Select-control { background-color: #0A1625 !important; border-color: #203653 !important; color: white !important; border-radius: 9px !important; }
.Select-menu-outer { background-color: #0D1B2A !important; color: white !important; z-index: 9999 !important; }
.Select-value-label { color: white !important; }
.Select-placeholder { color: #94A3B8 !important; }
.Select-input input { color: white !important; }

@media(max-width: 1200px) {
    .kpi-grid { grid-template-columns: repeat(2, 1fr); }
    .filter-bar { grid-template-columns: repeat(2, 1fr); }
}
@media(max-width: 850px) {
    .chart-grid { grid-template-columns: 1fr; }
    .chart-card.full { grid-column: auto; }
}
@media(max-width: 600px) {
    .app-container { padding: 9px; }
    .header { padding: 15px; }
    .title { font-size: 18px; }
    .kpi-grid { grid-template-columns: 1fr; }
    .filter-bar { grid-template-columns: 1fr; }
}
</style>
</head>
<body>
{%app_entry%}
<footer>
{%config%}
{%scripts%}
{%renderer%}
</footer>
</body>
</html>
"""


# ================================================================
# 16. KPI CALC HELPER (shared by initial render + callback — avoids drift/bugs)
# ================================================================

def compute_kpis(filtered):
    total_customers = filtered[CUSTOMER_COL].nunique() if CUSTOMER_COL else len(filtered)
    churned_customers = int(filtered["_IsChurned"].sum())
    retained_customers = int(filtered["_IsRetained"].sum())

    churn_rate = (churned_customers / total_customers * 100) if total_customers else 0
    retention_rate = (retained_customers / total_customers * 100) if total_customers else 0  # FIX #4

    total_revenue = filtered["_Revenue"].sum()
    net_revenue = filtered["_NetRevenue"].sum()
    revenue_per_customer = (total_revenue / total_customers) if total_customers else 0
    average_tenure = filtered["_Tenure"].mean() if len(filtered) else 0  # FIX #1 makes this real now

    return dict(total_customers=total_customers, churned_customers=churned_customers,
                retained_customers=retained_customers, churn_rate=churn_rate,
                retention_rate=retention_rate, total_revenue=total_revenue, net_revenue=net_revenue,
                revenue_per_customer=revenue_per_customer, average_tenure=average_tenure)


def build_kpi_cards(k):
    return [
        kpi_card("👥", "Total Customers", f"{k['total_customers']:,}", "Current customer base", BLUE),
        kpi_card("⚠️", "Churned Customers", f"{k['churned_customers']:,}", "Customers lost", RED),
        kpi_card("📉", "Churn Rate", f"{k['churn_rate']:.2f}%", "Customer churn", ORANGE),
        kpi_card("🔒", "Retention Rate", f"{k['retention_rate']:.2f}%", "Customer retention", GREEN),
        kpi_card("💰", "Total Revenue", f"₹{k['total_revenue']:,.0f}", "Revenue generated", CYAN),
        kpi_card("💎", "Net Revenue", f"₹{k['net_revenue']:,.0f}", "Net business revenue", PURPLE),
        kpi_card("🧾", "Revenue / Customer", f"₹{k['revenue_per_customer']:,.0f}", "Average customer value", BLUE),
        kpi_card("⏳", "Average Tenure", f"{k['average_tenure']:.1f} Months", "Customer longevity", GREEN),
    ]


initial_kpis = compute_kpis(work_df)


# ================================================================
# 17. LAYOUT
# ================================================================

app.layout = html.Div(className="app-container", children=[

    html.Div(className="header", children=[
        html.Div(className="brand", children=[
            html.Div("🤖", className="brand-icon"),
            html.Div(children=[
                html.Div(className="title", children=[
                    html.Span("AI-Powered Customer Churn & ", className="title-gradient"),
                    html.Span("Retention Analytics Dashboard", className="title-gradient"),
                ]),
                html.Div("Customer Health • Revenue Impact • Churn Drivers • Geographic Risk • "
                         "Retention Opportunities", className="subtitle"),
                html.Div(
                    ("Live billing dates detected — monthly trend charts use real calendar months."
                     if HAS_REAL_DATE else
                     "No billing/date column found — Revenue & Churn trend charts show tenure-cohort "
                     "trend instead (0-6, 7-12, 13-24, 25-48 months)."),
                    className="data-note"
                ),
            ]),
        ]),
    ]),

    html.Div(className="filter-bar", children=[
        html.Div(children=[
            html.Label("State", className="filter-label"),
            dcc.Dropdown(id="state-filter", options=STATE_OPTIONS, value="__ALL__",
                         clearable=False, searchable=True),
        ]),
        html.Div(children=[
            html.Label("Contract", className="filter-label"),
            dcc.Dropdown(id="contract-filter", options=CONTRACT_OPTIONS, value="__ALL__",
                         clearable=False, searchable=True),
        ]),
        html.Div(children=[
            html.Label("Internet Type", className="filter-label"),
            dcc.Dropdown(id="internet-filter", options=INTERNET_OPTIONS, value="__ALL__",
                         clearable=False, searchable=True),
        ]),
        html.Div(children=[
            html.Label("Customer Status", className="filter-label"),
            dcc.Dropdown(id="status-filter", options=STATUS_OPTIONS, value="__ALL__",
                         clearable=False, searchable=True),
        ]),
        html.Button("↻  RESET FILTERS", id="reset-button", className="reset-button"),
    ]),

    html.Div(id="kpi-container", className="kpi-grid", children=build_kpi_cards(initial_kpis)),

    html.Div(className="chart-grid", children=[
        html.Div(className="chart-card", children=[
            dcc.Graph(id="monthly-revenue-chart", config={"displayModeBar": True, "responsive": True},
                      style={"height": "400px"}),
        ]),
        html.Div(className="chart-card", children=[
            dcc.Graph(id="monthly-churn-chart", config={"displayModeBar": True, "responsive": True},
                      style={"height": "400px"}),
        ]),
        html.Div(className="chart-card", children=[
            dcc.Graph(id="state-map-chart", config={"displayModeBar": True, "responsive": True},
                      style={"height": "470px"}),
        ]),
        html.Div(className="chart-card", children=[
            dcc.Graph(id="status-donut-chart", config={"displayModeBar": True, "responsive": True},
                      style={"height": "470px"}),
        ]),
        html.Div(className="chart-card", children=[
            dcc.Graph(id="contract-churn-chart", config={"displayModeBar": True, "responsive": True},
                      style={"height": "430px"}),
        ]),
        html.Div(className="chart-card", children=[
            dcc.Graph(id="churn-reason-chart", config={"displayModeBar": True, "responsive": True},
                      style={"height": "430px"}),
        ]),
        html.Div(className="chart-card", children=[
            dcc.Graph(id="tenure-revenue-chart", config={"displayModeBar": True, "responsive": True},
                      style={"height": "450px"}),
        ]),
        html.Div(className="chart-card", children=[
            dcc.Graph(id="contract-internet-chart", config={"displayModeBar": True, "responsive": True},
                      style={"height": "450px"}),
        ]),
    ]),

    html.Div("AI-Powered Customer Churn & Retention Analytics • Interactive Business Intelligence Dashboard",
             className="footer"),
])


# ================================================================
# 18. CALLBACK
# ================================================================

@app.callback(
    [
        Output("state-filter", "value"),
        Output("contract-filter", "value"),
        Output("internet-filter", "value"),
        Output("status-filter", "value"),
        Output("kpi-container", "children"),
        Output("monthly-revenue-chart", "figure"),
        Output("monthly-churn-chart", "figure"),
        Output("state-map-chart", "figure"),
        Output("status-donut-chart", "figure"),
        Output("contract-churn-chart", "figure"),
        Output("churn-reason-chart", "figure"),
        Output("tenure-revenue-chart", "figure"),
        Output("contract-internet-chart", "figure"),
    ],
    [
        Input("state-filter", "value"),
        Input("contract-filter", "value"),
        Input("internet-filter", "value"),
        Input("status-filter", "value"),
        Input("reset-button", "n_clicks"),
    ],
    [
        State("state-filter", "value"),
        State("contract-filter", "value"),
        State("internet-filter", "value"),
        State("status-filter", "value"),
    ],
    prevent_initial_call=False,
)
def update_dashboard(state_value, contract_value, internet_value, status_value, reset_clicks,
                      state_state, contract_state, internet_state, status_state):

    reset_triggered = ctx.triggered_id == "reset-button"

    if reset_triggered:
        selected_state = selected_contract = selected_internet = selected_status = "__ALL__"
    else:
        selected_state, selected_contract = state_value, contract_value
        selected_internet, selected_status = internet_value, status_value

    # ---------------- FILTER DATA ----------------
    filtered = work_df.copy()

    if STATE_COL and selected_state != "__ALL__":
        filtered = filtered[filtered[STATE_COL].astype(str).str.strip() == str(selected_state)]
    if CONTRACT_COL and selected_contract != "__ALL__":
        filtered = filtered[filtered[CONTRACT_COL].astype(str).str.strip() == str(selected_contract)]
    if INTERNET_COL and selected_internet != "__ALL__":
        filtered = filtered[filtered[INTERNET_COL].astype(str).str.strip() == str(selected_internet)]
    if STATUS_COL and selected_status != "__ALL__":
        filtered = filtered[filtered[STATUS_COL].astype(str).str.strip() == str(selected_status)]

    empty_after_filter = filtered.empty

    # ---------------- KPIs ----------------
    k = compute_kpis(filtered)
    kpis = build_kpi_cards(k)

    if empty_after_filter:
        empty_msg = "No data available for the selected filters."
        blank = tuple(empty_figure(t, empty_msg) for t in [
            "Monthly Revenue Trend", "Monthly Churn Rate Trend", "State Revenue Analysis",
            "Customer Health", "Contract Churn Risk", "Churn Reasons",
            "Customer Tenure vs Revenue", "Contract x Internet Churn Risk"])
        return (selected_state, selected_contract, selected_internet, selected_status, kpis, *blank)

    # ============================================================
    # MONTHLY / COHORT TREND — FIX #3
    # ============================================================
    if HAS_REAL_DATE and filtered["_Date"].notna().any():
        date_df = filtered[filtered["_Date"].notna()].copy()
        date_df["_Month"] = date_df["_Date"].dt.to_period("M").dt.to_timestamp()

        monthly_revenue = (date_df.groupby("_Month", as_index=False)
                            .agg(Total_Revenue=("_Revenue", "sum"), Customers=("_Month", "size"))
                            .sort_values("_Month"))

        fig_revenue = px.line(monthly_revenue, x="_Month", y="Total_Revenue", markers=True,
                               labels={"_Month": "Month", "Total_Revenue": "Total Revenue"})
        fig_revenue.update_traces(line=dict(width=3), marker=dict(size=8),
                                   hovertemplate="<b>%{x|%b %Y}</b><br>Revenue: ₹%{y:,.0f}<extra></extra>")
        fig_revenue.update_xaxes(tickformat="%b %Y", tickangle=-45)
        fig_revenue.update_yaxes(tickprefix="₹", separatethousands=True)
        fig_revenue.update_layout(**graph_layout("Monthly Revenue Trend"), height=400, hovermode="x unified")

        monthly_churn = (date_df.groupby("_Month", as_index=False)
                          .agg(Churned_Customers=("_IsChurned", "sum"), Total_Customers=("_Month", "size"))
                          .sort_values("_Month"))
        monthly_churn["Churn_Rate"] = np.where(
            monthly_churn["Total_Customers"] > 0,
            monthly_churn["Churned_Customers"] / monthly_churn["Total_Customers"] * 100, 0)

        fig_churn = px.line(monthly_churn, x="_Month", y="Churn_Rate", markers=True,
                             labels={"_Month": "Month", "Churn_Rate": "Churn Rate (%)"},
                             custom_data=["Churned_Customers", "Total_Customers"])
        fig_churn.update_traces(line=dict(width=3), marker=dict(size=8),
                                 hovertemplate="<b>%{x|%b %Y}</b><br>Churn Rate: %{y:.2f}%<br>"
                                               "Churned: %{customdata[0]:,}<br>Total: %{customdata[1]:,}"
                                               "<extra></extra>")
        fig_churn.update_xaxes(tickformat="%b %Y", tickangle=-45)
        fig_churn.update_yaxes(ticksuffix="%")
        fig_churn.update_layout(**graph_layout("Monthly Churn Rate Trend"), height=400, hovermode="x unified")

    elif TENURE_GROUP_COL:
        # No genuine date column in this dataset -> documented, meaningful fallback:
        # trend across customer-tenure cohorts instead of calendar months.
        present_order = [t for t in TENURE_ORDER if t in filtered[TENURE_GROUP_COL].unique()]

        rev_by_cohort = filtered.groupby(TENURE_GROUP_COL)["_Revenue"].mean().reindex(present_order)
        fig_revenue = go.Figure(go.Scatter(
            x=rev_by_cohort.index, y=rev_by_cohort.values, mode="lines+markers", fill="tozeroy",
            line=dict(width=3, color=CYAN), marker=dict(size=9, color=CYAN),
            fillcolor="rgba(34,211,238,0.15)",
            hovertemplate="<b>%{x}</b><br>Avg Revenue: ₹%{y:,.0f}<extra></extra>",
        ))
        fig_revenue.update_layout(**graph_layout("Avg Revenue by Tenure Cohort"), height=400)

        churn_by_cohort = filtered.groupby(TENURE_GROUP_COL).apply(
            lambda g: g["_IsChurned"].sum() / len(g) * 100 if len(g) else 0).reindex(present_order)
        fig_churn = go.Figure(go.Scatter(
            x=churn_by_cohort.index, y=churn_by_cohort.values, mode="lines+markers",
            line=dict(width=3, color=RED), marker=dict(size=9, color=RED),
            hovertemplate="<b>%{x}</b><br>Churn Rate: %{y:.2f}%<extra></extra>",
        ))
        fig_churn.update_layout(**graph_layout("Churn Rate by Tenure Cohort"), height=400)

    else:
        fig_revenue = empty_figure("Monthly Revenue Trend", "No date or tenure-cohort column found")
        fig_churn = empty_figure("Monthly Churn Rate Trend", "No date or tenure-cohort column found")

    # ============================================================
    # STATE MAP — FIX #5 (real choropleth attempt + automatic fallback)
    # ============================================================
    if STATE_COL:
        agg_kwargs = dict(Total_Revenue=("_Revenue", "sum"))
        if CUSTOMER_COL:
            agg_kwargs["Customers"] = (CUSTOMER_COL, "nunique")
        else:
            agg_kwargs["Customers"] = ("_Revenue", "size")
        agg_kwargs["Churned"] = ("_IsChurned", "sum")

        state_summary = filtered.groupby(STATE_COL, as_index=False).agg(**agg_kwargs)
        state_summary["Churn_Rate"] = np.where(
            state_summary["Customers"] > 0, state_summary["Churned"] / state_summary["Customers"] * 100, 0)
        state_summary["Revenue_Per_Customer"] = np.where(
            state_summary["Customers"] > 0, state_summary["Total_Revenue"] / state_summary["Customers"], 0)

        geojson = get_india_geojson()
        fig_map = None
        if geojson is not None:
            try:
                fig_map = px.choropleth(
                    state_summary, geojson=geojson, featureidkey="properties.NAME_1",
                    locations=STATE_COL, color="Total_Revenue",
                    color_continuous_scale="Turbo",
                    custom_data=["Customers", "Churn_Rate", "Revenue_Per_Customer"],
                )
                fig_map.update_traces(
                    hovertemplate="<b>%{location}</b><br>Revenue: ₹%{z:,.0f}<br>"
                                  "Customers: %{customdata[0]:,}<br>Churn Rate: %{customdata[1]:.2f}%<br>"
                                  "Revenue/Customer: ₹%{customdata[2]:,.0f}<extra></extra>")
                fig_map.update_geos(fitbounds="locations", visible=False, bgcolor="rgba(0,0,0,0)")
                fig_map.update_layout(**graph_layout("Revenue by State — India"), height=470)
            except Exception:
                fig_map = None

        if fig_map is None:
            fig_map = px.bar(
                state_summary.sort_values("Total_Revenue", ascending=False).head(20),
                x="Total_Revenue", y=STATE_COL, orientation="h", color="Total_Revenue",
                color_continuous_scale="Turbo",
                custom_data=["Customers", "Churn_Rate", "Revenue_Per_Customer"],
                labels={STATE_COL: "State", "Total_Revenue": "Total Revenue"},
            )
            fig_map.update_traces(
                hovertemplate="<b>%{y}</b><br>Revenue: ₹%{x:,.0f}<br>Customers: %{customdata[0]:,}<br>"
                              "Churn Rate: %{customdata[1]:.2f}%<br>Revenue/Customer: ₹%{customdata[2]:,.0f}"
                              "<extra></extra>")
            fig_map.update_layout(**graph_layout("Top States by Revenue (map unavailable — ranked bar shown)"),
                                   height=470, yaxis=dict(categoryorder="total ascending"))
    else:
        fig_map = empty_figure("State Revenue Analysis", "State column not available")

    # ============================================================
    # CUSTOMER STATUS DONUT — FIX #4 (real category breakdown, not 100-churn)
    # ============================================================
    if STATUS_COL:
        status_counts = filtered[STATUS_COL].astype(str).str.strip().value_counts().reset_index()
        status_counts.columns = ["Status", "Customers"]
        color_map = {}
        for s in status_counts["Status"]:
            s_low = s.lower()
            if s_low in CHURN_TOKENS: color_map[s] = RED
            elif s_low in STAYED_TOKENS: color_map[s] = GREEN
            else: color_map[s] = BLUE   # e.g. "Joined" / other statuses

        fig_status = px.pie(status_counts, names="Status", values="Customers", hole=0.62,
                             color="Status", color_discrete_map=color_map)
        fig_status.update_traces(textinfo="percent+label",
                                  hovertemplate="<b>%{label}</b><br>Customers: %{value:,}<br>"
                                                "Share: %{percent}<extra></extra>")
        fig_status.update_layout(**graph_layout("Customer Health"), height=470, showlegend=True)
    else:
        fig_status = empty_figure("Customer Health", "Customer status column not available")

    # ============================================================
    # CONTRACT CHURN
    # ============================================================
    if CONTRACT_COL:
        contract_summary = filtered.groupby(CONTRACT_COL, as_index=False).agg(
            Customers=("_IsChurned", "size"), Churned=("_IsChurned", "sum"))
        contract_summary["Churn_Rate"] = np.where(
            contract_summary["Customers"] > 0, contract_summary["Churned"] / contract_summary["Customers"] * 100, 0)
        contract_summary = contract_summary.sort_values("Churn_Rate", ascending=False)

        fig_contract = px.bar(contract_summary, x="Churn_Rate", y=CONTRACT_COL, orientation="h",
                               text="Churn_Rate", color="Churn_Rate", color_continuous_scale="Turbo",
                               custom_data=["Customers", "Churned"])
        fig_contract.update_traces(texttemplate="%{text:.2f}%",
                                    hovertemplate="<b>%{y}</b><br>Churn Rate: %{x:.2f}%<br>"
                                                  "Customers: %{customdata[0]:,}<br>Churned: %{customdata[1]:,}"
                                                  "<extra></extra>")
        fig_contract.update_layout(**graph_layout("Contract Churn Risk"), height=430)
    else:
        fig_contract = empty_figure("Contract Churn Risk", "Contract column not available")

    # ============================================================
    # CHURN REASON LOLLIPOP
    # ============================================================
    if CHURN_REASON_COL:
        reason_df = (filtered[filtered["_IsChurned"]]
                     .groupby(CHURN_REASON_COL, as_index=False)
                     .agg(Churned_Customers=("_IsChurned", "sum"), Revenue_Exposure=("_Revenue", "sum"))
                     .sort_values("Revenue_Exposure", ascending=False).head(12))

        if not reason_df.empty:
            fig_reason = go.Figure()
            for _, row in reason_df.iterrows():
                fig_reason.add_trace(go.Scatter(
                    x=[0, row["Revenue_Exposure"]], y=[row[CHURN_REASON_COL], row[CHURN_REASON_COL]],
                    mode="lines", line=dict(width=3, color="#334155"), hoverinfo="skip", showlegend=False))
            fig_reason.add_trace(go.Scatter(
                x=reason_df["Revenue_Exposure"], y=reason_df[CHURN_REASON_COL], mode="markers",
                marker=dict(size=13, color=reason_df["Revenue_Exposure"], colorscale="Turbo",
                            showscale=True, colorbar=dict(title="Revenue")),
                customdata=reason_df[["Churned_Customers"]],
                hovertemplate="<b>%{y}</b><br>Revenue Exposure: ₹%{x:,.0f}<br>"
                              "Churned Customers: %{customdata[0]:,}<extra></extra>",
                showlegend=False))
            fig_reason.update_layout(**graph_layout("Churn Reasons — Revenue Exposure"), height=430)
            fig_reason.update_xaxes(tickprefix="₹")
        else:
            fig_reason = empty_figure("Churn Reasons", "No churned customers in the current filter selection")
    else:
        fig_reason = empty_figure("Churn Reasons", "Churn Reason column not available")

    # ============================================================
    # TENURE VS REVENUE — now populated correctly (FIX #1)
    # ============================================================
    if TENURE_COL:
        scatter_df = filtered.copy()
        if CUSTOMER_COL:
            customer_level = scatter_df.groupby(CUSTOMER_COL, as_index=False).agg(
                Tenure=("_Tenure", "mean"), Total_Revenue=("_Revenue", "sum"),
                Monthly_Charge=("_MonthlyCharge", "mean"), Churned=("_IsChurned", "max"))
        else:
            customer_level = pd.DataFrame({
                "Tenure": scatter_df["_Tenure"], "Total_Revenue": scatter_df["_Revenue"],
                "Monthly_Charge": scatter_df["_MonthlyCharge"], "Churned": scatter_df["_IsChurned"]})

        customer_level = (customer_level.replace([np.inf, -np.inf], np.nan)
                           .dropna(subset=["Tenure", "Total_Revenue"]))
        # FIX: marker size must never be negative
        customer_level["Monthly_Charge"] = customer_level["Monthly_Charge"].clip(lower=0)

        if not customer_level.empty:
            customer_level["Status"] = np.where(customer_level["Churned"], "Churned", "Active")
            fig_scatter = px.scatter(
                customer_level, x="Tenure", y="Total_Revenue", color="Status", size="Monthly_Charge",
                size_max=25, opacity=.72, color_discrete_map={"Churned": RED, "Active": CYAN},
                labels={"Tenure": "Tenure (Months)", "Total_Revenue": "Total Revenue",
                        "Monthly_Charge": "Monthly Charge"})
            fig_scatter.update_traces(hovertemplate="Tenure: %{x:.1f} months<br>Revenue: ₹%{y:,.0f}<extra></extra>")
            fig_scatter.update_layout(**graph_layout("Customer Tenure vs Revenue"), height=450)
        else:
            fig_scatter = empty_figure("Customer Tenure vs Revenue", "No valid tenure/revenue data")
    else:
        fig_scatter = empty_figure("Customer Tenure vs Revenue", "Tenure column not available")

    # ============================================================
    # CONTRACT x INTERNET TYPE HEATMAP — now uses real Internet_Type (FIX #2)
    # ============================================================
    if CONTRACT_COL and INTERNET_COL:
        heat_df = filtered.groupby([CONTRACT_COL, INTERNET_COL], as_index=False).agg(
            Customers=("_IsChurned", "size"), Churned=("_IsChurned", "sum"))
        heat_df["Churn_Rate"] = np.where(heat_df["Customers"] > 0,
                                          heat_df["Churned"] / heat_df["Customers"] * 100, 0)
        pivot = heat_df.pivot(index=CONTRACT_COL, columns=INTERNET_COL, values="Churn_Rate")

        if not pivot.empty:
            fig_heatmap = px.imshow(pivot, text_auto=".2f", aspect="auto", color_continuous_scale="Turbo",
                                     labels={"color": "Churn Rate %"})
            fig_heatmap.update_layout(**graph_layout("Contract x Internet Type Churn Risk"), height=450)
        else:
            fig_heatmap = empty_figure("Contract x Internet Type Churn Risk")
    else:
        fig_heatmap = empty_figure("Contract x Internet Type Churn Risk",
                                    "Contract or Internet Type column not available")

    return (selected_state, selected_contract, selected_internet, selected_status, kpis,
            fig_revenue, fig_churn, fig_map, fig_status, fig_contract, fig_reason, fig_scatter, fig_heatmap)


# ================================================================
# 19. RUN
# ================================================================

if __name__ == "__main__":
    app.run(debug=True, port=8050)
````

# 6. Business Recommendations

Based on the analytical framework, the business should focus on:

### 1. High-Value Customer Retention

Prioritize customers with high revenue contribution who are at risk of churn.

### 2. Contract Optimization

Identify high-churn contract types and develop targeted offers or improved contract incentives.

### 3. Churn Reason Reduction

Use churn-reason analysis to address major causes such as service, pricing, support, or competitive issues where those categories exist in the data.

### 4. Geographic Strategy

States with high revenue but weak retention should receive greater attention because they represent potentially significant revenue risk.

### 5. Early Customer Retention

Analyze low-tenure customers separately and improve onboarding and early customer engagement.

### 6. Revenue Protection

Instead of focusing only on the number of churned customers, prioritize **revenue exposure** so that the most financially important churn is addressed first.
