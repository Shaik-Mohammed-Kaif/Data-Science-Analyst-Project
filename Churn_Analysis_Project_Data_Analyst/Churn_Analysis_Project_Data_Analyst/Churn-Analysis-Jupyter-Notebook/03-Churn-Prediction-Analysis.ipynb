# 📊 Customer Churn Prediction — Business Understanding

## 🎯 Business Objective

The objective of this project is to **predict which customers are likely to churn** based on their demographic information, subscription details, service usage, billing behavior, tenure, and customer engagement patterns.

Customer churn directly impacts business revenue because acquiring a new customer is generally more expensive than retaining an existing customer.

Therefore, this machine learning solution will help the business:

* Identify customers who are at high risk of churn.
* Understand the major factors contributing to customer churn.
* Segment customers based on their churn probability.
* Enable proactive retention campaigns.
* Reduce customer loss and improve customer lifetime value.
* Support data-driven customer retention strategies.

---

## 💼 Business Problem

The business wants to answer:

> **"Which customers are most likely to leave, and what factors are driving their decision to leave?"**

Instead of waiting until customers cancel their services, the company can use a predictive model to identify **high-risk customers in advance**.

# 🎯 Target Variable

### `Churn`

The target variable represents whether the customer has churned.

| Churn       | Meaning           |
| ----------- | ----------------- |
| `0` / `No`  | Customer retained |
| `1` / `Yes` | Customer churned  |

The machine learning model will learn patterns from historical customer data and predict the probability of churn for new customers.

This is a **Binary Classification Problem**.

The model will predict:

```text

```

```text

Prediction: HIGH RISK
```

## Data Preprocessing & Feature Encoding

In this step, we prepare the customer churn dataset for machine learning by removing irrelevant columns, encoding categorical features, separating the target variable, and splitting the data into training and testing sets.

````python
# ============================================================
# Copy Dataset
# ============================================================

data = df.copy()


# ============================================================
# Data Processing
# ============================================================

# Drop columns that won't be used for prediction

data = data.drop(
    [
        'Customer_ID',
        'Churn_Category',
        'Churn_Reason',
        'churn_flag'
    ],
    axis=1,
    errors='ignore'
)


# ============================================================
# Clean Customer_Status
# ============================================================

# Remove extra spaces from Customer_Status

data['Customer_Status'] = (
    data['Customer_Status']
    .astype(str)
    .str.strip()
)


# Check unique values

print("Customer_Status values:")
print(data['Customer_Status'].value_counts(dropna=False))


# ============================================================
# List of Categorical Columns
# ============================================================

columns_to_encode = [
    'Gender',
    'Married',
    'State',
    'Value_Deal',
    'Phone_Service',
    'Multiple_Lines',
    'Internet_Service',
    'Internet_Type',
    'Online_Security',
    'Online_Backup',
    'Device_Protection_Plan',
    'Premium_Support',
    'Streaming_TV',
    'Streaming_Movies',
    'Streaming_Music',
    'Unlimited_Data',
    'Contract',
    'Paperless_Billing',
    'Payment_Method',

    # Derived categorical features
    'Age_Group',
    'Tenure_Group',
    'Customer_Lifecycle',
    'Referral_Group',
    'Revenue_Segment'
]


# ============================================================
# Encode Categorical Variables
# ============================================================

label_encoders = {}

for column in columns_to_encode:

    label_encoders[column] = LabelEncoder()

    data[column] = label_encoders[column].fit_transform(
        data[column].astype(str)
    )


# ============================================================
# Encode Target Variable
# ============================================================

# Stayed  = 0
# Churned = 1

data['Churn_Status'] = data['Customer_Status'].map({
    'Stayed': 0,
    'Churned': 1
})


# ============================================================
# Check Target Missing Values
# ============================================================

print("\nMissing values in Churn_Status:")
print(data['Churn_Status'].isna().sum())


# ============================================================
# Remove Rows Where Target Is Missing
# ============================================================

data = data.dropna(
    subset=['Churn_Status']
)


# ============================================================
# Convert Target to Integer
# ============================================================

data['Churn_Status'] = data['Churn_Status'].astype(int)


# ============================================================
# Drop Original Customer_Status
# ============================================================

# Customer_Status is already converted into Churn_Status.
# Therefore, it should not remain inside X.

data = data.drop(
    'Customer_Status',
    axis=1
)


# ============================================================
# Split Features and Target
# ============================================================

X = data.drop(
    'Churn_Status',
    axis=1
)

y = data['Churn_Status']


# ============================================================
# Check Feature Data Types
# ============================================================

print("\nRemaining Object Columns:")
print(X.select_dtypes(include='object').columns.tolist())


# ============================================================
# Train-Test Split
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# ============================================================
# Check Final Data
# ============================================================

print("\nX Train Shape:", X_train.shape)
print("X Test Shape :", X_test.shape)

print("\ny Train Shape:", y_train.shape)
print("y Test Shape :", y_test.shape)


# ============================================================
# Target Distribution
# ============================================================

print("\nTarget Distribution:")
print(y.value_counts())


# ============================================================
# Target Distribution Percentage
# ============================================================

print("\nTarget Distribution Percentage:")
print(
    y.value_counts(normalize=True).mul(100).round(2)
)


# ============================================================
# Missing Values Check
# ============================================================

print("\nMissing Values in X_train:")
print(X_train.isna().sum().sum())

print("\nMissing Values in X_test:")
print(X_test.isna().sum().sum())

print("\nMissing Values in y_train:")
print(y_train.isna().sum())

print("\nMissing Values in y_test:")
print(y_test.isna().sum())
````

## Random Forest Model Training & Evaluation

In this step, we train a Random Forest classification model on the training data and evaluate its performance on the test data.

The model is evaluated using a confusion matrix and classification report. We also analyze feature importance to identify the variables that contribute most to the churn prediction.

````python
# ============================================================
# Random Forest Churn Prediction Model
# ============================================================

from sklearn.ensemble import RandomForestClassifier

# Initialize Random Forest model

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight='balanced',
    n_jobs=-1
)

# Train the model

rf_model.fit(
    X_train,
    y_train
)

print("Random Forest Model trained successfully!")
````

````python
# ============================================================
# Random Forest Model Evaluation
# ============================================================

# Make predictions
y_pred = rf_model.predict(X_test)


# ============================================================
# Confusion Matrix
# ============================================================

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))


# ============================================================
# Classification Report
# ============================================================

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=['Stayed', 'Churned']
))
````

````python
# ============================================================
# Model Performance Metrics
# ============================================================

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)

roc_auc = roc_auc_score(
    y_test,
    rf_model.predict_proba(X_test)[:, 1]
)


print("Model Performance")
print("=" * 40)

print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")
print(f"ROC-AUC   : {roc_auc:.4f}")
````

# 🔮 Predict Customer Churn on New Data

- In this step, the trained Random Forest Churn Prediction Model is applied to new customer data to identify customers who are likely to stay or churn. The same preprocessing and saved encoders used during model training are applied to maintain consistency.

- The final predictions are added back to the original customer data, allowing us to identify high-risk churn customers and export them for further business analysis, customer retention, and Power BI reporting.

````python
# ============================================================
# PREDICT CUSTOMER CHURN ON NEW DATA
# ============================================================

# ------------------------------------------------------------
# 1. Load New Customer Data
# ------------------------------------------------------------

# Example for Excel:
# new_data = pd.read_excel("New_Customer_Data.xlsx")

# Example for CSV:
# new_data = pd.read_csv("New_Customer_Data.csv")

# For testing with your current dataset:
new_data = df.copy()

print("New Data Shape:", new_data.shape)

display(new_data.head())


# ============================================================
# 2. Preserve Original Data
# ============================================================

# Keep the original unencoded data
# so that final results remain human-readable.

original_data = new_data.copy()


# ============================================================
# 3. Preserve Customer IDs
# ============================================================

customer_ids = new_data["Customer_ID"].copy()


# ============================================================
# 4. Remove Columns Not Required for Prediction
# ============================================================

new_data = new_data.drop(
    [
        "Customer_ID",
        "Customer_Status",
        "Churn_Category",
        "Churn_Reason",
        "churn_flag",
        "Churn_Status"
    ],
    axis=1,
    errors="ignore"
)


# ============================================================
# 5. Encode Categorical Variables
# ============================================================

# IMPORTANT:
# Use the SAME LabelEncoders used during model training.
# Do NOT fit new encoders here.

for column in columns_to_encode:

    if column in new_data.columns:

        new_data[column] = (
            label_encoders[column]
            .transform(
                new_data[column].astype(str)
            )
        )


# ============================================================
# 6. Match Training Feature Order
# ============================================================

# Ensure new data contains exactly the same
# feature columns used while training the model.

new_data = new_data[X_train.columns]


# ============================================================
# 7. Check Data Before Prediction
# ============================================================

print("\nPrediction Data Shape:", new_data.shape)

print("\nMissing Values:")
print(new_data.isna().sum().sum())


# ============================================================
# 8. Generate Churn Predictions
# ============================================================

new_predictions = rf_model.predict(new_data)


# ============================================================
# 9. Add Numeric Prediction to Original Data
# ============================================================

original_data["Churn_Status_Predicted"] = new_predictions


# ============================================================
# 10. Convert Prediction into Business Labels
# ============================================================

original_data["Churn_Prediction"] = (
    original_data["Churn_Status_Predicted"]
    .map({
        0: "Stayed",
        1: "Churned"
    })
)


# ============================================================
# 11. Display Prediction Summary
# ============================================================

print("\n============================================================")
print("CHURN PREDICTION SUMMARY")
print("============================================================")

print(
    original_data["Churn_Prediction"]
    .value_counts()
)


# ============================================================
# 12. Calculate Churn Percentage
# ============================================================

total_customers = len(original_data)

churned_count = (
    original_data["Churn_Status_Predicted"] == 1
).sum()

churn_rate = (
    churned_count / total_customers
) * 100


print("\nTotal Customers :", total_customers)
print("Predicted Churn :", churned_count)
print(f"Predicted Churn Rate : {churn_rate:.2f}%")


# ============================================================
# 13. Filter Only Predicted Churned Customers
# ============================================================

churned_customers = original_data[
    original_data["Churn_Status_Predicted"] == 1
].copy()


# ============================================================
# 14. Display Churned Customers
# ============================================================

print("\n============================================================")
print("PREDICTED CHURNED CUSTOMERS")
print("============================================================")

print(
    "Number of Churned Customers:",
    len(churned_customers)
)

display(
    churned_customers.head(20)
)


# ============================================================
# 15. Save All Prediction Results
# ============================================================

original_data.to_csv(
    "Customer_Churn_Predictions.csv",
    index=False
)


# ============================================================
# 16. Save Only Churned Customers
# ============================================================

churned_customers.to_csv(
    "Predicted_Churned_Customers.csv",
    index=False
)


print("\n============================================================")
print("FILES SAVED SUCCESSFULLY")
print("============================================================")

print("1. Customer_Churn_Predictions.csv")
print("2. Predicted_Churned_Customers.csv")
````

````python
# ============================================================
# XGBoost Churn Prediction Model
# ============================================================

# If XGBoost is not installed, run once:
# !pip install xgboost

# ============================================================
# Import Libraries
# ============================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

from xgboost import XGBClassifier

import matplotlib.pyplot as plt


# ============================================================
# Copy Dataset
# ============================================================

data = df.copy()


# ============================================================
# Clean Column Names
# ============================================================

data.columns = data.columns.str.strip()


# ============================================================
# Create Target Variable
# Stayed = 0
# Churned = 1
# ============================================================

data["Customer_Status"] = (
    data["Customer_Status"]
    .astype(str)
    .str.strip()
)

data["Churn_Status"] = data["Customer_Status"].map({
    "Stayed": 0,
    "Churned": 1
})


# ============================================================
# Remove Rows Where Target Is Missing
# ============================================================

data = data.dropna(
    subset=["Churn_Status"]
).copy()


# ============================================================
# Convert Target to Integer
# ============================================================

data["Churn_Status"] = data["Churn_Status"].astype(int)


# ============================================================
# Columns NOT Used for Training
# ============================================================

columns_to_drop = [
    "Customer_ID",
    "Customer_Status",
    "Churn_Category",
    "Churn_Reason",
    "churn_flag",
    "Churn_Status_Predicted",
    "Churn_Prediction"
]


# ============================================================
# Drop Existing Prediction / Leakage Columns
# ============================================================

data = data.drop(
    columns=columns_to_drop,
    errors="ignore"
)


# ============================================================
# Separate Features and Target
# ============================================================

X = data.drop(
    "Churn_Status",
    axis=1
)

y = data["Churn_Status"]


# ============================================================
# Identify Categorical Columns
# ============================================================

categorical_columns = X.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("Categorical Columns:")
print(categorical_columns)


# ============================================================
# Label Encoding
# ============================================================

label_encoders = {}

for column in categorical_columns:

    label_encoders[column] = LabelEncoder()

    X[column] = label_encoders[column].fit_transform(
        X[column].astype(str)
    )


# ============================================================
# Handle Missing Values in Features
# ============================================================

for column in X.columns:

    if X[column].dtype == "object":

        X[column] = X[column].fillna(
            X[column].mode()[0]
        )

    else:

        X[column] = X[column].fillna(
            X[column].median()
        )


# ============================================================
# Final Data Check
# ============================================================

print("\nFinal X Shape:")
print(X.shape)

print("\nFinal Target Distribution:")
print(y.value_counts())

print("\nMissing Values:")
print(X.isna().sum().sum())


# ============================================================
# Train-Test Split
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# ============================================================
# Calculate Class Imbalance Ratio
# ============================================================

negative_count = (y_train == 0).sum()
positive_count = (y_train == 1).sum()

scale_pos_weight = negative_count / positive_count

print("\nScale Pos Weight:")
print(scale_pos_weight)


# ============================================================
# Create XGBoost Model
# ============================================================

xgb_model = XGBClassifier(

    n_estimators=300,

    max_depth=5,

    learning_rate=0.05,

    subsample=0.8,

    colsample_bytree=0.8,

    objective="binary:logistic",

    eval_metric="logloss",

    scale_pos_weight=scale_pos_weight,

    random_state=42,

    n_jobs=-1
)


# ============================================================
# Train Model
# ============================================================

xgb_model.fit(
    X_train,
    y_train
)

print("\nXGBoost Model trained successfully! ✅")


# ============================================================
# Predictions
# ============================================================

y_pred = xgb_model.predict(
    X_test
)


# ============================================================
# Prediction Probability
# ============================================================

y_probability = xgb_model.predict_proba(
    X_test
)[:, 1]


# ============================================================
# Accuracy
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\nXGBoost Accuracy:")
print(f"{accuracy:.4f}")


# ============================================================
# Confusion Matrix
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\nConfusion Matrix:")
print(cm)


# ============================================================
# Plot Confusion Matrix
# ============================================================

plt.figure(figsize=(7, 6))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Stayed", "Churned"]
)

disp.plot(
    cmap="Blues",
    values_format="d"
)

plt.title(
    "XGBoost - Customer Churn Confusion Matrix",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()
plt.show()


# ============================================================
# Classification Report
# ============================================================

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=[
            "Stayed",
            "Churned"
        ],
        digits=2
    )
)
````

````python
# ============================================================
# XGBoost Feature Importance
# ============================================================

feature_importance_df = pd.DataFrame({

    "Feature": X.columns,

    "Importance": xgb_model.feature_importances_

})


# Sort by Importance

feature_importance_df = (
    feature_importance_df
    .sort_values(
        by="Importance",
        ascending=False
    )
    .reset_index(drop=True)
)


# Top 20 Features

top_features = (
    feature_importance_df
    .head(20)
    .sort_values(
        by="Importance",
        ascending=True
    )
)


# ============================================================
# Plot
# ============================================================

plt.figure(figsize=(10, 7))

plt.barh(
    top_features["Feature"],
    top_features["Importance"],
    color="royalblue"
)

plt.xlabel(
    "Feature Importance"
)

plt.ylabel(
    "Features"
)

plt.title(
    "Top 20 Features Influencing Customer Churn - XGBoost",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
````

````python
# ============================================================
# Save XGBoost Predictions
# ============================================================

xgb_predictions = X_test.copy()

xgb_predictions["Actual_Churn_Status"] = y_test.values

xgb_predictions["Predicted_Churn_Status"] = y_pred

xgb_predictions["Churn_Probability"] = y_probability

xgb_predictions["Churn_Prediction"] = np.where(
    y_pred == 1,
    "Churned",
    "Stayed"
)

print(xgb_predictions.head())
````
