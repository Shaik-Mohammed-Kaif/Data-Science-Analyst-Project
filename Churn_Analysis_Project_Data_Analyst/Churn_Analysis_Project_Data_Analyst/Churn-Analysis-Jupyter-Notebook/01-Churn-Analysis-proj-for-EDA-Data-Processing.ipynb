# 📊 Customer Churn Analysis & Customer Intelligence

## 📌 Project Overview

Customer churn refers to the phenomenon where customers stop using a company's product or service over a specific period. In highly competitive industries such as OTT platforms, telecommunications, banking, SaaS, e-commerce, and subscription-based businesses, retaining existing customers is often more cost-effective than acquiring new ones.

This project focuses on analyzing customer behavior, identifying churn patterns, and discovering the key factors that influence customer attrition. By leveraging SQL, Python, and data visualization techniques, the project aims to generate actionable business insights that help organizations improve customer retention and reduce revenue loss.

---

# 🎯 Business Problem

Customer retention is one of the biggest challenges for subscription-based businesses. Every lost customer directly impacts revenue, customer lifetime value (CLTV), and long-term business growth.

The objective of this project is to answer three fundamental business questions:

- **Who** are the customers likely to churn?
- **Why** are they leaving?
- **When** is the highest risk period for churn?

Answering these questions enables businesses to proactively engage at-risk customers through personalized retention strategies.

## Automation Script for Large - Scale of Dataset Handles for Filling Values in Auomate way

````python
# ==========================================================
# Automate Script for Categorical/Object Missing Values
# ==========================================================

# Custom replacements
custom_fill = {
    "Value_Deal": "Not Deal",
    "Internet_Type": "No Type",
    "Churn_Category": "Not Applicable",
    "Churn_Reason": "Not Applicable"
}

# Apply custom replacements
for col, value in custom_fill.items():
    obj_val[col] = obj_val[col].fillna(value)


# Remaining categorical/object columns → "No"
remaining_obj_cols = [
    col for col in obj_val.columns
    if col not in custom_fill
]

obj_val[remaining_obj_cols] = obj_val[remaining_obj_cols].fillna("No")


# ==========================================================
# Verify Remaining Null Values
# ==========================================================

print("Remaining Null Values in Categorical Columns:")

display(obj_val.isnull().sum())
````

````python
print("""
========================================
       CUSTOMER CHURN VISUALIZATION
========================================

1. Bar Plot
2. Scatter Plot
3. Histogram
4. Area Plot
5. Horizontal Bar Plot
6. KDE Plot
7. Pie
8. Exit
""")


choice = input("Enter choice: ")


# ======================================
# COMMON DATA
# ======================================

state_churn = (
    pd.crosstab(
        df["State"],
        df["Customer_Status"],
        normalize="index"
    ) * 100
)


# ======================================
# 1. BAR PLOT
# ======================================

if choice == "1":

    display(state_churn.head(15))

    state_churn.head(15).plot(
        kind="bar",
        figsize=(12, 6)
    )

    plt.title("Customer Status by State")
    plt.ylabel("Percentage (%)")
    plt.xlabel("State")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


# ======================================
# 2. SCATTER PLOT
# ======================================

elif choice == "2":

    # Churn column select karo
    if "Churned" in state_churn.columns:

        scatter_data = state_churn[["Churned"]].head(15).reset_index()

        scatter_data["State_Number"] = range(
            1,
            len(scatter_data) + 1
        )

        display(scatter_data)

        scatter_data.plot(
            kind="scatter",
            x="State_Number",
            y="Churned",
            figsize=(12, 6)
        )

        plt.title("Top States by Churn Rate")
        plt.ylabel("Churn Rate (%)")
        plt.xlabel("State Number")
        plt.tight_layout()
        plt.show()

    else:
        print("❌ 'Churned' status not found.")


# ======================================
# 3. HISTOGRAM
# ======================================

elif choice == "3":

    if "Churned" in state_churn.columns:

        state_churn["Churned"].plot(
            kind="hist",
            bins=10,
            figsize=(10, 6)
        )

        plt.title("Distribution of State Churn Rate")
        plt.xlabel("Churn Rate (%)")
        plt.ylabel("Number of States")
        plt.tight_layout()
        plt.show()

    else:
        print("❌ 'Churned' status not found.")


# ======================================
# 4. AREA PLOT
# ======================================

elif choice == "4":

    state_churn.head(15).plot(
        kind="area",
        figsize=(12, 6),
        alpha=0.7
    )

    plt.title("Customer Status Percentage by State")
    plt.ylabel("Percentage (%)")
    plt.xlabel("State")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


# ======================================
# 5. HORIZONTAL BAR
# ======================================

elif choice == "5":

    if "Churned" in state_churn.columns:

        state_churn["Churned"].head(15).sort_values().plot(
            kind="barh",
            figsize=(10, 7)
        )

        plt.title("Top States by Churn Rate")
        plt.xlabel("Churn Rate (%)")
        plt.ylabel("State")
        plt.tight_layout()
        plt.show()

    else:
        print("❌ 'Churned' status not found.")


# ======================================
# 6. KDE PLOT
# ======================================

elif choice == "6":

    if "Churned" in state_churn.columns:

        state_churn["Churned"].plot(
            kind="kde",
            figsize=(10, 6)
        )

        plt.title("KDE Distribution of State Churn Rate")
        plt.xlabel("Churn Rate (%)")
        plt.tight_layout()
        plt.show()

    else:
        print("❌ 'Churned' status not found.")

# ======================================
# 7. Pie
# ======================================

elif choice == "7":
    plt.pie(
    df["Gender"].value_counts(),
    labels=df["Gender"].value_counts().index,
    autopct="%1.1f%%"
    )

    plt.show()


# ======================================
# 8. EXIT
# ======================================

elif choice == "8":

    print("Thank you! 👋")


# ======================================
# INVALID CHOICE
# ======================================

else:

    print("❌ Invalid choice!")
````

# 📊 STATISTICS

## 1. What is Statistics?

**Statistics** is the science of **collecting, organizing, summarizing, analyzing, interpreting, and presenting data** to discover patterns, understand uncertainty, and support decision-making.

### Simple Definition

> **Statistics is the process of understanding data and using it to make meaningful conclusions.**

### Example

Suppose a company has data for **10,000 customers**.

Statistics can help answer:

* What is the average customer age?
* What is the average monthly charge?
* How much variation exists in customer spending?
* What percentage of customers have churned?
* Are there unusual or extreme values?
* Can a sample of customers tell us something about the entire customer population?

---

# 🧠 2. Main Types of Statistics

Statistics is broadly divided into **two major branches**:

```text
                         STATISTICS
                             │
              ┌──────────────┴──────────────┐
              ↓                             ↓
      DESCRIPTIVE STATISTICS        INFERENTIAL STATISTICS
              │                             │
       "What does the data          "What can we conclude
        tell us?"                   about the population?"
```

---

# 1️⃣ DESCRIPTIVE STATISTICS

## Definition

**Descriptive Statistics** is the branch of statistics that focuses on **collecting, organizing, summarizing, and presenting data** in a meaningful and understandable form.

### Simple Definition

> **Descriptive Statistics summarizes and describes the data that we already have.**

It does not, by itself, attempt to generalize the results from a sample to an entire population.

### Example

Suppose we have data for **1,000 customers**.

We calculate:

```text
Average Age            = 35 years
Median Age             = 34 years
Minimum Age            = 18 years
Maximum Age            = 72 years
Standard Deviation     = 8.5
Churn Rate             = 22%
```

These are examples of **descriptive statistics** because they describe the observed dataset.

---

# 🏗️ Descriptive Statistics Structure

```text
                 DESCRIPTIVE STATISTICS
                         │
        ┌────────────────┼─────────────────┐
        ↓                ↓                 ↓
 CENTRAL TENDENCY    DISPERSION        DISTRIBUTION
        │                │                 │
        ↓                ↓                 ↓
      Mean             Range            Skewness
      Median           Variance         Kurtosis
      Mode             Standard
                        Deviation
                       IQR
                       CV
                         │
                         ↓
                  MEASURES OF POSITION
                         │
                    Percentile
                    Quartile
                    Z-Score
```

---

# A. Measures of Central Tendency

## Definition

**Measures of Central Tendency** describe the **central or typical value** of a dataset.

They answer:

> **"Where is the center of the data?"**

Main measures:

```text
Central Tendency
      │
      ├── Mean
      ├── Median
      └── Mode
```

---

## 1. Mean

### Definition

**Mean** is the arithmetic average of all observations in a dataset.

### Formula

```text
Mean = Sum of all values / Number of values
```

### Example

```text
10, 20, 30, 40, 50

Mean = 30
```

### Important Point

Mean can be strongly affected by **extreme values or outliers**.

---

## 2. Median

### Definition

**Median** is the middle value of a dataset after the observations have been arranged in ascending or descending order.

### Example — Odd Number of Values

```text
10, 20, 30, 40, 50

Median = 30
```

### Example — Even Number of Values

```text
10, 20, 30, 40

Middle values = 20, 30

Median = (20 + 30) / 2

Median = 25
```

### Important Point

Median is generally **less affected by extreme values than the mean**.

Therefore, it is often useful for highly skewed data such as:

* Income
* House prices
* Customer spending

---

## 3. Mode

### Definition

**Mode** is the value or category that occurs **most frequently** in a dataset.

### Example

```text
10, 20, 20, 30, 40

Mode = 20
```

Mode is particularly useful for:

* Categorical data
* Most common category
* Most frequently occurring value

---

# B. Measures of Dispersion

## Definition

**Measures of Dispersion** describe how much the observations are **spread out or varied** around their central value.

### Simple Definition

> **Dispersion tells us how spread out the data is.**

Central tendency tells us:

```text
Where is the center?
```

Dispersion tells us:

```text
How widely are the values spread?
```

### Main Measures

```text
                 DISPERSION
                     │
       ┌─────────────┼─────────────┐
       ↓             ↓             ↓
     Range        Variance         IQR
                     │
                     ↓
              Standard Deviation
```

---

# 1. Range

### Definition

**Range** is the difference between the maximum and minimum values.

### Formula

```text
Range = Maximum Value - Minimum Value
```

### Example

```text
10, 20, 30, 40, 50

Range = 50 - 10
      = 40
```

### Limitation

Range only uses the **minimum and maximum values**, so it can be heavily affected by extreme observations.

---

# 2. Variance

### Definition

**Variance** measures the average squared deviation of observations from the mean.

### Simple Definition

> **Variance measures how much the data varies around its mean using squared deviations.**

### Interpretation

```text
Higher Variance
      ↓
Greater spread

Lower Variance
      ↓
Less spread
```

Variance is expressed in **squared units** of the original variable.

---

# 3. Standard Deviation

### Definition

**Standard Deviation (SD)** measures the typical spread of observations around the mean and is expressed in the **same units as the original data**.

### Simple Definition

> **Standard deviation tells us approximately how far observations tend to be from the mean.**

### Interpretation

```text
Higher Standard Deviation
          ↓
      More spread

Lower Standard Deviation
          ↓
     More concentrated
```

### Example

If customer ages have:

```text
Mean Age = 35 years
SD = 5 years
```

then the data has a typical spread of roughly **5 years around the mean**, although SD should not be interpreted as saying every observation lies exactly 5 years from the mean.

---

# 4. IQR — Interquartile Range

### Definition

**Interquartile Range (IQR)** measures the spread of the **middle 50% of observations**.

### Formula

```text
IQR = Q3 - Q1
```

Where:

```text
Q1 = 25th Percentile
Q3 = 75th Percentile
```

### Why is IQR useful?

IQR is relatively resistant to extreme values and is commonly used for:

* Measuring spread
* Box plots
* Detecting potential outliers

---

# C. Measures of Distribution Shape

These measures help us understand the **shape of a distribution**.

```text
Distribution Shape
       │
       ├── Skewness
       └── Kurtosis
```

---

# 1. Skewness

### Definition

**Skewness** measures the **asymmetry of a probability distribution** around its center.

### Simple Definition

> **Skewness tells us whether a distribution is approximately symmetric or has a longer tail on one side.**

There are three common cases:

```text
Skewness
    │
    ├── Negative Skew
    ├── Approximately Symmetric
    └── Positive Skew
```

### Positive Skewness

The distribution has a **longer right tail**.

```text
         /
        / \
       /   \________
```

### Negative Skewness

The distribution has a **longer left tail**.

```text
 ________/
       / \
      /   \
```

### Approximately Symmetric

Both sides are relatively balanced.

```text
       /\
      /  \
     /    \
    /      \
```

---

# 2. Kurtosis

### Definition

**Kurtosis** describes the **tail heaviness of a distribution**, particularly its tendency to produce extreme observations, relative to a reference distribution.

### Simple Definition

> **Kurtosis helps us understand the behavior of the tails and extreme values of a distribution.**

In many statistical software packages, including pandas, `.kurt()` reports **excess kurtosis**.

For excess kurtosis:

```text
Positive  → Heavier tails
≈ 0       → Similar tail behavior to normal distribution
Negative  → Lighter tails
```

### Important

Kurtosis should primarily be interpreted in terms of **tail behavior and extreme observations**, rather than simply saying "peak height."

---

# D. Measures of Position

## Definition

**Measures of Position** describe where an observation lies within a dataset or distribution.

```text
Measures of Position
        │
        ├── Percentile
        ├── Quartile
        └── Z-Score
```

---

# 1. Percentile

### Definition

A **percentile** indicates the position of a value relative to the observations in a dataset.

For example:

> A value at the **90th percentile** is greater than or equal to approximately 90% of the observations, depending on the percentile definition being used.

### Example

Suppose a student's score is at the:

```text
90th Percentile
```

This means the score is at or above roughly **90% of the observed scores**.

---

# 2. Quartile

### Definition

**Quartiles** divide ordered data into four sections based on percentiles.

```text
Q1 → 25th Percentile
Q2 → 50th Percentile
Q3 → 75th Percentile
```

And:

```text
Q2 = Median
```

### Structure

```text
0%──────25%──────50%──────75%──────100%
          Q1       Q2        Q3
                   ↑
                 Median
```

Quartiles are especially important for:

* Understanding distribution
* Measuring IQR
* Creating box plots
* Identifying potential outliers

---

# 3. Z-Score

### Definition

A **Z-score** indicates how many standard deviations an observation is away from the mean.

### Formula

```text
Z = (X - Mean) / Standard Deviation
```

### Interpretation

```text
Z = 0
→ Observation is at the mean

Z = +1
→ 1 standard deviation above the mean

Z = -1
→ 1 standard deviation below the mean

Z = +2
→ 2 standard deviations above the mean

Z = -2
→ 2 standard deviations below the mean
```

### Uses

Z-score is useful for:

* Standardization
* Comparing values on different scales
* Identifying unusual observations
* Statistical analysis

---

# 2️⃣ INFERENTIAL STATISTICS

## Definition

**Inferential Statistics** is the branch of statistics that uses **sample data to make estimates, predictions, or conclusions about a larger population**.

### Simple Definition

> **Inferential Statistics uses a sample to learn about a population.**

### Example

```text
Population
1,000,000 Customers
       ↓
Select a Sample
5,000 Customers
       ↓
Analyze the Sample
       ↓
Make an Inference
about the Population
```

---

# 🏗️ Inferential Statistics Structure

```text
                  INFERENTIAL STATISTICS
                           │
             ┌─────────────┼──────────────┐

          Sampling     Estimation    Hypothesis

             │             │              │

        Sample Data    Confidence      H₀ / H₁
                       Intervals       p-values
                           │
                           ↓
                    Statistical Tests
```

---

# A. Sampling

## Definition

**Sampling** is the process of selecting a subset of observations from a larger population for analysis.

```text
Population
     ↓
  Sampling
     ↓
  Sample
```

The goal is usually to obtain a sample that provides useful information about the population.

---

# B. Estimation

## Definition

**Estimation** uses sample information to estimate an unknown population parameter.

Two major types:

```text
Estimation
    │
    ├── Point Estimate
    └── Interval Estimate
```

---

## 1. Point Estimate

### Definition

A **point estimate** is a single numerical value used to estimate an unknown population parameter.

Example:

```text
Sample Mean = 35
```

The sample mean can be used as a point estimate of the population mean.

---

## 2. Confidence Interval

### Definition

A **confidence interval** provides a range of plausible values for an unknown population parameter, based on sample data and a specified confidence procedure.

Example:

```text
Estimated Mean = 35

95% Confidence Interval
= [34.2, 35.8]
```

The exact interpretation of a 95% confidence interval is based on the repeated-sampling procedure used to construct it; it should not simply be interpreted as a 95% probability that this particular fixed parameter lies inside the calculated interval.

---

# C. Hypothesis Testing

## Definition

**Hypothesis Testing** is a statistical method used to evaluate a claim about a population using evidence from sample data.

### Basic Structure

```text
Research Question
       ↓
Null Hypothesis (H₀)
       ↓
Alternative Hypothesis (H₁)
       ↓
Choose Statistical Test
       ↓
Calculate Test Statistic
       ↓
Calculate p-value
       ↓
Make Statistical Decision
       ↓
Interpret the Result
```

---

# Common Statistical Tests

```text
                 STATISTICAL TESTS
                        │
        ┌───────────────┼────────────────┐
        ↓               ↓                ↓
   Parametric      Categorical      Non-Parametric
        │               │                │
        ├── t-test      └── Chi-Square   ├── Mann-Whitney
        ├── z-test                       ├── Wilcoxon
        ├── ANOVA                         ├── Kruskal-Wallis
        └── F-test                        └── Spearman/Kendall
```

These tests are selected based on factors such as:

* Type of variables
* Research question
* Sample design
* Distributional assumptions
* Independence
* Measurement scale

---

# 🔥 Descriptive vs Inferential Statistics

| Descriptive Statistics     | Inferential Statistics                       |
| -------------------------- | -------------------------------------------- |
| Describes observed data    | Makes inferences about a population          |
| Summarizes the dataset     | Uses sample evidence                         |
| Mean, Median, Mode         | Hypothesis Testing                           |
| Range, Variance, SD        | Confidence Intervals                         |
| Skewness, Kurtosis         | t-test, z-test                               |
| Tables and Graphs          | p-values and test statistics                 |
| "What does the data show?" | "What can we conclude about the population?" |

# 📊 Inferential Statistics

## Definition

**Inferential Statistics** is the branch of statistics that uses **sample data to make estimates, predictions, or conclusions about a larger population**.

### Simple Definition

> **Inferential Statistics = Analyze a sample → use the evidence to make a conclusion about the population.**

---

# 🧠 First Understand Population and Sample

Before understanding inferential statistics, these two concepts are very important.

### Population

**Population** means the **complete group of individuals, objects, or observations** that we are interested in studying.

Example:

```text
All customers of a telecom company
= Population
```

Suppose the company has:

```text
1,000,000 customers
```

Then:

```text
Population = 1,000,000 customers
```

---

### Sample

**Sample** is a **subset of the population selected for analysis**.

Suppose we select:

```text
5,000 customers
```

from the 1,000,000 customers.

Then:

```text
Population = 1,000,000 customers
Sample     = 5,000 customers
```

---

# 🔥 Main Idea of Inferential Statistics

```text
                 POPULATION
              1,000,000 Customers
                       │
                       │ Select a sample
                       ↓
                    SAMPLE
                  5,000 Customers
                       │
                       ↓
                Analyze the Sample
                       │
                       ↓
              Statistical Inference
                       │
                       ↓
          Conclusion about Population
```

This is the core idea of **Inferential Statistics**.

---

# 🎯 Why Do We Need Inferential Statistics?

In many real-world situations, studying the entire population is:

* Expensive
* Time-consuming
* Difficult
* Sometimes impossible

So instead of analyzing everyone, we analyze a **sample**.

### Example

Imagine you want to know:

> **What is the average monthly spending of all customers in India?**

There could be millions of customers.

You may select a representative sample:

```text
Population
Millions of customers
        ↓
Sample
10,000 customers
        ↓
Analyze spending
        ↓
Estimate population average
```

That is **Inferential Statistics**.

---

# 🏗️ Structure of Inferential Statistics

```text
                 INFERENTIAL STATISTICS
                          │
          ┌───────────────┼────────────────┐
          ↓               ↓                ↓
       Sampling       Estimation      Hypothesis
                                         Testing
          │               │                │
          ↓               ↓                ↓
      Sample Data     Point Estimate    H₀ / H₁
                      Confidence        p-value
                      Interval
                          │
                          ↓
                   Statistical Tests
```

---

# 1️⃣ Sampling

## Definition

**Sampling** is the process of selecting a subset of observations from a population for analysis.

```text
Population
     ↓
Sampling
     ↓
Sample
```

The quality of the sample matters because a biased or unrepresentative sample can lead to misleading conclusions.

---

# 2️⃣ Estimation

## Definition

**Estimation** uses sample data to estimate an unknown characteristic of the population.

There are two major types:

```text
Estimation
    │
    ├── Point Estimation
    │
    └── Interval Estimation
```

---

## Point Estimation

A **point estimate** gives a **single value** as an estimate of a population parameter.

Example:

```text
Sample Mean = ₹2,500
```

We may use ₹2,500 as an estimate of the population's average spending.

```text
Sample
   ↓
Sample Mean
   ↓
Estimate Population Mean
```

---

## Interval Estimation

An **interval estimate** provides a **range of plausible values** for a population parameter.

The most common example is a **confidence interval**.

Example:

```text
Estimated Average Spending = ₹2,500

95% Confidence Interval
= ₹2,400 to ₹2,600
```

The interval provides information about the uncertainty associated with the estimate.

---

# 3️⃣ Hypothesis Testing

## Definition

**Hypothesis Testing** is a statistical procedure used to evaluate a claim about a population using sample evidence.

### Example

A telecom company claims:

> "Our average customer satisfaction score is 80."

We collect a sample of customers and test whether the sample provides sufficient evidence against that claim.

---

# 🧠 Hypothesis Testing Structure

```text
Research Question
       ↓
Null Hypothesis (H₀)
       ↓
Alternative Hypothesis (H₁)
       ↓
Choose Statistical Test
       ↓
Collect / Analyze Sample Data
       ↓
Calculate Test Statistic
       ↓
Calculate p-value
       ↓
Decision
       ↓
Conclusion
```

---

# H₀ — Null Hypothesis

The **Null Hypothesis (H₀)** represents the default claim or a statement of no effect/no difference, depending on the test.

Example:

```text
H₀:
Average customer satisfaction = 80
```

---

# H₁ — Alternative Hypothesis

The **Alternative Hypothesis (H₁ or Ha)** represents the competing claim that there is an effect, difference, or relationship.

Example:

```text
H₁:
Average customer satisfaction ≠ 80
```

---

# 4️⃣ p-value

## Definition

The **p-value** is the probability, assuming the null hypothesis is true, of obtaining a test result **at least as extreme as the observed result**, according to the statistical test.

### Important

A p-value is **not**:

```text
❌ Probability that H₀ is true
❌ Probability that H₁ is true
❌ Proof that a result is important
```

### Common Decision Rule

Using:

```text
α = 0.05
```

we often use:

```text
p-value < 0.05
        ↓
Reject H₀
```

and:

```text
p-value ≥ 0.05
        ↓
Fail to reject H₀
```

This is a statistical decision rule, not proof that the null hypothesis is true or false.

---

# 5️⃣ Statistical Tests

Inferential statistics includes many statistical tests.

```text
                 STATISTICAL TESTS
                        │
       ┌────────────────┼─────────────────┐
       ↓                ↓                 ↓
  Parametric       Categorical      Non-Parametric
       │                │                 │
       ├── t-test       └── Chi-Square    ├── Mann-Whitney
       ├── z-test                          ├── Wilcoxon
       ├── ANOVA                            ├── Kruskal-Wallis
       └── F-test                           ├── Spearman
                                            └── Kendall
```

The correct test depends on:

* Type of variables
* Research question
* Sample design
* Number of groups
* Distributional assumptions
* Independence of observations
* Measurement scale

---

# 🔥 Descriptive vs Inferential

| Descriptive Statistics         | Inferential Statistics                 |
| ------------------------------ | -------------------------------------- |
| Describes observed data        | Makes conclusions about a population   |
| Summarizes existing data       | Uses sample evidence                   |
| Mean                           | Confidence Interval                    |
| Median                         | Hypothesis Testing                     |
| Mode                           | p-value                                |
| Standard Deviation             | t-test                                 |
| Variance                       | z-test                                 |
| Skewness                       | Chi-Square                             |
| Kurtosis                       | ANOVA                                  |
| Tables and Graphs              | Statistical Tests                      |
| **"What does the data show?"** | **"What can we infer from the data?"** |

---

# 🧠 One Real-World Example

Suppose a company has:

```text
Population
100,000 customers
```

You cannot practically analyze every customer.

So you select:

```text
Sample
2,000 customers
```

You calculate:

```text
Sample Average Monthly Charge
= ₹1,850
```

Then you build a confidence interval:

```text
95% CI
= ₹1,810 – ₹1,890
```

You could also test a hypothesis such as:

```text
H₀: Average monthly charge = ₹1,800

H₁: Average monthly charge ≠ ₹1,800
```

Now you're using **sample evidence to learn about the population**.

That is **Inferential Statistics**.

---

# 🎯 Final Definition to Remember

> **Inferential Statistics is the branch of statistics that uses sample data, probability, estimation, and hypothesis testing to draw conclusions or make predictions about a larger population while accounting for uncertainty.**

### Easy Memory Trick

```text
DESCRIPTIVE
    ↓
Describe the data
    ↓
"What happened?"

INFERENTIAL
    ↓
Infer from the data
    ↓
"What can we conclude?"
```

**Descriptive = Describe what you observed.**
**Inferential = Use a sample to learn about the larger population.**

````python
churned_tenure = df.loc[
    df["Customer_Status"] == "Churned",
    "Tenure_in_Months"
].dropna()

stayed_tenure = df.loc[
    df["Customer_Status"] == "Stayed",
    "Tenure_in_Months"
].dropna()

def mean_difference_ci(
    group1,
    group2,
    confidence=0.95
):

    group1 = group1.dropna()
    group2 = group2.dropna()

    mean1 = group1.mean()
    mean2 = group2.mean()

    var1 = group1.var()
    var2 = group2.var()

    n1 = len(group1)
    n2 = len(group2)

    difference = mean1 - mean2

    se = np.sqrt(
        var1 / n1 +
        var2 / n2
    )

    df_welch = (
        (var1 / n1 + var2 / n2) ** 2
        /
        (
            (var1 / n1) ** 2 / (n1 - 1)
            +
            (var2 / n2) ** 2 / (n2 - 1)
        )
    )

    alpha = 1 - confidence

    t_critical = stats.t.ppf(
        1 - alpha / 2,
        df_welch
    )

    margin_error = t_critical * se

    lower = difference - margin_error
    upper = difference + margin_error

    return {
        "Group_1_Mean": mean1,
        "Group_2_Mean": mean2,
        "Mean_Difference": difference,
        "CI_Lower": lower,
        "CI_Upper": upper,
        "Confidence_Level": confidence
    }

result = mean_difference_ci(
    churned_tenure,
    stayed_tenure
)

display(
    pd.DataFrame([result]).round(3)
)
````

````python
def chi_square_test(
    df,
    categorical_col1,
    categorical_col2
):

    table = pd.crosstab(
        df[categorical_col1],
        df[categorical_col2]
    )

    chi2, p_value, dof, expected = chi2_contingency(
        table
    )

    return {
        "Variable 1": categorical_col1,
        "Variable 2": categorical_col2,
        "Chi-Square": chi2,
        "P-Value": p_value,
        "Degrees of Freedom": dof,
        "Significant": p_value < 0.05
    }
````

````python
def cohens_d(group1, group2):

    n1 = len(group1)
    n2 = len(group2)

    var1 = group1.var()
    var2 = group2.var()

    pooled_std = np.sqrt(
        (
            (n1 - 1) * var1 +
            (n2 - 1) * var2
        )
        /
        (n1 + n2 - 2)
    )

    return (
        (group1.mean() - group2.mean())
        / pooled_std
    )

d = cohens_d(
    churned,
    stayed
)

print("Cohen's d:", d)
````

````python
def cramers_v(x, y):

    table = pd.crosstab(x, y)

    chi2 = chi2_contingency(table)[0]

    n = table.sum().sum()

    phi2 = chi2 / n

    r, k = table.shape

    phi2corr = max(
        0,
        phi2 - ((k - 1) * (r - 1)) / (n - 1)
    )

    rcorr = r - (
        (r - 1) ** 2 / (n - 1)
    )

    kcorr = k - (
        (k - 1) ** 2 / (n - 1)
    )

    return np.sqrt(
        phi2corr /
        min(
            kcorr - 1,
            rcorr - 1
        )
    )

v = cramers_v(
    df["Contract"],
    df["Customer_Status"]
)

print("Cramer's V:", v)
````

# 🔥 Feature Engineering

## Definition

**Feature Engineering** is the process of **creating, transforming, or selecting useful input variables (features) from existing raw data** so that the data becomes more meaningful for analysis and machine learning.

### Simple Definition

> **Feature Engineering = Raw Data → Useful Business Features**

---

## 💼 Business Perspective

Feature engineering converts **raw customer information into business-relevant signals** that help answer questions and make better decisions.

For example:

```text
Raw Data
   ↓
Age
Tenure
Monthly Charge
Total Revenue
Services
   ↓
Feature Engineering
   ↓
Customer Segment
Customer Lifetime Value
Service Usage
Revenue per Month
Tenure Group
   ↓
Business Insight
   ↓
Churn Risk / Customer Value / Retention Strategy
```

### Example

Raw features:

```text
Monthly_Charge = ₹1,500
Total_Revenue = ₹18,000
Tenure = 12 months
```

Engineered feature:

```text
Average_Monthly_Revenue = Total Revenue / Tenure
```

This creates a more meaningful business metric for understanding **customer value and spending behavior**.

---

## 🎯 Why Feature Engineering?

Feature engineering helps us:

* Understand customers better
* Create meaningful business metrics
* Find hidden patterns
* Improve model performance
* Reduce unnecessary information
* Support better business decisions

### Core Idea

```text
RAW DATA
   ↓
TRANSFORM
   ↓
CREATE MEANINGFUL FEATURES
   ↓
BUSINESS INSIGHT
   ↓
BETTER DECISION
```

**In short: Feature Engineering is the bridge between raw data and useful business intelligence.**

````python
num_v = df.select_dtypes(include="number")

# ----------------------------------------------------------
# Basic Calculations
# ----------------------------------------------------------

q1 = num_v.quantile(0.25)
q2 = num_v.quantile(0.50)
q3 = num_v.quantile(0.75)

iqr = q3 - q1

mean_abs_deviation = num_v.apply(
    lambda x: (x - x.mean()).abs().mean()
)

# Z-Score
z_score = num_v.apply(
    lambda x: (x - x.mean()) / x.std()
)

# Coefficient of Variation
cv = num_v.std() / num_v.mean().replace(0, np.nan)

# IQR Outlier Count
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outlier_count = pd.Series({
    col: (
        (num_v[col] < lower_bound[col]) |
        (num_v[col] > upper_bound[col])
    ).sum()
    for col in num_v.columns
})

# ----------------------------------------------------------
# Summary Table
# ----------------------------------------------------------

summary_tables = {

    # 01–03 Basic
    "Count": num_v.count(),
    "Size": pd.Series(
        {col: num_v[col].size for col in num_v.columns}
    ),
    "Unique": num_v.nunique(),

    # 04–09 Central / Basic Statistics
    "Min": num_v.min(),
    "Max": num_v.max(),
    "Sum": num_v.sum(),
    "Mean": num_v.mean(),
    "Median": num_v.median(),
    "Mode": num_v.mode().iloc[0],

    # 10–15 Dispersion
    "Standard_Deviation": num_v.std(),
    "Variance": num_v.var(),
    "SEM": num_v.sem(),
    "MAD": mean_abs_deviation,
    "Product": num_v.prod(),
    "Absolute_Mean": num_v.abs().mean(),

    # 16–19 Range / Quantiles
    "Range": num_v.max() - num_v.min(),
    "Q1_25%": q1,
    "Q2_50%": q2,
    "Q3_75%": q3,
    "IQR": iqr,

    # 20–21 Distribution Shape
    "Skewness": num_v.skew(),
    "Kurtosis": num_v.kurt(),

    # 22–25 Cumulative
    "Cumsum_Final": num_v.cumsum().iloc[-1],
    "Cumprod_Final": num_v.cumprod().iloc[-1],
    "Cummin_Final": num_v.cummin().iloc[-1],
    "Cummax_Final": num_v.cummax().iloc[-1],

    # 26–30 Change / Ranking
    "Diff_Mean": num_v.diff().mean(),
    "Pct_Change_Mean": num_v.pct_change().mean(),
    "Rank_Mean": num_v.rank().mean(),

    "Largest_Value": num_v.max(),
    "Smallest_Value": num_v.min(),

    # 31 Describe
    "Describe_Count": num_v.count(),
    "Describe_Mean": num_v.mean(),
    "Describe_Std": num_v.std(),

    # 32–38 Missing / Unique
    "Value_Counts_Unique": num_v.nunique(),
    "Unique_Count": num_v.nunique(),
    "Missing": num_v.isna().sum(),
    "Not_Missing": num_v.notna().sum(),
    "Null_Count": num_v.isnull().sum(),
    "NotNull_Count": num_v.notnull().sum(),
    "Mode_Value": num_v.mode().iloc[0],

    # 39–40 Value Conditions
    "Clipped_Count": pd.Series({
        col: (
            num_v[col].clip(
                lower=lower_bound[col],
                upper=upper_bound[col]
            ) != num_v[col]
        ).sum()
        for col in num_v.columns
    }),

    "Between_Q1_Q3_Count": pd.Series({
        col: num_v[col].between(
            q1[col],
            q3[col]
        ).sum()
        for col in num_v.columns
    }),

    # 41–43 Relationships
    "Correlation_Mean": num_v.corr().mean(),
    "Covariance_Mean": num_v.cov().mean(),
    "Autocorrelation": pd.Series({
        col: num_v[col].autocorr()
        for col in num_v.columns
    }),

    # 44–47 Rolling / Expanding
    "Rolling_Mean_Final": num_v.rolling(
        window=3
    ).mean().iloc[-1],

    "Rolling_Std_Final": num_v.rolling(
        window=3
    ).std().iloc[-1],

    "Expanding_Mean_Final": num_v.expanding().mean().iloc[-1],

    "Expanding_Std_Final": num_v.expanding().std().iloc[-1],

    # 48–50 Advanced
    "Z_Score_Mean": z_score.mean(),

    "Coefficient_of_Variation": cv,

    "IQR_Outlier_Count": outlier_count
}

# ----------------------------------------------------------
# Convert to DataFrame
# ----------------------------------------------------------

data_table = pd.DataFrame(summary_tables)

display(data_table.round(3))
````
